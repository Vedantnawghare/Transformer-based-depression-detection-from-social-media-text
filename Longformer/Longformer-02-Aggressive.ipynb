```python
# ============================================================
# CELL 3 — LONGFORMER + AGGRESSIVE
# EXACT D1 ANCHOR DETECTION
# ============================================================

import os
import re
import json
import glob

from collections import defaultdict

print("=" * 60)
print("LONGFORMER + AGGRESSIVE — EXACT D1 ANCHOR DETECTION")
print("=" * 60)


# ============================================================
# D1 TIMELINE DIRECTORY
# ============================================================

DATASET_ROOT = (
    "/kaggle/input/datasets/"
    "vedantnawghare35/"
    "depression-detection-mddl-final-pbl/"
    "Dataset_MDDL"
)

D1_TIMELINE_DIR = os.path.join(
    DATASET_ROOT,
    "labeled",
    "positive",
    "data",
    "timeline"
)

assert os.path.exists(
    D1_TIMELINE_DIR
)


# ============================================================
# FIND D1 TIMELINE FILES
# ============================================================

d1_timeline_files = glob.glob(
    os.path.join(
        D1_TIMELINE_DIR,
        "*.json"
    )
)

print()
print(
    "D1 timeline files found:",
    len(d1_timeline_files)
)

assert len(
    d1_timeline_files
) == 2626


# ============================================================
# EXACT ANCHOR PATTERNS
# ============================================================

REFINED_ANCHOR_PATTERNS = [

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+have\s+(?:major\s+)?depression\b",

    r"\bi\s+suffer\s+from\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|am|'m)\s+diagnosed\b.{0,80}\bdepress",

    (
        r"\b(?:my|a)\s+(?:doctor|psychiatrist)"
        r"\s+(?:has\s+)?diagnosed\s+me\b"
        r".{0,80}\bdepress"
    )
]


compiled_anchor_patterns = [
    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )
    for pattern in REFINED_ANCHOR_PATTERNS
]


# ============================================================
# EXCLUSION PATTERNS
# ============================================================

EXCLUSION_PATTERNS = [

    r"\bmy\s+friend\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+sister\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+brother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+mother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+father\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+answer\s+to\s*:",

    r"\banswer\s+to\s*:",

    r"\bquestion\s*:",

    r"\bq\s*:"
]


compiled_exclusion_patterns = [
    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )
    for pattern in EXCLUSION_PATTERNS
]


# ============================================================
# EXACT ANCHOR DETECTOR
# ============================================================

def is_refined_anchor_tweet(text):

    if text is None:
        return False

    text = str(text).strip()

    if not text:
        return False

    for pattern in compiled_exclusion_patterns:

        if pattern.search(text):
            return False

    for pattern in compiled_anchor_patterns:

        if pattern.search(text):
            return True

    return False


# ============================================================
# POSITIVE USERS BY SPLIT
# ============================================================

train_positive_users = set(
    train_df.loc[
        train_df["label"] == 1,
        "username"
    ].astype(str)
)

val_positive_users = set(
    val_df.loc[
        val_df["label"] == 1,
        "username"
    ].astype(str)
)

test_positive_users = set(
    test_df.loc[
        test_df["label"] == 1,
        "username"
    ].astype(str)
)

all_positive_users = (
    train_positive_users
    |
    val_positive_users
    |
    test_positive_users
)


print()
print("=" * 60)
print("POSITIVE USERS BY SPLIT")
print("=" * 60)

print(
    "TRAIN:",
    len(train_positive_users)
)

print(
    "VALIDATION:",
    len(val_positive_users)
)

print(
    "TEST:",
    len(test_positive_users)
)

assert len(
    train_positive_users
) == 2013

assert len(
    val_positive_users
) == 252

assert len(
    test_positive_users
) == 251

assert len(
    all_positive_users
) == 2516


# ============================================================
# INDEX TIMELINE FILES
# ============================================================

timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:

    filename = os.path.basename(
        timeline_file
    )

    username = filename.split(
        "created_at"
    )[0]

    timeline_files_by_user[
        username
    ].append(
        timeline_file
    )


assert len(
    timeline_files_by_user
) == 2516


# ============================================================
# SCAN ALL POSITIVE D1 TIMELINES
# ============================================================

anchor_records = []

users_processed = set()

tweets_scanned = 0


for username in sorted(
    all_positive_users
):

    if username not in timeline_files_by_user:

        raise KeyError(
            f"Missing D1 timeline for user: {username}"
        )

    users_processed.add(
        username
    )

    for timeline_file in (
        timeline_files_by_user[username]
    ):

        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:

                line = line.strip()

                if not line:
                    continue

                record = json.loads(
                    line
                )

                tweets_scanned += 1

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get(
                            "id",
                            ""
                        )
                    )
                )

                tweet_text = str(
                    record.get(
                        "text",
                        ""
                    )
                )

                if (
                    tweet_id
                    and
                    is_refined_anchor_tweet(
                        tweet_text
                    )
                ):

                    anchor_records.append(
                        {
                            "username":
                                username,

                            "tweet_id":
                                tweet_id,

                            "created_at":
                                record.get(
                                    "created_at"
                                ),

                            "text":
                                tweet_text
                        }
                    )


# ============================================================
# UNIQUE ANCHOR PAIRS
# ============================================================

anchor_pairs = {
    (
        record["username"],
        record["tweet_id"]
    )
    for record in anchor_records
}


# ============================================================
# SPLIT-WISE UNIQUE ANCHORS
# ============================================================

train_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in train_positive_users
}

val_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in val_positive_users
}

test_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in test_positive_users
}


# ============================================================
# SUMMARY
# ============================================================

print()
print("=" * 60)
print("EXACT ANCHOR SUMMARY")
print("=" * 60)

print(
    "Positive users processed:",
    len(users_processed)
)

print(
    "Tweets scanned:",
    tweets_scanned
)

print(
    "Raw anchor records:",
    len(anchor_records)
)

print(
    "Unique anchor pairs:",
    len(anchor_pairs)
)

print()
print(
    "TRAIN unique anchors:",
    len(train_anchor_pairs)
)

print(
    "VAL unique anchors:",
    len(val_anchor_pairs)
)

print(
    "TEST unique anchors:",
    len(test_anchor_pairs)
)


# ============================================================
# EXACT CONTROL ASSERTIONS
# ============================================================

assert len(
    users_processed
) == 2516

assert len(
    anchor_records
) == 1920

assert len(
    anchor_pairs
) == 1847

assert len(
    train_anchor_pairs
) == 1476

assert len(
    val_anchor_pairs
) == 190

assert len(
    test_anchor_pairs
) == 181


# ============================================================
# SAVE
# ============================================================

longformer_aggressive_anchor_df = pd.DataFrame(
    anchor_records
)

longformer_aggressive_anchor_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "exact_anchor_records.csv"
    ),
    index=False
)


pd.DataFrame(
    list(anchor_pairs),
    columns=[
        "username",
        "tweet_id"
    ]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "unique_anchor_pairs.csv"
    ),
    index=False
)


# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 60)
print("CELL 3 COMPLETE")
print("=" * 60)

print(
    "✓ Exact anchor detector reproduced"
)

print(
    "✓ 1,847 unique anchors confirmed"
)

print(
    "✓ 1,920 anchor records confirmed"
)

print(
    "✓ TRAIN = 1,476 unique anchors"
)

print(
    "✓ VAL = 190 unique anchors"
)

print(
    "✓ TEST = 181 unique anchors"
)

print(
    "✓ No D2 users included"
)

print(
    "✓ Anchor records saved"
)

print(
    "✓ No timelines modified yet"
)
```

```python
# ============================================================
# CELL 5 — LONGFORMER + AGGRESSIVE
# TEXT CLEANING
# ============================================================

import re
import html

print("=" * 60)
print("LONGFORMER + AGGRESSIVE — TEXT CLEANING")
print("=" * 60)


# ============================================================
# AGGRESSIVE CLEANING FUNCTION
# ============================================================

def aggressive_clean(text):

    text = str(text)

    # HTML entities / markup
    text = html.unescape(text)

    # URLs
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )

    # User mentions
    text = re.sub(
        r"@\w+",
        " ",
        text
    )

    # Hashtag symbols
    text = re.sub(
        r"#",
        " ",
        text
    )

    # HTML entities / markup
    text = re.sub(
        r"&\w+;|<[^>]+>",
        " ",
        text
    )

    # Punctuation / emojis / emoticons / special symbols
    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    # Underscores
    text = re.sub(
        r"_+",
        " ",
        text
    )

    # Whitespace normalization
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    # Lowercase
    return text.strip().lower()


# ============================================================
# APPLY AGGRESSIVE CLEANING
# ============================================================

train_longformer_aggressive_df[
    "model_text"
] = (
    train_longformer_aggressive_df[
        "anchor_removed_text"
    ]
    .fillna("")
    .map(
        aggressive_clean
    )
)


val_longformer_aggressive_df[
    "model_text"
] = (
    val_longformer_aggressive_df[
        "anchor_removed_text"
    ]
    .fillna("")
    .map(
        aggressive_clean
    )
)


test_longformer_aggressive_df[
    "model_text"
] = (
    test_longformer_aggressive_df[
        "anchor_removed_text"
    ]
    .fillna("")
    .map(
        aggressive_clean
    )
)


# ============================================================
# CLEANING SUMMARY
# ============================================================

print()
print("Cleaning applied:")

print("✓ URLs removed")

print("✓ Mentions removed")

print("✓ Hashtag symbols removed")

print("✓ HTML / markup removed")

print("✓ Punctuation removed")

print("✓ Emojis / emoticons removed")

print("✓ Special symbols removed")

print("✓ Whitespace normalized")

print("✓ Text lowercased")


# ============================================================
# MODEL TEXT INTEGRITY
# ============================================================

print()
print("=" * 60)
print("MODEL TEXT INTEGRITY")
print("=" * 60)


for name, df in [

    (
        "TRAIN",
        train_longformer_aggressive_df
    ),

    (
        "VALIDATION",
        val_longformer_aggressive_df
    ),

    (
        "TEST",
        test_longformer_aggressive_df
    )

]:

    empty_count = int(
        df[
            "model_text"
        ]
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"{name}: "
        f"empty model texts = "
        f"{empty_count}"
    )


# ============================================================
# USER + LABEL PRESERVATION
# ============================================================

for (
    original,
    result,
    name
) in [

    (
        train_df,
        train_longformer_aggressive_df,
        "TRAIN"
    ),

    (
        val_df,
        val_longformer_aggressive_df,
        "VALIDATION"
    ),

    (
        test_df,
        test_longformer_aggressive_df,
        "TEST"
    )

]:

    assert (
        original["user_id"].tolist()
        ==
        result["user_id"].tolist()
    )

    assert (
        original["username"].tolist()
        ==
        result["username"].tolist()
    )

    assert (
        original["label"].tolist()
        ==
        result["label"].tolist()
    )

    print(
        f"✓ {name}: "
        "users and labels unchanged"
    )


# ============================================================
# SAVE
# ============================================================

train_longformer_aggressive_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "train_cleaned.pkl"
    )
)

val_longformer_aggressive_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "val_cleaned.pkl"
    )
)

test_longformer_aggressive_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "test_cleaned.pkl"
    )
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 5 COMPLETE")
print("=" * 60)

print(
    "✓ Aggressive cleaning applied"
)

print(
    "✓ Anchor-removed source used"
)

print(
    "✓ Model text created"
)

print(
    "✓ Users preserved"
)

print(
    "✓ Labels preserved"
)

print(
    "✓ Cleaned data saved"
)
```

```python
# ============================================================
# CELL 6 — LONGFORMER + AGGRESSIVE
# TOKEN LENGTH ANALYSIS
# ============================================================

from transformers import AutoTokenizer
import numpy as np
import pandas as pd

print("=" * 60)
print("LONGFORMER + AGGRESSIVE — TOKEN LENGTH ANALYSIS")
print("=" * 60)


# ============================================================
# TOKENIZER
# ============================================================

print()
print(
    "Tokenizer:",
    MODEL_NAME
)

print(
    "Experimental context length:",
    MAX_LENGTH
)

print(
    "Longformer architectural capacity:",
    MODEL_MAX_LENGTH
)


# ============================================================
# HUGGING FACE TOKEN
# ============================================================

from kaggle_secrets import UserSecretsClient

user_secrets = UserSecretsClient()

hf_token = user_secrets.get_secret(
    "HF_TOKEN"
)

print()
print("✓ HF_TOKEN retrieved")


# ============================================================
# LOAD TOKENIZER
# ============================================================

longformer_aggressive_tokenizer = (
    AutoTokenizer.from_pretrained(
        MODEL_NAME,
        token=hf_token
    )
)

print()
print(
    "Tokenizer max length:",
    longformer_aggressive_tokenizer.model_max_length
)

print(
    "✓ Longformer tokenizer loaded"
)


# ============================================================
# TOKEN LENGTH FUNCTION
# ============================================================

def get_token_lengths(
    texts,
    split_name
):

    print()
    print(
        f"Processing {split_name}..."
    )

    lengths = []

    for i, text in enumerate(texts):

        encoded = (
            longformer_aggressive_tokenizer(
                str(text),
                add_special_tokens=True,
                truncation=False,
                padding=False
            )
        )

        lengths.append(
            len(
                encoded["input_ids"]
            )
        )

        if (
            (i + 1) % 1000 == 0
            or
            (i + 1) == len(texts)
        ):

            print(
                f"{split_name}: "
                f"{i + 1}/{len(texts)}"
            )

    return np.array(
        lengths,
        dtype=np.int64
    )


# ============================================================
# TRAIN TOKEN LENGTHS
# ============================================================

train_token_lengths = (
    get_token_lengths(
        train_longformer_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "TRAIN"
    )
)


# ============================================================
# VALIDATION TOKEN LENGTHS
# ============================================================

val_token_lengths = (
    get_token_lengths(
        val_longformer_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "VALIDATION"
    )
)


# ============================================================
# TEST TOKEN LENGTHS
# ============================================================

test_token_lengths = (
    get_token_lengths(
        test_longformer_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "TEST"
    )
)


# ============================================================
# TOKEN STATISTICS
# ============================================================

def show_stats(
    lengths,
    split_name
):

    print()
    print("=" * 60)
    print(
        f"{split_name} TOKEN STATISTICS"
    )
    print("=" * 60)

    print(
        "Minimum:",
        int(lengths.min())
    )

    print(
        "Maximum:",
        int(lengths.max())
    )

    print(
        "Mean:",
        round(
            float(lengths.mean()),
            2
        )
    )

    print(
        "Median:",
        round(
            float(np.median(lengths)),
            2
        )
    )

    print(
        "25th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    25
                )
            ),
            2
        )
    )

    print(
        "75th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    75
                )
            ),
            2
        )
    )

    print(
        "90th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    90
                )
            ),
            2
        )
    )

    print(
        "95th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    95
                )
            ),
            2
        )
    )

    print(
        "99th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    99
                )
            ),
            2
        )
    )


    # ========================================================
    # 512-TOKEN COVERAGE
    # ========================================================

    within_512 = int(
        np.sum(
            lengths <= 512
        )
    )

    above_512 = int(
        np.sum(
            lengths > 512
        )
    )

    print()
    print(
        "512-TOKEN COVERAGE"
    )

    print(
        "Users <= 512:",
        within_512
    )

    print(
        "Users > 512:",
        above_512
    )

    print(
        "Percentage <= 512:",
        round(
            100 * within_512 / len(lengths),
            2
        ),
        "%"
    )

    print(
        "Percentage > 512:",
        round(
            100 * above_512 / len(lengths),
            2
        ),
        "%"
    )


    # ========================================================
    # 4096-TOKEN REFERENCE
    # ========================================================

    within_4096 = int(
        np.sum(
            lengths <= 4096
        )
    )

    above_4096 = int(
        np.sum(
            lengths > 4096
        )
    )

    print()
    print(
        "4096-TOKEN REFERENCE"
    )

    print(
        "Users <= 4096:",
        within_4096
    )

    print(
        "Users > 4096:",
        above_4096
    )

    print(
        "Percentage <= 4096:",
        round(
            100 * within_4096 / len(lengths),
            2
        ),
        "%"
    )

    print(
        "Percentage > 4096:",
        round(
            100 * above_4096 / len(lengths),
            2
        ),
        "%"
    )


# ============================================================
# SHOW STATISTICS
# ============================================================

show_stats(
    train_token_lengths,
    "TRAIN"
)

show_stats(
    val_token_lengths,
    "VALIDATION"
)

show_stats(
    test_token_lengths,
    "TEST"
)


# ============================================================
# ESTIMATED CONTEXT CHUNKS
# ============================================================

print()
print("=" * 60)
print("ESTIMATED CONTEXT CHUNKS")
print("=" * 60)


for context_length in [
    512,
    4096
]:

    print()
    print(
        "Context length:",
        context_length
    )


    train_chunks = np.ceil(
        train_token_lengths
        /
        context_length
    )

    val_chunks = np.ceil(
        val_token_lengths
        /
        context_length
    )

    test_chunks = np.ceil(
        test_token_lengths
        /
        context_length
    )


    print(
        "TRAIN mean chunks:",
        round(
            float(
                train_chunks.mean()
            ),
            2
        )
    )

    print(
        "TRAIN maximum chunks:",
        int(
            train_chunks.max()
        )
    )

    print(
        "VALIDATION mean chunks:",
        round(
            float(
                val_chunks.mean()
            ),
            2
        )
    )

    print(
        "VALIDATION maximum chunks:",
        int(
            val_chunks.max()
        )
    )

    print(
        "TEST mean chunks:",
        round(
            float(
                test_chunks.mean()
            ),
            2
        )
    )

    print(
        "TEST maximum chunks:",
        int(
            test_chunks.max()
        )
    )


# ============================================================
# TOKEN-LENGTH SUMMARY
# ============================================================

token_summary = pd.DataFrame([

    {
        "Split": "Train",
        "Users":
            len(train_token_lengths),
        "Mean":
            train_token_lengths.mean(),
        "Median":
            np.median(
                train_token_lengths
            ),
        "<=512 (%)":
            100 * np.mean(
                train_token_lengths <= 512
            ),
        "<=4096 (%)":
            100 * np.mean(
                train_token_lengths <= 4096
            )
    },

    {
        "Split": "Validation",
        "Users":
            len(val_token_lengths),
        "Mean":
            val_token_lengths.mean(),
        "Median":
            np.median(
                val_token_lengths
            ),
        "<=512 (%)":
            100 * np.mean(
                val_token_lengths <= 512
            ),
        "<=4096 (%)":
            100 * np.mean(
                val_token_lengths <= 4096
            )
    },

    {
        "Split": "Test",
        "Users":
            len(test_token_lengths),
        "Mean":
            test_token_lengths.mean(),
        "Median":
            np.median(
                test_token_lengths
            ),
        "<=512 (%)":
            100 * np.mean(
                test_token_lengths <= 512
            ),
        "<=4096 (%)":
            100 * np.mean(
                test_token_lengths <= 4096
            )
    }
])


print()
print("=" * 60)
print("TOKEN-LENGTH SUMMARY")
print("=" * 60)

print(
    token_summary.to_string(
        index=False,
        float_format=lambda x:
            f"{x:.2f}"
    )
)


# ============================================================
# SAVE TOKEN LENGTHS
# ============================================================

np.save(
    os.path.join(
        RESULT_DIR,
        "train_token_lengths.npy"
    ),
    train_token_lengths
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_token_lengths.npy"
    ),
    val_token_lengths
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_token_lengths.npy"
    ),
    test_token_lengths
)


token_summary.to_csv(
    os.path.join(
        RESULT_DIR,
        "token_length_summary.csv"
    ),
    index=False
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 6 COMPLETE")
print("=" * 60)

print(
    "✓ Authenticated Longformer tokenizer loaded"
)

print(
    "✓ Aggressive cleaned text analyzed"
)

print(
    "✓ No truncation used during analysis"
)

print(
    "✓ 512-token coverage measured"
)

print(
    "✓ 4096-token reference measured"
)

print(
    "✓ Context chunk estimates calculated"
)

print(
    "✓ Token-length files saved"
)
```

```python
# ============================================================
# CELL 8 — LONGFORMER + AGGRESSIVE
# TOKENIZATION
# ============================================================

from transformers import AutoTokenizer

print("=" * 60)
print("LONGFORMER + AGGRESSIVE — TOKENIZATION")
print("=" * 60)


# ============================================================
# LOAD TOKENIZER
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    token=hf_token
)


print()
print(
    "Model:",
    MODEL_NAME
)

print(
    "Maximum sequence length:",
    MAX_LENGTH
)

print(
    "Longformer model capacity:",
    MODEL_MAX_LENGTH
)

print(
    "Truncation: ENABLED"
)

print(
    "Padding: MAX_LENGTH"
)

print(
    "Text source: Aggressive cleaned model_text"
)

print()
print(
    "✓ Longformer tokenizer available"
)


# ============================================================
# TOKENIZATION FUNCTION
# ============================================================

def tokenize_texts(
    texts,
    name
):

    print()
    print(
        f"Tokenizing {name}..."
    )

    enc = tokenizer(
        texts,
        max_length=MAX_LENGTH,
        truncation=True,
        padding="max_length",
        return_attention_mask=True
    )


    input_ids = np.asarray(
        enc["input_ids"],
        dtype=np.int64
    )

    attention_masks = np.asarray(
        enc["attention_mask"],
        dtype=np.int64
    )


    print(
        f"{name} input IDs:",
        input_ids.shape
    )

    print(
        f"{name} attention masks:",
        attention_masks.shape
    )


    return (
        input_ids,
        attention_masks
    )


# ============================================================
# TRAIN
# ============================================================

train_input_ids, train_attention_masks = (
    tokenize_texts(
        train_longformer_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "TRAIN"
    )
)


# ============================================================
# VALIDATION
# ============================================================

val_input_ids, val_attention_masks = (
    tokenize_texts(
        val_longformer_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "VALIDATION"
    )
)


# ============================================================
# TEST
# ============================================================

test_input_ids, test_attention_masks = (
    tokenize_texts(
        test_longformer_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "TEST"
    )
)


# ============================================================
# TOKENIZATION VERIFICATION
# ============================================================

print()
print("=" * 60)
print("TOKENIZATION VERIFICATION")
print("=" * 60)

print(
    "Train:",
    train_input_ids.shape
)

print(
    "Validation:",
    val_input_ids.shape
)

print(
    "Test:",
    test_input_ids.shape
)


assert train_input_ids.shape == (
    6135,
    512
)

assert val_input_ids.shape == (
    767,
    512
)

assert test_input_ids.shape == (
    767,
    512
)


assert train_attention_masks.shape == (
    6135,
    512
)

assert val_attention_masks.shape == (
    767,
    512
)

assert test_attention_masks.shape == (
    767,
    512
)


print(
    "✓ All token shapes correct"
)


# ============================================================
# ATTENTION MASK CHECK
# ============================================================

print()
print("=" * 60)
print("ATTENTION MASK CHECK")
print("=" * 60)


for (
    name,
    masks
) in [

    (
        "TRAIN",
        train_attention_masks
    ),

    (
        "VALIDATION",
        val_attention_masks
    ),

    (
        "TEST",
        test_attention_masks
    )

]:

    active = masks.sum(
        axis=1
    )


    print()
    print(name)

    print(
        "Minimum active tokens:",
        int(active.min())
    )

    print(
        "Maximum active tokens:",
        int(active.max())
    )

    print(
        "Mean active tokens:",
        round(
            float(active.mean()),
            2
        )
    )


    assert np.all(
        active >= 1
    )


print()
print(
    "✓ Labels aligned"
)


# ============================================================
# SAVE
# ============================================================

np.save(
    os.path.join(
        RESULT_DIR,
        "train_input_ids.npy"
    ),
    train_input_ids
)

np.save(
    os.path.join(
        RESULT_DIR,
        "train_attention_masks.npy"
    ),
    train_attention_masks
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_input_ids.npy"
    ),
    val_input_ids
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_attention_masks.npy"
    ),
    val_attention_masks
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_input_ids.npy"
    ),
    test_input_ids
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_attention_masks.npy"
    ),
    test_attention_masks
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 8 COMPLETE")
print("=" * 60)

print(
    "✓ Aggressive cleaned text tokenized"
)

print(
    "✓ Longformer maximum experimental length = 512"
)

print(
    "✓ Truncation enabled"
)

print(
    "✓ Padding = MAX_LENGTH"
)

print(
    "✓ Attention masks generated"
)

print(
    "✓ Labels aligned"
)

print(
    "✓ Tokenized arrays saved"
)
```

```python
# ============================================================
# CELL 11 — LONGFORMER + AGGRESSIVE
# BUILDING FROZEN LONGFORMER + METADATA MODEL
# ============================================================

from transformers import AutoModel
import torch.nn as nn

print("=" * 60)
print("LONGFORMER + AGGRESSIVE — BUILDING FROZEN LONGFORMER + METADATA MODEL")
print("=" * 60)

longformer_encoder = AutoModel.from_pretrained(
    MODEL_NAME,
    token=hf_token
).to(device)

for p in longformer_encoder.parameters():
    p.requires_grad = False


class LongformerMetadataClassifier(nn.Module):

    def __init__(
        self,
        encoder,
        metadata_dim=53
    ):
        super().__init__()

        self.encoder = encoder
        self.text_hidden_size = 768

        self.attention_projection = nn.Linear(
            768,
            256
        )

        self.attention_score = nn.Linear(
            256,
            1
        )

        self.metadata_branch = nn.Sequential(
            nn.Linear(
                metadata_dim,
                128
            ),
            nn.ReLU(),
            nn.Dropout(
                0.3
            ),
            nn.Linear(
                128,
                64
            ),
            nn.ReLU()
        )

        self.metadata_beta = nn.Parameter(
            torch.tensor(
                0.0
            )
        )

        self.classifier = nn.Sequential(
            nn.Linear(
                832,
                512
            ),
            nn.ReLU(),
            nn.Dropout(
                0.3
            ),
            nn.Linear(
                512,
                256
            ),
            nn.ReLU(),
            nn.Dropout(
                0.3
            ),
            nn.Linear(
                256,
                2
            )
        )

    def forward(
        self,
        input_ids,
        attention_mask,
        metadata
    ):

        with torch.no_grad():

            outputs = self.encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

        hidden_states = outputs.last_hidden_state

        projected = torch.tanh(
            self.attention_projection(
                hidden_states
            )
        )

        scores = self.attention_score(
            projected
        ).squeeze(-1)

        scores = scores.masked_fill(
            attention_mask == 0,
            -1e9
        )

        attention_weights = torch.softmax(
            scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        metadata_representation = (
            self.metadata_branch(
                metadata
            )
        )

        metadata_gate = torch.sigmoid(
            self.metadata_beta
        )

        gated_metadata = (
            metadata_gate *
            metadata_representation
        )

        fused = torch.cat(
            [
                text_representation,
                gated_metadata
            ],
            dim=1
        )

        logits = self.classifier(
            fused
        )

        return {
            "logits": logits,
            "text_representation":
                text_representation,
            "metadata_representation":
                metadata_representation,
            "attention_weights":
                attention_weights,
            "metadata_gate":
                metadata_gate
        }


model_longformer_aggressive = (
    LongformerMetadataClassifier(
        longformer_encoder,
        METADATA_DIM
    ).to(device)
)


# ============================================================
# PARAMETER COUNT
# ============================================================

total_params = sum(
    p.numel()
    for p in model_longformer_aggressive.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model_longformer_aggressive.parameters()
    if p.requires_grad
)

frozen_params = sum(
    p.numel()
    for p in model_longformer_aggressive.parameters()
    if not p.requires_grad
)


print()
print("=" * 60)
print("PARAMETER COUNT")
print("=" * 60)

print(
    "Total parameters:",
    total_params
)

print(
    "Trainable parameters:",
    trainable_params
)

print(
    "Frozen parameters:",
    frozen_params
)

print(
    "Total parameters (M):",
    round(
        total_params / 1e6,
        3
    )
)

print(
    "Trainable parameters (M):",
    round(
        trainable_params / 1e6,
        3
    )
)

print(
    "Frozen parameters (M):",
    round(
        frozen_params / 1e6,
        3
    )
)


encoder_trainable = sum(
    p.numel()
    for p in model_longformer_aggressive.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print(
    "✓ All Longformer backbone parameters frozen"
)


# ============================================================
# TRAINABLE COMPONENTS
# ============================================================

print()
print("=" * 60)
print("TRAINABLE COMPONENTS")
print("=" * 60)

for name, p in (
    model_longformer_aggressive.named_parameters()
):

    if p.requires_grad:
        print(
            "✓",
            name
        )


# ============================================================
# ARCHITECTURE
# ============================================================

print()
print("=" * 60)
print("ARCHITECTURE CONFIGURATION")
print("=" * 60)

print(
    "Transformer:",
    MODEL_NAME
)

print(
    "Base architecture:",
    "Longformer-base-4096"
)

print(
    "Experimental context length:",
    MAX_LENGTH
)

print(
    "Text hidden size:",
    768
)

print(
    "Attention projection:",
    256
)

print(
    "Metadata:",
    "53 -> 128 -> 64"
)

print(
    "Gate:",
    "g = sigmoid(beta)"
)

print(
    "Initial beta:",
    0.0
)

print(
    "Initial gate:",
    0.5
)

print(
    "Fusion:",
    "768 + 64 = 832"
)

print(
    "Classifier:",
    "832 -> 512 -> 256 -> 2"
)

print(
    "Dropout:",
    0.3
)

print(
    "Transformer frozen:",
    True
)


# ============================================================
# FORWARD PASS SANITY CHECK
# ============================================================

print()
print("=" * 60)
print("FORWARD PASS SANITY CHECK")
print("=" * 60)

model_longformer_aggressive.eval()

batch = next(
    iter(train_loader)
)

input_ids = batch[
    "input_ids"
].to(device)

attention_mask = batch[
    "attention_mask"
].to(device)

metadata = batch[
    "metadata"
].to(device)


with torch.no_grad():

    output = (
        model_longformer_aggressive(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )
    )


print(
    "Input IDs:",
    tuple(
        input_ids.shape
    )
)

print(
    "Attention mask:",
    tuple(
        attention_mask.shape
    )
)

print(
    "Metadata:",
    tuple(
        metadata.shape
    )
)

print(
    "Logits:",
    tuple(
        output[
            "logits"
        ].shape
    )
)

print(
    "Text representation:",
    tuple(
        output[
            "text_representation"
        ].shape
    )
)

print(
    "Metadata representation:",
    tuple(
        output[
            "metadata_representation"
        ].shape
    )
)

print(
    "Attention weights:",
    tuple(
        output[
            "attention_weights"
        ].shape
    )
)


gate = float(
    output[
        "metadata_gate"
    ].item()
)

print(
    "Metadata gate:",
    gate
)


padding_attention = output[
    "attention_weights"
][
    attention_mask == 0
]


max_padding_attention = (
    float(
        padding_attention.max().item()
    )
    if padding_attention.numel()
    else 0.0
)


print(
    "Maximum attention on padding:",
    max_padding_attention
)


# ============================================================
# EXACT ASSERTIONS
# ============================================================

assert (
    output[
        "logits"
    ].shape
    ==
    (16, 2)
)

assert (
    output[
        "text_representation"
    ].shape
    ==
    (16, 768)
)

assert (
    output[
        "metadata_representation"
    ].shape
    ==
    (16, 64)
)

assert (
    output[
        "attention_weights"
    ].shape
    ==
    (16, 512)
)

assert (
    max_padding_attention
    ==
    0.0
)

assert (
    abs(
        gate - 0.5
    )
    <
    1e-6
)


encoder_trainable = sum(
    p.numel()
    for p in model_longformer_aggressive.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0


print()
print(
    "✓ Transformer freeze verified"
)


# ============================================================
# SAVE CONFIG
# ============================================================

architecture_config = {

    "model_name":
        MODEL_NAME,

    "base_architecture":
        "Longformer-base-4096",

    "experimental_context_length":
        MAX_LENGTH,

    "text_hidden_size":
        768,

    "attention_projection":
        256,

    "metadata_dimension":
        53,

    "metadata_branch":
        "53 -> 128 -> 64",

    "initial_beta":
        0.0,

    "initial_gate":
        0.5,

    "fusion":
        "768 + 64 = 832",

    "classifier":
        "832 -> 512 -> 256 -> 2",

    "dropout":
        0.3,

    "transformer_frozen":
        True
}


with open(
    os.path.join(
        RESULT_DIR,
        "architecture_config.json"
    ),
    "w"
) as f:

    json.dump(
        architecture_config,
        f,
        indent=2
    )


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 11 COMPLETE")
print("=" * 60)

print(
    "✓ Longformer-base-4096 loaded"
)

print(
    "✓ Transformer backbone frozen"
)

print(
    "✓ Attention pooling active"
)

print(
    "✓ Metadata branch = 53 -> 128 -> 64"
)

print(
    "✓ Single metadata gate active"
)

print(
    "✓ Text representation = 768"
)

print(
    "✓ Fusion = 768 + 64 = 832"
)

print(
    "✓ Classifier = 832 -> 512 -> 256 -> 2"
)

print(
    "✓ Padding attention masked"
)

print(
    "✓ Forward pass successful"
)

print(
    "✓ Frozen Longformer + Aggressive model ready for training"
)
```

```python
# ============================================================
# CELL 12 — LONGFORMER + AGGRESSIVE
# MODEL TRAINING
# ============================================================

import os
import time
import copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

MAX_EPOCHS = 10
EARLY_STOPPING_PATIENCE = 2
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1

print("=" * 60)
print("STARTING LONGFORMER + AGGRESSIVE TRAINING")
print("=" * 60)

print()
print("Training users:", len(train_dataset))
print("Validation users:", len(val_dataset))
print("Test users:", len(test_dataset))
print("Maximum epochs:", MAX_EPOCHS)
print("Early stopping patience:", EARLY_STOPPING_PATIENCE)
print("Checkpoint metric: Validation F1")
print("Learning rate:", LEARNING_RATE)
print("Weight decay:", WEIGHT_DECAY)
print("Batch size:", BATCH_SIZE)
print("Transformer: FROZEN")

encoder_trainable = sum(
    p.numel()
    for p in model_longformer_aggressive.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0
print()
print("✓ LONGFORMER backbone confirmed frozen")

trainable_params = [
    p for p in model_longformer_aggressive.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_params,
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

total_steps = len(train_loader) * MAX_EPOCHS
warmup_steps = int(total_steps * WARMUP_RATIO)

def lr_lambda(step):
    if step < warmup_steps:
        return (step + 1) / max(1, warmup_steps)

    return max(
        0.0,
        (total_steps - step) /
        max(1, total_steps - warmup_steps)
    )

scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda
)

criterion = nn.CrossEntropyLoss()


def evaluate_model(model, loader):
    model.eval()

    total_loss = 0.0
    labels_all = []
    preds_all = []
    probs_all = []

    with torch.no_grad():
        for batch in loader:
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            metadata = batch["metadata"].to(device)
            labels = batch["labels"].to(device)

            output = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                metadata=metadata
            )

            logits = output["logits"]

            loss = criterion(
                logits,
                labels
            )

            total_loss += (
                loss.item() * labels.size(0)
            )

            probs = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            preds = torch.argmax(
                logits,
                dim=1
            )

            labels_all.extend(
                labels.cpu().numpy()
            )

            preds_all.extend(
                preds.cpu().numpy()
            )

            probs_all.extend(
                probs.cpu().numpy()
            )

    metrics = {
        "loss":
            total_loss / len(loader.dataset),

        "accuracy":
            accuracy_score(
                labels_all,
                preds_all
            ),

        "precision":
            precision_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "recall":
            recall_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "f1":
            f1_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "roc_auc":
            roc_auc_score(
                labels_all,
                probs_all
            )
    }

    return metrics


history = []
best_val_f1 = -np.inf
best_epoch = None
best_state_dict = None
epochs_without_improvement = 0

training_start = time.time()

for epoch in range(1, MAX_EPOCHS + 1):

    epoch_start = time.time()

    print()
    print("=" * 60)
    print(f"EPOCH {epoch}/{MAX_EPOCHS}")
    print("=" * 60)

    model_longformer_aggressive.train()

    running_loss = 0.0
    train_labels = []
    train_preds = []

    for batch_idx, batch in enumerate(
        train_loader,
        start=1
    ):

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        metadata = batch["metadata"].to(device)
        labels = batch["labels"].to(device)

        optimizer.zero_grad()

        output = model_longformer_aggressive(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = output["logits"]

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            trainable_params,
            max_norm=1.0
        )

        optimizer.step()
        scheduler.step()

        running_loss += (
            loss.item() * labels.size(0)
        )

        preds = torch.argmax(
            logits.detach(),
            dim=1
        )

        train_labels.extend(
            labels.detach().cpu().numpy()
        )

        train_preds.extend(
            preds.cpu().numpy()
        )

        if (
            batch_idx == 1
            or batch_idx % 50 == 0
            or batch_idx == len(train_loader)
        ):

            batch_acc = accuracy_score(
                labels.detach().cpu().numpy(),
                preds.cpu().numpy()
            )

            elapsed = time.time() - epoch_start

            print(
                f"Epoch {epoch} | "
                f"Batch {batch_idx}/{len(train_loader)} | "
                f"Loss: {loss.item():.4f} | "
                f"Accuracy: {batch_acc:.4f} | "
                f"LR: {optimizer.param_groups[0]['lr']:.2e} | "
                f"Time: {elapsed:.1f}s"
            )

    train_loss = (
        running_loss / len(train_loader.dataset)
    )

    train_accuracy = accuracy_score(
        train_labels,
        train_preds
    )

    train_precision = precision_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_recall = recall_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_f1 = f1_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    print()
    print(f"Epoch {epoch} — Validation")

    val_metrics = evaluate_model(
        model_longformer_aggressive,
        val_loader
    )

    with torch.no_grad():
        metadata_gate = float(
            torch.sigmoid(
                model_longformer_aggressive.metadata_beta
            ).item()
        )

    epoch_time = time.time() - epoch_start

    print()
    print("=" * 60)
    print(f"EPOCH {epoch} SUMMARY")
    print("=" * 60)

    print(f"Train Loss: {train_loss:.4f}")
    print(f"Train Accuracy: {train_accuracy:.4f}")
    print(f"Train Precision: {train_precision:.4f}")
    print(f"Train Recall: {train_recall:.4f}")
    print(f"Train F1: {train_f1:.4f}")

    print()
    print(f"Val Loss: {val_metrics['loss']:.4f}")
    print(f"Val Accuracy: {val_metrics['accuracy']:.4f}")
    print(f"Val Precision: {val_metrics['precision']:.4f}")
    print(f"Val Recall: {val_metrics['recall']:.4f}")
    print(f"Val F1: {val_metrics['f1']:.4f}")
    print(f"Val ROC-AUC: {val_metrics['roc_auc']:.4f}")
    print(f"Metadata gate: {metadata_gate:.6f}")
    print(f"Epoch Time: {epoch_time:.2f}s")

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "train_precision": train_precision,
        "train_recall": train_recall,
        "train_f1": train_f1,
        "val_loss": val_metrics["loss"],
        "val_accuracy": val_metrics["accuracy"],
        "val_precision": val_metrics["precision"],
        "val_recall": val_metrics["recall"],
        "val_f1": val_metrics["f1"],
        "val_roc_auc": val_metrics["roc_auc"],
        "metadata_gate": metadata_gate,
        "learning_rate":
            optimizer.param_groups[0]["lr"],
        "epoch_time": epoch_time
    })

    if val_metrics["f1"] > best_val_f1:

        best_val_f1 = val_metrics["f1"]
        best_epoch = epoch

        best_state_dict = {
            k: v.detach().cpu().clone()
            for k, v in model_longformer_aggressive.state_dict().items()
        }

        epochs_without_improvement = 0

        checkpoint_path = os.path.join(
            CHECKPOINT_DIR,
            "best_longformer_aggressive.pt"
        )

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict":
                    best_state_dict,
                "optimizer_state_dict":
                    optimizer.state_dict(),
                "scheduler_state_dict":
                    scheduler.state_dict(),
                "best_val_f1":
                    best_val_f1,
                "experiment_name":
                    EXPERIMENT_NAME,
                "model_name":
                    MODEL_NAME,
                "cleaning_method":
                    CLEANING_METHOD
            },
            checkpoint_path
        )

        print()
        print("✓ NEW BEST CHECKPOINT")
        print(
            "Best Validation F1:",
            f"{best_val_f1:.4f}"
        )
        print(
            "Saved:",
            checkpoint_path
        )

    else:
        epochs_without_improvement += 1

        print()
        print("No validation F1 improvement.")
        print(
            "Epochs without improvement:",
            f"{epochs_without_improvement} / "
            f"{EARLY_STOPPING_PATIENCE}"
        )

    if (
        epochs_without_improvement
        >= EARLY_STOPPING_PATIENCE
    ):
        print()
        print("EARLY STOPPING TRIGGERED")
        break


assert best_state_dict is not None
assert best_epoch is not None

model_longformer_aggressive.load_state_dict(best_state_dict)
model_longformer_aggressive.to(device)
model_longformer_aggressive.eval()

total_training_time = (
    time.time() - training_start
)

history_df = pd.DataFrame(history)

history_path = os.path.join(
    RESULT_DIR,
    "longformer_aggressive_training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)

with torch.no_grad():
    learned_beta = float(
        model_longformer_aggressive.metadata_beta.item()
    )

    learned_gate = float(
        torch.sigmoid(
            model_longformer_aggressive.metadata_beta
        ).item()
    )

print()
print("=" * 60)
print("LONGFORMER + AGGRESSIVE TRAINING COMPLETE")
print("=" * 60)

print("Epochs completed:", len(history))
print("Best epoch:", best_epoch)
print(
    "Best Validation F1:",
    f"{best_val_f1:.4f}"
)
print(
    "Total training time:",
    f"{total_training_time:.2f} seconds"
)
print(
    "Best checkpoint:",
    os.path.join(
        CHECKPOINT_DIR,
        "best_longformer_aggressive.pt"
    )
)
print(
    "Training history saved:",
    history_path
)

print()
print("Learned beta:", learned_beta)
print(
    "Metadata gate:",
    f"{learned_gate:.6f}"
)

print()
print("✓ LONGFORMER transformer remained frozen")
print("✓ Downstream layers trained")

print()
print("=" * 60)
print("CELL 12 COMPLETE")
print("=" * 60)
```
