```python
# ============================================================
# CELL 3 — LONGFORMER + SELECTIVE
# ============================================================

print("=" * 60)
print("LONGFORMER + SELECTIVE — EXACT D1 ANCHOR DETECTION")
print("=" * 60)

import os
import re
import json
import glob

from collections import defaultdict

DATASET_ROOT = (
    "/kaggle/input/datasets/"
    "vedantnawghare35/"
    "depression-detection-mddl-final-pbl/"
    "Dataset_MDDL"
)

D1_TIMELINE_DIR = os.path.join(
    DATASET_ROOT,
    "labeled",
    "positive",
    "data",
    "timeline"
)

assert os.path.exists(
    D1_TIMELINE_DIR
)

d1_timeline_files = glob.glob(
    os.path.join(
        D1_TIMELINE_DIR,
        "*.json"
    )
)

print()
print(
    "D1 timeline files found:",
    len(d1_timeline_files)
)

assert len(d1_timeline_files) == 2626

# ============================================================
# EXACT ANCHOR PATTERNS
# ============================================================

REFINED_ANCHOR_PATTERNS = [

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+have\s+(?:major\s+)?depression\b",

    r"\bi\s+suffer\s+from\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|am|'m)\s+diagnosed\b.{0,80}\bdepress",

    (
        r"\b(?:my|a)\s+(?:doctor|psychiatrist)"
        r"\s+(?:has\s+)?diagnosed\s+me\b"
        r".{0,80}\bdepress"
    )
]

compiled_anchor_patterns = [
    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )
    for pattern in REFINED_ANCHOR_PATTERNS
]

# ============================================================
# EXCLUSION PATTERNS
# ============================================================

EXCLUSION_PATTERNS = [

    r"\bmy\s+friend\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+sister\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+brother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+mother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+father\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+answer\s+to\s*:",

    r"\banswer\s+to\s*:",

    r"\bquestion\s*:",

    r"\bq\s*:"
]

compiled_exclusion_patterns = [
    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )
    for pattern in EXCLUSION_PATTERNS
]

# ============================================================
# EXACT ANCHOR DETECTOR
# ============================================================

def is_refined_anchor_tweet(text):

    if text is None:
        return False

    text = str(text).strip()

    if not text:
        return False

    for pattern in compiled_exclusion_patterns:

        if pattern.search(text):
            return False

    for pattern in compiled_anchor_patterns:

        if pattern.search(text):
            return True

    return False

# ============================================================
# POSITIVE USERS BY SPLIT
# ============================================================

train_positive_users = set(
    train_df.loc[
        train_df["label"] == 1,
        "username"
    ].astype(str)
)

val_positive_users = set(
    val_df.loc[
        val_df["label"] == 1,
        "username"
    ].astype(str)
)

test_positive_users = set(
    test_df.loc[
        test_df["label"] == 1,
        "username"
    ].astype(str)
)

all_positive_users = (
    train_positive_users
    |
    val_positive_users
    |
    test_positive_users
)

print()
print("=" * 60)
print("POSITIVE USERS BY SPLIT")
print("=" * 60)

print(
    "TRAIN:",
    len(train_positive_users)
)

print(
    "VALIDATION:",
    len(val_positive_users)
)

print(
    "TEST:",
    len(test_positive_users)
)

assert len(train_positive_users) == 2013
assert len(val_positive_users) == 252
assert len(test_positive_users) == 251
assert len(all_positive_users) == 2516

# ============================================================
# INDEX TIMELINE FILES
# ============================================================

timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:

    filename = os.path.basename(
        timeline_file
    )

    username = filename.split(
        "created_at"
    )[0]

    timeline_files_by_user[
        username
    ].append(
        timeline_file
    )

assert len(
    timeline_files_by_user
) == 2516

# ============================================================
# SCAN ALL POSITIVE D1 TIMELINES
# ============================================================

anchor_records = []

users_processed = set()
tweets_scanned = 0

for username in sorted(
    all_positive_users
):

    if username not in timeline_files_by_user:

        raise KeyError(
            f"Missing D1 timeline for user: {username}"
        )

    users_processed.add(
        username
    )

    for timeline_file in (
        timeline_files_by_user[username]
    ):

        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:

                line = line.strip()

                if not line:
                    continue

                record = json.loads(
                    line
                )

                tweets_scanned += 1

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get(
                            "id",
                            ""
                        )
                    )
                )

                tweet_text = str(
                    record.get(
                        "text",
                        ""
                    )
                )

                if (
                    tweet_id
                    and
                    is_refined_anchor_tweet(
                        tweet_text
                    )
                ):

                    anchor_records.append(
                        {
                            "username":
                                username,

                            "tweet_id":
                                tweet_id,

                            "created_at":
                                record.get(
                                    "created_at"
                                ),

                            "text":
                                tweet_text
                        }
                    )

# ============================================================
# UNIQUE ANCHOR PAIRS
# ============================================================

anchor_pairs = {
    (
        record["username"],
        record["tweet_id"]
    )
    for record in anchor_records
}

# ============================================================
# SPLIT-WISE UNIQUE ANCHORS
# ============================================================

train_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in train_positive_users
}

val_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in val_positive_users
}

test_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in test_positive_users
}

# ============================================================
# SUMMARY
# ============================================================

print()
print("=" * 60)
print("EXACT ANCHOR SUMMARY")
print("=" * 60)

print(
    "Positive users processed:",
    len(users_processed)
)

print(
    "Tweets scanned:",
    tweets_scanned
)

print(
    "Raw anchor records:",
    len(anchor_records)
)

print(
    "Unique anchor pairs:",
    len(anchor_pairs)
)

print()
print(
    "TRAIN unique anchors:",
    len(train_anchor_pairs)
)

print(
    "VAL unique anchors:",
    len(val_anchor_pairs)
)

print(
    "TEST unique anchors:",
    len(test_anchor_pairs)
)

# ============================================================
# EXACT CONTROL ASSERTIONS
# ============================================================

assert len(users_processed) == 2516
assert len(anchor_records) == 1920
assert len(anchor_pairs) == 1847
assert len(train_anchor_pairs) == 1476
assert len(val_anchor_pairs) == 190
assert len(test_anchor_pairs) == 181

# ============================================================
# SAVE
# ============================================================

longformer_selective_anchor_df = pd.DataFrame(
    anchor_records
)

longformer_selective_anchor_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "exact_anchor_records.csv"
    ),
    index=False
)

pd.DataFrame(
    list(anchor_pairs),
    columns=[
        "username",
        "tweet_id"
    ]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "unique_anchor_pairs.csv"
    ),
    index=False
)

print()
print("=" * 60)
print("CELL 3 COMPLETE")
print("=" * 60)

print(
    "✓ Exact anchor detector reproduced"
)

print(
    "✓ 1,847 unique anchors confirmed"
)

print(
    "✓ TRAIN = 1,476"
)

print(
    "✓ VALIDATION = 190"
)

print(
    "✓ TEST = 181"
)

print(
    "✓ Anchor records saved"
)

print(
    "✓ Unique anchor pairs saved"
)
```

```python
# ============================================================
# CELL 4 — LONGFORMER + SELECTIVE
# EXACT D1 ANCHOR REMOVAL
# ============================================================

print("=" * 60)
print("LONGFORMER + SELECTIVE — EXACT D1 ANCHOR REMOVAL")
print("=" * 60)

EXACT_ANCHOR_SET = set(anchor_pairs)

train_anchor_records = [
    r for r in anchor_records
    if r["username"] in train_positive_users
]

val_anchor_records = [
    r for r in anchor_records
    if r["username"] in val_positive_users
]

test_anchor_records = [
    r for r in anchor_records
    if r["username"] in test_positive_users
]

print()
print("=" * 60)
print("PRE-REMOVAL ANCHOR VERIFICATION")
print("=" * 60)

duplicate_anchor_occurrences = (
    len(anchor_records) - len(anchor_pairs)
)

print("Raw anchor records:", len(anchor_records))
print("Unique anchor pairs:", len(anchor_pairs))
print("Duplicate anchor occurrences:", duplicate_anchor_occurrences)

assert len(anchor_records) == 1920
assert len(anchor_pairs) == 1847
assert duplicate_anchor_occurrences == 73

print("✓ 1,847 unique anchors verified")
print("✓ 73 duplicate occurrences verified")
print("✓ 1,920 total anchor records verified")

print()
print("=" * 60)
print("SPLIT-LEVEL ANCHOR RECORD COUNTS")
print("=" * 60)

print("TRAIN:", len(train_anchor_records), "| expected: 1535")
print("VAL:", len(val_anchor_records), "| expected: 199")
print("TEST:", len(test_anchor_records), "| expected: 186")

assert len(train_anchor_records) == 1535
assert len(val_anchor_records) == 199
assert len(test_anchor_records) == 186

# Build timeline index
timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:
    filename = os.path.basename(timeline_file)
    username = filename.split("created_at")[0]
    timeline_files_by_user[username].append(timeline_file)

assert len(timeline_files_by_user) == 2516


def build_anchor_removed_text(username):
    kept_texts = []
    removed_count = 0

    for timeline_file in timeline_files_by_user[username]:
        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:
                line = line.strip()

                if not line:
                    continue

                record = json.loads(line)

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get("id", "")
                    )
                )

                if (
                    str(username),
                    tweet_id
                ) in EXACT_ANCHOR_SET:

                    removed_count += 1
                    continue

                tweet_text = str(
                    record.get("text", "")
                ).strip()

                if tweet_text:
                    kept_texts.append(tweet_text)

    return " ".join(kept_texts), removed_count


# Copy original data
train_longformer_selective_df = train_df.copy(deep=True)
val_longformer_selective_df = val_df.copy(deep=True)
test_longformer_selective_df = test_df.copy(deep=True)

positive_anchor_removed_text = {}

train_removed = 0
val_removed = 0
test_removed = 0

for username in sorted(all_positive_users):

    text, removed = build_anchor_removed_text(username)

    positive_anchor_removed_text[username] = text

    if username in train_positive_users:
        train_removed += removed
    elif username in val_positive_users:
        val_removed += removed
    elif username in test_positive_users:
        test_removed += removed

print()
print("=" * 60)
print("FINAL ANCHOR REMOVAL COUNTS")
print("=" * 60)

print("TRAIN removed:", train_removed)
print("VAL removed:", val_removed)
print("TEST removed:", test_removed)

total_removed = (
    train_removed +
    val_removed +
    test_removed
)

print("TOTAL removed:", total_removed)

assert train_removed == 1535
assert val_removed == 199
assert test_removed == 186
assert total_removed == 1920

print(
    "✓ Record-level removal exactly matches previous experiments"
)

# Assign anchor-removed text
for df in [
    train_longformer_selective_df,
    val_longformer_selective_df,
    test_longformer_selective_df
]:

    positive_mask = df["label"] == 1

    df.loc[
        positive_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            positive_mask,
            "username"
        ]
        .astype(str)
        .map(positive_anchor_removed_text)
        .fillna("")
        .values
    )

    negative_mask = df["label"] == 0

    df.loc[
        negative_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            negative_mask,
            "timeline_text"
        ]
        .fillna("")
        .astype(str)
        .values
    )

# D2 preservation
print()
print("=" * 60)
print("D2 PRESERVATION CHECK")
print("=" * 60)

for original_df, result_df, name in [
    (
        train_df,
        train_longformer_selective_df,
        "TRAIN"
    ),
    (
        val_df,
        val_longformer_selective_df,
        "VALIDATION"
    ),
    (
        test_df,
        test_longformer_selective_df,
        "TEST"
    )
]:

    original_d2 = (
        original_df.loc[
            original_df["label"] == 0,
            ["username", "timeline_text"]
        ]
        .set_index("username")
        .sort_index()
    )

    result_d2 = (
        result_df.loc[
            result_df["label"] == 0,
            ["username", "anchor_removed_text"]
        ]
        .rename(
            columns={
                "anchor_removed_text": "timeline_text"
            }
        )
        .set_index("username")
        .sort_index()
    )

    assert original_d2.equals(result_d2)
    print(f"✓ {name} D2 unchanged")

# User and label preservation
print()
print("=" * 60)
print("USER + LABEL PRESERVATION")
print("=" * 60)

for original_df, result_df, name in [
    (
        train_df,
        train_longformer_selective_df,
        "TRAIN"
    ),
    (
        val_df,
        val_longformer_selective_df,
        "VALIDATION"
    ),
    (
        test_df,
        test_longformer_selective_df,
        "TEST"
    )
]:

    assert (
        original_df["user_id"].tolist()
        ==
        result_df["user_id"].tolist()
    )

    assert (
        original_df["username"].tolist()
        ==
        result_df["username"].tolist()
    )

    assert (
        original_df["label"].tolist()
        ==
        result_df["label"].tolist()
    )

    print(
        f"✓ {name}: users and labels unchanged"
    )

# Empty text check
print()
print("=" * 60)
print("ANCHOR-REMOVED TEXT INTEGRITY")
print("=" * 60)

for name, df in [
    (
        "TRAIN",
        train_longformer_selective_df
    ),
    (
        "VALIDATION",
        val_longformer_selective_df
    ),
    (
        "TEST",
        test_longformer_selective_df
    )
]:

    empty_count = int(
        (
            df["anchor_removed_text"]
            .fillna("")
            .astype(str)
            .str.strip()
            .eq("")
        ).sum()
    )

    print(
        f"{name}: empty anchor-removed texts = {empty_count}"
    )

# Save
train_longformer_selective_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "train_anchor_removed.pkl"
    )
)

val_longformer_selective_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "val_anchor_removed.pkl"
    )
)

test_longformer_selective_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "test_anchor_removed.pkl"
    )
)

print()
print("=" * 60)
print("CELL 4 COMPLETE")
print("=" * 60)

print("✓ 1,847 unique D1 anchors removed")
print("✓ 73 duplicate occurrences handled")
print("✓ 1,920 total anchor records removed")
print("✓ TRAIN = 1,535 removed")
print("✓ VAL = 199 removed")
print("✓ TEST = 186 removed")
print("✓ D2 completely preserved")
print("✓ User IDs preserved")
print("✓ Labels preserved")
print("✓ Anchor-removed source saved")
```

```python
# ============================================================
# CELL 5 — LONGFORMER + SELECTIVE
# ============================================================

import re
import html

print("=" * 60)
print("LONGFORMER + SELECTIVE — TEXT CLEANING")
print("=" * 60)


def selective_clean(text):
    text = str(text)
    text = html.unescape(text)

    # Remove URLs and mentions
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"@\w+", " ", text)

    # Remove hashtag symbol but keep the word
    text = re.sub(r"#", "", text)

    # Remove HTML / markup
    text = re.sub(r"<[^>]+>", " ", text)

    # Keep punctuation, emojis and emoticons
    # Remove only unwanted control characters
    text = re.sub(r"[\r\n\t]+", " ", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    return text.strip().lower()


# Apply selective cleaning
train_longformer_selective_df["model_text"] = (
    train_longformer_selective_df["anchor_removed_text"]
    .fillna("")
    .map(selective_clean)
)

val_longformer_selective_df["model_text"] = (
    val_longformer_selective_df["anchor_removed_text"]
    .fillna("")
    .map(selective_clean)
)

test_longformer_selective_df["model_text"] = (
    test_longformer_selective_df["anchor_removed_text"]
    .fillna("")
    .map(selective_clean)
)

print()
print("Cleaning applied:")
print("✓ URLs removed")
print("✓ Mentions removed")
print("✓ Hashtag symbols removed")
print("✓ HTML / markup removed")
print("✓ Normal punctuation preserved")
print("✓ Emojis preserved")
print("✓ Emoticons preserved")
print("✓ Expressive symbols preserved")
print("✓ Whitespace normalized")
print("✓ Text lowercased")


# ============================================================
# TEXT INTEGRITY
# ============================================================

print()
print("=" * 60)
print("MODEL TEXT INTEGRITY")
print("=" * 60)

for name, df in [
    ("TRAIN", train_longformer_selective_df),
    ("VALIDATION", val_longformer_selective_df),
    ("TEST", test_longformer_selective_df)
]:

    empty_count = int(
        df["model_text"]
        .fillna("")
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"{name}: empty model texts = {empty_count}"
    )


# ============================================================
# USER + LABEL PRESERVATION
# ============================================================

print()
print("=" * 60)
print("USER + LABEL PRESERVATION")
print("=" * 60)

for original, result, name in [
    (train_df, train_longformer_selective_df, "TRAIN"),
    (val_df, val_longformer_selective_df, "VALIDATION"),
    (test_df, test_longformer_selective_df, "TEST")
]:

    assert (
        original["user_id"].tolist()
        ==
        result["user_id"].tolist()
    )

    assert (
        original["username"].tolist()
        ==
        result["username"].tolist()
    )

    assert (
        original["label"].tolist()
        ==
        result["label"].tolist()
    )

    print(
        f"✓ {name}: users and labels unchanged"
    )


# ============================================================
# SAVE
# ============================================================

train_longformer_selective_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "train_cleaned.pkl"
    )
)

val_longformer_selective_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "val_cleaned.pkl"
    )
)

test_longformer_selective_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "test_cleaned.pkl"
    )
)


print()
print("=" * 60)
print("CELL 5 COMPLETE")
print("=" * 60)

print("✓ Selective cleaning applied")
print("✓ Anchor-removed source used")
print("✓ Model text created")
print("✓ Users preserved")
print("✓ Labels preserved")
print("✓ Cleaned data saved")
```

```python
# ============================================================
# CELL 6 — LONGFORMER + SELECTIVE
# ============================================================

from transformers import AutoTokenizer
import numpy as np
import pandas as pd

print("=" * 60)
print("LONGFORMER + SELECTIVE — TOKEN LENGTH ANALYSIS")
print("=" * 60)

print()
print("Tokenizer:", MODEL_NAME)
print("Experimental context length:", MAX_LENGTH)

from kaggle_secrets import UserSecretsClient

user_secrets = UserSecretsClient()
hf_token = user_secrets.get_secret("HF_TOKEN")

print()
print("✓ HF_TOKEN retrieved")

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    token=hf_token
)

print()
print("Tokenizer max length:", tokenizer.model_max_length)
print("✓ Longformer tokenizer loaded")


def get_token_lengths(texts, name):
    print()
    print(f"Processing {name}...")

    lengths = []

    for i, text in enumerate(texts):
        encoded = tokenizer(
            str(text),
            add_special_tokens=True,
            truncation=False,
            padding=False
        )

        lengths.append(len(encoded["input_ids"]))

        if (i + 1) % 1000 == 0 or (i + 1) == len(texts):
            print(f"{name}: {i + 1}/{len(texts)}")

    return np.array(lengths, dtype=np.int64)


train_token_lengths = get_token_lengths(
    train_longformer_selective_df["model_text"].fillna("").tolist(),
    "TRAIN"
)

val_token_lengths = get_token_lengths(
    val_longformer_selective_df["model_text"].fillna("").tolist(),
    "VALIDATION"
)

test_token_lengths = get_token_lengths(
    test_longformer_selective_df["model_text"].fillna("").tolist(),
    "TEST"
)


def show_stats(lengths, name):
    print()
    print("=" * 60)
    print(f"{name} TOKEN STATISTICS")
    print("=" * 60)

    print("Minimum:", int(lengths.min()))
    print("Maximum:", int(lengths.max()))
    print("Mean:", round(float(lengths.mean()), 2))
    print("Median:", round(float(np.median(lengths)), 2))
    print("25th percentile:", round(float(np.percentile(lengths, 25)), 2))
    print("75th percentile:", round(float(np.percentile(lengths, 75)), 2))
    print("90th percentile:", round(float(np.percentile(lengths, 90)), 2))
    print("95th percentile:", round(float(np.percentile(lengths, 95)), 2))
    print("99th percentile:", round(float(np.percentile(lengths, 99)), 2))

    within_512 = int(np.sum(lengths <= 512))
    above_512 = int(np.sum(lengths > 512))

    print()
    print("512-TOKEN COVERAGE")
    print("Users <= 512:", within_512)
    print("Users > 512:", above_512)
    print(
        "Percentage <= 512:",
        round(100 * within_512 / len(lengths), 2),
        "%"
    )
    print(
        "Percentage > 512:",
        round(100 * above_512 / len(lengths), 2),
        "%"
    )

    within_4096 = int(np.sum(lengths <= 4096))
    above_4096 = int(np.sum(lengths > 4096))

    print()
    print("4096-TOKEN REFERENCE")
    print("Users <= 4096:", within_4096)
    print("Users > 4096:", above_4096)
    print(
        "Percentage <= 4096:",
        round(100 * within_4096 / len(lengths), 2),
        "%"
    )
    print(
        "Percentage > 4096:",
        round(100 * above_4096 / len(lengths), 2),
        "%"
    )


show_stats(train_token_lengths, "TRAIN")
show_stats(val_token_lengths, "VALIDATION")
show_stats(test_token_lengths, "TEST")


print()
print("=" * 60)
print("ESTIMATED CONTEXT CHUNKS")
print("=" * 60)

for context_length in [512, 4096]:
    print()
    print("Context length:", context_length)

    train_chunks = np.ceil(
        train_token_lengths / context_length
    )

    val_chunks = np.ceil(
        val_token_lengths / context_length
    )

    test_chunks = np.ceil(
        test_token_lengths / context_length
    )

    print(
        "TRAIN mean chunks:",
        round(float(train_chunks.mean()), 2)
    )
    print(
        "TRAIN maximum chunks:",
        int(train_chunks.max())
    )

    print(
        "VALIDATION mean chunks:",
        round(float(val_chunks.mean()), 2)
    )
    print(
        "VALIDATION maximum chunks:",
        int(val_chunks.max())
    )

    print(
        "TEST mean chunks:",
        round(float(test_chunks.mean()), 2)
    )
    print(
        "TEST maximum chunks:",
        int(test_chunks.max())
    )


token_summary = pd.DataFrame([
    {
        "Split": "Train",
        "Users": len(train_token_lengths),
        "Mean": train_token_lengths.mean(),
        "Median": np.median(train_token_lengths),
        "<=512 (%)": 100 * np.mean(train_token_lengths <= 512),
        "<=4096 (%)": 100 * np.mean(train_token_lengths <= 4096)
    },
    {
        "Split": "Validation",
        "Users": len(val_token_lengths),
        "Mean": val_token_lengths.mean(),
        "Median": np.median(val_token_lengths),
        "<=512 (%)": 100 * np.mean(val_token_lengths <= 512),
        "<=4096 (%)": 100 * np.mean(val_token_lengths <= 4096)
    },
    {
        "Split": "Test",
        "Users": len(test_token_lengths),
        "Mean": test_token_lengths.mean(),
        "Median": np.median(test_token_lengths),
        "<=512 (%)": 100 * np.mean(test_token_lengths <= 512),
        "<=4096 (%)": 100 * np.mean(test_token_lengths <= 4096)
    }
])


print()
print("=" * 60)
print("TOKEN-LENGTH SUMMARY")
print("=" * 60)

print(
    token_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.2f}"
    )
)


np.save(
    os.path.join(
        RESULT_DIR,
        "train_token_lengths.npy"
    ),
    train_token_lengths
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_token_lengths.npy"
    ),
    val_token_lengths
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_token_lengths.npy"
    ),
    test_token_lengths
)

token_summary.to_csv(
    os.path.join(
        RESULT_DIR,
        "token_length_summary.csv"
    ),
    index=False
)


print()
print("=" * 60)
print("CELL 6 COMPLETE")
print("=" * 60)

print("✓ Authenticated Longformer tokenizer loaded")
print("✓ Selective cleaned text analyzed")
print("✓ No truncation used during analysis")
print("✓ 512-token coverage measured")
print("✓ 4096-token reference measured")
print("✓ Context chunk estimates calculated")
print("✓ Token-length files saved")
```

```python
# ============================================================
# CELL 11 — LONGFORMER + SELECTIVE
# ============================================================

from transformers import AutoModel
import torch.nn as nn

print("=" * 60)
print("LONGFORMER + SELECTIVE — BUILDING FROZEN LONGFORMER + METADATA MODEL")
print("=" * 60)

longformer_encoder = AutoModel.from_pretrained(
    MODEL_NAME,
    token=hf_token
).to(device)

for p in longformer_encoder.parameters():
    p.requires_grad = False


class LongformerMetadataClassifier(nn.Module):
    def __init__(self, encoder, metadata_dim=53):
        super().__init__()

        self.encoder = encoder

        hidden_size = encoder.config.hidden_size

        self.attention_projection = nn.Linear(
            hidden_size,
            256
        )

        self.attention_score = nn.Linear(
            256,
            1
        )

        self.metadata_branch = nn.Sequential(
            nn.Linear(metadata_dim, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, 64),
            nn.ReLU()
        )

        self.metadata_beta = nn.Parameter(
            torch.tensor(0.0)
        )

        self.classifier = nn.Sequential(
            nn.Linear(
                hidden_size + 64,
                512
            ),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(
                512,
                256
            ),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(
                256,
                2
            )
        )

    def forward(
        self,
        input_ids,
        attention_mask,
        metadata
    ):
        with torch.no_grad():

            outputs = self.encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

        last_hidden = outputs.last_hidden_state

        projected = torch.tanh(
            self.attention_projection(
                last_hidden
            )
        )

        scores = self.attention_score(
            projected
        ).squeeze(-1)

        scores = scores.masked_fill(
            attention_mask == 0,
            -1e9
        )

        attention = torch.softmax(
            scores,
            dim=1
        )

        text_rep = torch.bmm(
            attention.unsqueeze(1),
            last_hidden
        ).squeeze(1)

        metadata_rep = self.metadata_branch(
            metadata
        )

        gate = torch.sigmoid(
            self.metadata_beta
        )

        fused = torch.cat(
            [
                text_rep,
                gate * metadata_rep
            ],
            dim=1
        )

        logits = self.classifier(
            fused
        )

        return {
            "logits": logits,
            "text_rep": text_rep,
            "metadata_rep": metadata_rep,
            "attention_weights": attention,
            "metadata_gate": gate
        }


model = LongformerMetadataClassifier(
    longformer_encoder,
    METADATA_DIM
).to(device)

hidden_size = model.encoder.config.hidden_size


# ============================================================
# PARAMETER COUNT
# ============================================================

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

frozen_params = sum(
    p.numel()
    for p in model.parameters()
    if not p.requires_grad
)

print()
print("=" * 60)
print("PARAMETER COUNT")
print("=" * 60)

print("Total parameters:", total_params)
print("Trainable parameters:", trainable_params)
print("Frozen parameters:", frozen_params)

print(
    "Total parameters (M):",
    round(total_params / 1e6, 3)
)

print(
    "Trainable parameters (M):",
    round(trainable_params / 1e6, 3)
)

print(
    "Frozen parameters (M):",
    round(frozen_params / 1e6, 3)
)

encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print("✓ All Longformer backbone parameters frozen")


# ============================================================
# TRAINABLE COMPONENTS
# ============================================================

print()
print("=" * 60)
print("TRAINABLE COMPONENTS")
print("=" * 60)

for name, p in model.named_parameters():

    if p.requires_grad:
        print("✓", name)


# ============================================================
# ARCHITECTURE
# ============================================================

print()
print("=" * 60)
print("ARCHITECTURE CONFIGURATION")
print("=" * 60)

print("Transformer:", MODEL_NAME)
print("Base architecture: Longformer-base-4096")
print("Experimental context length:", MAX_LENGTH)
print("Text hidden size:", hidden_size)

print("Attention projection:", 256)

print("Metadata: 53 -> 128 -> 64")

print("Gate: g = sigmoid(beta)")
print("Initial beta: 0.0")
print("Initial gate: 0.5")

print(
    "Fusion:",
    f"{hidden_size} + 64 = {hidden_size + 64}"
)

print(
    "Classifier:",
    f"{hidden_size + 64} -> 512 -> 256 -> 2"
)

print("Dropout: 0.3")
print("Transformer frozen: True")


# ============================================================
# FORWARD PASS SANITY CHECK
# ============================================================

print()
print("=" * 60)
print("FORWARD PASS SANITY CHECK")
print("=" * 60)

model.eval()

batch = next(iter(train_loader))

input_ids = batch["input_ids"].to(device)
attention_mask = batch["attention_mask"].to(device)
metadata = batch["metadata"].to(device)

with torch.no_grad():

    output = model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        metadata=metadata
    )

print(
    "Input IDs:",
    tuple(input_ids.shape)
)

print(
    "Attention mask:",
    tuple(attention_mask.shape)
)

print(
    "Metadata:",
    tuple(metadata.shape)
)

print(
    "Logits:",
    tuple(output["logits"].shape)
)

print(
    "Text representation:",
    tuple(output["text_rep"].shape)
)

print(
    "Metadata representation:",
    tuple(output["metadata_rep"].shape)
)

print(
    "Attention weights:",
    tuple(output["attention_weights"].shape)
)

print(
    "Metadata gate:",
    float(output["metadata_gate"])
)

assert (
    tuple(output["logits"].shape)
    ==
    (16, 2)
)

assert (
    tuple(output["text_rep"].shape)
    ==
    (16, hidden_size)
)

assert (
    tuple(output["metadata_rep"].shape)
    ==
    (16, 64)
)

assert (
    tuple(output["attention_weights"].shape)
    ==
    (16, 512)
)

assert (
    0.0
    <=
    float(output["metadata_gate"])
    <=
    1.0
)


# ============================================================
# PADDING ATTENTION CHECK
# ============================================================

print()
print("=" * 60)
print("PADDING ATTENTION CHECK")
print("=" * 60)

padding_attention = (
    output["attention_weights"]
    .masked_select(
        attention_mask == 0
    )
)

if padding_attention.numel() > 0:

    print(
        "Maximum attention on padding:",
        float(
            padding_attention.abs().max()
        )
    )

    assert float(
        padding_attention.abs().max()
    ) < 1e-6

else:

    print(
        "No padding tokens in sanity-check batch"
    )


# ============================================================
# TRANSFORMER FREEZE VERIFICATION
# ============================================================

print()
print("=" * 60)
print("TRANSFORMER FREEZE VERIFICATION")
print("=" * 60)

for name, p in model.encoder.named_parameters():

    assert p.requires_grad is False

print("✓ Transformer freeze verified")


print()
print("=" * 60)
print("CELL 11 COMPLETE")
print("=" * 60)

print("✓ Longformer-base-4096 encoder loaded")
print("✓ Experimental context length = 512")
print("✓ Longformer backbone frozen")
print("✓ Attention pooling configured")
print("✓ Metadata branch configured")
print("✓ Metadata gate initialized at 0.5")
print("✓ Fusion layer configured")
print("✓ Classifier configured")
print("✓ Forward pass verified")
print("✓ Padding attention verified")
print("✓ Transformer freeze verified")
```

```python
# ============================================================
# CELL 12 — LONGFORMER + SELECTIVE
# MODEL TRAINING
# ============================================================

import os
import time
import copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

MAX_EPOCHS = 10
EARLY_STOPPING_PATIENCE = 2
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1

print("=" * 60)
print("STARTING LONGFORMER + SELECTIVE TRAINING")
print("=" * 60)

print()
print("Training users:", len(train_dataset))
print("Validation users:", len(val_dataset))
print("Test users:", len(test_dataset))
print("Maximum epochs:", MAX_EPOCHS)
print("Early stopping patience:", EARLY_STOPPING_PATIENCE)
print("Checkpoint metric: Validation F1")
print("Learning rate:", LEARNING_RATE)
print("Weight decay:", WEIGHT_DECAY)
print("Batch size:", BATCH_SIZE)
print("Transformer: FROZEN")

encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0
print()
print("✓ Longformer backbone confirmed frozen")

trainable_params = [
    p for p in model.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_params,
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

total_steps = len(train_loader) * MAX_EPOCHS
warmup_steps = int(
    total_steps * WARMUP_RATIO
)

def lr_lambda(step):
    if step < warmup_steps:
        return (step + 1) / max(
            1,
            warmup_steps
        )

    return max(
        0.0,
        (total_steps - step) /
        max(
            1,
            total_steps - warmup_steps
        )
    )

scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda
)

criterion = nn.CrossEntropyLoss()


def evaluate_model(model, loader):
    model.eval()

    total_loss = 0.0
    labels_all = []
    preds_all = []
    probs_all = []

    with torch.no_grad():
        for batch in loader:
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            metadata = batch["metadata"].to(device)
            labels = batch["labels"].to(device)

            output = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                metadata=metadata
            )

            logits = output["logits"]

            loss = criterion(
                logits,
                labels
            )

            total_loss += (
                loss.item() * labels.size(0)
            )

            probs = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            preds = torch.argmax(
                logits,
                dim=1
            )

            labels_all.extend(
                labels.cpu().numpy()
            )

            preds_all.extend(
                preds.cpu().numpy()
            )

            probs_all.extend(
                probs.cpu().numpy()
            )

    return {
        "loss":
            total_loss / len(loader.dataset),

        "accuracy":
            accuracy_score(
                labels_all,
                preds_all
            ),

        "precision":
            precision_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "recall":
            recall_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "f1":
            f1_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "roc_auc":
            roc_auc_score(
                labels_all,
                probs_all
            )
    }


history = []
best_val_f1 = -np.inf
best_epoch = None
best_state_dict = None
epochs_without_improvement = 0

training_start = time.time()

for epoch in range(1, MAX_EPOCHS + 1):

    epoch_start = time.time()

    print()
    print("=" * 60)
    print(f"EPOCH {epoch}/{MAX_EPOCHS}")
    print("=" * 60)

    model.train()

    running_loss = 0.0
    train_labels = []
    train_preds = []

    for batch_idx, batch in enumerate(
        train_loader,
        start=1
    ):

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        metadata = batch["metadata"].to(device)
        labels = batch["labels"].to(device)

        optimizer.zero_grad()

        output = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = output["logits"]

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            trainable_params,
            max_norm=1.0
        )

        optimizer.step()
        scheduler.step()

        running_loss += (
            loss.item() * labels.size(0)
        )

        preds = torch.argmax(
            logits.detach(),
            dim=1
        )

        train_labels.extend(
            labels.detach().cpu().numpy()
        )

        train_preds.extend(
            preds.cpu().numpy()
        )

        if (
            batch_idx == 1
            or batch_idx % 50 == 0
            or batch_idx == len(train_loader)
        ):

            batch_acc = accuracy_score(
                labels.detach().cpu().numpy(),
                preds.cpu().numpy()
            )

            elapsed = time.time() - epoch_start

            print(
                f"Epoch {epoch} | "
                f"Batch {batch_idx}/{len(train_loader)} | "
                f"Loss: {loss.item():.4f} | "
                f"Accuracy: {batch_acc:.4f} | "
                f"LR: {optimizer.param_groups[0]['lr']:.2e} | "
                f"Time: {elapsed:.1f}s"
            )

    train_loss = (
        running_loss / len(train_loader.dataset)
    )

    train_accuracy = accuracy_score(
        train_labels,
        train_preds
    )

    train_precision = precision_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_recall = recall_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_f1 = f1_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    print()
    print(f"Epoch {epoch} — Validation")

    val_metrics = evaluate_model(
        model,
        val_loader
    )

    with torch.no_grad():
        metadata_gate = float(
            torch.sigmoid(
                model.metadata_beta
            ).item()
        )

    epoch_time = time.time() - epoch_start

    print()
    print("=" * 60)
    print(f"EPOCH {epoch} SUMMARY")
    print("=" * 60)

    print(f"Train Loss: {train_loss:.4f}")
    print(f"Train Accuracy: {train_accuracy:.4f}")
    print(f"Train Precision: {train_precision:.4f}")
    print(f"Train Recall: {train_recall:.4f}")
    print(f"Train F1: {train_f1:.4f}")

    print()
    print(f"Val Loss: {val_metrics['loss']:.4f}")
    print(f"Val Accuracy: {val_metrics['accuracy']:.4f}")
    print(f"Val Precision: {val_metrics['precision']:.4f}")
    print(f"Val Recall: {val_metrics['recall']:.4f}")
    print(f"Val F1: {val_metrics['f1']:.4f}")
    print(f"Val ROC-AUC: {val_metrics['roc_auc']:.4f}")
    print(f"Metadata gate: {metadata_gate:.6f}")
    print(f"Epoch Time: {epoch_time:.2f}s")

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "train_precision": train_precision,
        "train_recall": train_recall,
        "train_f1": train_f1,
        "val_loss": val_metrics["loss"],
        "val_accuracy": val_metrics["accuracy"],
        "val_precision": val_metrics["precision"],
        "val_recall": val_metrics["recall"],
        "val_f1": val_metrics["f1"],
        "val_roc_auc": val_metrics["roc_auc"],
        "metadata_gate": metadata_gate,
        "learning_rate": optimizer.param_groups[0]["lr"],
        "epoch_time": epoch_time
    })

    if val_metrics["f1"] > best_val_f1:

        best_val_f1 = val_metrics["f1"]
        best_epoch = epoch

        best_state_dict = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }

        epochs_without_improvement = 0

        checkpoint_path = os.path.join(
            CHECKPOINT_DIR,
            "best_longformer_selective.pt"
        )

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict": best_state_dict,
                "optimizer_state_dict":
                    optimizer.state_dict(),
                "scheduler_state_dict":
                    scheduler.state_dict(),
                "best_val_f1": best_val_f1,
                "experiment_name":
                    EXPERIMENT_NAME,
                "model_name":
                    MODEL_NAME,
                "cleaning_method":
                    CLEANING_METHOD
            },
            checkpoint_path
        )

        print()
        print("✓ NEW BEST CHECKPOINT")
        print(
            "Best Validation F1:",
            f"{best_val_f1:.4f}"
        )
        print("Saved:", checkpoint_path)

    else:

        epochs_without_improvement += 1

        print()
        print("No validation F1 improvement.")
        print(
            "Epochs without improvement:",
            f"{epochs_without_improvement} / "
            f"{EARLY_STOPPING_PATIENCE}"
        )

    if (
        epochs_without_improvement
        >= EARLY_STOPPING_PATIENCE
    ):

        print()
        print("EARLY STOPPING TRIGGERED")
        break


assert best_state_dict is not None
assert best_epoch is not None

model.load_state_dict(best_state_dict)
model.to(device)
model.eval()

total_training_time = (
    time.time() - training_start
)

history_df = pd.DataFrame(history)

history_path = os.path.join(
    RESULT_DIR,
    "longformer_selective_training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)

with torch.no_grad():

    learned_beta = float(
        model.metadata_beta.item()
    )

    learned_gate = float(
        torch.sigmoid(
            model.metadata_beta
        ).item()
    )

print()
print("=" * 60)
print("LONGFORMER + SELECTIVE TRAINING COMPLETE")
print("=" * 60)

print("Epochs completed:", len(history))
print("Best epoch:", best_epoch)
print(
    "Best Validation F1:",
    f"{best_val_f1:.4f}"
)
print(
    "Total training time:",
    f"{total_training_time:.2f} seconds"
)
print(
    "Best checkpoint:",
    os.path.join(
        CHECKPOINT_DIR,
        "best_longformer_selective.pt"
    )
)
print(
    "Training history saved:",
    history_path
)

print()
print("Learned beta:", learned_beta)
print(
    "Metadata gate:",
    f"{learned_gate:.6f}"
)

print()
print("✓ Longformer transformer remained frozen")
print("✓ Downstream layers trained")

print()
print("=" * 60)
print("CELL 12 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# CELL 15 — LONGFORMER + SELECTIVE
# t-SNE REPRESENTATION ANALYSIS
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

print("=" * 70)
print("t-SNE REPRESENTATION ANALYSIS")
print("LONGFORMER + SELECTIVE")
print("=" * 70)

# ============================================================
# FIXED EXPLAINABILITY SETTINGS
# ============================================================

EXPLAINABILITY_SEED = 42

PCA_COMPONENTS = 50

TSNE_COMPONENTS = 2
TSNE_PERPLEXITY = 30
TSNE_LEARNING_RATE = 200
TSNE_MAX_ITER = 1000

np.random.seed(
    EXPLAINABILITY_SEED
)

torch.manual_seed(
    EXPLAINABILITY_SEED
)

# ============================================================
# OUTPUT DIRECTORIES
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

TSNE_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "tsne"
)

os.makedirs(
    TSNE_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model.eval()

print()
print("✓ Longformer model in evaluation mode")

# ============================================================
# EXTRACT EXACT FUSED REPRESENTATION
# ============================================================

fused_representations = []
test_labels = []

print()
print("=" * 70)
print("EXTRACTING FUSED REPRESENTATIONS")
print("=" * 70)

with torch.no_grad():

    for batch_idx, batch in enumerate(
        test_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)

        # ----------------------------------------------------
        # LONGFORMER ENCODER
        # ----------------------------------------------------

        encoder_output = model.encoder(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        hidden_states = (
            encoder_output.last_hidden_state
        )

        # ----------------------------------------------------
        # EXACT TEXT ATTENTION POOLING
        # ----------------------------------------------------

        projected = torch.tanh(
            model.attention_projection(
                hidden_states
            )
        )

        attention_scores = (
            model.attention_score(
                projected
            )
            .squeeze(-1)
        )

        attention_scores = (
            attention_scores.masked_fill(
                attention_mask == 0,
                -1e9
            )
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states
            *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # ----------------------------------------------------
        # EXACT METADATA BRANCH
        # ----------------------------------------------------

        metadata_representation = (
            model.metadata_branch(
                metadata
            )
        )

        # ----------------------------------------------------
        # EXACT LEARNED METADATA GATE
        # ----------------------------------------------------

        metadata_gate = torch.sigmoid(
            model.metadata_beta
        )

        gated_metadata = (
            metadata_gate
            *
            metadata_representation
        )

        # ----------------------------------------------------
        # EXACT FUSED REPRESENTATION
        # ----------------------------------------------------

        fused = torch.cat(
            [
                text_representation,
                gated_metadata
            ],
            dim=1
        )

        fused_representations.append(
            fused.cpu().numpy()
        )

        test_labels.extend(
            labels.cpu().numpy().tolist()
        )

        if (
            batch_idx == 1
            or batch_idx % 10 == 0
            or batch_idx == len(test_loader)
        ):

            print(
                f"Processed batch "
                f"{batch_idx}/{len(test_loader)}"
            )

# ============================================================
# COMBINE
# ============================================================

fused_representations = np.concatenate(
    fused_representations,
    axis=0
)

test_labels = np.array(
    test_labels
)

# ============================================================
# REPRESENTATION SUMMARY
# ============================================================

print()
print("=" * 70)
print("FUSED REPRESENTATION SUMMARY")
print("=" * 70)

print(
    "Fused representation shape:",
    fused_representations.shape
)

print(
    "Test users:",
    len(test_labels)
)

print(
    "Fused dimensions:",
    fused_representations.shape[1]
)

print(
    "Negative users:",
    int(
        (test_labels == 0).sum()
    )
)

print(
    "Positive users:",
    int(
        (test_labels == 1).sum()
    )
)

print(
    "Metadata gate:",
    f"{float(metadata_gate.item()):.6f}"
)

assert (
    fused_representations.shape[0]
    ==
    len(test_labels)
)

assert np.isfinite(
    fused_representations
).all()

# ============================================================
# PCA
# ============================================================

print()
print("=" * 70)
print("PCA REDUCTION")
print("=" * 70)

pca_dimensions = min(
    PCA_COMPONENTS,
    fused_representations.shape[1]
)

pca = PCA(
    n_components=pca_dimensions,
    random_state=EXPLAINABILITY_SEED
)

pca_representation = pca.fit_transform(
    fused_representations
)

pca_variance_retained = (
    pca.explained_variance_ratio_.sum()
)

print(
    "Original dimensions:",
    fused_representations.shape[1]
)

print(
    "PCA dimensions:",
    pca_representation.shape[1]
)

print(
    "Variance retained:",
    f"{pca_variance_retained * 100:.2f}%"
)

# ============================================================
# t-SNE
# ============================================================

print()
print("=" * 70)
print("t-SNE REDUCTION")
print("=" * 70)

tsne = TSNE(
    n_components=TSNE_COMPONENTS,
    perplexity=TSNE_PERPLEXITY,
    learning_rate=TSNE_LEARNING_RATE,
    init="pca",
    random_state=EXPLAINABILITY_SEED,
    max_iter=TSNE_MAX_ITER
)

tsne_representation = tsne.fit_transform(
    pca_representation
)

print(
    "t-SNE output shape:",
    tsne_representation.shape
)

# ============================================================
# SAVE COORDINATES
# ============================================================

tsne_df = pd.DataFrame({

    "tsne_1":
        tsne_representation[:, 0],

    "tsne_2":
        tsne_representation[:, 1],

    "label":
        test_labels

})

tsne_df["class"] = tsne_df[
    "label"
].map({
    0: "Negative",
    1: "Positive"
})

tsne_csv_path = os.path.join(
    TSNE_DIR,
    "longformer_selective_tsne_coordinates.csv"
)

tsne_df.to_csv(
    tsne_csv_path,
    index=False
)

# ============================================================
# PUBLICATION-QUALITY t-SNE PLOT
# ============================================================

plt.figure(
    figsize=(10, 8)
)

negative_mask = (
    test_labels == 0
)

positive_mask = (
    test_labels == 1
)

plt.scatter(
    tsne_representation[
        negative_mask,
        0
    ],
    tsne_representation[
        negative_mask,
        1
    ],
    s=40,
    alpha=0.72,
    label="Negative",
    color="#4C78A8",
    edgecolors="white",
    linewidths=0.4
)

plt.scatter(
    tsne_representation[
        positive_mask,
        0
    ],
    tsne_representation[
        positive_mask,
        1
    ],
    s=40,
    alpha=0.72,
    label="Positive",
    color="#E45756",
    edgecolors="white",
    linewidths=0.4
)

plt.xlabel(
    "t-SNE Dimension 1",
    fontsize=12
)

plt.ylabel(
    "t-SNE Dimension 2",
    fontsize=12
)

plt.title(
    "Longformer + Selective — "
    "t-SNE of Fused Test Representations",
    fontsize=14,
    fontweight="bold"
)

plt.legend(
    fontsize=11
)

plt.grid(
    alpha=0.20
)

plt.tight_layout()

tsne_plot_path = os.path.join(
    TSNE_DIR,
    "longformer_selective_tsne.png"
)

# SAVE
plt.savefig(
    tsne_plot_path,
    dpi=300,
    bbox_inches="tight"
)

# DISPLAY
plt.show()

plt.close()

# ============================================================
# SAVE CONFIGURATION
# ============================================================

tsne_config = {

    "model":
        "Longformer",

    "cleaning":
        "Selective",

    "dataset":
        "Held-out test set",

    "test_users":
        int(len(test_labels)),

    "input_representation":
        "Fused classifier representation",

    "text_representation_dimension":
        int(
            text_representation.shape[1]
        ),

    "metadata_representation_dimension":
        int(
            metadata_representation.shape[1]
        ),

    "fused_representation_dimension":
        int(
            fused_representations.shape[1]
        ),

    "pca_dimensions":
        int(
            pca_dimensions
        ),

    "pca_variance_retained":
        float(
            pca_variance_retained
        ),

    "tsne_dimensions":
        int(
            TSNE_COMPONENTS
        ),

    "tsne_perplexity":
        int(
            TSNE_PERPLEXITY
        ),

    "tsne_learning_rate":
        TSNE_LEARNING_RATE,

    "tsne_max_iterations":
        int(
            TSNE_MAX_ITER
        ),

    "random_seed":
        EXPLAINABILITY_SEED,

    "metadata_gate":
        float(
            metadata_gate.item()
        )

}

with open(
    os.path.join(
        TSNE_DIR,
        "longformer_selective_tsne_config.json"
    ),
    "w"
) as f:

    json.dump(
        tsne_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("t-SNE ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ Exact Longformer fused representation extracted"
)

print(
    "✓ PCA reduction completed"
)

print(
    "✓ t-SNE reduction completed"
)

print(
    "✓ t-SNE graph saved:"
)

print(
    tsne_plot_path
)

print(
    "✓ t-SNE graph displayed above"
)

print(
    "✓ Coordinates saved:"
)

print(
    tsne_csv_path
)

print(
    "✓ Configuration saved"
)
```

```python
# ============================================================
# EXPLAINABILITY — GRAD-CAM
# LONGFORMER + SELECTIVE
# TOKEN-LEVEL EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

print("=" * 70)
print("GRAD-CAM ANALYSIS")
print("LONGFORMER + SELECTIVE")
print("=" * 70)


# ============================================================
# FIXED SETTINGS
# ============================================================

GRADCAM_SEED = 42
TOP_TOKENS = 20

np.random.seed(GRADCAM_SEED)
torch.manual_seed(GRADCAM_SEED)


# ============================================================
# OUTPUT DIRECTORY
# ============================================================

GRADCAM_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "gradcam"
)

os.makedirs(
    GRADCAM_DIR,
    exist_ok=True
)


# ============================================================
# MODEL
# ============================================================

model.eval()

for parameter in model.parameters():
    parameter.requires_grad = False

print()
print("✓ Longformer Selective model ready")


# ============================================================
# COLLECT TEST PREDICTIONS
# ============================================================

test_true = []
test_pred = []
test_prob = []

with torch.no_grad():

    for batch in test_loader:

        input_ids = (
            batch["input_ids"]
            .to(device)
        )

        attention_mask = (
            batch["attention_mask"]
            .to(device)
        )

        metadata = (
            batch["metadata"]
            .to(device)
        )

        labels = (
            batch["labels"]
            .to(device)
        )

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = outputs["logits"]

        probabilities = torch.softmax(
            logits.float(),
            dim=1
        )[:, 1]

        predictions = (
            probabilities >= 0.5
        ).long()

        test_true.extend(
            labels.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        test_pred.extend(
            predictions.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        test_prob.extend(
            probabilities.detach()
            .cpu()
            .numpy()
            .tolist()
        )


test_true = np.asarray(
    test_true
)

test_pred = np.asarray(
    test_pred
)

test_prob = np.asarray(
    test_prob
)

print()
print(
    "Test users:",
    len(test_true)
)


# ============================================================
# DETERMINISTIC TP / TN / FP / FN CASES
# ============================================================

tp_candidates = np.where(
    (test_true == 1) &
    (test_pred == 1)
)[0]

tn_candidates = np.where(
    (test_true == 0) &
    (test_pred == 0)
)[0]

fp_candidates = np.where(
    (test_true == 0) &
    (test_pred == 1)
)[0]

fn_candidates = np.where(
    (test_true == 1) &
    (test_pred == 0)
)[0]


assert len(tp_candidates) > 0
assert len(tn_candidates) > 0
assert len(fp_candidates) > 0
assert len(fn_candidates) > 0


selected_cases = {

    "TP":
        int(tp_candidates[0]),

    "TN":
        int(tn_candidates[0]),

    "FP":
        int(fp_candidates[0]),

    "FN":
        int(fn_candidates[0])
}


print()
print("=" * 70)
print("SELECTED CASES")
print("=" * 70)

for case_name, index in selected_cases.items():

    print(
        f"{case_name}: "
        f"Test Index = {index} | "
        f"True = {test_true[index]} | "
        f"Pred = {test_pred[index]} | "
        f"Positive Probability = {test_prob[index]:.4f}"
    )


# ============================================================
# GRAD-CAM FUNCTION
# ============================================================

def compute_gradcam(
    sample_index
):

    sample = test_dataset[
        sample_index
    ]

    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
        .to(device)
    )

    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
        .to(device)
    )

    metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
    )


    # --------------------------------------------------------
    # ENABLE GRADIENTS
    # --------------------------------------------------------

    model.zero_grad(
        set_to_none=True
    )


    # --------------------------------------------------------
    # LONGFORMER ENCODER
    # --------------------------------------------------------

    encoder_output = model.encoder(
        input_ids=input_ids,
        attention_mask=attention_mask
    )

    hidden_states = (
        encoder_output.last_hidden_state
    )
    
    # Enable gradients on final hidden states
    # while keeping model weights frozen
    hidden_states.requires_grad_(True)
    hidden_states.retain_grad()


    # --------------------------------------------------------
    # ATTENTION POOLING
    # --------------------------------------------------------

    projected = torch.tanh(
        model.attention_projection(
            hidden_states
        )
    )

    attention_scores = (
        model.attention_score(
            projected
        )
        .squeeze(-1)
    )

    attention_scores = (
        attention_scores.masked_fill(
            attention_mask == 0,
            -1e9
        )
    )

    attention_weights = torch.softmax(
        attention_scores,
        dim=1
    )


    text_representation = torch.sum(
        hidden_states *
        attention_weights.unsqueeze(-1),
        dim=1
    )


    # --------------------------------------------------------
    # METADATA BRANCH
    # --------------------------------------------------------

    metadata_representation = (
        model.metadata_branch(
            metadata
        )
    )


    metadata_gate = torch.sigmoid(
        model.metadata_beta
    )

    gated_metadata = (
        metadata_gate *
        metadata_representation
    )


    # --------------------------------------------------------
    # FUSION
    # --------------------------------------------------------

    fused = torch.cat(
        [
            text_representation,
            gated_metadata
        ],
        dim=1
    )


    # --------------------------------------------------------
    # CLASSIFIER
    # --------------------------------------------------------

    logits = model.classifier(
        fused
    )


    positive_logit = (
        logits[:, 1]
    )

    positive_logit.backward()


    # --------------------------------------------------------
    # GRAD-CAM
    # --------------------------------------------------------

    gradients = hidden_states.grad

    channel_weights = gradients.mean(
        dim=1
    )


    cam = (
        (
            hidden_states *
            channel_weights.unsqueeze(1)
        )
        .sum(dim=2)
        .squeeze(0)
    )


    # ReLU
    cam = torch.relu(
        cam
    )


    # Normalize
    cam = (
        cam /
        (
            cam.max() +
            1e-12
        )
    )


    cam = cam.detach().cpu().numpy()

    tokens = tokenizer.convert_ids_to_tokens(
        input_ids.squeeze(0)
        .detach()
        .cpu()
        .tolist()
    )

    mask = (
        attention_mask.squeeze(0)
        .detach()
        .cpu()
        .numpy()
        .astype(bool)
    )


    tokens = [
        token
        for token, valid in zip(
            tokens,
            mask
        )
        if valid
    ]

    cam = cam[
        mask
    ]


    # --------------------------------------------------------
    # REMOVE SPECIAL TOKENS
    # --------------------------------------------------------

    valid_pairs = []

    special_tokens = {
        tokenizer.cls_token,
        tokenizer.sep_token,
        tokenizer.pad_token
    }

    for token, importance in zip(
        tokens,
        cam
    ):

        if token in special_tokens:
            continue

        valid_pairs.append(
            (
                token,
                float(importance)
            )
        )


    # --------------------------------------------------------
    # TOP TOKENS
    # --------------------------------------------------------

    valid_pairs = sorted(
        valid_pairs,
        key=lambda x: x[1],
        reverse=True
    )

    top_pairs = valid_pairs[
        :TOP_TOKENS
    ]

    return (
        top_pairs,
        tokens,
        cam,
        int(test_true[sample_index]),
        int(test_pred[sample_index]),
        float(test_prob[sample_index])
    )


# ============================================================
# RUN FOUR CASES
# ============================================================

gradcam_results = {}

print()
print("=" * 70)
print("COMPUTING GRAD-CAM")
print("=" * 70)


for case_name, sample_index in selected_cases.items():

    print()
    print(
        f"Processing {case_name} "
        f"(Test Index {sample_index})"
    )

    result = compute_gradcam(
        sample_index
    )

    gradcam_results[
        case_name
    ] = result

    print(
        f"✓ {case_name} complete"
    )


# ============================================================
# SAVE TOKEN IMPORTANCE CSV
# ============================================================

all_rows = []

for case_name, result in gradcam_results.items():

    top_pairs = result[0]

    true_label = result[3]
    predicted_label = result[4]
    positive_probability = result[5]

    for rank, (
        token,
        importance
    ) in enumerate(
        top_pairs,
        start=1
    ):

        all_rows.append({

            "case":
                case_name,

            "rank":
                rank,

            "token":
                token,

            "importance":
                importance,

            "true_label":
                true_label,

            "predicted_label":
                predicted_label,

            "positive_probability":
                positive_probability
        })


gradcam_df = pd.DataFrame(
    all_rows
)

gradcam_csv_path = os.path.join(
    GRADCAM_DIR,
    "longformer_selective_gradcam_token_importance.csv"
)

gradcam_df.to_csv(
    gradcam_csv_path,
    index=False
)


# ============================================================
# PLOT FOUR CASES
# ============================================================

case_titles = {

    "TP":
        "True Positive",

    "TN":
        "True Negative",

    "FP":
        "False Positive",

    "FN":
        "False Negative"
}


for case_name, result in gradcam_results.items():

    top_pairs = result[0]

    tokens_plot = [
        pair[0]
        for pair in top_pairs
    ][::-1]

    importance_plot = [
        pair[1]
        for pair in top_pairs
    ][::-1]


    plt.figure(
        figsize=(10, 7)
    )

    bars = plt.barh(
        tokens_plot,
        importance_plot
    )


    # Highlight highest-importance token
    if len(bars) > 0:

        max_index = int(
            np.argmax(
                importance_plot
            )
        )

        bars[max_index].set_color(
            "#F28E2B"
        )

        for index, bar in enumerate(
            bars
        ):

            if index != max_index:

                bar.set_color(
                    "#59A14F"
                )


    plt.xlabel(
        "Grad-CAM Token Importance"
    )

    plt.ylabel(
        "Token"
    )

    plt.title(
        "Longformer + Selective — "
        f"{case_titles[case_name]}"
    )

    plt.grid(
        axis="x",
        alpha=0.20
    )

    plt.tight_layout()


    plot_path = os.path.join(
        GRADCAM_DIR,
        f"longformer_selective_gradcam_{case_name.lower()}.png"
    )

    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# ============================================================
# SAVE CONFIGURATION
# ============================================================

gradcam_config = {

    "model":
        "Longformer",

    "cleaning":
        "Selective",

    "method":
        "Gradient-weighted Class Activation Mapping",

    "target":
        "Positive/depression-class logit",

    "representation":
        "Final Transformer hidden states",

    "channel_weight":
        "Mean gradient across token positions",

    "activation":
        "ReLU",

    "normalization":
        "Per-user max normalization",

    "top_tokens":
        TOP_TOKENS,

    "seed":
        GRADCAM_SEED,

    "cases":
        selected_cases,

    "test_users":
        len(test_true)

}


with open(
    os.path.join(
        GRADCAM_DIR,
        "longformer_selective_gradcam_config.json"
    ),
    "w"
) as f:

    json.dump(
        gradcam_config,
        f,
        indent=2
    )


# ============================================================
# PRINT RESULTS
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM ANALYSIS COMPLETE")
print("=" * 70)

for case_name, result in gradcam_results.items():

    print()
    print(
        f"{case_name} — "
        f"True: {result[3]} | "
        f"Predicted: {result[4]} | "
        f"P(Positive): {result[5]:.4f}"
    )

    print(
        "Top tokens:",
        result[0]
    )

print()
print(
    "✓ TP, TN, FP, FN cases explained"
)

print(
    "✓ Final Longformer hidden states used"
)

print(
    "✓ Positive/depression-class logit used"
)

print(
    "✓ Top 20 tokens extracted"
)

print(
    "✓ Token importance CSV saved"
)

print(
    "✓ Four paper-ready plots saved and displayed"
)

print(
    "✓ Configuration saved"
)

print()
print(
    "Grad-CAM output directory:",
    GRADCAM_DIR
)
```

```python
# ============================================================
# EXPLAINABILITY — LIME
# LONGFORMER + SELECTIVE
# LOCAL TOKEN-LEVEL EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from lime.lime_text import LimeTextExplainer
from IPython.display import display, HTML


print("=" * 70)
print("LIME ANALYSIS")
print("LONGFORMER + SELECTIVE")
print("=" * 70)


# ============================================================
# FIXED LIME SETTINGS
# ============================================================

LIME_SEED = 42
LIME_NUM_FEATURES = 15
LIME_NUM_SAMPLES = 1000
LIME_BATCH_SIZE = 8

np.random.seed(
    LIME_SEED
)

torch.manual_seed(
    LIME_SEED
)


# ============================================================
# OUTPUT DIRECTORY
# ============================================================

LIME_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "lime"
)

os.makedirs(
    LIME_DIR,
    exist_ok=True
)


# ============================================================
# MODEL
# ============================================================

model.eval()

for parameter in model.parameters():
    parameter.requires_grad = False

print()
print("✓ Longformer Selective model ready")


# ============================================================
# TEST PREDICTIONS
# ============================================================

labels_np = []
preds_np = []
probs_np = []

with torch.no_grad():

    for batch in test_loader:

        input_ids = (
            batch["input_ids"]
            .to(device)
        )

        attention_mask = (
            batch["attention_mask"]
            .to(device)
        )

        metadata = (
            batch["metadata"]
            .to(device)
        )

        labels = (
            batch["labels"]
            .to(device)
        )

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = outputs["logits"]

        probabilities = torch.softmax(
            logits.float(),
            dim=1
        )[:, 1]

        predictions = (
            probabilities >= 0.5
        ).long()

        labels_np.extend(
            labels.cpu()
            .numpy()
            .tolist()
        )

        preds_np.extend(
            predictions.cpu()
            .numpy()
            .tolist()
        )

        probs_np.extend(
            probabilities.cpu()
            .numpy()
            .tolist()
        )


labels_np = np.asarray(
    labels_np
)

preds_np = np.asarray(
    preds_np
)

probs_np = np.asarray(
    probs_np
)


print()
print(
    "Test users:",
    len(labels_np)
)


# ============================================================
# DETERMINISTIC TP / TN / FP / FN
# ============================================================

tp_candidates = np.where(
    (labels_np == 1) &
    (preds_np == 1)
)[0]

tn_candidates = np.where(
    (labels_np == 0) &
    (preds_np == 0)
)[0]

fp_candidates = np.where(
    (labels_np == 0) &
    (preds_np == 1)
)[0]

fn_candidates = np.where(
    (labels_np == 1) &
    (preds_np == 0)
)[0]


assert len(tp_candidates) > 0
assert len(tn_candidates) > 0
assert len(fp_candidates) > 0
assert len(fn_candidates) > 0


selected_cases = {

    "TP":
        int(
            tp_candidates[0]
        ),

    "TN":
        int(
            tn_candidates[0]
        ),

    "FP":
        int(
            fp_candidates[0]
        ),

    "FN":
        int(
            fn_candidates[0]
        )
}


print()
print("=" * 70)
print("SELECTED CASES")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print(
        f"{case_name}: "
        f"Test Index = {test_index} | "
        f"True = {labels_np[test_index]} | "
        f"Pred = {preds_np[test_index]} | "
        f"Positive Probability = "
        f"{probs_np[test_index]:.4f}"
    )


# ============================================================
# LIME TEXT ACCESS
# ============================================================
# The same cleaned model_text that was used for Longformer
# tokenization is used for LIME explanations.
# ============================================================

def get_test_text(
    test_index
):

    return str(
        test_longformer_selective_df.iloc[
            int(test_index)
        ]["model_text"]
    )


# ============================================================
# LIME PREDICTOR
# ============================================================
# Text is perturbed by LIME.
# Metadata remains fixed for the explained user.
# ============================================================

def make_lime_predictor(
    test_index
):

    sample = test_dataset[
        int(test_index)
    ]

    fixed_metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .float()
        .to(device)
    )

    def lime_predict(
        texts
    ):

        all_probabilities = []

        for start in range(
            0,
            len(texts),
            LIME_BATCH_SIZE
        ):

            text_batch = texts[
                start:
                start + LIME_BATCH_SIZE
            ]

            encoded = tokenizer(
                text_batch,
                max_length=MAX_LENGTH,
                truncation=True,
                padding="max_length",
                return_attention_mask=True
            )

            input_ids = torch.tensor(
                encoded["input_ids"],
                dtype=torch.long
            ).to(device)

            attention_mask = torch.tensor(
                encoded["attention_mask"],
                dtype=torch.long
            ).to(device)

            metadata_batch = (
                fixed_metadata
                .expand(
                    len(text_batch),
                    -1
                )
            )

            with torch.no_grad():

                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    metadata=metadata_batch
                )

                logits = outputs["logits"]

                probabilities = torch.softmax(
                    logits.float(),
                    dim=1
                )

            all_probabilities.append(
                probabilities
                .cpu()
                .numpy()
            )

            del input_ids
            del attention_mask
            del metadata_batch
            del outputs
            del logits
            del probabilities

            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        return np.vstack(
            all_probabilities
        )

    return lime_predict


# ============================================================
# LIME EXPLAINER
# ============================================================

explainer = LimeTextExplainer(
    class_names=[
        "Negative",
        "Positive"
    ],
    split_expression=r"\s+",
    random_state=LIME_SEED
)

print()
print("✓ LIME explainer ready")


# ============================================================
# RUN LIME FOR FOUR CASES
# ============================================================

all_explanations = {}

print()
print("=" * 70)
print("GENERATING LIME EXPLANATIONS")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print()
    print(
        f"Processing {case_name} "
        f"(test index {test_index})..."
    )

    text = get_test_text(
        test_index
    )

    predictor = make_lime_predictor(
        test_index
    )

    explanation = explainer.explain_instance(
        text,
        predictor,
        labels=(1,),
        num_features=LIME_NUM_FEATURES,
        num_samples=LIME_NUM_SAMPLES
    )

    all_explanations[
        case_name
    ] = explanation

    print(
        f"✓ {case_name} explanation complete"
    )


# ============================================================
# SAVE LIME RESULTS
# ============================================================

all_lime_rows = []

for case_name, explanation in all_explanations.items():

    explanation_list = explanation.as_list(
        label=1
    )

    for rank, (
        word,
        weight
    ) in enumerate(
        explanation_list,
        start=1
    ):

        all_lime_rows.append({

            "case":
                case_name,

            "test_index":
                selected_cases[
                    case_name
                ],

            "true_label":
                int(
                    labels_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "predicted_label":
                int(
                    preds_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "positive_probability":
                float(
                    probs_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "rank":
                rank,

            "word":
                word,

            "lime_weight":
                float(weight)

        })


lime_results_df = pd.DataFrame(
    all_lime_rows
)

lime_csv_path = os.path.join(
    LIME_DIR,
    "longformer_selective_lime_results.csv"
)

lime_results_df.to_csv(
    lime_csv_path,
    index=False
)


# ============================================================
# DISPLAY + SAVE HTML
# ============================================================

print()
print("=" * 70)
print("LIME EXPLANATIONS")
print("=" * 70)

for case_name, explanation in all_explanations.items():

    print()
    print(
        "=" * 60
    )

    print(
        f"{case_name} — "
        f"LIME LOCAL EXPLANATION"
    )

    print(
        "=" * 60
    )

    display(
        HTML(
            explanation.as_html(
                labels=[1]
            )
        )
    )

    html_path = os.path.join(
        LIME_DIR,
        f"longformer_selective_lime_{case_name}.html"
    )

    with open(
        html_path,
        "w",
        encoding="utf-8"
    ) as f:

        f.write(
            explanation.as_html(
                labels=[1]
            )
        )


# ============================================================
# PAPER-FRIENDLY LIME BAR PLOTS
# ============================================================

for case_name, explanation in all_explanations.items():

    explanation_list = explanation.as_list(
        label=1
    )

    plot_df = pd.DataFrame(
        explanation_list,
        columns=[
            "word",
            "weight"
        ]
    )

    plot_df = plot_df.sort_values(
        "weight"
    )

    plt.figure(
        figsize=(10, 7)
    )

    bars = plt.barh(
        plot_df["word"],
        plot_df["weight"]
    )

    # Positive contribution
    for bar, weight in zip(
        bars,
        plot_df["weight"]
    ):

        if weight > 0:
            bar.set_color(
                "#E45756"
            )
        else:
            bar.set_color(
                "#4C78A8"
            )

    plt.axvline(
        0,
        linewidth=0.8
    )

    plt.xlabel(
        "LIME Weight",
        fontsize=12
    )

    plt.ylabel(
        "Token / Word",
        fontsize=12
    )

    plt.title(
        f"Longformer + Selective — "
        f"LIME Explanation ({case_name})",
        fontsize=14
    )

    plt.grid(
        axis="x",
        alpha=0.20
    )

    plt.tight_layout()

    plot_path = os.path.join(
        LIME_DIR,
        f"longformer_selective_lime_{case_name}.png"
    )

    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# ============================================================
# SAVE CASE INFORMATION
# ============================================================

case_info_rows = []

for case_name, test_index in selected_cases.items():

    case_info_rows.append({

        "case":
            case_name,

        "test_index":
            int(test_index),

        "true_label":
            int(
                labels_np[
                    test_index
                ]
            ),

        "predicted_label":
            int(
                preds_np[
                    test_index
                ]
            ),

        "positive_probability":
            float(
                probs_np[
                    test_index
                ]
            ),

        "text_length_characters":
            len(
                get_test_text(
                    test_index
                )
            )

    })


case_info_df = pd.DataFrame(
    case_info_rows
)

case_info_df.to_csv(
    os.path.join(
        LIME_DIR,
        "longformer_selective_lime_case_information.csv"
    ),
    index=False
)


# ============================================================
# SAVE CONFIGURATION
# ============================================================

lime_config = {

    "model":
        "Longformer",

    "cleaning":
        "Selective",

    "dataset":
        "Held-out test set",

    "target":
        "Positive/depression class",

    "representative_cases":
        [
            "TP",
            "TN",
            "FP",
            "FN"
        ],

    "case_selection":
        "First matching test example in fixed test order",

    "random_seed":
        LIME_SEED,

    "num_samples":
        LIME_NUM_SAMPLES,

    "num_features":
        LIME_NUM_FEATURES,

    "batch_size":
        LIME_BATCH_SIZE,

    "metadata":
        "Fixed for each explained user",

    "text_perturbation":
        "LIME word-level perturbations",

    "model_training":
        "No additional training"
}


with open(
    os.path.join(
        LIME_DIR,
        "longformer_selective_lime_config.json"
    ),
    "w"
) as f:

    json.dump(
        lime_config,
        f,
        indent=2
    )


# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("LIME ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ TP, TN, FP, FN cases explained"
)

print(
    "✓ Same trained Longformer model used"
)

print(
    "✓ Metadata held fixed per user"
)

print(
    "✓ 1000 LIME perturbations per case"
)

print(
    "✓ Top 15 features extracted"
)

print(
    "✓ LIME results saved"
)

print(
    "✓ HTML explanations saved"
)

print(
    "✓ Paper-ready LIME plots saved and displayed"
)

print()
print(
    "LIME output directory:",
    LIME_DIR
)
```
