```python
# ============================================================
# CELL 3 — SENTENCE-BERT + EMOTION-BASED
# ============================================================

print("=" * 60)
print("SENTENCE-BERT + EMOTION-BASED — EXACT D1 ANCHOR DETECTION")
print("=" * 60)

import os
import re
import json
import glob
from collections import defaultdict

DATASET_ROOT = (
    "/kaggle/input/datasets/"
    "vedantnawghare35/"
    "depression-detection-mddl-final-pbl/"
    "Dataset_MDDL"
)

D1_TIMELINE_DIR = os.path.join(
    DATASET_ROOT,
    "labeled",
    "positive",
    "data",
    "timeline"
)

assert os.path.exists(D1_TIMELINE_DIR)

d1_timeline_files = glob.glob(
    os.path.join(D1_TIMELINE_DIR, "*.json")
)

print()
print("D1 timeline files found:", len(d1_timeline_files))

assert len(d1_timeline_files) == 2626

REFINED_ANCHOR_PATTERNS = [
    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",
    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",
    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depression\b",
    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",
    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",
    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",
    r"\bi\s+have\s+(?:major\s+)?depression\b",
    r"\bi\s+suffer\s+from\s+(?:major\s+)?depression\b",
    r"\bi\s+(?:was|am|'m)\s+diagnosed\b.{0,80}\bdepress",
    r"\b(?:my|a)\s+(?:doctor|psychiatrist)\s+(?:has\s+)?diagnosed\s+me\b.{0,80}\bdepress"
]

EXCLUSION_PATTERNS = [
    r"\bmy\s+friend\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+sister\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+brother\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+mother\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+father\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+answer\s+to\s*:",
    r"\banswer\s+to\s*:",
    r"\bquestion\s*:",
    r"\bq\s*:"
]

anchor_patterns = [
    re.compile(
        p,
        re.IGNORECASE | re.DOTALL
    )
    for p in REFINED_ANCHOR_PATTERNS
]

exclusion_patterns = [
    re.compile(
        p,
        re.IGNORECASE | re.DOTALL
    )
    for p in EXCLUSION_PATTERNS
]


def is_refined_anchor_tweet(text):
    if text is None:
        return False

    text = str(text).strip()

    if not text:
        return False

    for pattern in exclusion_patterns:
        if pattern.search(text):
            return False

    for pattern in anchor_patterns:
        if pattern.search(text):
            return True

    return False


train_positive_users = set(
    train_df.loc[
        train_df["label"] == 1,
        "username"
    ].astype(str)
)

val_positive_users = set(
    val_df.loc[
        val_df["label"] == 1,
        "username"
    ].astype(str)
)

test_positive_users = set(
    test_df.loc[
        test_df["label"] == 1,
        "username"
    ].astype(str)
)

all_positive_users = (
    train_positive_users
    | val_positive_users
    | test_positive_users
)

print()
print("=" * 60)
print("POSITIVE USERS BY SPLIT")
print("=" * 60)

print("TRAIN:", len(train_positive_users))
print("VALIDATION:", len(val_positive_users))
print("TEST:", len(test_positive_users))

assert len(train_positive_users) == 2013
assert len(val_positive_users) == 252
assert len(test_positive_users) == 251
assert len(all_positive_users) == 2516


timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:
    filename = os.path.basename(timeline_file)
    username = filename.split("created_at")[0]

    timeline_files_by_user[
        username
    ].append(timeline_file)

assert len(timeline_files_by_user) == 2516


anchor_records = []
users_processed = set()
tweets_scanned = 0

for username in sorted(all_positive_users):

    if username not in timeline_files_by_user:
        raise KeyError(
            f"Missing D1 timeline for user: {username}"
        )

    users_processed.add(username)

    for timeline_file in timeline_files_by_user[username]:

        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:
                line = line.strip()

                if not line:
                    continue

                record = json.loads(line)
                tweets_scanned += 1

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get("id", "")
                    )
                )

                tweet_text = str(
                    record.get("text", "")
                )

                if (
                    tweet_id
                    and
                    is_refined_anchor_tweet(
                        tweet_text
                    )
                ):

                    anchor_records.append({
                        "username": username,
                        "tweet_id": tweet_id,
                        "created_at":
                            record.get("created_at"),
                        "text": tweet_text
                    })


anchor_pairs = {
    (
        record["username"],
        record["tweet_id"]
    )
    for record in anchor_records
}

train_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in train_positive_users
}

val_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in val_positive_users
}

test_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in test_positive_users
}


print()
print("=" * 60)
print("EXACT ANCHOR SUMMARY")
print("=" * 60)

print(
    "Positive users processed:",
    len(users_processed)
)

print(
    "Tweets scanned:",
    tweets_scanned
)

print(
    "Raw anchor records:",
    len(anchor_records)
)

print(
    "Unique anchor pairs:",
    len(anchor_pairs)
)

print()
print(
    "TRAIN unique anchors:",
    len(train_anchor_pairs)
)

print(
    "VAL unique anchors:",
    len(val_anchor_pairs)
)

print(
    "TEST unique anchors:",
    len(test_anchor_pairs)
)

assert len(users_processed) == 2516
assert tweets_scanned == 508806
assert len(anchor_records) == 1920
assert len(anchor_pairs) == 1847
assert len(train_anchor_pairs) == 1476
assert len(val_anchor_pairs) == 190
assert len(test_anchor_pairs) == 181


# Save anchor records
sentencebert_emotion_anchor_df = pd.DataFrame(
    anchor_records
)

sentencebert_emotion_anchor_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "exact_anchor_records.csv"
    ),
    index=False
)

pd.DataFrame(
    list(anchor_pairs),
    columns=["username", "tweet_id"]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "unique_anchor_pairs.csv"
    ),
    index=False
)


print()
print("=" * 60)
print("CELL 3 COMPLETE")
print("=" * 60)

print("✓ Exact anchor detector reproduced")
print("✓ 1,847 unique anchors confirmed")
print("✓ 1,920 anchor records confirmed")
print("✓ TRAIN = 1,476 unique anchors")
print("✓ VAL = 190 unique anchors")
print("✓ TEST = 181 unique anchors")
print("✓ No D2 users included")
print("✓ Anchor records saved")
print("✓ No timelines modified yet")
```

```python
# ============================================================
# CELL 4 — SENTENCE-BERT + EMOTION-BASED
# ============================================================

print("=" * 60)
print("SENTENCE-BERT + EMOTION-BASED — EXACT D1 ANCHOR REMOVAL")
print("=" * 60)

EXACT_ANCHOR_SET = set(anchor_pairs)

train_anchor_records = [
    r for r in anchor_records
    if r["username"] in train_positive_users
]

val_anchor_records = [
    r for r in anchor_records
    if r["username"] in val_positive_users
]

test_anchor_records = [
    r for r in anchor_records
    if r["username"] in test_positive_users
]

print()
print("=" * 60)
print("PRE-REMOVAL ANCHOR VERIFICATION")
print("=" * 60)

duplicate_anchor_occurrences = (
    len(anchor_records) - len(anchor_pairs)
)

print("Raw anchor records:", len(anchor_records))
print("Unique anchor pairs:", len(anchor_pairs))
print(
    "Duplicate anchor occurrences:",
    duplicate_anchor_occurrences
)

assert len(anchor_records) == 1920
assert len(anchor_pairs) == 1847
assert duplicate_anchor_occurrences == 73

print("✓ 1,847 unique anchors verified")
print("✓ 73 duplicate occurrences verified")
print("✓ 1,920 total anchor records verified")

print()
print("=" * 60)
print("SPLIT-LEVEL ANCHOR RECORD COUNTS")
print("=" * 60)

print("TRAIN:", len(train_anchor_records), "| expected: 1535")
print("VAL:", len(val_anchor_records), "| expected: 199")
print("TEST:", len(test_anchor_records), "| expected: 186")

assert len(train_anchor_records) == 1535
assert len(val_anchor_records) == 199
assert len(test_anchor_records) == 186

# Timeline index
timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:
    filename = os.path.basename(timeline_file)
    username = filename.split("created_at")[0]
    timeline_files_by_user[username].append(
        timeline_file
    )

assert len(timeline_files_by_user) == 2516


def build_anchor_removed_text(username):
    kept_texts = []
    removed_count = 0

    for timeline_file in timeline_files_by_user[username]:
        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:
                line = line.strip()

                if not line:
                    continue

                record = json.loads(line)

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get("id", "")
                    )
                )

                if (
                    str(username),
                    tweet_id
                ) in EXACT_ANCHOR_SET:

                    removed_count += 1
                    continue

                tweet_text = str(
                    record.get("text", "")
                ).strip()

                if tweet_text:
                    kept_texts.append(tweet_text)

    return " ".join(kept_texts), removed_count


# Copy source data
train_sentencebert_emotion_df = train_df.copy(deep=True)
val_sentencebert_emotion_df = val_df.copy(deep=True)
test_sentencebert_emotion_df = test_df.copy(deep=True)

positive_anchor_removed_text = {}

train_removed = 0
val_removed = 0
test_removed = 0

for username in sorted(all_positive_users):

    text, removed = build_anchor_removed_text(username)

    positive_anchor_removed_text[username] = text

    if username in train_positive_users:
        train_removed += removed
    elif username in val_positive_users:
        val_removed += removed
    elif username in test_positive_users:
        test_removed += removed

print()
print("=" * 60)
print("FINAL ANCHOR REMOVAL COUNTS")
print("=" * 60)

print("TRAIN removed:", train_removed)
print("VAL removed:", val_removed)
print("TEST removed:", test_removed)
print(
    "TOTAL removed:",
    train_removed + val_removed + test_removed
)

assert train_removed == 1535
assert val_removed == 199
assert test_removed == 186
assert (
    train_removed +
    val_removed +
    test_removed
) == 1920

print(
    "✓ Record-level removal exactly matches previous experiments"
)

# Assign anchor-removed text
for df in [
    train_sentencebert_emotion_df,
    val_sentencebert_emotion_df,
    test_sentencebert_emotion_df
]:

    positive_mask = df["label"] == 1

    df.loc[
        positive_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            positive_mask,
            "username"
        ]
        .astype(str)
        .map(positive_anchor_removed_text)
        .fillna("")
        .values
    )

    negative_mask = df["label"] == 0

    df.loc[
        negative_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            negative_mask,
            "timeline_text"
        ]
        .fillna("")
        .astype(str)
        .values
    )


print()
print("=" * 60)
print("D2 PRESERVATION CHECK")
print("=" * 60)

for original_df, result_df, name in [
    (train_df, train_sentencebert_emotion_df, "TRAIN"),
    (val_df, val_sentencebert_emotion_df, "VALIDATION"),
    (test_df, test_sentencebert_emotion_df, "TEST")
]:

    original_d2 = (
        original_df.loc[
            original_df["label"] == 0,
            ["username", "timeline_text"]
        ]
        .set_index("username")
        .sort_index()
    )

    result_d2 = (
        result_df.loc[
            result_df["label"] == 0,
            ["username", "anchor_removed_text"]
        ]
        .rename(
            columns={
                "anchor_removed_text":
                    "timeline_text"
            }
        )
        .set_index("username")
        .sort_index()
    )

    assert original_d2.equals(result_d2)
    print(f"✓ {name} D2 unchanged")


print()
print("=" * 60)
print("USER + LABEL PRESERVATION")
print("=" * 60)

for original_df, result_df, name in [
    (train_df, train_sentencebert_emotion_df, "TRAIN"),
    (val_df, val_sentencebert_emotion_df, "VALIDATION"),
    (test_df, test_sentencebert_emotion_df, "TEST")
]:

    assert (
        original_df["user_id"].tolist()
        ==
        result_df["user_id"].tolist()
    )

    assert (
        original_df["username"].tolist()
        ==
        result_df["username"].tolist()
    )

    assert (
        original_df["label"].tolist()
        ==
        result_df["label"].tolist()
    )

    print(
        f"✓ {name}: users and labels unchanged"
    )


print()
print("=" * 60)
print("ANCHOR-REMOVED TEXT INTEGRITY")
print("=" * 60)

for name, df in [
    ("TRAIN", train_sentencebert_emotion_df),
    ("VALIDATION", val_sentencebert_emotion_df),
    ("TEST", test_sentencebert_emotion_df)
]:

    empty_count = int(
        (
            df["anchor_removed_text"]
            .fillna("")
            .astype(str)
            .str.strip()
            .eq("")
        ).sum()
    )

    print(
        f"{name}: empty anchor-removed texts = "
        f"{empty_count}"
    )


train_sentencebert_emotion_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "train_anchor_removed.pkl"
    )
)

val_sentencebert_emotion_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "val_anchor_removed.pkl"
    )
)

test_sentencebert_emotion_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "test_anchor_removed.pkl"
    )
)

print()
print("=" * 60)
print("CELL 4 COMPLETE")
print("=" * 60)

print("✓ 1,847 unique D1 anchors removed")
print("✓ 73 duplicate occurrences handled")
print("✓ 1,920 total anchor records removed")
print("✓ TRAIN = 1,535 removed")
print("✓ VAL = 199 removed")
print("✓ TEST = 186 removed")
print("✓ D2 completely preserved")
print("✓ User IDs preserved")
print("✓ Labels preserved")
print("✓ Anchor-removed source saved")
```

```python
# ============================================================
# CELL 5 — SENTENCE-BERT + EMOTION-BASED
# ============================================================

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)
from kaggle_secrets import UserSecretsClient
import torch.nn.functional as F
from tqdm.auto import tqdm

print("=" * 60)
print("SENTENCE-BERT + EMOTION-BASED — EMOTION EXTRACTION")
print("=" * 60)

EMOTION_MODEL_NAME = (
    "cardiffnlp/twitter-roberta-base-emotion-latest"
)

user_secrets = UserSecretsClient()
hf_token = user_secrets.get_secret("HF_TOKEN")

print()
print("Emotion model:", EMOTION_MODEL_NAME)
print("Emotion features:", len(EMOTION_LABELS))
print("✓ HF_TOKEN retrieved")

emotion_tokenizer = AutoTokenizer.from_pretrained(
    EMOTION_MODEL_NAME,
    token=hf_token
)

emotion_model = (
    AutoModelForSequenceClassification
    .from_pretrained(
        EMOTION_MODEL_NAME,
        token=hf_token
    )
    .to(device)
)

emotion_model.eval()

for p in emotion_model.parameters():
    p.requires_grad = False

print("✓ Emotion tokenizer loaded")
print("✓ Emotion model loaded")
print("✓ Emotion model frozen")

model_labels = [
    emotion_model.config.id2label[i]
    for i in range(
        emotion_model.config.num_labels
    )
]

print()
print("Model emotion labels:")
print(model_labels)

assert emotion_model.config.num_labels == 11
assert model_labels == EMOTION_LABELS


def extract_emotion_features(
    texts,
    split_name,
    batch_size=32
):
    features = []
    empty_count = 0
    nonempty_indices = []

    for i, text in enumerate(texts):
        if str(text).strip() == "":
            empty_count += 1
        else:
            nonempty_indices.append(i)

    output = np.zeros(
        (len(texts), 11),
        dtype=np.float32
    )

    print()
    print(f"Processing {split_name}...")
    print("Users:", len(texts))
    print("Empty texts:", empty_count)
    print("Non-empty texts:", len(nonempty_indices))

    total_batches = (
        len(nonempty_indices) + batch_size - 1
    ) // batch_size

    with torch.no_grad():

        for start in tqdm(
            range(
                0,
                len(nonempty_indices),
                batch_size
            ),
            desc=split_name
        ):

            batch_indices = nonempty_indices[
                start:start + batch_size
            ]

            batch_texts = [
                str(texts[i])
                for i in batch_indices
            ]

            encoded = emotion_tokenizer(
                batch_texts,
                padding=True,
                truncation=True,
                max_length=512,
                return_tensors="pt"
            )

            encoded = {
                k: v.to(device)
                for k, v in encoded.items()
            }

            logits = emotion_model(
                **encoded
            ).logits.float()

            # CardiffNLP emotion model is multilabel.
            probabilities = torch.sigmoid(logits)

            output[
                batch_indices
            ] = probabilities.cpu().numpy()

    assert output.shape == (
        len(texts),
        11
    )

    assert np.isfinite(output).all()
    assert output.min() >= 0.0
    assert output.max() <= 1.0

    return output


train_emotion_features = extract_emotion_features(
    train_sentencebert_emotion_df[
        "anchor_removed_text"
    ].fillna("").tolist(),
    "TRAIN"
)

val_emotion_features = extract_emotion_features(
    val_sentencebert_emotion_df[
        "anchor_removed_text"
    ].fillna("").tolist(),
    "VALIDATION"
)

test_emotion_features = extract_emotion_features(
    test_sentencebert_emotion_df[
        "anchor_removed_text"
    ].fillna("").tolist(),
    "TEST"
)

print()
print("=" * 60)
print("EMOTION FEATURE SHAPES")
print("=" * 60)

print("Train:", train_emotion_features.shape)
print("Validation:", val_emotion_features.shape)
print("Test:", test_emotion_features.shape)

assert train_emotion_features.shape == (6135, 11)
assert val_emotion_features.shape == (767, 11)
assert test_emotion_features.shape == (767, 11)

print("✓ Exact 11 emotion features confirmed")

print()
print("=" * 60)
print("EMOTION FEATURE INTEGRITY")
print("=" * 60)

for features, name in [
    (train_emotion_features, "TRAIN"),
    (val_emotion_features, "VALIDATION"),
    (test_emotion_features, "TEST")
]:
    print(
        f"{name}: "
        f"NaN={np.isnan(features).sum()}, "
        f"Inf={np.isinf(features).sum()}"
    )

    print(
        f"{name}: "
        f"min={features.min():.6f}, "
        f"max={features.max():.6f}, "
        f"mean={features.mean():.6f}"
    )

    assert np.isnan(features).sum() == 0
    assert np.isinf(features).sum() == 0

emotion_columns = [
    f"emotion_{label}"
    for label in EMOTION_LABELS
]

print()
print("=" * 60)
print("EMOTION FEATURE COLUMNS")
print("=" * 60)

for i, column in enumerate(
    emotion_columns,
    start=1
):
    print(
        f"{i:02d}. {column}"
    )

# Save raw emotion features
np.save(
    os.path.join(
        RESULT_DIR,
        "train_emotion_features.npy"
    ),
    train_emotion_features
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_emotion_features.npy"
    ),
    val_emotion_features
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_emotion_features.npy"
    ),
    test_emotion_features
)

pd.DataFrame(
    train_emotion_features,
    columns=emotion_columns
).to_csv(
    os.path.join(
        RESULT_DIR,
        "train_emotion_features.csv"
    ),
    index=False
)

pd.DataFrame(
    val_emotion_features,
    columns=emotion_columns
).to_csv(
    os.path.join(
        RESULT_DIR,
        "val_emotion_features.csv"
    ),
    index=False
)

pd.DataFrame(
    test_emotion_features,
    columns=emotion_columns
).to_csv(
    os.path.join(
        RESULT_DIR,
        "test_emotion_features.csv"
    ),
    index=False
)

print()
print("=" * 60)
print("CELL 5 COMPLETE")
print("=" * 60)

print("✓ Emotion model applied to anchor-removed text")
print("✓ Multilabel sigmoid probabilities used")
print("✓ 11 emotion features extracted")
print("✓ Train = 6135 × 11")
print("✓ Validation = 767 × 11")
print("✓ Test = 767 × 11")
print("✓ No NaN / Inf")
print("✓ Emotion features saved")
```

```python
# ============================================================
# CELL 7A — SENTENCE-BERT + EMOTION-BASED
# ============================================================

from transformers import AutoTokenizer
from tqdm.auto import tqdm

print("=" * 60)
print("SENTENCE-BERT + EMOTION-BASED — TOKENIZATION")
print("=" * 60)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    token=hf_token
)

print()
print("Tokenizer:", MODEL_NAME)
print("Max length:", MAX_LENGTH)

def tokenize_texts(
    texts,
    split_name,
    batch_size=32
):

    all_input_ids = []
    all_attention_masks = []

    print()
    print(f"Processing {split_name}...")
    print("Users:", len(texts))

    for start in tqdm(
        range(
            0,
            len(texts),
            batch_size
        ),
        desc=split_name
    ):

        batch_texts = [
            str(text)
            for text in texts[
                start:start + batch_size
            ]
        ]

        encoded = tokenizer(
            batch_texts,
            padding="max_length",
            truncation=True,
            max_length=MAX_LENGTH,
            return_attention_mask=True
        )

        all_input_ids.extend(
            encoded["input_ids"]
        )

        all_attention_masks.extend(
            encoded["attention_mask"]
        )

    input_ids = np.asarray(
        all_input_ids,
        dtype=np.int64
    )

    attention_masks = np.asarray(
        all_attention_masks,
        dtype=np.int64
    )

    return input_ids, attention_masks


train_input_ids, train_attention_masks = (
    tokenize_texts(
        train_sentencebert_emotion_df[
            "anchor_removed_text"
        ].fillna("").tolist(),
        "TRAIN"
    )
)

val_input_ids, val_attention_masks = (
    tokenize_texts(
        val_sentencebert_emotion_df[
            "anchor_removed_text"
        ].fillna("").tolist(),
        "VALIDATION"
    )
)

test_input_ids, test_attention_masks = (
    tokenize_texts(
        test_sentencebert_emotion_df[
            "anchor_removed_text"
        ].fillna("").tolist(),
        "TEST"
    )
)

print()
print("=" * 60)
print("TOKENIZATION SHAPES")
print("=" * 60)

print(
    "TRAIN input IDs:",
    train_input_ids.shape
)

print(
    "TRAIN attention masks:",
    train_attention_masks.shape
)

print(
    "VALIDATION input IDs:",
    val_input_ids.shape
)

print(
    "VALIDATION attention masks:",
    val_attention_masks.shape
)

print(
    "TEST input IDs:",
    test_input_ids.shape
)

print(
    "TEST attention masks:",
    test_attention_masks.shape
)

assert train_input_ids.shape == (
    6135,
    512
)

assert train_attention_masks.shape == (
    6135,
    512
)

assert val_input_ids.shape == (
    767,
    512
)

assert val_attention_masks.shape == (
    767,
    512
)

assert test_input_ids.shape == (
    767,
    512
)

assert test_attention_masks.shape == (
    767,
    512
)

print()
print("=" * 60)
print("TOKEN VALUE INTEGRITY")
print("=" * 60)

for array, name in [
    (train_input_ids, "TRAIN input IDs"),
    (train_attention_masks, "TRAIN attention masks"),
    (val_input_ids, "VAL input IDs"),
    (val_attention_masks, "VAL attention masks"),
    (test_input_ids, "TEST input IDs"),
    (test_attention_masks, "TEST attention masks")
]:

    assert np.isfinite(array).all()

    print(
        f"✓ {name}: finite values only"
    )

print()
print("=" * 60)
print("ACTIVE TOKEN CHECK")
print("=" * 60)

print(
    "TRAIN mean active tokens:",
    float(train_attention_masks.sum(axis=1).mean())
)

print(
    "VALIDATION mean active tokens:",
    float(val_attention_masks.sum(axis=1).mean())
)

print(
    "TEST mean active tokens:",
    float(test_attention_masks.sum(axis=1).mean())
)

print()
print("=" * 60)
print("CELL 7A COMPLETE")
print("=" * 60)

print("✓ Sentence-BERT tokenizer loaded")
print("✓ Max length = 512")
print("✓ Truncation enabled")
print("✓ Padding = max_length")
print("✓ TRAIN token inputs created")
print("✓ VALIDATION token inputs created")
print("✓ TEST token inputs created")
print("✓ All token shapes verified")
```

```python
# ============================================================
# CELL 9 — SENTENCE-BERT + EMOTION-BASED
# ============================================================

from torch.utils.data import Dataset, DataLoader

print("=" * 60)
print("SENTENCE-BERT + EMOTION-BASED — BUILDING PYTORCH DATASETS")
print("=" * 60)

# Map original TRAIN users to their row positions
user_to_index = {
    user_id: i
    for i, user_id in enumerate(
        train_sentencebert_emotion_df["user_id"]
    )
}

balanced_indices = np.array(
    [
        user_to_index[user_id]
        for user_id in
        balanced_train_sentencebert_emotion_df["user_id"]
    ],
    dtype=np.int64
)

# Balanced TRAIN inputs
train_input_ids_balanced = train_input_ids[
    balanced_indices
]

train_attention_masks_balanced = (
    train_attention_masks[
        balanced_indices
    ]
)

train_metadata_balanced = (
    train_sentencebert_emotion_metadata_scaled[
        balanced_indices
    ]
)

train_emotion_balanced = (
    train_emotion_scaled[
        balanced_indices
    ]
)

train_labels_balanced = (
    balanced_train_sentencebert_emotion_df["label"]
    .to_numpy(dtype=np.int64)
)

# Validation labels
val_labels = (
    val_sentencebert_emotion_df["label"]
    .to_numpy(dtype=np.int64)
)

# Test labels
test_labels = (
    test_sentencebert_emotion_df["label"]
    .to_numpy(dtype=np.int64)
)


class EmotionSentenceBERTDataset(Dataset):

    def __init__(
        self,
        input_ids,
        attention_masks,
        metadata,
        emotions,
        labels
    ):
        self.input_ids = torch.tensor(
            input_ids,
            dtype=torch.long
        )

        self.attention_masks = torch.tensor(
            attention_masks,
            dtype=torch.long
        )

        self.metadata = torch.tensor(
            metadata,
            dtype=torch.float32
        )

        self.emotions = torch.tensor(
            emotions,
            dtype=torch.float32
        )

        self.labels = torch.tensor(
            labels,
            dtype=torch.long
        )

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, index):
        return {
            "input_ids": self.input_ids[index],
            "attention_mask": self.attention_masks[index],
            "metadata": self.metadata[index],
            "emotions": self.emotions[index],
            "labels": self.labels[index]
        }


train_dataset = EmotionSentenceBERTDataset(
    train_input_ids_balanced,
    train_attention_masks_balanced,
    train_metadata_balanced,
    train_emotion_balanced,
    train_labels_balanced
)

val_dataset = EmotionSentenceBERTDataset(
    val_input_ids,
    val_attention_masks,
    val_sentencebert_emotion_metadata_scaled,
    val_emotion_scaled,
    val_labels
)

test_dataset = EmotionSentenceBERTDataset(
    test_input_ids,
    test_attention_masks,
    test_sentencebert_emotion_metadata_scaled,
    test_emotion_scaled,
    test_labels
)


BATCH_SIZE = 16

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)


print()
print("=" * 60)
print("DATASET SIZES")
print("=" * 60)

print("Balanced Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))

assert len(train_dataset) == 4026
assert len(val_dataset) == 767
assert len(test_dataset) == 767


print()
print("=" * 60)
print("DATALOADER CONFIGURATION")
print("=" * 60)

print("Batch size:", BATCH_SIZE)
print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))

assert len(train_loader) == 252
assert len(val_loader) == 48
assert len(test_loader) == 48


# Batch sanity check
batch = next(iter(train_loader))

print()
print("=" * 60)
print("BATCH SANITY CHECK")
print("=" * 60)

print(
    "Input IDs:",
    tuple(batch["input_ids"].shape)
)

print(
    "Attention mask:",
    tuple(batch["attention_mask"].shape)
)

print(
    "Metadata:",
    tuple(batch["metadata"].shape)
)

print(
    "Emotions:",
    tuple(batch["emotions"].shape)
)

print(
    "Labels:",
    tuple(batch["labels"].shape)
)

assert (
    tuple(batch["input_ids"].shape)
    == (16, 512)
)

assert (
    tuple(batch["attention_mask"].shape)
    == (16, 512)
)

assert (
    tuple(batch["metadata"].shape)
    == (16, 53)
)

assert (
    tuple(batch["emotions"].shape)
    == (16, 11)
)

assert (
    tuple(batch["labels"].shape)
    == (16,)
)


print()
print("=" * 60)
print("BATCH DATA TYPES")
print("=" * 60)

print(
    "Input IDs:",
    batch["input_ids"].dtype
)

print(
    "Attention mask:",
    batch["attention_mask"].dtype
)

print(
    "Metadata:",
    batch["metadata"].dtype
)

print(
    "Emotions:",
    batch["emotions"].dtype
)

print(
    "Labels:",
    batch["labels"].dtype
)

assert batch["input_ids"].dtype == torch.int64
assert batch["attention_mask"].dtype == torch.int64
assert batch["metadata"].dtype == torch.float32
assert batch["emotions"].dtype == torch.float32
assert batch["labels"].dtype == torch.int64


print()
print("=" * 60)
print("TRAINING LABEL BALANCE")
print("=" * 60)

balanced_distribution = np.bincount(
    train_labels_balanced,
    minlength=2
)

print(
    "Balanced TRAIN:",
    balanced_distribution
)

assert np.array_equal(
    balanced_distribution,
    np.array([2013, 2013])
)


print()
print("=" * 60)
print("USER ALIGNMENT")
print("=" * 60)

assert (
    len(
        balanced_train_sentencebert_emotion_df
    )
    ==
    len(train_dataset)
)

assert (
    len(train_emotion_balanced)
    ==
    len(train_labels_balanced)
)

print(
    "Balanced training users:",
    len(balanced_train_sentencebert_emotion_df)
)

print("✓ User alignment verified")


dataset_info = {
    "train_users": 4026,
    "validation_users": 767,
    "test_users": 767,
    "batch_size": 16,
    "train_batches": 252,
    "validation_batches": 48,
    "test_batches": 48,
    "context_length": 512,
    "metadata_features": 53,
    "emotion_features": 11
}

with open(
    os.path.join(
        RESULT_DIR,
        "dataset_info.json"
    ),
    "w"
) as f:
    json.dump(
        dataset_info,
        f,
        indent=2
    )


print()
print("=" * 60)
print("CELL 9 COMPLETE")
print("=" * 60)

print("✓ Balanced TRAIN Dataset created")
print("✓ Validation Dataset unchanged")
print("✓ Test Dataset unchanged")
print("✓ TRAIN = 4,026 users")
print("✓ TRAIN = 2,013 negative + 2,013 positive")
print("✓ Batch size = 16")
print("✓ Training shuffle enabled")
print("✓ Validation shuffle disabled")
print("✓ Test shuffle disabled")
print("✓ 512-token inputs aligned")
print("✓ 53 metadata features aligned")
print("✓ 11 emotion features aligned")
print("✓ Labels aligned")
print("✓ DataLoader sanity check passed")
```

```python
# ============================================================
# CELL 10 — SENTENCE-BERT + EMOTION-BASED
# ============================================================

from transformers import AutoModel
import torch.nn as nn

print("=" * 60)
print("SENTENCE-BERT + EMOTION-BASED — BUILDING FROZEN SENTENCE-BERT + METADATA + EMOTION MODEL")
print("=" * 60)

sentencebert_encoder = AutoModel.from_pretrained(
    MODEL_NAME,
    token=hf_token
).to(device)

for param in sentencebert_encoder.parameters():
    param.requires_grad = False


class SentenceBERTEmotionClassifier(nn.Module):

    def __init__(
        self,
        encoder,
        metadata_dim=53,
        emotion_dim=11
    ):
        super().__init__()

        self.encoder = encoder
        self.text_hidden_size = 768

        # Text attention pooling
        self.attention_projection = nn.Linear(
            768,
            256
        )

        self.attention_score = nn.Linear(
            256,
            1
        )

        # Metadata branch
        self.metadata_branch = nn.Sequential(
            nn.Linear(
                metadata_dim,
                128
            ),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(
                128,
                64
            ),
            nn.ReLU()
        )

        # Emotion branch
        self.emotion_branch = nn.Sequential(
            nn.Linear(
                emotion_dim,
                64
            ),
            nn.ReLU(),
            nn.Dropout(0.3)
        )

        # Learnable gates
        self.metadata_beta = nn.Parameter(
            torch.tensor(0.0)
        )

        self.emotion_beta = nn.Parameter(
            torch.tensor(0.0)
        )

        # 768 text + 64 metadata + 64 emotion
        self.classifier = nn.Sequential(
            nn.Linear(
                896,
                512
            ),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(
                512,
                256
            ),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(
                256,
                2
            )
        )

    def forward(
        self,
        input_ids,
        attention_mask,
        metadata,
        emotions
    ):

        # Frozen Sentence-BERT
        with torch.no_grad():

            outputs = self.encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

        hidden_states = outputs.last_hidden_state

        # Attention pooling
        projected = torch.tanh(
            self.attention_projection(
                hidden_states
            )
        )

        scores = self.attention_score(
            projected
        ).squeeze(-1)

        scores = scores.masked_fill(
            attention_mask == 0,
            -1e9
        )

        attention_weights = torch.softmax(
            scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # Metadata
        metadata_representation = (
            self.metadata_branch(metadata)
        )

        metadata_gate = torch.sigmoid(
            self.metadata_beta
        )

        gated_metadata = (
            metadata_gate *
            metadata_representation
        )

        # Emotion
        emotion_representation = (
            self.emotion_branch(emotions)
        )

        emotion_gate = torch.sigmoid(
            self.emotion_beta
        )

        gated_emotion = (
            emotion_gate *
            emotion_representation
        )

        # Fusion
        fused = torch.cat(
            [
                text_representation,
                gated_metadata,
                gated_emotion
            ],
            dim=1
        )

        logits = self.classifier(
            fused
        )

        return {
            "logits": logits,
            "text_representation":
                text_representation,
            "metadata_representation":
                metadata_representation,
            "emotion_representation":
                emotion_representation,
            "attention_weights":
                attention_weights,
            "metadata_gate":
                metadata_gate,
            "emotion_gate":
                emotion_gate
        }


model = SentenceBERTEmotionClassifier(
    sentencebert_encoder,
    metadata_dim=53,
    emotion_dim=11
).to(device)


# ============================================================
# PARAMETER COUNT
# ============================================================

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

frozen_params = sum(
    p.numel()
    for p in model.parameters()
    if not p.requires_grad
)

print()
print("=" * 60)
print("PARAMETER COUNT")
print("=" * 60)

print(
    "Total parameters:",
    total_params
)

print(
    "Trainable parameters:",
    trainable_params
)

print(
    "Frozen parameters:",
    frozen_params
)

print(
    "Total parameters (M):",
    round(
        total_params / 1e6,
        3
    )
)

print(
    "Trainable parameters (M):",
    round(
        trainable_params / 1e6,
        3
    )
)

print(
    "Frozen parameters (M):",
    round(
        frozen_params / 1e6,
        3
    )
)

encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print(
    "✓ All Sentence-BERT backbone parameters frozen"
)


# ============================================================
# TRAINABLE COMPONENTS
# ============================================================

print()
print("=" * 60)
print("TRAINABLE COMPONENTS")
print("=" * 60)

for name, param in model.named_parameters():

    if param.requires_grad:
        print(
            "✓",
            name
        )


# ============================================================
# ARCHITECTURE
# ============================================================

print()
print("=" * 60)
print("ARCHITECTURE CONFIGURATION")
print("=" * 60)

print(
    "Transformer:",
    MODEL_NAME
)

print(
    "Base architecture: Sentence-BERT (MPNet-base)"
)

print(
    "Text hidden size: 768"
)

print(
    "Attention projection: 256"
)

print(
    "Metadata: 53 -> 128 -> 64"
)

print(
    "Emotion: 11 -> 64"
)

print(
    "Metadata gate: g_m = sigmoid(beta_m)"
)

print(
    "Emotion gate: g_e = sigmoid(beta_e)"
)

print(
    "Initial metadata beta: 0.0"
)

print(
    "Initial emotion beta: 0.0"
)

print(
    "Initial metadata gate: 0.5"
)

print(
    "Initial emotion gate: 0.5"
)

print(
    "Fusion: 768 + 64 + 64 = 896"
)

print(
    "Classifier: 896 -> 512 -> 256 -> 2"
)

print(
    "Dropout: 0.3"
)

print(
    "Transformer frozen: True"
)


# ============================================================
# FORWARD PASS SANITY CHECK
# ============================================================

print()
print("=" * 60)
print("FORWARD PASS SANITY CHECK")
print("=" * 60)

model.eval()

batch = next(
    iter(train_loader)
)

input_ids = batch[
    "input_ids"
].to(device)

attention_mask = batch[
    "attention_mask"
].to(device)

metadata = batch[
    "metadata"
].to(device)

emotions = batch[
    "emotions"
].to(device)

with torch.no_grad():

    output = model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        metadata=metadata,
        emotions=emotions
    )

print(
    "Input IDs:",
    tuple(
        input_ids.shape
    )
)

print(
    "Attention mask:",
    tuple(
        attention_mask.shape
    )
)

print(
    "Metadata:",
    tuple(
        metadata.shape
    )
)

print(
    "Emotions:",
    tuple(
        emotions.shape
    )
)

print(
    "Logits:",
    tuple(
        output["logits"].shape
    )
)

print(
    "Text representation:",
    tuple(
        output[
            "text_representation"
        ].shape
    )
)

print(
    "Metadata representation:",
    tuple(
        output[
            "metadata_representation"
        ].shape
    )
)

print(
    "Emotion representation:",
    tuple(
        output[
            "emotion_representation"
        ].shape
    )
)

print(
    "Attention weights:",
    tuple(
        output[
            "attention_weights"
        ].shape
    )
)

metadata_gate = float(
    output[
        "metadata_gate"
    ].item()
)

emotion_gate = float(
    output[
        "emotion_gate"
    ].item()
)

print(
    "Metadata gate:",
    metadata_gate
)

print(
    "Emotion gate:",
    emotion_gate
)

padding_attention = output[
    "attention_weights"
][attention_mask == 0]

max_padding_attention = (
    float(
        padding_attention.max().item()
    )
    if padding_attention.numel()
    else 0.0
)

print(
    "Maximum attention on padding:",
    max_padding_attention
)

assert output[
    "logits"
].shape == (16, 2)

assert output[
    "text_representation"
].shape == (16, 768)

assert output[
    "metadata_representation"
].shape == (16, 64)

assert output[
    "emotion_representation"
].shape == (16, 64)

assert output[
    "attention_weights"
].shape == (16, 512)

assert abs(
    metadata_gate - 0.5
) < 1e-6

assert abs(
    emotion_gate - 0.5
) < 1e-6

assert max_padding_attention == 0.0

assert encoder_trainable == 0

print()
print(
    "✓ Transformer freeze verified"
)

print(
    "✓ Metadata gate verified"
)

print(
    "✓ Emotion gate verified"
)


# ============================================================
# SAVE CONFIGURATION
# ============================================================

architecture_config = {
    "model_name": MODEL_NAME,
    "base_architecture": "Sentence-BERT (MPNet-base)",
    "text_hidden_size": 768,
    "attention_projection": 256,
    "metadata_branch": "53 -> 128 -> 64",
    "emotion_branch": "11 -> 64",
    "metadata_gate": "sigmoid(metadata_beta)",
    "emotion_gate": "sigmoid(emotion_beta)",
    "initial_metadata_beta": 0.0,
    "initial_emotion_beta": 0.0,
    "initial_metadata_gate": 0.5,
    "initial_emotion_gate": 0.5,
    "fusion": "768 + 64 + 64 = 896",
    "classifier": "896 -> 512 -> 256 -> 2",
    "dropout": 0.3,
    "transformer_frozen": True,
    "total_parameters": total_params,
    "trainable_parameters": trainable_params,
    "frozen_parameters": frozen_params
}

with open(
    os.path.join(
        RESULT_DIR,
        "architecture_config.json"
    ),
    "w"
) as f:

    json.dump(
        architecture_config,
        f,
        indent=2
    )


print()
print("=" * 60)
print("CELL 10 COMPLETE")
print("=" * 60)

print(
    "✓ Sentence-BERT loaded"
)

print(
    "✓ Transformer backbone frozen"
)

print(
    "✓ Attention pooling active"
)

print(
    "✓ Metadata branch = 53 -> 128 -> 64"
)

print(
    "✓ Emotion branch = 11 -> 64"
)

print(
    "✓ Metadata gate active"
)

print(
    "✓ Emotion gate active"
)

print(
    "✓ Text representation = 768"
)

print(
    "✓ Fusion = 768 + 64 + 64 = 896"
)

print(
    "✓ Classifier = 896 -> 512 -> 256 -> 2"
)

print(
    "✓ Padding attention masked"
)

print(
    "✓ Forward pass successful"
)

print(
    "✓ Frozen Sentence-BERT + metadata + emotion model ready"
)
```

```python
# ============================================================
# CELL 11 — SENTENCE-BERT + EMOTION-BASED
# ============================================================

import os
import time
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

MAX_EPOCHS = 10
EARLY_STOPPING_PATIENCE = 2
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1

print("=" * 60)
print("STARTING SENTENCE-BERT + EMOTION-BASED TRAINING")
print("=" * 60)

print()
print("Training users:", len(train_dataset))
print("Validation users:", len(val_dataset))
print("Test users:", len(test_dataset))
print("Maximum epochs:", MAX_EPOCHS)
print("Early stopping patience:", EARLY_STOPPING_PATIENCE)
print("Checkpoint metric: Validation F1")
print("Learning rate:", LEARNING_RATE)
print("Weight decay:", WEIGHT_DECAY)
print("Batch size:", BATCH_SIZE)
print("Transformer: FROZEN")

# Verify Sentence-BERT remains frozen
encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print("✓ Sentence-BERT backbone confirmed frozen")

trainable_params = [
    p for p in model.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_params,
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

total_steps = (
    len(train_loader)
    * MAX_EPOCHS
)

warmup_steps = int(
    total_steps * WARMUP_RATIO
)


def lr_lambda(step):

    if step < warmup_steps:
        return (
            (step + 1)
            / max(1, warmup_steps)
        )

    return max(
        0.0,
        (total_steps - step)
        / max(
            1,
            total_steps - warmup_steps
        )
    )


scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda
)

criterion = nn.CrossEntropyLoss()


def evaluate_model(model, loader):

    model.eval()

    total_loss = 0.0
    labels_all = []
    preds_all = []
    probs_all = []

    with torch.no_grad():

        for batch in loader:

            input_ids = batch[
                "input_ids"
            ].to(device)

            attention_mask = batch[
                "attention_mask"
            ].to(device)

            metadata = batch[
                "metadata"
            ].to(device)

            emotions = batch[
                "emotions"
            ].to(device)

            labels = batch[
                "labels"
            ].to(device)

            output = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                metadata=metadata,
                emotions=emotions
            )

            logits = output["logits"]

            loss = criterion(
                logits,
                labels
            )

            total_loss += (
                loss.item()
                * labels.size(0)
            )

            probs = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            preds = torch.argmax(
                logits,
                dim=1
            )

            labels_all.extend(
                labels.cpu().numpy()
            )

            preds_all.extend(
                preds.cpu().numpy()
            )

            probs_all.extend(
                probs.cpu().numpy()
            )

    return {
        "loss":
            total_loss
            / len(loader.dataset),

        "accuracy":
            accuracy_score(
                labels_all,
                preds_all
            ),

        "precision":
            precision_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "recall":
            recall_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "f1":
            f1_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "roc_auc":
            roc_auc_score(
                labels_all,
                probs_all
            )
    }


history = []

best_val_f1 = -np.inf
best_epoch = None
best_state_dict = None

epochs_without_improvement = 0

training_start = time.time()


for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    epoch_start = time.time()

    print()
    print("=" * 60)
    print(
        f"EPOCH {epoch}/{MAX_EPOCHS}"
    )
    print("=" * 60)

    model.train()

    running_loss = 0.0
    train_labels = []
    train_preds = []

    for batch_idx, batch in enumerate(
        train_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        emotions = batch[
            "emotions"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)

        optimizer.zero_grad()

        output = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata,
            emotions=emotions
        )

        logits = output["logits"]

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            trainable_params,
            max_norm=1.0
        )

        optimizer.step()
        scheduler.step()

        running_loss += (
            loss.item()
            * labels.size(0)
        )

        preds = torch.argmax(
            logits.detach(),
            dim=1
        )

        train_labels.extend(
            labels.detach()
            .cpu()
            .numpy()
        )

        train_preds.extend(
            preds.cpu().numpy()
        )

        if (
            batch_idx == 1
            or batch_idx % 50 == 0
            or batch_idx == len(train_loader)
        ):

            batch_accuracy = accuracy_score(
                labels.detach()
                .cpu()
                .numpy(),
                preds.cpu().numpy()
            )

            elapsed = (
                time.time()
                - epoch_start
            )

            print(
                f"Epoch {epoch} | "
                f"Batch "
                f"{batch_idx}/"
                f"{len(train_loader)} | "
                f"Loss: "
                f"{loss.item():.4f} | "
                f"Accuracy: "
                f"{batch_accuracy:.4f} | "
                f"LR: "
                f"{optimizer.param_groups[0]['lr']:.2e} | "
                f"Time: "
                f"{elapsed:.1f}s"
            )

    train_loss = (
        running_loss
        / len(train_loader.dataset)
    )

    train_accuracy = accuracy_score(
        train_labels,
        train_preds
    )

    train_precision = precision_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_recall = recall_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_f1 = f1_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    print()
    print(
        f"Epoch {epoch} — Validation"
    )

    val_metrics = evaluate_model(
        model,
        val_loader
    )

    with torch.no_grad():

        metadata_gate = float(
            torch.sigmoid(
                model.metadata_beta
            ).item()
        )

        emotion_gate = float(
            torch.sigmoid(
                model.emotion_beta
            ).item()
        )

    epoch_time = (
        time.time()
        - epoch_start
    )

    print()
    print("=" * 60)
    print(f"EPOCH {epoch} SUMMARY")
    print("=" * 60)

    print(
        f"Train Loss: "
        f"{train_loss:.4f}"
    )

    print(
        f"Train Accuracy: "
        f"{train_accuracy:.4f}"
    )

    print(
        f"Train Precision: "
        f"{train_precision:.4f}"
    )

    print(
        f"Train Recall: "
        f"{train_recall:.4f}"
    )

    print(
        f"Train F1: "
        f"{train_f1:.4f}"
    )

    print()
    print(
        f"Val Loss: "
        f"{val_metrics['loss']:.4f}"
    )

    print(
        f"Val Accuracy: "
        f"{val_metrics['accuracy']:.4f}"
    )

    print(
        f"Val Precision: "
        f"{val_metrics['precision']:.4f}"
    )

    print(
        f"Val Recall: "
        f"{val_metrics['recall']:.4f}"
    )

    print(
        f"Val F1: "
        f"{val_metrics['f1']:.4f}"
    )

    print(
        f"Val ROC-AUC: "
        f"{val_metrics['roc_auc']:.4f}"
    )

    print(
        f"Metadata gate: "
        f"{metadata_gate:.6f}"
    )

    print(
        f"Emotion gate: "
        f"{emotion_gate:.6f}"
    )

    print(
        f"Epoch Time: "
        f"{epoch_time:.2f}s"
    )

    history.append({
        "epoch": epoch,

        "train_loss":
            train_loss,

        "train_accuracy":
            train_accuracy,

        "train_precision":
            train_precision,

        "train_recall":
            train_recall,

        "train_f1":
            train_f1,

        "val_loss":
            val_metrics["loss"],

        "val_accuracy":
            val_metrics["accuracy"],

        "val_precision":
            val_metrics["precision"],

        "val_recall":
            val_metrics["recall"],

        "val_f1":
            val_metrics["f1"],

        "val_roc_auc":
            val_metrics["roc_auc"],

        "metadata_gate":
            metadata_gate,

        "emotion_gate":
            emotion_gate,

        "learning_rate":
            optimizer.param_groups[0]["lr"],

        "epoch_time":
            epoch_time
    })

    # --------------------------------------------------------
    # BEST CHECKPOINT
    # --------------------------------------------------------

    if val_metrics["f1"] > best_val_f1:

        best_val_f1 = (
            val_metrics["f1"]
        )

        best_epoch = epoch

        best_state_dict = {
            key:
                value.detach()
                .cpu()
                .clone()
            for key, value
            in model.state_dict().items()
        }

        epochs_without_improvement = 0

        checkpoint_path = os.path.join(
            CHECKPOINT_DIR,
            "best_sentencebert_emotion.pt"
        )

        torch.save(
            {
                "epoch":
                    epoch,

                "model_state_dict":
                    best_state_dict,

                "optimizer_state_dict":
                    optimizer.state_dict(),

                "scheduler_state_dict":
                    scheduler.state_dict(),

                "best_val_f1":
                    best_val_f1,

                "experiment_name":
                    EXPERIMENT_NAME,

                "model_name":
                    MODEL_NAME,

                "cleaning_method":
                    CLEANING_METHOD
            },
            checkpoint_path
        )

        print()
        print("✓ NEW BEST CHECKPOINT")
        print(
            "Best Validation F1:",
            f"{best_val_f1:.4f}"
        )

        print(
            "Saved:",
            checkpoint_path
        )

    else:

        epochs_without_improvement += 1

        print()
        print(
            "No validation F1 improvement."
        )

        print(
            "Epochs without improvement:",
            f"{epochs_without_improvement} / "
            f"{EARLY_STOPPING_PATIENCE}"
        )

    # --------------------------------------------------------
    # EARLY STOPPING
    # --------------------------------------------------------

    if (
        epochs_without_improvement
        >= EARLY_STOPPING_PATIENCE
    ):

        print()
        print(
            "EARLY STOPPING TRIGGERED"
        )

        break


assert best_state_dict is not None
assert best_epoch is not None

model.load_state_dict(
    best_state_dict
)

model.to(device)
model.eval()

total_training_time = (
    time.time()
    - training_start
)

history_df = pd.DataFrame(
    history
)

history_path = os.path.join(
    RESULT_DIR,
    "sentencebert_emotion_training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)

with torch.no_grad():

    learned_metadata_beta = float(
        model.metadata_beta.item()
    )

    learned_emotion_beta = float(
        model.emotion_beta.item()
    )

    learned_metadata_gate = float(
        torch.sigmoid(
            model.metadata_beta
        ).item()
    )

    learned_emotion_gate = float(
        torch.sigmoid(
            model.emotion_beta
        ).item()
    )


print()
print("=" * 60)
print("SENTENCE-BERT + EMOTION-BASED TRAINING COMPLETE")
print("=" * 60)

print(
    "Epochs completed:",
    len(history)
)

print(
    "Best epoch:",
    best_epoch
)

print(
    "Best Validation F1:",
    f"{best_val_f1:.4f}"
)

print(
    "Total training time:",
    f"{total_training_time:.2f} seconds"
)

print(
    "Best checkpoint:",
    os.path.join(
        CHECKPOINT_DIR,
        "best_sentencebert_emotion.pt"
    )
)

print(
    "Training history saved:",
    history_path
)

print()
print(
    "Learned metadata beta:",
    learned_metadata_beta
)

print(
    "Learned emotion beta:",
    learned_emotion_beta
)

print(
    "Metadata gate:",
    f"{learned_metadata_gate:.6f}"
)

print(
    "Emotion gate:",
    f"{learned_emotion_gate:.6f}"
)

print()
print("✓ Sentence-BERT transformer remained frozen")
print("✓ Metadata branch trained")
print("✓ Emotion branch trained")
print("✓ Metadata gate learned")
print("✓ Emotion gate learned")

print()
print("=" * 60)
print("CELL 11 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# CELL 12 — SENTENCE-BERT + EMOTION-BASED
# FINAL TEST EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

print("=" * 60)
print("SENTENCE-BERT + EMOTION-BASED — FINAL TEST EVALUATION")
print("=" * 60)

# ------------------------------------------------------------
# CHECKPOINT
# ------------------------------------------------------------

checkpoint_path = os.path.join(
    CHECKPOINT_DIR,
    "best_sentencebert_emotion.pt"
)

assert os.path.exists(checkpoint_path)

print()
print("✓ Best checkpoint found")

checkpoint = torch.load(
    checkpoint_path,
    map_location=device
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.to(device)
model.eval()

best_epoch = checkpoint["epoch"]
best_val_f1 = checkpoint["best_val_f1"]

print()
print("Best epoch:", best_epoch)
print(
    "Best validation F1:",
    f"{best_val_f1:.4f}"
)
print(
    "Checkpoint:",
    checkpoint_path
)

# ------------------------------------------------------------
# GATE VERIFICATION
# ------------------------------------------------------------

with torch.no_grad():

    learned_metadata_beta = float(
        model.metadata_beta.item()
    )

    learned_emotion_beta = float(
        model.emotion_beta.item()
    )

    learned_metadata_gate = float(
        torch.sigmoid(
            model.metadata_beta
        ).item()
    )

    learned_emotion_gate = float(
        torch.sigmoid(
            model.emotion_beta
        ).item()
    )

print()
print("=" * 60)
print("GATE VERIFICATION")
print("=" * 60)

print(
    "Metadata beta:",
    learned_metadata_beta
)

print(
    "Metadata gate:",
    f"{learned_metadata_gate:.6f}"
)

print(
    "Emotion beta:",
    learned_emotion_beta
)

print(
    "Emotion gate:",
    f"{learned_emotion_gate:.6f}"
)

# ------------------------------------------------------------
# FROZEN BACKBONE CHECK
# ------------------------------------------------------------

encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print("✓ Sentence-BERT backbone remains frozen")

# ------------------------------------------------------------
# TEST SET
# ------------------------------------------------------------

assert len(test_dataset) == 767

print()
print("=" * 60)
print("TEST SET INTEGRITY")
print("=" * 60)

print(
    "Expected test users:",
    767
)

print(
    "Actual test users:",
    len(test_dataset)
)

print("✓ Test set size correct")

# ------------------------------------------------------------
# EVALUATION
# ------------------------------------------------------------

labels_all = []
preds_all = []
probs_all = []

total_test_loss = 0.0

print()
print("=" * 60)
print("EVALUATING TEST SET")
print("=" * 60)

with torch.no_grad():

    for batch_idx, batch in enumerate(
        test_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        emotions = batch[
            "emotions"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)

        output = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata,
            emotions=emotions
        )

        logits = output["logits"]

        loss = criterion(
            logits,
            labels
        )

        total_test_loss += (
            loss.item()
            * labels.size(0)
        )

        probs = torch.softmax(
            logits,
            dim=1
        )[:, 1]

        preds = torch.argmax(
            logits,
            dim=1
        )

        labels_all.extend(
            labels.cpu().numpy()
        )

        preds_all.extend(
            preds.cpu().numpy()
        )

        probs_all.extend(
            probs.cpu().numpy()
        )

        if (
            batch_idx == 1
            or batch_idx % 10 == 0
            or batch_idx == len(test_loader)
        ):

            print(
                f"Test batch "
                f"{batch_idx}/{len(test_loader)}"
            )

# ------------------------------------------------------------
# METRICS
# ------------------------------------------------------------

test_loss = (
    total_test_loss
    /
    len(test_loader.dataset)
)

test_accuracy = accuracy_score(
    labels_all,
    preds_all
)

test_precision = precision_score(
    labels_all,
    preds_all,
    zero_division=0
)

test_recall = recall_score(
    labels_all,
    preds_all,
    zero_division=0
)

test_f1 = f1_score(
    labels_all,
    preds_all,
    zero_division=0
)

test_roc_auc = roc_auc_score(
    labels_all,
    probs_all
)

cm = confusion_matrix(
    labels_all,
    preds_all
)

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print()
print("=" * 60)
print("SENTENCE-BERT + EMOTION-BASED — FINAL TEST RESULTS")
print("=" * 60)

print(
    f"Test Loss: {test_loss:.4f}"
)

print(
    f"Test Accuracy: "
    f"{test_accuracy * 100:.2f}%"
)

print(
    f"Test Precision: "
    f"{test_precision * 100:.2f}%"
)

print(
    f"Test Recall: "
    f"{test_recall * 100:.2f}%"
)

print(
    f"Test F1: "
    f"{test_f1 * 100:.2f}%"
)

print(
    f"Test ROC-AUC: "
    f"{test_roc_auc * 100:.2f}%"
)

# ------------------------------------------------------------
# CONFUSION MATRIX
# ------------------------------------------------------------

print()
print("=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

print(
    "                 Predicted"
)

print(
    "                 0       1"
)

print(
    f"Actual 0        "
    f"{cm[0,0]:<7}{cm[0,1]}"
)

print(
    f"Actual 1        "
    f"{cm[1,0]:<7}{cm[1,1]}"
)

# ------------------------------------------------------------
# CLASSIFICATION REPORT
# ------------------------------------------------------------

print()
print("=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        labels_all,
        preds_all,
        target_names=[
            "Negative",
            "Positive"
        ],
        digits=4,
        zero_division=0
    )
)

# ------------------------------------------------------------
# INTEGRITY
# ------------------------------------------------------------

assert len(labels_all) == 767
assert len(preds_all) == 767
assert len(probs_all) == 767

print()
print("=" * 60)
print("FINAL TEST INTEGRITY CHECK")
print("=" * 60)

print(
    "✓ All test users evaluated exactly once"
)

print(
    "✓ Predictions aligned with labels"
)

print(
    "✓ Best Sentence-BERT checkpoint used"
)

print(
    "✓ Training data not used for evaluation"
)

print(
    "✓ Validation data not used for evaluation"
)

print(
    "✓ Sentence-BERT backbone remained frozen"
)

# ------------------------------------------------------------
# SAVE PREDICTIONS
# ------------------------------------------------------------

predictions_df = pd.DataFrame({

    "label":
        labels_all,

    "prediction":
        preds_all,

    "probability_positive":
        probs_all
})

predictions_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "sentencebert_emotion_test_predictions.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# SAVE METRICS
# ------------------------------------------------------------

metrics = {

    "test_loss":
        float(test_loss),

    "accuracy":
        float(test_accuracy),

    "precision":
        float(test_precision),

    "recall":
        float(test_recall),

    "f1":
        float(test_f1),

    "roc_auc":
        float(test_roc_auc),

    "best_epoch":
        int(best_epoch),

    "best_validation_f1":
        float(best_val_f1),

    "metadata_beta":
        float(learned_metadata_beta),

    "metadata_gate":
        float(learned_metadata_gate),

    "emotion_beta":
        float(learned_emotion_beta),

    "emotion_gate":
        float(learned_emotion_gate),

    "test_users":
        767,

    "confusion_matrix":
        cm.tolist()
}

with open(
    os.path.join(
        RESULT_DIR,
        "sentencebert_emotion_test_metrics.json"
    ),
    "w"
) as f:

    json.dump(
        metrics,
        f,
        indent=2
    )

# ------------------------------------------------------------
# SAVE CONFUSION MATRIX
# ------------------------------------------------------------

pd.DataFrame(
    cm,
    index=[
        "Actual 0",
        "Actual 1"
    ],
    columns=[
        "Predicted 0",
        "Predicted 1"
    ]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "sentencebert_emotion_confusion_matrix.csv"
    )
)

# ------------------------------------------------------------
# SAVE CLASSIFICATION REPORT
# ------------------------------------------------------------

report = classification_report(
    labels_all,
    preds_all,
    target_names=[
        "Negative",
        "Positive"
    ],
    output_dict=True,
    zero_division=0
)

pd.DataFrame(
    report
).transpose().to_csv(
    os.path.join(
        RESULT_DIR,
        "sentencebert_emotion_classification_report.csv"
    )
)

print()
print("=" * 60)
print("CELL 12 COMPLETE")
print("=" * 60)

print("✓ Final test evaluation complete")
print("✓ Metrics saved")
print("✓ Predictions saved")
print("✓ Confusion matrix saved")
print("✓ Classification report saved")
print("✓ Frozen backbone verified")
```

```python
# ============================================================
# CELL 15 — SENTENCE-BERT + EMOTION-BASED
# t-SNE REPRESENTATION ANALYSIS
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

print("=" * 70)
print("t-SNE REPRESENTATION ANALYSIS")
print("SENTENCE-BERT + EMOTION-BASED")
print("=" * 70)

# ============================================================
# FIXED EXPLAINABILITY SETTINGS
# ============================================================

EXPLAINABILITY_SEED = 42

PCA_COMPONENTS = 50

TSNE_COMPONENTS = 2
TSNE_PERPLEXITY = 30
TSNE_LEARNING_RATE = 200
TSNE_MAX_ITER = 1000

np.random.seed(
    EXPLAINABILITY_SEED
)

torch.manual_seed(
    EXPLAINABILITY_SEED
)

# ============================================================
# OUTPUT DIRECTORIES
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

TSNE_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "tsne"
)

os.makedirs(
    TSNE_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model.eval()

print()
print("✓ Sentence-BERT model in evaluation mode")

# ============================================================
# EXTRACT FUSED REPRESENTATIONS
# ============================================================

fused_representations = []
test_labels = []

print()
print("=" * 70)
print("EXTRACTING FUSED REPRESENTATIONS")
print("=" * 70)

with torch.no_grad():

    for batch_idx, batch in enumerate(
        test_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        emotions = batch[
            "emotions"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)

        output = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata,
            emotions=emotions
        )

        # ----------------------------------------------------
        # ACTUAL REPRESENTATIONS FROM THIS MODEL
        # ----------------------------------------------------

        text_representation = output[
            "text_representation"
        ]

        metadata_representation = output[
            "metadata_representation"
        ]

        emotion_representation = output[
            "emotion_representation"
        ]

        # ----------------------------------------------------
        # LEARNED GATES
        # ----------------------------------------------------

        metadata_gate = torch.sigmoid(
            model.metadata_beta
        )

        emotion_gate = torch.sigmoid(
            model.emotion_beta
        )

        gated_metadata = (
            metadata_gate
            *
            metadata_representation
        )

        gated_emotion = (
            emotion_gate
            *
            emotion_representation
        )

        # ----------------------------------------------------
        # EXACT FUSED REPRESENTATION USED BY CLASSIFIER
        # ----------------------------------------------------

        fused = torch.cat(
            [
                text_representation,
                gated_metadata,
                gated_emotion
            ],
            dim=1
        )

        fused_representations.append(
            fused.cpu().numpy()
        )

        test_labels.extend(
            labels.cpu().numpy().tolist()
        )

        if (
            batch_idx == 1
            or batch_idx % 10 == 0
            or batch_idx == len(test_loader)
        ):

            print(
                f"Processed batch "
                f"{batch_idx}/{len(test_loader)}"
            )

# ============================================================
# COMBINE
# ============================================================

fused_representations = np.concatenate(
    fused_representations,
    axis=0
)

test_labels = np.array(
    test_labels
)

# ============================================================
# REPRESENTATION CHECK
# ============================================================

print()
print("=" * 70)
print("FUSED REPRESENTATION SUMMARY")
print("=" * 70)

print(
    "Fused representation shape:",
    fused_representations.shape
)

print(
    "Test users:",
    len(test_labels)
)

print(
    "Fused dimensions:",
    fused_representations.shape[1]
)

print(
    "Negative users:",
    int(
        (test_labels == 0).sum()
    )
)

print(
    "Positive users:",
    int(
        (test_labels == 1).sum()
    )
)

print(
    "Metadata gate:",
    f"{float(metadata_gate.item()):.6f}"
)

print(
    "Emotion gate:",
    f"{float(emotion_gate.item()):.6f}"
)

assert (
    fused_representations.shape[0]
    ==
    len(test_labels)
)

assert np.isfinite(
    fused_representations
).all()

# ============================================================
# PCA
# ============================================================

print()
print("=" * 70)
print("PCA REDUCTION")
print("=" * 70)

pca_dimensions = min(
    PCA_COMPONENTS,
    fused_representations.shape[1]
)

pca = PCA(
    n_components=pca_dimensions,
    random_state=EXPLAINABILITY_SEED
)

pca_representation = pca.fit_transform(
    fused_representations
)

pca_variance_retained = (
    pca.explained_variance_ratio_.sum()
)

print(
    "Original dimensions:",
    fused_representations.shape[1]
)

print(
    "PCA dimensions:",
    pca_representation.shape[1]
)

print(
    "Variance retained:",
    f"{pca_variance_retained * 100:.2f}%"
)

# ============================================================
# t-SNE
# ============================================================

print()
print("=" * 70)
print("t-SNE REDUCTION")
print("=" * 70)

tsne = TSNE(
    n_components=TSNE_COMPONENTS,
    perplexity=TSNE_PERPLEXITY,
    learning_rate=TSNE_LEARNING_RATE,
    init="pca",
    random_state=EXPLAINABILITY_SEED,
    max_iter=TSNE_MAX_ITER
)

tsne_representation = tsne.fit_transform(
    pca_representation
)

print(
    "t-SNE output shape:",
    tsne_representation.shape
)

# ============================================================
# SAVE t-SNE COORDINATES
# ============================================================

tsne_df = pd.DataFrame({

    "tsne_1":
        tsne_representation[:, 0],

    "tsne_2":
        tsne_representation[:, 1],

    "label":
        test_labels

})

tsne_df["class"] = tsne_df[
    "label"
].map({
    0: "Negative",
    1: "Positive"
})

tsne_csv_path = os.path.join(
    TSNE_DIR,
    "sentencebert_emotion_tsne_coordinates.csv"
)

tsne_df.to_csv(
    tsne_csv_path,
    index=False
)

# ============================================================
# PUBLICATION-QUALITY t-SNE PLOT
# ============================================================

plt.figure(
    figsize=(10, 8)
)

negative_mask = (
    test_labels == 0
)

positive_mask = (
    test_labels == 1
)

plt.scatter(
    tsne_representation[
        negative_mask,
        0
    ],
    tsne_representation[
        negative_mask,
        1
    ],
    s=35,
    alpha=0.70,
    label="Negative",
    edgecolors="none"
)

plt.scatter(
    tsne_representation[
        positive_mask,
        0
    ],
    tsne_representation[
        positive_mask,
        1
    ],
    s=35,
    alpha=0.70,
    label="Positive",
    edgecolors="none"
)

plt.xlabel(
    "t-SNE Dimension 1",
    fontsize=12
)

plt.ylabel(
    "t-SNE Dimension 2",
    fontsize=12
)

plt.title(
    "Sentence-BERT + Emotion-Based — "
    "t-SNE of Fused Test Representations",
    fontsize=14
)

plt.legend(
    fontsize=11
)

plt.grid(
    alpha=0.20
)

plt.tight_layout()

tsne_plot_path = os.path.join(
    TSNE_DIR,
    "sentencebert_emotion_tsne.png"
)

# SAVE
plt.savefig(
    tsne_plot_path,
    dpi=300,
    bbox_inches="tight"
)

# DISPLAY
plt.show()

plt.close()

# ============================================================
# SAVE CONFIGURATION
# ============================================================

tsne_config = {

    "model":
        "Sentence-BERT",

    "base_encoder":
        MODEL_NAME,

    "cleaning":
        "Emotion-Based",

    "dataset":
        "Held-out test set",

    "test_users":
        int(len(test_labels)),

    "input_representation":
        "Fused classifier representation",

    "text_representation_dimension":
        int(
            text_representation.shape[1]
        ),

    "metadata_representation_dimension":
        int(
            metadata_representation.shape[1]
        ),

    "emotion_representation_dimension":
        int(
            emotion_representation.shape[1]
        ),

    "fused_representation_dimension":
        int(
            fused_representations.shape[1]
        ),

    "pca_dimensions":
        int(pca_dimensions),

    "pca_variance_retained":
        float(
            pca_variance_retained
        ),

    "tsne_dimensions":
        int(TSNE_COMPONENTS),

    "tsne_perplexity":
        int(TSNE_PERPLEXITY),

    "tsne_learning_rate":
        TSNE_LEARNING_RATE,

    "tsne_max_iterations":
        int(TSNE_MAX_ITER),

    "random_seed":
        EXPLAINABILITY_SEED,

    "metadata_gate":
        float(
            metadata_gate.item()
        ),

    "emotion_gate":
        float(
            emotion_gate.item()
        )

}

with open(
    os.path.join(
        TSNE_DIR,
        "sentencebert_emotion_tsne_config.json"
    ),
    "w"
) as f:

    json.dump(
        tsne_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("t-SNE ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ 896D fused representation extracted"
)

print(
    "✓ PCA reduction completed"
)

print(
    "✓ t-SNE reduction completed"
)

print(
    "✓ t-SNE graph saved:"
)

print(
    tsne_plot_path
)

print(
    "✓ t-SNE graph displayed above"
)

print(
    "✓ Coordinates saved:"
)

print(
    tsne_csv_path
)

print(
    "✓ Configuration saved"
)
```

```python
# ============================================================
# CELL 18 — SENTENCE-BERT + EMOTION-BASED
# GRAD-CAM TOKEN-LEVEL EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

print("=" * 70)
print("GRAD-CAM ANALYSIS")
print("SENTENCE-BERT + EMOTION-BASED")
print("=" * 70)

# ============================================================
# FIXED GRAD-CAM SETTINGS
# ============================================================

GRADCAM_SEED = 42
GRADCAM_TOP_TOKENS = 20

np.random.seed(
    GRADCAM_SEED
)

torch.manual_seed(
    GRADCAM_SEED
)

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

GRADCAM_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "gradcam"
)

os.makedirs(
    GRADCAM_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model.eval()

for parameter in model.parameters():
    parameter.requires_grad = False

print()
print("✓ Sentence-BERT model in evaluation mode")
print("✓ All model parameters remain frozen")
print("✓ No model training will occur")

# ============================================================
# CHECK REPRESENTATIVE CASES
# ============================================================

assert "selected_cases" in globals()

print()
print("=" * 70)
print("REPRESENTATIVE TEST CASES")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print(
        f"{case_name}: "
        f"test index = {test_index}"
    )

# ============================================================
# GRAD-CAM FUNCTION
# ============================================================

def compute_sentencebert_gradcam(
    test_index
):

    sample = test_dataset[
        int(test_index)
    ]

    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
        .to(device)
    )

    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
        .to(device)
    )

    metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
    )

    emotions = (
        sample["emotions"]
        .unsqueeze(0)
        .to(device)
    )

    # --------------------------------------------------------
    # FORWARD PASS WITH GRADIENTS
    # --------------------------------------------------------

    with torch.enable_grad():

        encoder_output = (
            model.encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )
        )

        hidden_states = (
            encoder_output.last_hidden_state
        )

        # Frozen encoder weights but gradients are retained
        # with respect to the activation tensor.
        hidden_states = (
            hidden_states
            .detach()
            .requires_grad_(True)
        )

        hidden_states.retain_grad()

        # ----------------------------------------------------
        # EXACT TEXT ATTENTION POOLING
        # ----------------------------------------------------

        projected = torch.tanh(
            model.attention_projection(
                hidden_states
            )
        )

        attention_scores = (
            model.attention_score(
                projected
            )
            .squeeze(-1)
        )

        attention_scores = (
            attention_scores.masked_fill(
                attention_mask == 0,
                -1e9
            )
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states
            *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # ----------------------------------------------------
        # EXACT METADATA BRANCH
        # ----------------------------------------------------

        metadata_representation = (
            model.metadata_branch(
                metadata
            )
        )

        metadata_gate = torch.sigmoid(
            model.metadata_beta
        )

        gated_metadata = (
            metadata_gate
            *
            metadata_representation
        )

        # ----------------------------------------------------
        # EXACT EMOTION BRANCH
        # ----------------------------------------------------

        emotion_representation = (
            model.emotion_branch(
                emotions
            )
        )

        emotion_gate = torch.sigmoid(
            model.emotion_beta
        )

        gated_emotion = (
            emotion_gate
            *
            emotion_representation
        )

        # ----------------------------------------------------
        # EXACT FUSION
        # ----------------------------------------------------

        fused_representation = torch.cat(
            [
                text_representation,
                gated_metadata,
                gated_emotion
            ],
            dim=1
        )

        logits = model.classifier(
            fused_representation
        )

        # Positive/depression class
        target_score = logits[
            0,
            1
        ]

        # ----------------------------------------------------
        # BACKWARD
        # ----------------------------------------------------

        model.zero_grad(
            set_to_none=True
        )

        target_score.backward()

        gradients = (
            hidden_states.grad
            .detach()
            .clone()
        )

        activations = (
            hidden_states.detach()
            .clone()
        )

    # ========================================================
    # GRAD-CAM
    # ========================================================

    # Channel-wise mean gradient
    channel_weights = gradients.mean(
        dim=1
    )

    # Weighted activation map
    cam = torch.sum(
        activations
        *
        channel_weights.unsqueeze(1),
        dim=-1
    )

    # ReLU
    cam = torch.relu(
        cam
    )

    cam = (
        cam
        .squeeze(0)
        .cpu()
        .numpy()
    )

    attention_weights_np = (
        attention_weights
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
    )

    input_ids_np = (
        input_ids
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
    )

    valid_mask = (
        attention_mask
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
        .astype(bool)
    )

    # --------------------------------------------------------
    # NORMALIZATION
    # --------------------------------------------------------

    valid_cam = cam[
        valid_mask
    ]

    if (
        valid_cam.max()
        -
        valid_cam.min()
    ) > 1e-12:

        cam_normalized = (
            cam
            -
            valid_cam.min()
        ) / (
            valid_cam.max()
            -
            valid_cam.min()
        )

    else:

        cam_normalized = (
            np.zeros_like(cam)
        )

    # Padding receives zero importance
    cam_normalized[
        ~valid_mask
    ] = 0.0

    return {

        "cam":
            cam_normalized,

        "attention_weights":
            attention_weights_np,

        "input_ids":
            input_ids_np,

        "attention_mask":
            valid_mask,

        "positive_logit":
            float(
                target_score
                .detach()
                .cpu()
                .item()
            ),

        "logits":
            logits
            .detach()
            .cpu()
            .numpy()
            .squeeze(0)

    }


# ============================================================
# TOKEN INFORMATION
# ============================================================

def get_token_information(
    result
):

    tokens = tokenizer.convert_ids_to_tokens(
        result["input_ids"].tolist()
    )

    cam = result[
        "cam"
    ]

    valid_mask = result[
        "attention_mask"
    ]

    rows = []

    for position, (
        token,
        importance,
        valid
    ) in enumerate(
        zip(
            tokens,
            cam,
            valid_mask
        )
    ):

        if not valid:
            continue

        if token in tokenizer.all_special_tokens:
            continue

        rows.append({

            "position":
                position,

            "token":
                token,

            "importance":
                float(
                    importance
                )

        })

    token_df = pd.DataFrame(
        rows
    )

    token_df = (
        token_df
        .sort_values(
            "importance",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )

    token_df["rank"] = (
        np.arange(
            len(token_df)
        ) + 1
    )

    return token_df


# ============================================================
# RUN FOUR CASES
# ============================================================

gradcam_results = {}

print()
print("=" * 70)
print("GENERATING GRAD-CAM EXPLANATIONS")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print()
    print(
        f"Processing {case_name} "
        f"(test index {test_index})..."
    )

    result = compute_sentencebert_gradcam(
        test_index
    )

    token_df = get_token_information(
        result
    )

    gradcam_results[
        case_name
    ] = {

        "test_index":
            test_index,

        "result":
            result,

        "token_df":
            token_df

    }

    print(
        f"✓ {case_name} Grad-CAM complete"
    )

# ============================================================
# SAVE ALL TOKEN IMPORTANCE
# ============================================================

all_gradcam_rows = []

for case_name, data in gradcam_results.items():

    token_df = data[
        "token_df"
    ]

    for _, row in token_df.iterrows():

        test_index = data[
            "test_index"
        ]

        all_gradcam_rows.append({

            "case":
                case_name,

            "test_index":
                int(test_index),

            "true_label":
                int(
                    labels_np[test_index]
                ),

            "predicted_label":
                int(
                    preds_np[test_index]
                ),

            "positive_probability":
                float(
                    probs_np[test_index]
                ),

            "position":
                int(
                    row["position"]
                ),

            "token":
                row["token"],

            "gradcam_importance":
                float(
                    row["importance"]
                ),

            "rank":
                int(
                    row["rank"]
                )

        })

gradcam_results_df = pd.DataFrame(
    all_gradcam_rows
)

gradcam_results_df.to_csv(
    os.path.join(
        GRADCAM_DIR,
        "sentencebert_emotion_gradcam_token_importance.csv"
    ),
    index=False
)

# ============================================================
# TOP-20 TOKEN PLOTS
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM TOKEN IMPORTANCE")
print("=" * 70)

for case_name, data in gradcam_results.items():

    token_df = (
        data["token_df"]
        .head(
            GRADCAM_TOP_TOKENS
        )
        .copy()
    )

    token_df = token_df.iloc[
        ::-1
    ]

    plt.figure(
        figsize=(10, 8)
    )

    plt.barh(
        token_df["token"],
        token_df["importance"]
    )

    plt.xlabel(
        "Grad-CAM Importance",
        fontsize=12
    )

    plt.ylabel(
        "Token",
        fontsize=12
    )

    plt.title(
        f"Sentence-BERT + Emotion-Based — "
        f"Grad-CAM Token Importance ({case_name})",
        fontsize=14
    )

    plt.grid(
        axis="x",
        alpha=0.20
    )

    plt.tight_layout()

    plot_path = os.path.join(
        GRADCAM_DIR,
        f"sentencebert_emotion_gradcam_{case_name}.png"
    )

    # SAVE
    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )

    # DISPLAY
    plt.show()

    plt.close()

# ============================================================
# SAVE TOP TOKENS TABLE
# ============================================================

top_token_rows = []

for case_name, data in gradcam_results.items():

    top_tokens = (
        data["token_df"]
        .head(
            GRADCAM_TOP_TOKENS
        )
    )

    for _, row in top_tokens.iterrows():

        top_token_rows.append({

            "case":
                case_name,

            "rank":
                int(
                    row["rank"]
                ),

            "token":
                row["token"],

            "gradcam_importance":
                float(
                    row["importance"]
                )

        })

top_token_df = pd.DataFrame(
    top_token_rows
)

top_token_df.to_csv(
    os.path.join(
        GRADCAM_DIR,
        "sentencebert_emotion_gradcam_top_tokens.csv"
    ),
    index=False
)

# ============================================================
# SAVE CASE INFORMATION
# ============================================================

case_rows = []

for case_name, data in gradcam_results.items():

    test_index = data[
        "test_index"
    ]

    case_rows.append({

        "case":
            case_name,

        "test_index":
            int(test_index),

        "true_label":
            int(
                labels_np[test_index]
            ),

        "predicted_label":
            int(
                preds_np[test_index]
            ),

        "positive_probability":
            float(
                probs_np[test_index]
            )

    })

case_df = pd.DataFrame(
    case_rows
)

case_df.to_csv(
    os.path.join(
        GRADCAM_DIR,
        "sentencebert_emotion_gradcam_case_information.csv"
    ),
    index=False
)

# ============================================================
# SAVE CONFIGURATION
# ============================================================

gradcam_config = {

    "model":
        "Sentence-BERT",

    "base_encoder":
        MODEL_NAME,

    "cleaning":
        "Emotion-Based",

    "dataset":
        "Held-out test set",

    "target":
        "Positive/depression class logit",

    "layer":
        "Final Transformer hidden states",

    "method":
        "Transformer Grad-CAM",

    "hidden_dimension":
        768,

    "cases":
        [
            "TP",
            "TN",
            "FP",
            "FN"
        ],

    "top_tokens_displayed":
        GRADCAM_TOP_TOKENS,

    "random_seed":
        GRADCAM_SEED,

    "model_training":
        "No additional training",

    "transformer_weights":
        "Frozen"

}

with open(
    os.path.join(
        GRADCAM_DIR,
        "sentencebert_emotion_gradcam_config.json"
    ),
    "w"
) as f:

    json.dump(
        gradcam_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ TP, TN, FP, FN explained"
)

print(
    "✓ Final Transformer hidden states used"
)

print(
    "✓ Positive-class gradients computed"
)

print(
    "✓ Sentence-BERT backbone remained frozen"
)

print(
    "✓ No additional model training"
)

print(
    "✓ Token importance calculated"
)

print(
    "✓ Top-20 token plots saved and displayed"
)

print(
    "✓ Complete token importance CSV saved"
)

print(
    "✓ Top-token table saved"
)

print(
    "✓ Case information saved"
)

print(
    "✓ Configuration saved"
)

print()
print(
    "Grad-CAM output directory:",
    GRADCAM_DIR
)
```

```python
# ============================================================
# CELL 18 — SENTENCE-BERT + EMOTION-BASED
# GRAD-CAM TOKEN-LEVEL EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

print("=" * 70)
print("GRAD-CAM ANALYSIS")
print("SENTENCE-BERT + EMOTION-BASED")
print("=" * 70)

# ============================================================
# FIXED GRAD-CAM SETTINGS
# ============================================================

GRADCAM_SEED = 42
GRADCAM_TOP_TOKENS = 20

np.random.seed(
    GRADCAM_SEED
)

torch.manual_seed(
    GRADCAM_SEED
)

# ============================================================
# FIXED COLOR PALETTE
# ============================================================

POSITIVE_TOKEN_COLOR = "#59A14F"
TOP_TOKEN_COLOR = "#F28E2B"

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

GRADCAM_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "gradcam"
)

os.makedirs(
    GRADCAM_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model.eval()

for parameter in model.parameters():
    parameter.requires_grad = False

print()
print("✓ Sentence-BERT model in evaluation mode")
print("✓ All model parameters remain frozen")
print("✓ No model training will occur")

# ============================================================
# CHECK REPRESENTATIVE CASES
# ============================================================

assert "selected_cases" in globals()

print()
print("=" * 70)
print("REPRESENTATIVE TEST CASES")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print(
        f"{case_name}: "
        f"test index = {test_index}"
    )

# ============================================================
# GRAD-CAM FUNCTION
# ============================================================

def compute_sentencebert_gradcam(
    test_index
):

    sample = test_dataset[
        int(test_index)
    ]

    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
        .to(device)
    )

    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
        .to(device)
    )

    metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
    )

    emotions = (
        sample["emotions"]
        .unsqueeze(0)
        .to(device)
    )

    # --------------------------------------------------------
    # FORWARD PASS WITH GRADIENTS
    # --------------------------------------------------------

    with torch.enable_grad():

        encoder_output = (
            model.encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )
        )

        hidden_states = (
            encoder_output.last_hidden_state
        )

        # Frozen encoder weights but gradients are retained
        # with respect to the activation tensor.
        hidden_states = (
            hidden_states
            .detach()
            .requires_grad_(True)
        )

        hidden_states.retain_grad()

        # ----------------------------------------------------
        # EXACT TEXT ATTENTION POOLING
        # ----------------------------------------------------

        projected = torch.tanh(
            model.attention_projection(
                hidden_states
            )
        )

        attention_scores = (
            model.attention_score(
                projected
            )
            .squeeze(-1)
        )

        attention_scores = (
            attention_scores.masked_fill(
                attention_mask == 0,
                -1e9
            )
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states
            *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # ----------------------------------------------------
        # EXACT METADATA BRANCH
        # ----------------------------------------------------

        metadata_representation = (
            model.metadata_branch(
                metadata
            )
        )

        metadata_gate = torch.sigmoid(
            model.metadata_beta
        )

        gated_metadata = (
            metadata_gate
            *
            metadata_representation
        )

        # ----------------------------------------------------
        # EXACT EMOTION BRANCH
        # ----------------------------------------------------

        emotion_representation = (
            model.emotion_branch(
                emotions
            )
        )

        emotion_gate = torch.sigmoid(
            model.emotion_beta
        )

        gated_emotion = (
            emotion_gate
            *
            emotion_representation
        )

        # ----------------------------------------------------
        # EXACT FUSION
        # ----------------------------------------------------

        fused_representation = torch.cat(
            [
                text_representation,
                gated_metadata,
                gated_emotion
            ],
            dim=1
        )

        logits = model.classifier(
            fused_representation
        )

        # Positive/depression class
        target_score = logits[
            0,
            1
        ]

        # ----------------------------------------------------
        # BACKWARD
        # ----------------------------------------------------

        model.zero_grad(
            set_to_none=True
        )

        target_score.backward()

        gradients = (
            hidden_states.grad
            .detach()
            .clone()
        )

        activations = (
            hidden_states.detach()
            .clone()
        )

    # ========================================================
    # GRAD-CAM
    # ========================================================

    # Channel-wise mean gradient
    channel_weights = gradients.mean(
        dim=1
    )

    # Weighted activation map
    cam = torch.sum(
        activations
        *
        channel_weights.unsqueeze(1),
        dim=-1
    )

    # ReLU
    cam = torch.relu(
        cam
    )

    cam = (
        cam
        .squeeze(0)
        .cpu()
        .numpy()
    )

    attention_weights_np = (
        attention_weights
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
    )

    input_ids_np = (
        input_ids
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
    )

    valid_mask = (
        attention_mask
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
        .astype(bool)
    )

    # --------------------------------------------------------
    # NORMALIZATION
    # --------------------------------------------------------

    valid_cam = cam[
        valid_mask
    ]

    if (
        valid_cam.max()
        -
        valid_cam.min()
    ) > 1e-12:

        cam_normalized = (
            cam
            -
            valid_cam.min()
        ) / (
            valid_cam.max()
            -
            valid_cam.min()
        )

    else:

        cam_normalized = (
            np.zeros_like(cam)
        )

    # Padding receives zero importance
    cam_normalized[
        ~valid_mask
    ] = 0.0

    return {

        "cam":
            cam_normalized,

        "attention_weights":
            attention_weights_np,

        "input_ids":
            input_ids_np,

        "attention_mask":
            valid_mask,

        "positive_logit":
            float(
                target_score
                .detach()
                .cpu()
                .item()
            ),

        "logits":
            logits
            .detach()
            .cpu()
            .numpy()
            .squeeze(0)

    }


# ============================================================
# TOKEN INFORMATION
# ============================================================

def get_token_information(
    result
):

    tokens = tokenizer.convert_ids_to_tokens(
        result["input_ids"].tolist()
    )

    cam = result[
        "cam"
    ]

    valid_mask = result[
        "attention_mask"
    ]

    rows = []

    for position, (
        token,
        importance,
        valid
    ) in enumerate(
        zip(
            tokens,
            cam,
            valid_mask
        )
    ):

        if not valid:
            continue

        if token in tokenizer.all_special_tokens:
            continue

        rows.append({

            "position":
                position,

            "token":
                token,

            "importance":
                float(
                    importance
                )

        })

    token_df = pd.DataFrame(
        rows
    )

    token_df = (
        token_df
        .sort_values(
            "importance",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )

    token_df["rank"] = (
        np.arange(
            len(token_df)
        ) + 1
    )

    return token_df


# ============================================================
# RUN FOUR CASES
# ============================================================

gradcam_results = {}

print()
print("=" * 70)
print("GENERATING GRAD-CAM EXPLANATIONS")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print()
    print(
        f"Processing {case_name} "
        f"(test index {test_index})..."
    )

    result = compute_sentencebert_gradcam(
        test_index
    )

    token_df = get_token_information(
        result
    )

    gradcam_results[
        case_name
    ] = {

        "test_index":
            test_index,

        "result":
            result,

        "token_df":
            token_df

    }

    print(
        f"✓ {case_name} Grad-CAM complete"
    )

# ============================================================
# SAVE ALL TOKEN IMPORTANCE
# ============================================================

all_gradcam_rows = []

for case_name, data in gradcam_results.items():

    token_df = data[
        "token_df"
    ]

    for _, row in token_df.iterrows():

        test_index = data[
            "test_index"
        ]

        all_gradcam_rows.append({

            "case":
                case_name,

            "test_index":
                int(test_index),

            "true_label":
                int(
                    labels_np[test_index]
                ),

            "predicted_label":
                int(
                    preds_np[test_index]
                ),

            "positive_probability":
                float(
                    probs_np[test_index]
                ),

            "position":
                int(
                    row["position"]
                ),

            "token":
                row["token"],

            "gradcam_importance":
                float(
                    row["importance"]
                ),

            "rank":
                int(
                    row["rank"]
                )

        })

gradcam_results_df = pd.DataFrame(
    all_gradcam_rows
)

gradcam_results_df.to_csv(
    os.path.join(
        GRADCAM_DIR,
        "sentencebert_emotion_gradcam_token_importance.csv"
    ),
    index=False
)

# ============================================================
# TOP-20 TOKEN PLOTS
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM TOKEN IMPORTANCE")
print("=" * 70)

for case_name, data in gradcam_results.items():

    token_df = (
        data["token_df"]
        .head(
            GRADCAM_TOP_TOKENS
        )
        .copy()
    )

    token_df = token_df.iloc[::-1]

    plt.figure(
        figsize=(10, 8)
    )

    bars = plt.barh(
        token_df["token"],
        token_df["importance"],
        color=POSITIVE_TOKEN_COLOR,
        alpha=0.85
    )

    # Highlight highest-importance token
    if len(bars) > 0:

        bars[-1].set_color(
            TOP_TOKEN_COLOR
        )

    plt.xlabel(
        "Grad-CAM Importance",
        fontsize=12
    )

    plt.ylabel(
        "Token",
        fontsize=12
    )

    plt.title(
        f"Sentence-BERT + Emotion-Based — "
        f"Grad-CAM Token Importance ({case_name})",
        fontsize=14,
        fontweight="bold"
    )

    plt.grid(
        axis="x",
        alpha=0.20
    )

    plt.tight_layout()

    plot_path = os.path.join(
        GRADCAM_DIR,
        f"sentencebert_emotion_gradcam_{case_name}.png"
    )

    # SAVE
    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )

    # DISPLAY
    plt.show()

    plt.close()

# ============================================================
# SAVE TOP TOKENS TABLE
# ============================================================

top_token_rows = []

for case_name, data in gradcam_results.items():

    top_tokens = (
        data["token_df"]
        .head(
            GRADCAM_TOP_TOKENS
        )
    )

    for _, row in top_tokens.iterrows():

        top_token_rows.append({

            "case":
                case_name,

            "rank":
                int(
                    row["rank"]
                ),

            "token":
                row["token"],

            "gradcam_importance":
                float(
                    row["importance"]
                )

        })

top_token_df = pd.DataFrame(
    top_token_rows
)

top_token_df.to_csv(
    os.path.join(
        GRADCAM_DIR,
        "sentencebert_emotion_gradcam_top_tokens.csv"
    ),
    index=False
)

# ============================================================
# SAVE CASE INFORMATION
# ============================================================

case_rows = []

for case_name, data in gradcam_results.items():

    test_index = data[
        "test_index"
    ]

    case_rows.append({

        "case":
            case_name,

        "test_index":
            int(test_index),

        "true_label":
            int(
                labels_np[test_index]
            ),

        "predicted_label":
            int(
                preds_np[test_index]
            ),

        "positive_probability":
            float(
                probs_np[test_index]
            )

    })

case_df = pd.DataFrame(
    case_rows
)

case_df.to_csv(
    os.path.join(
        GRADCAM_DIR,
        "sentencebert_emotion_gradcam_case_information.csv"
    ),
    index=False
)

# ============================================================
# SAVE CONFIGURATION
# ============================================================

gradcam_config = {

    "model":
        "Sentence-BERT",

    "base_encoder":
        MODEL_NAME,

    "cleaning":
        "Emotion-Based",

    "dataset":
        "Held-out test set",

    "target":
        "Positive/depression class logit",

    "layer":
        "Final Transformer hidden states",

    "method":
        "Transformer Grad-CAM",

    "hidden_dimension":
        768,

    "cases":
        [
            "TP",
            "TN",
            "FP",
            "FN"
        ],

    "top_tokens_displayed":
        GRADCAM_TOP_TOKENS,

    "random_seed":
        GRADCAM_SEED,

    "model_training":
        "No additional training",

    "transformer_weights":
        "Frozen",

    "plot_colors":
        {
            "token_importance":
                POSITIVE_TOKEN_COLOR,

            "highest_importance":
                TOP_TOKEN_COLOR
        }

}

with open(
    os.path.join(
        GRADCAM_DIR,
        "sentencebert_emotion_gradcam_config.json"
    ),
    "w"
) as f:

    json.dump(
        gradcam_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ TP, TN, FP, FN explained"
)

print(
    "✓ Final Transformer hidden states used"
)

print(
    "✓ Positive-class gradients computed"
)

print(
    "✓ Sentence-BERT backbone remained frozen"
)

print(
    "✓ No additional model training"
)

print(
    "✓ Token importance calculated"
)

print(
    "✓ Top-20 colorful token plots saved and displayed"
)

print(
    "✓ Complete token importance CSV saved"
)

print(
    "✓ Top-token table saved"
)

print(
    "✓ Case information saved"
)

print(
    "✓ Configuration saved"
)

print()
print(
    "Grad-CAM output directory:",
    GRADCAM_DIR
)
```
