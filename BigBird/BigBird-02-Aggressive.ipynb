```python
# ============================================================
# CELL 3 — BIGBIRD + AGGRESSIVE
# EXACT D1 ANCHOR DETECTION
# ============================================================

import os
import re
import json
import glob

from collections import defaultdict

print("=" * 60)
print("BIGBIRD + AGGRESSIVE — EXACT D1 ANCHOR DETECTION")
print("=" * 60)


# ============================================================
# D1 TIMELINE DIRECTORY
# ============================================================

DATASET_ROOT = (
    "/kaggle/input/datasets/"
    "vedantnawghare35/"
    "depression-detection-mddl-final-pbl/"
    "Dataset_MDDL"
)

D1_TIMELINE_DIR = os.path.join(
    DATASET_ROOT,
    "labeled",
    "positive",
    "data",
    "timeline"
)

assert os.path.exists(
    D1_TIMELINE_DIR
)


# ============================================================
# FIND D1 TIMELINE FILES
# ============================================================

d1_timeline_files = glob.glob(
    os.path.join(
        D1_TIMELINE_DIR,
        "*.json"
    )
)

print()
print(
    "D1 timeline files found:",
    len(d1_timeline_files)
)

assert len(
    d1_timeline_files
) == 2626


# ============================================================
# EXACT ANCHOR PATTERNS
# ============================================================

REFINED_ANCHOR_PATTERNS = [

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+have\s+(?:major\s+)?depression\b",

    r"\bi\s+suffer\s+from\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|am|'m)\s+diagnosed\b.{0,80}\bdepress",

    (
        r"\b(?:my|a)\s+(?:doctor|psychiatrist)"
        r"\s+(?:has\s+)?diagnosed\s+me\b"
        r".{0,80}\bdepress"
    )
]


compiled_anchor_patterns = [
    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )
    for pattern in REFINED_ANCHOR_PATTERNS
]


# ============================================================
# EXCLUSION PATTERNS
# ============================================================

EXCLUSION_PATTERNS = [

    r"\bmy\s+friend\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+sister\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+brother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+mother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+father\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+answer\s+to\s*:",

    r"\banswer\s+to\s*:",

    r"\bquestion\s*:",

    r"\bq\s*:"
]


compiled_exclusion_patterns = [
    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )
    for pattern in EXCLUSION_PATTERNS
]


# ============================================================
# EXACT ANCHOR DETECTOR
# ============================================================

def is_refined_anchor_tweet(text):

    if text is None:
        return False

    text = str(text).strip()

    if not text:
        return False

    for pattern in compiled_exclusion_patterns:

        if pattern.search(text):
            return False

    for pattern in compiled_anchor_patterns:

        if pattern.search(text):
            return True

    return False


# ============================================================
# POSITIVE USERS BY SPLIT
# ============================================================

train_positive_users = set(
    train_df.loc[
        train_df["label"] == 1,
        "username"
    ].astype(str)
)

val_positive_users = set(
    val_df.loc[
        val_df["label"] == 1,
        "username"
    ].astype(str)
)

test_positive_users = set(
    test_df.loc[
        test_df["label"] == 1,
        "username"
    ].astype(str)
)

all_positive_users = (
    train_positive_users
    |
    val_positive_users
    |
    test_positive_users
)


print()
print("=" * 60)
print("POSITIVE USERS BY SPLIT")
print("=" * 60)

print(
    "TRAIN:",
    len(train_positive_users)
)

print(
    "VALIDATION:",
    len(val_positive_users)
)

print(
    "TEST:",
    len(test_positive_users)
)

assert len(
    train_positive_users
) == 2013

assert len(
    val_positive_users
) == 252

assert len(
    test_positive_users
) == 251

assert len(
    all_positive_users
) == 2516


# ============================================================
# INDEX TIMELINE FILES
# ============================================================

timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:

    filename = os.path.basename(
        timeline_file
    )

    username = filename.split(
        "created_at"
    )[0]

    timeline_files_by_user[
        username
    ].append(
        timeline_file
    )


assert len(
    timeline_files_by_user
) == 2516


# ============================================================
# SCAN ALL POSITIVE D1 TIMELINES
# ============================================================

anchor_records = []

users_processed = set()

tweets_scanned = 0


for username in sorted(
    all_positive_users
):

    if username not in timeline_files_by_user:

        raise KeyError(
            f"Missing D1 timeline for user: {username}"
        )

    users_processed.add(
        username
    )

    for timeline_file in (
        timeline_files_by_user[username]
    ):

        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:

                line = line.strip()

                if not line:
                    continue

                record = json.loads(
                    line
                )

                tweets_scanned += 1

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get(
                            "id",
                            ""
                        )
                    )
                )

                tweet_text = str(
                    record.get(
                        "text",
                        ""
                    )
                )

                if (
                    tweet_id
                    and
                    is_refined_anchor_tweet(
                        tweet_text
                    )
                ):

                    anchor_records.append(
                        {
                            "username":
                                username,

                            "tweet_id":
                                tweet_id,

                            "created_at":
                                record.get(
                                    "created_at"
                                ),

                            "text":
                                tweet_text
                        }
                    )


# ============================================================
# UNIQUE ANCHOR PAIRS
# ============================================================

anchor_pairs = {
    (
        record["username"],
        record["tweet_id"]
    )
    for record in anchor_records
}


# ============================================================
# SPLIT-WISE UNIQUE ANCHORS
# ============================================================

train_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in train_positive_users
}

val_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in val_positive_users
}

test_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in test_positive_users
}


# ============================================================
# SUMMARY
# ============================================================

print()
print("=" * 60)
print("EXACT ANCHOR SUMMARY")
print("=" * 60)

print(
    "Positive users processed:",
    len(users_processed)
)

print(
    "Tweets scanned:",
    tweets_scanned
)

print(
    "Raw anchor records:",
    len(anchor_records)
)

print(
    "Unique anchor pairs:",
    len(anchor_pairs)
)

print()
print(
    "TRAIN unique anchors:",
    len(train_anchor_pairs)
)

print(
    "VAL unique anchors:",
    len(val_anchor_pairs)
)

print(
    "TEST unique anchors:",
    len(test_anchor_pairs)
)


# ============================================================
# EXACT CONTROL ASSERTIONS
# ============================================================

assert len(
    users_processed
) == 2516

assert len(
    anchor_records
) == 1920

assert len(
    anchor_pairs
) == 1847

assert len(
    train_anchor_pairs
) == 1476

assert len(
    val_anchor_pairs
) == 190

assert len(
    test_anchor_pairs
) == 181


# ============================================================
# SAVE
# ============================================================

bigbird_aggressive_anchor_df = pd.DataFrame(
    anchor_records
)

bigbird_aggressive_anchor_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "exact_anchor_records.csv"
    ),
    index=False
)


pd.DataFrame(
    list(anchor_pairs),
    columns=[
        "username",
        "tweet_id"
    ]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "unique_anchor_pairs.csv"
    ),
    index=False
)


# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 60)
print("CELL 3 COMPLETE")
print("=" * 60)

print(
    "✓ Exact anchor detector reproduced"
)

print(
    "✓ 1,847 unique anchors confirmed"
)

print(
    "✓ 1,920 anchor records confirmed"
)

print(
    "✓ TRAIN = 1,476 unique anchors"
)

print(
    "✓ VAL = 190 unique anchors"
)

print(
    "✓ TEST = 181 unique anchors"
)

print(
    "✓ No D2 users included"
)

print(
    "✓ Anchor records saved"
)

print(
    "✓ No timelines modified yet"
)
```

```python
# ============================================================
# CELL 5 — BIGBIRD + AGGRESSIVE
# TEXT CLEANING
# ============================================================

import re
import html

print("=" * 60)
print("BIGBIRD + AGGRESSIVE — TEXT CLEANING")
print("=" * 60)


# ============================================================
# AGGRESSIVE CLEANING FUNCTION
# ============================================================

def aggressive_clean(text):

    text = str(text)

    # HTML entities / markup
    text = html.unescape(text)

    # URLs
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )

    # User mentions
    text = re.sub(
        r"@\w+",
        " ",
        text
    )

    # Hashtag symbols
    text = re.sub(
        r"#",
        " ",
        text
    )

    # HTML entities / markup
    text = re.sub(
        r"&\w+;|<[^>]+>",
        " ",
        text
    )

    # Punctuation / emojis / emoticons / special symbols
    text = re.sub(
        r"[^\w\s]",
        " ",
        text,
        flags=re.UNICODE
    )

    # Underscores
    text = re.sub(
        r"_+",
        " ",
        text
    )

    # Whitespace normalization
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    # Lowercase
    return text.strip().lower()


# ============================================================
# APPLY AGGRESSIVE CLEANING
# ============================================================

train_bigbird_aggressive_df[
    "model_text"
] = (
    train_bigbird_aggressive_df[
        "anchor_removed_text"
    ]
    .fillna("")
    .map(
        aggressive_clean
    )
)


val_bigbird_aggressive_df[
    "model_text"
] = (
    val_bigbird_aggressive_df[
        "anchor_removed_text"
    ]
    .fillna("")
    .map(
        aggressive_clean
    )
)


test_bigbird_aggressive_df[
    "model_text"
] = (
    test_bigbird_aggressive_df[
        "anchor_removed_text"
    ]
    .fillna("")
    .map(
        aggressive_clean
    )
)


# ============================================================
# CLEANING SUMMARY
# ============================================================

print()
print("Cleaning applied:")

print("✓ URLs removed")

print("✓ Mentions removed")

print("✓ Hashtag symbols removed")

print("✓ HTML / markup removed")

print("✓ Punctuation removed")

print("✓ Emojis / emoticons removed")

print("✓ Special symbols removed")

print("✓ Whitespace normalized")

print("✓ Text lowercased")


# ============================================================
# MODEL TEXT INTEGRITY
# ============================================================

print()
print("=" * 60)
print("MODEL TEXT INTEGRITY")
print("=" * 60)


for name, df in [

    (
        "TRAIN",
        train_bigbird_aggressive_df
    ),

    (
        "VALIDATION",
        val_bigbird_aggressive_df
    ),

    (
        "TEST",
        test_bigbird_aggressive_df
    )

]:

    empty_count = int(
        df[
            "model_text"
        ]
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"{name}: "
        f"empty model texts = "
        f"{empty_count}"
    )


# ============================================================
# USER + LABEL PRESERVATION
# ============================================================

for (
    original,
    result,
    name
) in [

    (
        train_df,
        train_bigbird_aggressive_df,
        "TRAIN"
    ),

    (
        val_df,
        val_bigbird_aggressive_df,
        "VALIDATION"
    ),

    (
        test_df,
        test_bigbird_aggressive_df,
        "TEST"
    )

]:

    assert (
        original["user_id"].tolist()
        ==
        result["user_id"].tolist()
    )

    assert (
        original["username"].tolist()
        ==
        result["username"].tolist()
    )

    assert (
        original["label"].tolist()
        ==
        result["label"].tolist()
    )

    print(
        f"✓ {name}: "
        "users and labels unchanged"
    )


# ============================================================
# SAVE
# ============================================================

train_bigbird_aggressive_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "train_cleaned.pkl"
    )
)

val_bigbird_aggressive_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "val_cleaned.pkl"
    )
)

test_bigbird_aggressive_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "test_cleaned.pkl"
    )
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 5 COMPLETE")
print("=" * 60)

print(
    "✓ Aggressive cleaning applied"
)

print(
    "✓ Anchor-removed source used"
)

print(
    "✓ Model text created"
)

print(
    "✓ Users preserved"
)

print(
    "✓ Labels preserved"
)

print(
    "✓ Cleaned data saved"
)
```

```python
# ============================================================
# CELL 6 — BIGBIRD + AGGRESSIVE
# TOKEN LENGTH ANALYSIS
# ============================================================

from transformers import AutoTokenizer
import numpy as np
import pandas as pd

print("=" * 60)
print("BIGBIRD + AGGRESSIVE — TOKEN LENGTH ANALYSIS")
print("=" * 60)


# ============================================================
# TOKENIZER
# ============================================================

print()
print(
    "Tokenizer:",
    MODEL_NAME
)

print(
    "Experimental context length:",
    MAX_LENGTH
)

print(
    "BigBird architectural capacity:",
    MODEL_MAX_LENGTH
)


# ============================================================
# HUGGING FACE TOKEN
# ============================================================

from kaggle_secrets import UserSecretsClient

user_secrets = UserSecretsClient()

hf_token = user_secrets.get_secret(
    "HF_TOKEN"
)

print()
print("✓ HF_TOKEN retrieved")


# ============================================================
# LOAD TOKENIZER
# ============================================================

bigbird_aggressive_tokenizer = (
    AutoTokenizer.from_pretrained(
        MODEL_NAME,
        token=hf_token
    )
)

print()
print(
    "Tokenizer max length:",
    bigbird_aggressive_tokenizer.model_max_length
)

print(
    "✓ BigBird tokenizer loaded"
)


# ============================================================
# TOKEN LENGTH FUNCTION
# ============================================================

def get_token_lengths(
    texts,
    split_name
):

    print()
    print(
        f"Processing {split_name}..."
    )

    lengths = []

    for i, text in enumerate(texts):

        encoded = (
            bigbird_aggressive_tokenizer(
                str(text),
                add_special_tokens=True,
                truncation=False,
                padding=False
            )
        )

        lengths.append(
            len(
                encoded["input_ids"]
            )
        )

        if (
            (i + 1) % 1000 == 0
            or
            (i + 1) == len(texts)
        ):

            print(
                f"{split_name}: "
                f"{i + 1}/{len(texts)}"
            )

    return np.array(
        lengths,
        dtype=np.int64
    )


# ============================================================
# TRAIN TOKEN LENGTHS
# ============================================================

train_token_lengths = (
    get_token_lengths(
        train_bigbird_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "TRAIN"
    )
)


# ============================================================
# VALIDATION TOKEN LENGTHS
# ============================================================

val_token_lengths = (
    get_token_lengths(
        val_bigbird_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "VALIDATION"
    )
)


# ============================================================
# TEST TOKEN LENGTHS
# ============================================================

test_token_lengths = (
    get_token_lengths(
        test_bigbird_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "TEST"
    )
)


# ============================================================
# TOKEN STATISTICS
# ============================================================

def show_stats(
    lengths,
    split_name
):

    print()
    print("=" * 60)
    print(
        f"{split_name} TOKEN STATISTICS"
    )
    print("=" * 60)

    print(
        "Minimum:",
        int(lengths.min())
    )

    print(
        "Maximum:",
        int(lengths.max())
    )

    print(
        "Mean:",
        round(
            float(lengths.mean()),
            2
        )
    )

    print(
        "Median:",
        round(
            float(np.median(lengths)),
            2
        )
    )

    print(
        "25th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    25
                )
            ),
            2
        )
    )

    print(
        "75th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    75
                )
            ),
            2
        )
    )

    print(
        "90th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    90
                )
            ),
            2
        )
    )

    print(
        "95th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    95
                )
            ),
            2
        )
    )

    print(
        "99th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    99
                )
            ),
            2
        )
    )


    # ========================================================
    # 512-TOKEN COVERAGE
    # ========================================================

    within_512 = int(
        np.sum(
            lengths <= 512
        )
    )

    above_512 = int(
        np.sum(
            lengths > 512
        )
    )

    print()
    print(
        "512-TOKEN COVERAGE"
    )

    print(
        "Users <= 512:",
        within_512
    )

    print(
        "Users > 512:",
        above_512
    )

    print(
        "Percentage <= 512:",
        round(
            100 * within_512 / len(lengths),
            2
        ),
        "%"
    )

    print(
        "Percentage > 512:",
        round(
            100 * above_512 / len(lengths),
            2
        ),
        "%"
    )


    # ========================================================
    # 4096-TOKEN REFERENCE
    # ========================================================

    within_4096 = int(
        np.sum(
            lengths <= 4096
        )
    )

    above_4096 = int(
        np.sum(
            lengths > 4096
        )
    )

    print()
    print(
        "4096-TOKEN REFERENCE"
    )

    print(
        "Users <= 4096:",
        within_4096
    )

    print(
        "Users > 4096:",
        above_4096
    )

    print(
        "Percentage <= 4096:",
        round(
            100 * within_4096 / len(lengths),
            2
        ),
        "%"
    )

    print(
        "Percentage > 4096:",
        round(
            100 * above_4096 / len(lengths),
            2
        ),
        "%"
    )


# ============================================================
# SHOW STATISTICS
# ============================================================

show_stats(
    train_token_lengths,
    "TRAIN"
)

show_stats(
    val_token_lengths,
    "VALIDATION"
)

show_stats(
    test_token_lengths,
    "TEST"
)


# ============================================================
# ESTIMATED CONTEXT CHUNKS
# ============================================================

print()
print("=" * 60)
print("ESTIMATED CONTEXT CHUNKS")
print("=" * 60)


for context_length in [
    512,
    4096
]:

    print()
    print(
        "Context length:",
        context_length
    )


    train_chunks = np.ceil(
        train_token_lengths
        /
        context_length
    )

    val_chunks = np.ceil(
        val_token_lengths
        /
        context_length
    )

    test_chunks = np.ceil(
        test_token_lengths
        /
        context_length
    )


    print(
        "TRAIN mean chunks:",
        round(
            float(
                train_chunks.mean()
            ),
            2
        )
    )

    print(
        "TRAIN maximum chunks:",
        int(
            train_chunks.max()
        )
    )

    print(
        "VALIDATION mean chunks:",
        round(
            float(
                val_chunks.mean()
            ),
            2
        )
    )

    print(
        "VALIDATION maximum chunks:",
        int(
            val_chunks.max()
        )
    )

    print(
        "TEST mean chunks:",
        round(
            float(
                test_chunks.mean()
            ),
            2
        )
    )

    print(
        "TEST maximum chunks:",
        int(
            test_chunks.max()
        )
    )


# ============================================================
# TOKEN-LENGTH SUMMARY
# ============================================================

token_summary = pd.DataFrame([

    {
        "Split": "Train",
        "Users":
            len(train_token_lengths),
        "Mean":
            train_token_lengths.mean(),
        "Median":
            np.median(
                train_token_lengths
            ),
        "<=512 (%)":
            100 * np.mean(
                train_token_lengths <= 512
            ),
        "<=4096 (%)":
            100 * np.mean(
                train_token_lengths <= 4096
            )
    },

    {
        "Split": "Validation",
        "Users":
            len(val_token_lengths),
        "Mean":
            val_token_lengths.mean(),
        "Median":
            np.median(
                val_token_lengths
            ),
        "<=512 (%)":
            100 * np.mean(
                val_token_lengths <= 512
            ),
        "<=4096 (%)":
            100 * np.mean(
                val_token_lengths <= 4096
            )
    },

    {
        "Split": "Test",
        "Users":
            len(test_token_lengths),
        "Mean":
            test_token_lengths.mean(),
        "Median":
            np.median(
                test_token_lengths
            ),
        "<=512 (%)":
            100 * np.mean(
                test_token_lengths <= 512
            ),
        "<=4096 (%)":
            100 * np.mean(
                test_token_lengths <= 4096
            )
    }
])


print()
print("=" * 60)
print("TOKEN-LENGTH SUMMARY")
print("=" * 60)

print(
    token_summary.to_string(
        index=False,
        float_format=lambda x:
            f"{x:.2f}"
    )
)


# ============================================================
# SAVE TOKEN LENGTHS
# ============================================================

np.save(
    os.path.join(
        RESULT_DIR,
        "train_token_lengths.npy"
    ),
    train_token_lengths
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_token_lengths.npy"
    ),
    val_token_lengths
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_token_lengths.npy"
    ),
    test_token_lengths
)


token_summary.to_csv(
    os.path.join(
        RESULT_DIR,
        "token_length_summary.csv"
    ),
    index=False
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 6 COMPLETE")
print("=" * 60)

print(
    "✓ Authenticated BigBird tokenizer loaded"
)

print(
    "✓ Aggressive cleaned text analyzed"
)

print(
    "✓ No truncation used during analysis"
)

print(
    "✓ 512-token coverage measured"
)

print(
    "✓ 4096-token reference measured"
)

print(
    "✓ Context chunk estimates calculated"
)

print(
    "✓ Token-length files saved"
)
```

```python
# ============================================================
# CELL 8 — BIGBIRD + AGGRESSIVE
# TOKENIZATION
# ============================================================

from transformers import AutoTokenizer

print("=" * 60)
print("BIGBIRD + AGGRESSIVE — TOKENIZATION")
print("=" * 60)


# ============================================================
# LOAD TOKENIZER
# ============================================================

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    token=hf_token
)


print()
print(
    "Model:",
    MODEL_NAME
)

print(
    "Maximum sequence length:",
    MAX_LENGTH
)

print(
    "BigBird model capacity:",
    MODEL_MAX_LENGTH
)

print(
    "Truncation: ENABLED"
)

print(
    "Padding: MAX_LENGTH"
)

print(
    "Text source: Aggressive cleaned model_text"
)

print()
print(
    "✓ BigBird tokenizer available"
)


# ============================================================
# TOKENIZATION FUNCTION
# ============================================================

def tokenize_texts(
    texts,
    name
):

    print()
    print(
        f"Tokenizing {name}..."
    )

    enc = tokenizer(
        texts,
        max_length=MAX_LENGTH,
        truncation=True,
        padding="max_length",
        return_attention_mask=True
    )


    input_ids = np.asarray(
        enc["input_ids"],
        dtype=np.int64
    )

    attention_masks = np.asarray(
        enc["attention_mask"],
        dtype=np.int64
    )


    print(
        f"{name} input IDs:",
        input_ids.shape
    )

    print(
        f"{name} attention masks:",
        attention_masks.shape
    )


    return (
        input_ids,
        attention_masks
    )


# ============================================================
# TRAIN
# ============================================================

train_input_ids, train_attention_masks = (
    tokenize_texts(
        train_bigbird_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "TRAIN"
    )
)


# ============================================================
# VALIDATION
# ============================================================

val_input_ids, val_attention_masks = (
    tokenize_texts(
        val_bigbird_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "VALIDATION"
    )
)


# ============================================================
# TEST
# ============================================================

test_input_ids, test_attention_masks = (
    tokenize_texts(
        test_bigbird_aggressive_df[
            "model_text"
        ]
        .fillna("")
        .tolist(),
        "TEST"
    )
)


# ============================================================
# TOKENIZATION VERIFICATION
# ============================================================

print()
print("=" * 60)
print("TOKENIZATION VERIFICATION")
print("=" * 60)

print(
    "Train:",
    train_input_ids.shape
)

print(
    "Validation:",
    val_input_ids.shape
)

print(
    "Test:",
    test_input_ids.shape
)


assert train_input_ids.shape == (
    6135,
    512
)

assert val_input_ids.shape == (
    767,
    512
)

assert test_input_ids.shape == (
    767,
    512
)


assert train_attention_masks.shape == (
    6135,
    512
)

assert val_attention_masks.shape == (
    767,
    512
)

assert test_attention_masks.shape == (
    767,
    512
)


print(
    "✓ All token shapes correct"
)


# ============================================================
# ATTENTION MASK CHECK
# ============================================================

print()
print("=" * 60)
print("ATTENTION MASK CHECK")
print("=" * 60)


for (
    name,
    masks
) in [

    (
        "TRAIN",
        train_attention_masks
    ),

    (
        "VALIDATION",
        val_attention_masks
    ),

    (
        "TEST",
        test_attention_masks
    )

]:

    active = masks.sum(
        axis=1
    )


    print()
    print(name)

    print(
        "Minimum active tokens:",
        int(active.min())
    )

    print(
        "Maximum active tokens:",
        int(active.max())
    )

    print(
        "Mean active tokens:",
        round(
            float(active.mean()),
            2
        )
    )


    assert np.all(
        active >= 1
    )


print()
print(
    "✓ Labels aligned"
)


# ============================================================
# SAVE
# ============================================================

np.save(
    os.path.join(
        RESULT_DIR,
        "train_input_ids.npy"
    ),
    train_input_ids
)

np.save(
    os.path.join(
        RESULT_DIR,
        "train_attention_masks.npy"
    ),
    train_attention_masks
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_input_ids.npy"
    ),
    val_input_ids
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_attention_masks.npy"
    ),
    val_attention_masks
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_input_ids.npy"
    ),
    test_input_ids
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_attention_masks.npy"
    ),
    test_attention_masks
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 8 COMPLETE")
print("=" * 60)

print(
    "✓ Aggressive cleaned text tokenized"
)

print(
    "✓ BigBird maximum experimental length = 512"
)

print(
    "✓ Truncation enabled"
)

print(
    "✓ Padding = MAX_LENGTH"
)

print(
    "✓ Attention masks generated"
)

print(
    "✓ Labels aligned"
)

print(
    "✓ Tokenized arrays saved"
)
```

```python
# ============================================================
# CELL 11 — BIGBIRD + AGGRESSIVE
# BUILDING FROZEN BIGBIRD + METADATA MODEL
# ============================================================

from transformers import AutoModel
import torch.nn as nn

print("=" * 60)
print("BIGBIRD + AGGRESSIVE — BUILDING FROZEN BIGBIRD + METADATA MODEL")
print("=" * 60)

bigbird_encoder = AutoModel.from_pretrained(
    MODEL_NAME,
    token=hf_token
).to(device)

for p in bigbird_encoder.parameters():
    p.requires_grad = False


class BigBirdMetadataClassifier(nn.Module):

    def __init__(
        self,
        encoder,
        metadata_dim=53
    ):
        super().__init__()

        self.encoder = encoder
        self.text_hidden_size = 768

        self.attention_projection = nn.Linear(
            768,
            256
        )

        self.attention_score = nn.Linear(
            256,
            1
        )

        self.metadata_branch = nn.Sequential(
            nn.Linear(
                metadata_dim,
                128
            ),
            nn.ReLU(),
            nn.Dropout(
                0.3
            ),
            nn.Linear(
                128,
                64
            ),
            nn.ReLU()
        )

        self.metadata_beta = nn.Parameter(
            torch.tensor(
                0.0
            )
        )

        self.classifier = nn.Sequential(
            nn.Linear(
                832,
                512
            ),
            nn.ReLU(),
            nn.Dropout(
                0.3
            ),
            nn.Linear(
                512,
                256
            ),
            nn.ReLU(),
            nn.Dropout(
                0.3
            ),
            nn.Linear(
                256,
                2
            )
        )

    def forward(
        self,
        input_ids,
        attention_mask,
        metadata
    ):

        with torch.no_grad():

            outputs = self.encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

        hidden_states = outputs.last_hidden_state

        projected = torch.tanh(
            self.attention_projection(
                hidden_states
            )
        )

        scores = self.attention_score(
            projected
        ).squeeze(-1)

        scores = scores.masked_fill(
            attention_mask == 0,
            -1e9
        )

        attention_weights = torch.softmax(
            scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        metadata_representation = (
            self.metadata_branch(
                metadata
            )
        )

        metadata_gate = torch.sigmoid(
            self.metadata_beta
        )

        gated_metadata = (
            metadata_gate *
            metadata_representation
        )

        fused = torch.cat(
            [
                text_representation,
                gated_metadata
            ],
            dim=1
        )

        logits = self.classifier(
            fused
        )

        return {
            "logits": logits,
            "text_representation":
                text_representation,
            "metadata_representation":
                metadata_representation,
            "attention_weights":
                attention_weights,
            "metadata_gate":
                metadata_gate
        }


model_bigbird_aggressive = (
    BigBirdMetadataClassifier(
        bigbird_encoder,
        METADATA_DIM
    ).to(device)
)


# ============================================================
# PARAMETER COUNT
# ============================================================

total_params = sum(
    p.numel()
    for p in model_bigbird_aggressive.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model_bigbird_aggressive.parameters()
    if p.requires_grad
)

frozen_params = sum(
    p.numel()
    for p in model_bigbird_aggressive.parameters()
    if not p.requires_grad
)


print()
print("=" * 60)
print("PARAMETER COUNT")
print("=" * 60)

print(
    "Total parameters:",
    total_params
)

print(
    "Trainable parameters:",
    trainable_params
)

print(
    "Frozen parameters:",
    frozen_params
)

print(
    "Total parameters (M):",
    round(
        total_params / 1e6,
        3
    )
)

print(
    "Trainable parameters (M):",
    round(
        trainable_params / 1e6,
        3
    )
)

print(
    "Frozen parameters (M):",
    round(
        frozen_params / 1e6,
        3
    )
)


encoder_trainable = sum(
    p.numel()
    for p in model_bigbird_aggressive.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print(
    "✓ All BigBird backbone parameters frozen"
)


# ============================================================
# TRAINABLE COMPONENTS
# ============================================================

print()
print("=" * 60)
print("TRAINABLE COMPONENTS")
print("=" * 60)

for name, p in (
    model_bigbird_aggressive.named_parameters()
):

    if p.requires_grad:
        print(
            "✓",
            name
        )


# ============================================================
# ARCHITECTURE
# ============================================================

print()
print("=" * 60)
print("ARCHITECTURE CONFIGURATION")
print("=" * 60)

print(
    "Transformer:",
    MODEL_NAME
)

print(
    "Base architecture:",
    "BigBird-RoBERTa-base"
)

print(
    "Experimental context length:",
    MAX_LENGTH
)

print(
    "Text hidden size:",
    768
)

print(
    "Attention projection:",
    256
)

print(
    "Metadata:",
    "53 -> 128 -> 64"
)

print(
    "Gate:",
    "g = sigmoid(beta)"
)

print(
    "Initial beta:",
    0.0
)

print(
    "Initial gate:",
    0.5
)

print(
    "Fusion:",
    "768 + 64 = 832"
)

print(
    "Classifier:",
    "832 -> 512 -> 256 -> 2"
)

print(
    "Dropout:",
    0.3
)

print(
    "Transformer frozen:",
    True
)


# ============================================================
# FORWARD PASS SANITY CHECK
# ============================================================

print()
print("=" * 60)
print("FORWARD PASS SANITY CHECK")
print("=" * 60)

model_bigbird_aggressive.eval()

batch = next(
    iter(train_loader)
)

input_ids = batch[
    "input_ids"
].to(device)

attention_mask = batch[
    "attention_mask"
].to(device)

metadata = batch[
    "metadata"
].to(device)


with torch.no_grad():

    output = (
        model_bigbird_aggressive(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )
    )


print(
    "Input IDs:",
    tuple(
        input_ids.shape
    )
)

print(
    "Attention mask:",
    tuple(
        attention_mask.shape
    )
)

print(
    "Metadata:",
    tuple(
        metadata.shape
    )
)

print(
    "Logits:",
    tuple(
        output[
            "logits"
        ].shape
    )
)

print(
    "Text representation:",
    tuple(
        output[
            "text_representation"
        ].shape
    )
)

print(
    "Metadata representation:",
    tuple(
        output[
            "metadata_representation"
        ].shape
    )
)

print(
    "Attention weights:",
    tuple(
        output[
            "attention_weights"
        ].shape
    )
)


gate = float(
    output[
        "metadata_gate"
    ].item()
)

print(
    "Metadata gate:",
    gate
)


padding_attention = output[
    "attention_weights"
][
    attention_mask == 0
]


max_padding_attention = (
    float(
        padding_attention.max().item()
    )
    if padding_attention.numel()
    else 0.0
)


print(
    "Maximum attention on padding:",
    max_padding_attention
)


# ============================================================
# EXACT ASSERTIONS
# ============================================================

assert (
    output[
        "logits"
    ].shape
    ==
    (16, 2)
)

assert (
    output[
        "text_representation"
    ].shape
    ==
    (16, 768)
)

assert (
    output[
        "metadata_representation"
    ].shape
    ==
    (16, 64)
)

assert (
    output[
        "attention_weights"
    ].shape
    ==
    (16, 512)
)

assert (
    max_padding_attention
    ==
    0.0
)

assert (
    abs(
        gate - 0.5
    )
    <
    1e-6
)


encoder_trainable = sum(
    p.numel()
    for p in model_bigbird_aggressive.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0


print()
print(
    "✓ Transformer freeze verified"
)


# ============================================================
# SAVE CONFIG
# ============================================================

architecture_config = {

    "model_name":
        MODEL_NAME,

    "base_architecture":
        "BigBird-RoBERTa-base",

    "experimental_context_length":
        MAX_LENGTH,

    "text_hidden_size":
        768,

    "attention_projection":
        256,

    "metadata_dimension":
        53,

    "metadata_branch":
        "53 -> 128 -> 64",

    "initial_beta":
        0.0,

    "initial_gate":
        0.5,

    "fusion":
        "768 + 64 = 832",

    "classifier":
        "832 -> 512 -> 256 -> 2",

    "dropout":
        0.3,

    "transformer_frozen":
        True
}


with open(
    os.path.join(
        RESULT_DIR,
        "architecture_config.json"
    ),
    "w"
) as f:

    json.dump(
        architecture_config,
        f,
        indent=2
    )


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 11 COMPLETE")
print("=" * 60)

print(
    "✓ BigBird-RoBERTa-base loaded"
)

print(
    "✓ Transformer backbone frozen"
)

print(
    "✓ Attention pooling active"
)

print(
    "✓ Metadata branch = 53 -> 128 -> 64"
)

print(
    "✓ Single metadata gate active"
)

print(
    "✓ Text representation = 768"
)

print(
    "✓ Fusion = 768 + 64 = 832"
)

print(
    "✓ Classifier = 832 -> 512 -> 256 -> 2"
)

print(
    "✓ Padding attention masked"
)

print(
    "✓ Forward pass successful"
)

print(
    "✓ Frozen BigBird + Aggressive model ready for training"
)
```

```python
# ============================================================
# CELL 12 — BIGBIRD + AGGRESSIVE
# MODEL TRAINING
# ============================================================

import os
import time
import copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

MAX_EPOCHS = 10
EARLY_STOPPING_PATIENCE = 2
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1

print("=" * 60)
print("STARTING BIGBIRD + AGGRESSIVE TRAINING")
print("=" * 60)

print()
print("Training users:", len(train_dataset))
print("Validation users:", len(val_dataset))
print("Test users:", len(test_dataset))
print("Maximum epochs:", MAX_EPOCHS)
print("Early stopping patience:", EARLY_STOPPING_PATIENCE)
print("Checkpoint metric: Validation F1")
print("Learning rate:", LEARNING_RATE)
print("Weight decay:", WEIGHT_DECAY)
print("Batch size:", BATCH_SIZE)
print("Transformer: FROZEN")

encoder_trainable = sum(
    p.numel()
    for p in model_bigbird_aggressive.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0
print()
print("✓ BIGBIRD backbone confirmed frozen")

trainable_params = [
    p for p in model_bigbird_aggressive.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_params,
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

total_steps = len(train_loader) * MAX_EPOCHS
warmup_steps = int(total_steps * WARMUP_RATIO)

def lr_lambda(step):
    if step < warmup_steps:
        return (step + 1) / max(1, warmup_steps)

    return max(
        0.0,
        (total_steps - step) /
        max(1, total_steps - warmup_steps)
    )

scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda
)

criterion = nn.CrossEntropyLoss()


def evaluate_model(model, loader):
    model.eval()

    total_loss = 0.0
    labels_all = []
    preds_all = []
    probs_all = []

    with torch.no_grad():
        for batch in loader:

            input_ids = batch[
                "input_ids"
            ].to(device)

            attention_mask = batch[
                "attention_mask"
            ].to(device)

            metadata = batch[
                "metadata"
            ].to(device)

            labels = batch[
                "labels"
            ].to(device)

            output = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                metadata=metadata
            )

            logits = output["logits"]

            loss = criterion(
                logits,
                labels
            )

            total_loss += (
                loss.item() * labels.size(0)
            )

            probs = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            preds = torch.argmax(
                logits,
                dim=1
            )

            labels_all.extend(
                labels.cpu().numpy()
            )

            preds_all.extend(
                preds.cpu().numpy()
            )

            probs_all.extend(
                probs.cpu().numpy()
            )

    metrics = {
        "loss":
            total_loss / len(loader.dataset),

        "accuracy":
            accuracy_score(
                labels_all,
                preds_all
            ),

        "precision":
            precision_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "recall":
            recall_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "f1":
            f1_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "roc_auc":
            roc_auc_score(
                labels_all,
                probs_all
            )
    }

    return metrics


history = []
best_val_f1 = -np.inf
best_epoch = None
best_state_dict = None
epochs_without_improvement = 0

training_start = time.time()

for epoch in range(1, MAX_EPOCHS + 1):

    epoch_start = time.time()

    print()
    print("=" * 60)
    print(f"EPOCH {epoch}/{MAX_EPOCHS}")
    print("=" * 60)

    model_bigbird_aggressive.train()

    running_loss = 0.0
    train_labels = []
    train_preds = []

    for batch_idx, batch in enumerate(
        train_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)

        optimizer.zero_grad()

        output = model_bigbird_aggressive(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = output["logits"]

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            trainable_params,
            max_norm=1.0
        )

        optimizer.step()
        scheduler.step()

        running_loss += (
            loss.item() * labels.size(0)
        )

        preds = torch.argmax(
            logits.detach(),
            dim=1
        )

        train_labels.extend(
            labels.detach().cpu().numpy()
        )

        train_preds.extend(
            preds.cpu().numpy()
        )

        if (
            batch_idx == 1
            or batch_idx % 50 == 0
            or batch_idx == len(train_loader)
        ):

            batch_acc = accuracy_score(
                labels.detach().cpu().numpy(),
                preds.cpu().numpy()
            )

            elapsed = (
                time.time()
                - epoch_start
            )

            print(
                f"Epoch {epoch} | "
                f"Batch {batch_idx}/{len(train_loader)} | "
                f"Loss: {loss.item():.4f} | "
                f"Accuracy: {batch_acc:.4f} | "
                f"LR: "
                f"{optimizer.param_groups[0]['lr']:.2e} | "
                f"Time: {elapsed:.1f}s"
            )

    train_loss = (
        running_loss
        /
        len(train_loader.dataset)
    )

    train_accuracy = accuracy_score(
        train_labels,
        train_preds
    )

    train_precision = precision_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_recall = recall_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_f1 = f1_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    print()
    print(
        f"Epoch {epoch} — Validation"
    )

    val_metrics = evaluate_model(
        model_bigbird_aggressive,
        val_loader
    )

    with torch.no_grad():

        metadata_gate = float(
            torch.sigmoid(
                model_bigbird_aggressive.metadata_beta
            ).item()
        )

    epoch_time = (
        time.time()
        -
        epoch_start
    )

    print()
    print("=" * 60)
    print(f"EPOCH {epoch} SUMMARY")
    print("=" * 60)

    print(
        f"Train Loss: {train_loss:.4f}"
    )

    print(
        f"Train Accuracy: {train_accuracy:.4f}"
    )

    print(
        f"Train Precision: {train_precision:.4f}"
    )

    print(
        f"Train Recall: {train_recall:.4f}"
    )

    print(
        f"Train F1: {train_f1:.4f}"
    )

    print()

    print(
        f"Val Loss: {val_metrics['loss']:.4f}"
    )

    print(
        f"Val Accuracy: {val_metrics['accuracy']:.4f}"
    )

    print(
        f"Val Precision: {val_metrics['precision']:.4f}"
    )

    print(
        f"Val Recall: {val_metrics['recall']:.4f}"
    )

    print(
        f"Val F1: {val_metrics['f1']:.4f}"
    )

    print(
        f"Val ROC-AUC: {val_metrics['roc_auc']:.4f}"
    )

    print(
        f"Metadata gate: {metadata_gate:.6f}"
    )

    print(
        f"Epoch Time: {epoch_time:.2f}s"
    )

    history.append({
        "epoch": epoch,

        "train_loss":
            train_loss,

        "train_accuracy":
            train_accuracy,

        "train_precision":
            train_precision,

        "train_recall":
            train_recall,

        "train_f1":
            train_f1,

        "val_loss":
            val_metrics["loss"],

        "val_accuracy":
            val_metrics["accuracy"],

        "val_precision":
            val_metrics["precision"],

        "val_recall":
            val_metrics["recall"],

        "val_f1":
            val_metrics["f1"],

        "val_roc_auc":
            val_metrics["roc_auc"],

        "metadata_gate":
            metadata_gate,

        "learning_rate":
            optimizer.param_groups[0]["lr"],

        "epoch_time":
            epoch_time
    })

    if (
        val_metrics["f1"]
        >
        best_val_f1
    ):

        best_val_f1 = (
            val_metrics["f1"]
        )

        best_epoch = epoch

        best_state_dict = {
            k:
                v.detach().cpu().clone()

            for k, v
            in
            model_bigbird_aggressive.state_dict().items()
        }

        epochs_without_improvement = 0

        checkpoint_path = os.path.join(
            CHECKPOINT_DIR,
            "best_bigbird_aggressive.pt"
        )

        torch.save(
            {
                "epoch":
                    epoch,

                "model_state_dict":
                    best_state_dict,

                "optimizer_state_dict":
                    optimizer.state_dict(),

                "scheduler_state_dict":
                    scheduler.state_dict(),

                "best_val_f1":
                    best_val_f1,

                "experiment_name":
                    EXPERIMENT_NAME,

                "model_name":
                    MODEL_NAME,

                "cleaning_method":
                    CLEANING_METHOD
            },
            checkpoint_path
        )

        print()
        print(
            "✓ NEW BEST CHECKPOINT"
        )

        print(
            "Best Validation F1:",
            f"{best_val_f1:.4f}"
        )

        print(
            "Saved:",
            checkpoint_path
        )

    else:

        epochs_without_improvement += 1

        print()
        print(
            "No validation F1 improvement."
        )

        print(
            "Epochs without improvement:",
            f"{epochs_without_improvement} / "
            f"{EARLY_STOPPING_PATIENCE}"
        )

    if (
        epochs_without_improvement
        >=
        EARLY_STOPPING_PATIENCE
    ):

        print()
        print(
            "EARLY STOPPING TRIGGERED"
        )

        break


assert best_state_dict is not None
assert best_epoch is not None

model_bigbird_aggressive.load_state_dict(
    best_state_dict
)

model_bigbird_aggressive.to(
    device
)

model_bigbird_aggressive.eval()

total_training_time = (
    time.time()
    -
    training_start
)

history_df = pd.DataFrame(
    history
)

history_path = os.path.join(
    RESULT_DIR,
    "bigbird_aggressive_training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)

with torch.no_grad():

    learned_beta = float(
        model_bigbird_aggressive.metadata_beta.item()
    )

    learned_gate = float(
        torch.sigmoid(
            model_bigbird_aggressive.metadata_beta
        ).item()
    )


print()
print("=" * 60)
print("BIGBIRD + AGGRESSIVE TRAINING COMPLETE")
print("=" * 60)

print(
    "Epochs completed:",
    len(history)
)

print(
    "Best epoch:",
    best_epoch
)

print(
    "Best Validation F1:",
    f"{best_val_f1:.4f}"
)

print(
    "Total training time:",
    f"{total_training_time:.2f} seconds"
)

print(
    "Best checkpoint:",
    os.path.join(
        CHECKPOINT_DIR,
        "best_bigbird_aggressive.pt"
    )
)

print(
    "Training history saved:",
    history_path
)

print()
print(
    "Learned beta:",
    learned_beta
)

print(
    "Metadata gate:",
    f"{learned_gate:.6f}"
)

print()
print(
    "✓ BIGBIRD transformer remained frozen"
)

print(
    "✓ Downstream layers trained"
)

print()
print("=" * 60)
print("CELL 12 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# EXPLAINABILITY — CELL 1
# BIGBIRD + AGGRESSIVE
# t-SNE: FUSED REPRESENTATION → PCA → t-SNE
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

print("=" * 70)
print("t-SNE REPRESENTATION ANALYSIS")
print("BIGBIRD + AGGRESSIVE")
print("=" * 70)

# ============================================================
# FIXED EXPLAINABILITY SETTINGS
# ============================================================

EXPLAINABILITY_SEED = 42
PCA_COMPONENTS = 50
TSNE_COMPONENTS = 2
TSNE_PERPLEXITY = 30
TSNE_LEARNING_RATE = 200
TSNE_MAX_ITER = 1000

np.random.seed(EXPLAINABILITY_SEED)
torch.manual_seed(EXPLAINABILITY_SEED)

# ============================================================
# OUTPUT DIRECTORIES
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

TSNE_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "tsne"
)

os.makedirs(
    TSNE_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model_bigbird_aggressive.eval()

# ============================================================
# EXTRACT FUSED REPRESENTATIONS
# ============================================================

fused_representations = []
test_labels = []

print()
print("Extracting fused model representations...")

with torch.no_grad():

    for batch_idx, batch in enumerate(
        test_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)

        output = model_bigbird_aggressive(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        # ----------------------------------------------------
        # ACTUAL OUTPUT NAMES FROM CELL 11
        # ----------------------------------------------------

        text_representation = output[
            "text_representation"
        ]

        metadata_representation = output[
            "metadata_representation"
        ]

        # ----------------------------------------------------
        # SAME LEARNED METADATA GATE
        # ----------------------------------------------------

        metadata_gate = torch.sigmoid(
            model_bigbird_aggressive.metadata_beta
        )

        gated_metadata = (
            metadata_gate *
            metadata_representation
        )

        # ----------------------------------------------------
        # EXACT FUSED REPRESENTATION
        # ----------------------------------------------------

        fused = torch.cat(
            [
                text_representation,
                gated_metadata
            ],
            dim=1
        )

        fused_representations.append(
            fused.cpu().numpy()
        )

        test_labels.extend(
            labels.cpu().numpy().tolist()
        )

        if (
            batch_idx == 1
            or batch_idx % 10 == 0
            or batch_idx == len(test_loader)
        ):

            print(
                f"Processed batch "
                f"{batch_idx}/{len(test_loader)}"
            )

# ============================================================
# COMBINE
# ============================================================

fused_representations = np.concatenate(
    fused_representations,
    axis=0
)

test_labels = np.array(
    test_labels
)

# ============================================================
# INTEGRITY
# ============================================================

print()
print("=" * 70)
print("FUSED REPRESENTATION")
print("=" * 70)

print(
    "Representation shape:",
    fused_representations.shape
)

print(
    "Number of test users:",
    len(test_labels)
)

print(
    "Representation dimensions:",
    fused_representations.shape[1]
)

print(
    "Negative users:",
    int((test_labels == 0).sum())
)

print(
    "Positive users:",
    int((test_labels == 1).sum())
)

assert (
    fused_representations.shape[0]
    ==
    len(test_labels)
)

assert np.isfinite(
    fused_representations
).all()

# ============================================================
# PCA
# ============================================================

print()
print("=" * 70)
print("PCA REDUCTION")
print("=" * 70)

pca_dimensions = min(
    PCA_COMPONENTS,
    fused_representations.shape[1]
)

pca = PCA(
    n_components=pca_dimensions,
    random_state=EXPLAINABILITY_SEED
)

pca_representation = pca.fit_transform(
    fused_representations
)

explained_variance = (
    pca.explained_variance_ratio_.sum()
)

print(
    "Original dimensions:",
    fused_representations.shape[1]
)

print(
    "PCA dimensions:",
    pca_representation.shape[1]
)

print(
    "Variance retained:",
    f"{explained_variance * 100:.2f}%"
)

# ============================================================
# t-SNE
# ============================================================

print()
print("=" * 70)
print("t-SNE REDUCTION")
print("=" * 70)

tsne = TSNE(
    n_components=TSNE_COMPONENTS,
    perplexity=TSNE_PERPLEXITY,
    learning_rate=TSNE_LEARNING_RATE,
    init="pca",
    random_state=EXPLAINABILITY_SEED,
    max_iter=TSNE_MAX_ITER
)

tsne_representation = tsne.fit_transform(
    pca_representation
)

print(
    "t-SNE output shape:",
    tsne_representation.shape
)

# ============================================================
# SAVE COORDINATES
# ============================================================

tsne_df = pd.DataFrame({

    "tsne_1":
        tsne_representation[:, 0],

    "tsne_2":
        tsne_representation[:, 1],

    "label":
        test_labels

})

tsne_df["class"] = tsne_df[
    "label"
].map({
    0: "Negative",
    1: "Positive"
})

tsne_csv_path = os.path.join(
    TSNE_DIR,
    "bigbird_aggressive_tsne_coordinates.csv"
)

tsne_df.to_csv(
    tsne_csv_path,
    index=False
)

# ============================================================
# PUBLICATION-QUALITY PLOT
# ============================================================

plt.figure(
    figsize=(10, 8)
)

negative_mask = (
    test_labels == 0
)

positive_mask = (
    test_labels == 1
)

plt.scatter(
    tsne_representation[
        negative_mask,
        0
    ],
    tsne_representation[
        negative_mask,
        1
    ],
    s=35,
    alpha=0.70,
    label="Negative",
    edgecolors="none"
)

plt.scatter(
    tsne_representation[
        positive_mask,
        0
    ],
    tsne_representation[
        positive_mask,
        1
    ],
    s=35,
    alpha=0.70,
    label="Positive",
    edgecolors="none"
)

plt.xlabel(
    "t-SNE Dimension 1",
    fontsize=12
)

plt.ylabel(
    "t-SNE Dimension 2",
    fontsize=12
)

plt.title(
    "BigBird + Aggressive — "
    "t-SNE of Fused Test Representations",
    fontsize=14
)

plt.legend(
    fontsize=11
)

plt.grid(
    alpha=0.20
)

plt.tight_layout()

tsne_plot_path = os.path.join(
    TSNE_DIR,
    "bigbird_aggressive_tsne.png"
)

# SAVE
plt.savefig(
    tsne_plot_path,
    dpi=300,
    bbox_inches="tight"
)

# DISPLAY
plt.show()

plt.close()

# ============================================================
# SAVE CONFIGURATION
# ============================================================

tsne_config = {

    "model":
        "BigBird",

    "cleaning":
        "Aggressive",

    "dataset":
        "Held-out test set",

    "test_users":
        int(len(test_labels)),

    "input_representation":
        "Fused classifier representation",

    "text_representation_dimension":
        int(
            text_representation.shape[1]
        ),

    "metadata_representation_dimension":
        int(
            metadata_representation.shape[1]
        ),

    "fused_representation_dimension":
        int(
            fused_representations.shape[1]
        ),

    "pca_dimensions":
        int(pca_dimensions),

    "pca_variance_retained":
        float(explained_variance),

    "tsne_dimensions":
        int(TSNE_COMPONENTS),

    "tsne_perplexity":
        int(TSNE_PERPLEXITY),

    "tsne_learning_rate":
        TSNE_LEARNING_RATE,

    "tsne_max_iterations":
        int(TSNE_MAX_ITER),

    "random_seed":
        EXPLAINABILITY_SEED

}

with open(
    os.path.join(
        TSNE_DIR,
        "bigbird_aggressive_tsne_config.json"
    ),
    "w"
) as f:

    json.dump(
        tsne_config,
        f,
        indent=2
    )

# ============================================================
# FINAL OUTPUT
# ============================================================

print()
print("=" * 70)
print("t-SNE ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ Fused representation extracted"
)

print(
    "✓ PCA reduction completed"
)

print(
    "✓ t-SNE reduction completed"
)

print(
    "✓ Coordinates saved:"
)

print(
    tsne_csv_path
)

print(
    "✓ Graph saved:"
)

print(
    tsne_plot_path
)

print(
    "✓ Graph displayed above"
)

print(
    "✓ Configuration saved"
)
```

```python
# ============================================================
# EXPLAINABILITY — SHAP
# BIGBIRD + AGGRESSIVE
# GLOBAL METADATA FEATURE EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import shap

print("=" * 70)
print("SHAP ANALYSIS")
print("BIGBIRD + AGGRESSIVE")
print("=" * 70)

# ============================================================
# FIXED SHAP SETTINGS
# ============================================================

SHAP_SEED = 42
SHAP_BACKGROUND_SIZE = 50
SHAP_TEST_SIZE_PER_CLASS = 50
SHAP_TOP_FEATURES = 20

np.random.seed(SHAP_SEED)
torch.manual_seed(SHAP_SEED)

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

SHAP_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "shap"
)

os.makedirs(
    SHAP_DIR,
    exist_ok=True
)

# ============================================================
# FEATURE NAMES
# ============================================================

feature_names = list(
    selected_metadata_names
)

assert len(feature_names) == 53

print()
print(
    "Total metadata features:",
    len(feature_names)
)

# ============================================================
# FIXED BACKGROUND DATA
# TRAINING USERS ONLY
# ============================================================

background_metadata = (
    train_dataset.metadata[
        :SHAP_BACKGROUND_SIZE
    ]
    .clone()
    .detach()
    .to(device)
)

print()
print(
    "SHAP background samples:",
    background_metadata.shape[0]
)

print(
    "Background dimensions:",
    background_metadata.shape[1]
)

assert background_metadata.shape[1] == 53

# ============================================================
# FIXED TEST USERS
# 50 NEGATIVE + 50 POSITIVE
# ============================================================

test_labels_tensor = (
    test_dataset.labels
    .detach()
    .cpu()
)

negative_indices = torch.where(
    test_labels_tensor == 0
)[0].numpy()

positive_indices = torch.where(
    test_labels_tensor == 1
)[0].numpy()

assert len(negative_indices) >= SHAP_TEST_SIZE_PER_CLASS
assert len(positive_indices) >= SHAP_TEST_SIZE_PER_CLASS

shap_negative_indices = (
    negative_indices[:SHAP_TEST_SIZE_PER_CLASS]
)

shap_positive_indices = (
    positive_indices[:SHAP_TEST_SIZE_PER_CLASS]
)

shap_test_indices = np.concatenate(
    [
        shap_negative_indices,
        shap_positive_indices
    ]
)

print()
print(
    "SHAP test samples:",
    len(shap_test_indices)
)

print(
    "Negative:",
    len(shap_negative_indices)
)

print(
    "Positive:",
    len(shap_positive_indices)
)

# ============================================================
# MODEL WRAPPER
# ============================================================
# Explains metadata while keeping each user's
# Transformer text representation fixed.
# No model parameters are changed.

class MetadataSHAPWrapper(
    nn.Module
):

    def __init__(
        self,
        base_model,
        fixed_input_ids,
        fixed_attention_mask
    ):

        super().__init__()

        self.base_model = base_model

        self.fixed_input_ids = (
            fixed_input_ids
            .to(device)
        )

        self.fixed_attention_mask = (
            fixed_attention_mask
            .to(device)
        )

    def forward(
        self,
        metadata
    ):

        batch_size = metadata.size(0)

        input_ids = (
            self.fixed_input_ids
            .expand(
                batch_size,
                -1
            )
        )

        attention_mask = (
            self.fixed_attention_mask
            .expand(
                batch_size,
                -1
            )
        )

        output = self.base_model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        # Positive/depression class logit
        positive_logit = output[
            "logits"
        ][:, 1:2]

        return positive_logit


# ============================================================
# MODEL PREPARATION
# ============================================================

model_bigbird_aggressive.eval()

for parameter in (
    model_bigbird_aggressive.parameters()
):
    parameter.requires_grad = False

print()
print(
    "✓ BigBird model set to evaluation mode"
)

print(
    "✓ No parameters will be updated"
)

# ============================================================
# COMPUTE SHAP VALUES
# ============================================================

all_shap_values = []
all_metadata_values = []
all_explained_labels = []

print()
print("=" * 70)
print("COMPUTING SHAP VALUES")
print("=" * 70)

for count, test_index in enumerate(
    shap_test_indices,
    start=1
):

    sample = test_dataset[
        int(test_index)
    ]

    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
    )

    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
    )

    metadata_sample = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
    )

    label = int(
        sample["labels"].item()
    )

    # --------------------------------------------------------
    # User-specific wrapper
    # Text representation remains fixed for this user.
    # --------------------------------------------------------

    wrapper = MetadataSHAPWrapper(
        model_bigbird_aggressive,
        input_ids,
        attention_mask
    ).to(device)

    wrapper.eval()

    # --------------------------------------------------------
    # Gradient-based SHAP
    # --------------------------------------------------------

    explainer = shap.GradientExplainer(
        wrapper,
        background_metadata
    )

    shap_result = explainer.shap_values(
        metadata_sample
    )

    # --------------------------------------------------------
    # Handle SHAP version-dependent output format
    # --------------------------------------------------------

    if isinstance(
        shap_result,
        list
    ):

        shap_array = np.asarray(
            shap_result[0]
        )

    else:

        shap_array = np.asarray(
            shap_result
        )

    shap_array = np.squeeze(
        shap_array
    )

    assert shap_array.shape == (
        53,
    )

    all_shap_values.append(
        shap_array
    )

    all_metadata_values.append(
        metadata_sample
        .detach()
        .cpu()
        .numpy()
        .squeeze(0)
    )

    all_explained_labels.append(
        label
    )

    if (
        count == 1
        or count % 10 == 0
        or count == len(shap_test_indices)
    ):

        print(
            f"Explained "
            f"{count}/{len(shap_test_indices)} "
            f"test users"
        )

# ============================================================
# COMBINE RESULTS
# ============================================================

all_shap_values = np.vstack(
    all_shap_values
)

all_metadata_values = np.vstack(
    all_metadata_values
)

all_explained_labels = np.array(
    all_explained_labels
)

print()
print("=" * 70)
print("SHAP RESULT SUMMARY")
print("=" * 70)

print(
    "SHAP matrix:",
    all_shap_values.shape
)

print(
    "Metadata matrix:",
    all_metadata_values.shape
)

assert all_shap_values.shape == (
    len(shap_test_indices),
    53
)

assert all_metadata_values.shape == (
    len(shap_test_indices),
    53
)

assert np.isfinite(
    all_shap_values
).all()

# ============================================================
# GLOBAL FEATURE IMPORTANCE
# ============================================================

mean_abs_shap = np.mean(
    np.abs(all_shap_values),
    axis=0
)

mean_signed_shap = np.mean(
    all_shap_values,
    axis=0
)

feature_importance_df = pd.DataFrame({

    "feature":
        feature_names,

    "mean_abs_shap":
        mean_abs_shap,

    "mean_signed_shap":
        mean_signed_shap

})

feature_importance_df = (
    feature_importance_df
    .sort_values(
        "mean_abs_shap",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

print()
print("=" * 70)
print("TOP 20 SHAP FEATURES")
print("=" * 70)

print(
    feature_importance_df
    .head(SHAP_TOP_FEATURES)
    .to_string(
        index=False
    )
)

# ============================================================
# SAVE SHAP VALUES
# ============================================================

shap_values_df = pd.DataFrame(
    all_shap_values,
    columns=feature_names
)

shap_values_df.to_csv(
    os.path.join(
        SHAP_DIR,
        "bigbird_aggressive_shap_values.csv"
    ),
    index=False
)

# ============================================================
# SAVE FEATURE IMPORTANCE
# ============================================================

feature_importance_df.to_csv(
    os.path.join(
        SHAP_DIR,
        "bigbird_aggressive_shap_feature_importance.csv"
    ),
    index=False
)

# ============================================================
# SAVE EXPLAINED USER INFORMATION
# ============================================================

explained_users_df = pd.DataFrame({

    "test_index":
        shap_test_indices,

    "label":
        all_explained_labels

})

explained_users_df.to_csv(
    os.path.join(
        SHAP_DIR,
        "bigbird_aggressive_shap_test_users.csv"
    ),
    index=False
)

# ============================================================
# SHAP BEESWARM — PAPER FIGURE
# ============================================================

plt.figure(
    figsize=(11, 8)
)

shap.summary_plot(
    all_shap_values,
    all_metadata_values,
    feature_names=feature_names,
    max_display=SHAP_TOP_FEATURES,
    show=False
)

plt.title(
    "BigBird + Aggressive — SHAP Feature Importance",
    fontsize=14
)

plt.tight_layout()

shap_beeswarm_path = os.path.join(
    SHAP_DIR,
    "bigbird_aggressive_shap_beeswarm.png"
)

plt.savefig(
    shap_beeswarm_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

# ============================================================
# SHAP BAR — PAPER FIGURE
# ============================================================

top_features = (
    feature_importance_df
    .head(SHAP_TOP_FEATURES)
    .iloc[::-1]
)

plt.figure(
    figsize=(10, 8)
)

plt.barh(
    top_features["feature"],
    top_features["mean_abs_shap"]
)

plt.xlabel(
    "Mean Absolute SHAP Value",
    fontsize=12
)

plt.ylabel(
    "Metadata Feature",
    fontsize=12
)

plt.title(
    "BigBird + Aggressive — Global SHAP Feature Importance",
    fontsize=14
)

plt.tight_layout()

shap_bar_path = os.path.join(
    SHAP_DIR,
    "bigbird_aggressive_shap_bar.png"
)

plt.savefig(
    shap_bar_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

# ============================================================
# SAVE CONFIGURATION
# ============================================================

shap_config = {

    "model":
        "BigBird",

    "cleaning":
        "Aggressive",

    "dataset":
        "Held-out test set",

    "total_metadata_features":
        53,

    "background_samples":
        SHAP_BACKGROUND_SIZE,

    "test_samples_per_class":
        SHAP_TEST_SIZE_PER_CLASS,

    "total_explained_test_users":
        len(shap_test_indices),

    "target":
        "Positive/depression class logit",

    "explainer":
        "SHAP GradientExplainer",

    "text_representation":
        "Fixed for each explained user",

    "random_seed":
        SHAP_SEED,

    "top_features_displayed":
        SHAP_TOP_FEATURES

}

with open(
    os.path.join(
        SHAP_DIR,
        "bigbird_aggressive_shap_config.json"
    ),
    "w"
) as f:

    json.dump(
        shap_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("SHAP ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ 53 metadata features explained"
)

print(
    "✓ 100 fixed test users explained"
)

print(
    "✓ 50 negative + 50 positive"
)

print(
    "✓ Training-only background used"
)

print(
    "✓ Positive-class logit explained"
)

print(
    "✓ BigBird weights unchanged"
)

print(
    "✓ SHAP beeswarm saved and displayed"
)

print(
    "✓ SHAP bar plot saved and displayed"
)

print(
    "✓ SHAP values saved"
)

print(
    "✓ Feature importance ranking saved"
)

print(
    "✓ Configuration saved"
)

print()
print(
    "SHAP output directory:",
    SHAP_DIR
)
```

```python
# ============================================================
# EXPLAINABILITY — GRAD-CAM
# BIGBIRD + AGGRESSIVE
# TOKEN-LEVEL LOCAL EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from IPython.display import display, HTML

print("=" * 70)
print("GRAD-CAM ANALYSIS")
print("BIGBIRD + AGGRESSIVE")
print("=" * 70)

# ============================================================
# FIXED GRAD-CAM SETTINGS
# ============================================================

GRADCAM_SEED = 42
GRADCAM_TOP_TOKENS = 20

np.random.seed(GRADCAM_SEED)
torch.manual_seed(GRADCAM_SEED)

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

GRADCAM_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "gradcam"
)

os.makedirs(
    GRADCAM_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model_bigbird_aggressive.eval()

# Keep all model parameters frozen
for parameter in model_bigbird_aggressive.parameters():
    parameter.requires_grad = False

print()
print("✓ BigBird model loaded")
print("✓ Evaluation mode enabled")
print("✓ All model parameters remain frozen")

# ============================================================
# REUSE SAME REPRESENTATIVE CASES AS LIME
# ============================================================

assert "selected_cases" in globals()

print()
print("=" * 70)
print("FIXED REPRESENTATIVE TEST CASES")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print(
        f"{case_name}: "
        f"test index = {test_index}"
    )

# ============================================================
# GRAD-CAM FUNCTION
# ============================================================

def compute_bigbird_gradcam(
    test_index
):

    sample = test_dataset[
        int(test_index)
    ]

    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
        .to(device)
    )

    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
        .to(device)
    )

    metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
    )

    # --------------------------------------------------------
    # Encoder forward pass
    # --------------------------------------------------------
    # Encoder parameters are frozen, but gradients are enabled
    # with respect to the hidden-state activation.
    # --------------------------------------------------------

    with torch.enable_grad():

        encoder_output = (
            model_bigbird_aggressive.encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )
        )

        hidden_states = (
            encoder_output.last_hidden_state
        )

        # Required because encoder parameters are frozen
        # and therefore hidden_states initially has no grad.
        hidden_states = hidden_states.detach()
        hidden_states.requires_grad_(True)

        hidden_states.retain_grad()

        # ----------------------------------------------------
        # EXACT ATTENTION POOLING FROM MODEL
        # ----------------------------------------------------

        projected = torch.tanh(
            model_bigbird_aggressive.attention_projection(
                hidden_states
            )
        )

        scores = (
            model_bigbird_aggressive.attention_score(
                projected
            )
            .squeeze(-1)
        )

        scores = scores.masked_fill(
            attention_mask == 0,
            -1e9
        )

        attention_weights = torch.softmax(
            scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # ----------------------------------------------------
        # EXACT METADATA BRANCH FROM MODEL
        # ----------------------------------------------------

        metadata_representation = (
            model_bigbird_aggressive.metadata_branch(
                metadata
            )
        )

        metadata_gate = torch.sigmoid(
            model_bigbird_aggressive.metadata_beta
        )

        gated_metadata = (
            metadata_gate *
            metadata_representation
        )

        # ----------------------------------------------------
        # EXACT FUSION + CLASSIFIER
        # ----------------------------------------------------

        fused = torch.cat(
            [
                text_representation,
                gated_metadata
            ],
            dim=1
        )

        logits = (
            model_bigbird_aggressive.classifier(
                fused
            )
        )

        # Positive/depression class
        target_score = logits[
            0,
            1
        ]

        # ----------------------------------------------------
        # BACKWARD
        # ----------------------------------------------------

        model_bigbird_aggressive.zero_grad(
            set_to_none=True
        )

        target_score.backward()

        gradients = (
            hidden_states.grad
            .detach()
            .clone()
        )

        activations = (
            hidden_states.detach()
            .clone()
        )

    # ========================================================
    # GRAD-CAM WEIGHTS
    # ========================================================

    # Global average pooling over token positions
    channel_weights = gradients.mean(
        dim=1
    )

    # Weighted combination of hidden-state channels
    cam = torch.sum(
        activations *
        channel_weights.unsqueeze(1),
        dim=-1
    )

    # Standard Grad-CAM ReLU
    cam = torch.relu(
        cam
    )

    cam = cam.squeeze(
        0
    ).cpu().numpy()

    attention_weights_np = (
        attention_weights
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
    )

    input_ids_np = (
        input_ids
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
    )

    # --------------------------------------------------------
    # Normalize Grad-CAM
    # --------------------------------------------------------

    valid_mask = (
        attention_mask
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
        .astype(bool)
    )

    valid_cam = cam[
        valid_mask
    ]

    if (
        valid_cam.max()
        -
        valid_cam.min()
    ) > 1e-12:

        cam_normalized = (
            cam
            -
            valid_cam.min()
        ) / (
            valid_cam.max()
            -
            valid_cam.min()
        )

    else:

        cam_normalized = (
            np.zeros_like(cam)
        )

    # Padding tokens must never receive attribution
    cam_normalized[
        ~valid_mask
    ] = 0.0

    return {
        "cam": cam_normalized,
        "attention_weights":
            attention_weights_np,
        "input_ids":
            input_ids_np,
        "attention_mask":
            valid_mask,
        "positive_logit":
            float(
                target_score.detach().cpu().item()
            ),
        "logits":
            logits.detach()
            .cpu()
            .numpy()
            .squeeze(0)
    }


# ============================================================
# TOKEN PROCESSING
# ============================================================

def get_token_information(
    result
):

    token_ids = (
        result["input_ids"]
    )

    cam = (
        result["cam"]
    )

    attention_mask = (
        result["attention_mask"]
    )

    tokens = tokenizer.convert_ids_to_tokens(
        token_ids.tolist()
    )

    rows = []

    for position, (
        token,
        importance,
        valid
    ) in enumerate(
        zip(
            tokens,
            cam,
            attention_mask
        )
    ):

        if not valid:
            continue

        # Ignore special tokens
        if token in tokenizer.all_special_tokens:
            continue

        rows.append({

            "position":
                position,

            "token":
                token,

            "importance":
                float(importance)

        })

    token_df = pd.DataFrame(
        rows
    )

    token_df = token_df.sort_values(
        "importance",
        ascending=False
    ).reset_index(
        drop=True
    )

    token_df["rank"] = (
        np.arange(
            len(token_df)
        ) + 1
    )

    return token_df


# ============================================================
# RUN GRAD-CAM FOR TP/TN/FP/FN
# ============================================================

gradcam_results = {}

print()
print("=" * 70)
print("GENERATING GRAD-CAM EXPLANATIONS")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print()
    print(
        f"Processing {case_name} "
        f"(test index {test_index})..."
    )

    result = compute_bigbird_gradcam(
        test_index
    )

    token_df = get_token_information(
        result
    )

    gradcam_results[
        case_name
    ] = {

        "test_index":
            test_index,

        "result":
            result,

        "token_df":
            token_df

    }

    print(
        f"✓ {case_name} Grad-CAM complete"
    )

# ============================================================
# SAVE TOKEN IMPORTANCE
# ============================================================

all_gradcam_rows = []

for case_name, data in gradcam_results.items():

    token_df = data[
        "token_df"
    ]

    for _, row in token_df.iterrows():

        all_gradcam_rows.append({

            "case":
                case_name,

            "test_index":
                data["test_index"],

            "true_label":
                int(
                    labels_np[
                        data["test_index"]
                    ]
                ),

            "predicted_label":
                int(
                    preds_np[
                        data["test_index"]
                    ]
                ),

            "positive_probability":
                float(
                    probs_np[
                        data["test_index"]
                    ]
                ),

            "position":
                int(
                    row["position"]
                ),

            "token":
                row["token"],

            "gradcam_importance":
                row["importance"],

            "rank":
                int(
                    row["rank"]
                )

        })

gradcam_results_df = pd.DataFrame(
    all_gradcam_rows
)

gradcam_csv_path = os.path.join(
    GRADCAM_DIR,
    "bigbird_aggressive_gradcam_token_importance.csv"
)

gradcam_results_df.to_csv(
    gradcam_csv_path,
    index=False
)

# ============================================================
# PAPER-READY TOP TOKEN PLOTS
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM TOKEN IMPORTANCE")
print("=" * 70)

for case_name, data in gradcam_results.items():

    token_df = data[
        "token_df"
    ].head(
        GRADCAM_TOP_TOKENS
    ).copy()

    # Reverse order so most important is at top
    token_df = token_df.iloc[::-1]

    plt.figure(
        figsize=(10, 8)
    )

    plt.barh(
        token_df["token"],
        token_df["importance"]
    )

    plt.xlabel(
        "Grad-CAM Importance",
        fontsize=12
    )

    plt.ylabel(
        "Token",
        fontsize=12
    )

    plt.title(
        f"BigBird + Aggressive — "
        f"Grad-CAM Token Importance ({case_name})",
        fontsize=14
    )

    plt.grid(
        axis="x",
        alpha=0.20
    )

    plt.tight_layout()

    plot_path = os.path.join(
        GRADCAM_DIR,
        f"bigbird_aggressive_gradcam_{case_name}.png"
    )

    # SAVE
    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )

    # DISPLAY
    plt.show()

    plt.close()

# ============================================================
# SAVE TOP TOKEN TABLE
# ============================================================

top_token_rows = []

for case_name, data in gradcam_results.items():

    top_tokens = data[
        "token_df"
    ].head(
        GRADCAM_TOP_TOKENS
    )

    for _, row in top_tokens.iterrows():

        top_token_rows.append({

            "case":
                case_name,

            "rank":
                int(
                    row["rank"]
                ),

            "token":
                row["token"],

            "gradcam_importance":
                float(
                    row["importance"]
                )

        })

top_token_df = pd.DataFrame(
    top_token_rows
)

top_token_df.to_csv(
    os.path.join(
        GRADCAM_DIR,
        "bigbird_aggressive_gradcam_top_tokens.csv"
    ),
    index=False
)

# ============================================================
# SAVE CONFIGURATION
# ============================================================

gradcam_config = {

    "model":
        "BigBird",

    "cleaning":
        "Aggressive",

    "dataset":
        "Held-out test set",

    "target":
        "Positive/depression class logit",

    "layer":
        "Final Transformer hidden states",

    "hidden_dimension":
        768,

    "method":
        "Transformer Grad-CAM",

    "cases":
        [
            "TP",
            "TN",
            "FP",
            "FN"
        ],

    "top_tokens_displayed":
        GRADCAM_TOP_TOKENS,

    "random_seed":
        GRADCAM_SEED,

    "model_training":
        "No additional training",

    "transformer_weights":
        "Frozen"

}

with open(
    os.path.join(
        GRADCAM_DIR,
        "bigbird_aggressive_gradcam_config.json"
    ),
    "w"
) as f:

    json.dump(
        gradcam_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ Same TP/TN/FP/FN users as LIME"
)

print(
    "✓ Final Transformer hidden states used"
)

print(
    "✓ Positive-class gradients computed"
)

print(
    "✓ BigBird weights remained frozen"
)

print(
    "✓ No additional model training"
)

print(
    "✓ Token importance calculated"
)

print(
    "✓ Top-20 token plots saved and displayed"
)

print(
    "✓ Token importance CSV saved"
)

print(
    "✓ Configuration saved"
)

print()
print(
    "Grad-CAM output directory:",
    GRADCAM_DIR
)
```
