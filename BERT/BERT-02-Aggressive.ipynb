```python
# ============================================================
# CELL 3 — BERT + AGGRESSIVE
# EXACT D1 ANCHOR DETECTION
# ============================================================

import os
import re
import json
import glob

from collections import defaultdict

print("=" * 60)
print("BERT + AGGRESSIVE — EXACT D1 ANCHOR DETECTION")
print("=" * 60)

# ============================================================
# D1 TIMELINE DIRECTORY
# ============================================================

DATASET_ROOT = (
    "/kaggle/input/datasets/"
    "vedantnawghare35/"
    "depression-detection-mddl-final-pbl/"
    "Dataset_MDDL"
)

D1_TIMELINE_DIR = os.path.join(
    DATASET_ROOT,
    "labeled",
    "positive",
    "data",
    "timeline"
)

assert os.path.exists(D1_TIMELINE_DIR)

# ============================================================
# FIND D1 TIMELINE FILES
# ============================================================

d1_timeline_files = glob.glob(
    os.path.join(
        D1_TIMELINE_DIR,
        "*.json"
    )
)

print()
print(
    "D1 timeline files found:",
    len(d1_timeline_files)
)

assert len(d1_timeline_files) == 2626

# ============================================================
# EXACT ANCHOR PATTERNS
# ============================================================

REFINED_ANCHOR_PATTERNS = [

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+have\s+(?:major\s+)?depression\b",

    r"\bi\s+suffer\s+from\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|am|'m)\s+diagnosed\b.{0,80}\bdepress",

    (
        r"\b(?:my|a)\s+(?:doctor|psychiatrist)"
        r"\s+(?:has\s+)?diagnosed\s+me\b"
        r".{0,80}\bdepress"
    )
]

compiled_anchor_patterns = [
    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )
    for pattern in REFINED_ANCHOR_PATTERNS
]

# ============================================================
# EXCLUSION PATTERNS
# ============================================================

EXCLUSION_PATTERNS = [

    r"\bmy\s+friend\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+sister\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+brother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+mother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+father\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+answer\s+to\s*:",

    r"\banswer\s+to\s*:",

    r"\bquestion\s*:",

    r"\bq\s*:"
]

compiled_exclusion_patterns = [
    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )
    for pattern in EXCLUSION_PATTERNS
]

# ============================================================
# EXACT ANCHOR DETECTOR
# ============================================================

def is_refined_anchor_tweet(text):

    if text is None:
        return False

    text = str(text).strip()

    if not text:
        return False

    for pattern in compiled_exclusion_patterns:
        if pattern.search(text):
            return False

    for pattern in compiled_anchor_patterns:
        if pattern.search(text):
            return True

    return False

# ============================================================
# POSITIVE USERS BY SPLIT
# ============================================================

train_positive_users = set(
    train_df.loc[
        train_df["label"] == 1,
        "username"
    ].astype(str)
)

val_positive_users = set(
    val_df.loc[
        val_df["label"] == 1,
        "username"
    ].astype(str)
)

test_positive_users = set(
    test_df.loc[
        test_df["label"] == 1,
        "username"
    ].astype(str)
)

all_positive_users = (
    train_positive_users
    |
    val_positive_users
    |
    test_positive_users
)

print()
print("=" * 60)
print("POSITIVE USERS BY SPLIT")
print("=" * 60)

print("TRAIN:", len(train_positive_users))
print("VALIDATION:", len(val_positive_users))
print("TEST:", len(test_positive_users))

assert len(train_positive_users) == 2013
assert len(val_positive_users) == 252
assert len(test_positive_users) == 251
assert len(all_positive_users) == 2516

# ============================================================
# INDEX TIMELINE FILES
# ============================================================

timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:

    filename = os.path.basename(
        timeline_file
    )

    username = filename.split(
        "created_at"
    )[0]

    timeline_files_by_user[
        username
    ].append(
        timeline_file
    )

assert len(timeline_files_by_user) == 2516

# ============================================================
# SCAN ALL POSITIVE D1 TIMELINES
# ============================================================

anchor_records = []

users_processed = set()
tweets_scanned = 0

for username in sorted(
    all_positive_users
):

    if username not in timeline_files_by_user:
        raise KeyError(
            f"Missing D1 timeline for user: {username}"
        )

    users_processed.add(username)

    for timeline_file in timeline_files_by_user[username]:

        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:

                line = line.strip()

                if not line:
                    continue

                record = json.loads(line)

                tweets_scanned += 1

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get("id", "")
                    )
                )

                tweet_text = str(
                    record.get(
                        "text",
                        ""
                    )
                )

                if (
                    tweet_id
                    and
                    is_refined_anchor_tweet(
                        tweet_text
                    )
                ):

                    anchor_records.append(
                        {
                            "username": username,
                            "tweet_id": tweet_id,
                            "created_at":
                                record.get(
                                    "created_at"
                                ),
                            "text": tweet_text
                        }
                    )

# ============================================================
# UNIQUE ANCHOR PAIRS
# ============================================================

anchor_pairs = {
    (
        record["username"],
        record["tweet_id"]
    )
    for record in anchor_records
}

# ============================================================
# SPLIT-WISE UNIQUE ANCHORS
# ============================================================

train_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in train_positive_users
}

val_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in val_positive_users
}

test_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in test_positive_users
}

# ============================================================
# SUMMARY
# ============================================================

print()
print("=" * 60)
print("EXACT ANCHOR SUMMARY")
print("=" * 60)

print(
    "Positive users processed:",
    len(users_processed)
)

print(
    "Tweets scanned:",
    tweets_scanned
)

print(
    "Raw anchor records:",
    len(anchor_records)
)

print(
    "Unique anchor pairs:",
    len(anchor_pairs)
)

print()
print(
    "TRAIN unique anchors:",
    len(train_anchor_pairs)
)

print(
    "VAL unique anchors:",
    len(val_anchor_pairs)
)

print(
    "TEST unique anchors:",
    len(test_anchor_pairs)
)

# ============================================================
# EXACT CONTROL ASSERTIONS
# ============================================================

assert len(users_processed) == 2516
assert len(anchor_records) == 1920
assert len(anchor_pairs) == 1847
assert len(train_anchor_pairs) == 1476
assert len(val_anchor_pairs) == 190
assert len(test_anchor_pairs) == 181

# ============================================================
# SAVE
# ============================================================

bert_aggressive_anchor_df = pd.DataFrame(
    anchor_records
)

bert_aggressive_anchor_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "exact_anchor_records.csv"
    ),
    index=False
)

pd.DataFrame(
    list(anchor_pairs),
    columns=[
        "username",
        "tweet_id"
    ]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "unique_anchor_pairs.csv"
    ),
    index=False
)

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 60)
print("CELL 3 COMPLETE")
print("=" * 60)

print("✓ Exact anchor detector reproduced")
print("✓ 1,847 unique anchors confirmed")
print("✓ 1,920 anchor records confirmed")
print("✓ TRAIN = 1,476 unique anchors")
print("✓ VAL = 190 unique anchors")
print("✓ TEST = 181 unique anchors")
print("✓ No D2 users included")
print("✓ Anchor records saved")
print("✓ No timelines modified yet")
```

```python
# ============================================================
# CELL 4 — BERT + AGGRESSIVE
# EXACT D1 ANCHOR REMOVAL
# ============================================================

print("=" * 60)
print("BERT + AGGRESSIVE — EXACT D1 ANCHOR REMOVAL")
print("=" * 60)

# ============================================================
# PRE-REMOVAL ANCHOR VERIFICATION
# ============================================================

print()
print("=" * 60)
print("PRE-REMOVAL ANCHOR VERIFICATION")
print("=" * 60)

duplicate_anchor_occurrences = (
    len(anchor_records)
    - len(anchor_pairs)
)

print(
    "Raw anchor records:",
    len(anchor_records)
)

print(
    "Unique anchor pairs:",
    len(anchor_pairs)
)

print(
    "Duplicate anchor occurrences:",
    duplicate_anchor_occurrences
)

assert len(anchor_records) == 1920
assert len(anchor_pairs) == 1847
assert duplicate_anchor_occurrences == 73

print("✓ 1,847 unique anchors verified")
print("✓ 73 duplicate occurrences verified")
print("✓ 1,920 total anchor records verified")


# ============================================================
# SPLIT-LEVEL ANCHOR RECORD COUNTS
# ============================================================

train_anchor_records = [
    record
    for record in anchor_records
    if record["username"] in train_positive_users
]

val_anchor_records = [
    record
    for record in anchor_records
    if record["username"] in val_positive_users
]

test_anchor_records = [
    record
    for record in anchor_records
    if record["username"] in test_positive_users
]

print()
print("=" * 60)
print("SPLIT-LEVEL ANCHOR RECORD COUNTS")
print("=" * 60)

print(
    "TRAIN:",
    len(train_anchor_records),
    "| expected:",
    1535
)

print(
    "VAL:",
    len(val_anchor_records),
    "| expected:",
    199
)

print(
    "TEST:",
    len(test_anchor_records),
    "| expected:",
    186
)

assert len(train_anchor_records) == 1535
assert len(val_anchor_records) == 199
assert len(test_anchor_records) == 186


# ============================================================
# EXACT ANCHOR SET
# ============================================================

EXACT_ANCHOR_SET = set(anchor_pairs)


# ============================================================
# TIMELINE FILE INDEX
# ============================================================

timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:

    filename = os.path.basename(
        timeline_file
    )

    username = filename.split(
        "created_at"
    )[0]

    timeline_files_by_user[
        username
    ].append(
        timeline_file
    )

assert len(
    timeline_files_by_user
) == 2516


# ============================================================
# BUILD ANCHOR-REMOVED TEXT
# ============================================================

def build_anchor_removed_text(username):

    kept_texts = []
    removed_count = 0

    for timeline_file in (
        timeline_files_by_user[username]
    ):

        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:

                line = line.strip()

                if not line:
                    continue

                record = json.loads(line)

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get(
                            "id",
                            ""
                        )
                    )
                )

                pair = (
                    str(username),
                    tweet_id
                )

                if pair in EXACT_ANCHOR_SET:

                    removed_count += 1
                    continue

                tweet_text = str(
                    record.get(
                        "text",
                        ""
                    )
                ).strip()

                if tweet_text:

                    kept_texts.append(
                        tweet_text
                    )

    return (
        " ".join(kept_texts),
        removed_count
    )


# ============================================================
# COPY ORIGINAL DATAFRAMES
# ============================================================

train_bert_aggressive_df = (
    train_df.copy(deep=True)
)

val_bert_aggressive_df = (
    val_df.copy(deep=True)
)

test_bert_aggressive_df = (
    test_df.copy(deep=True)
)


# ============================================================
# BUILD ANCHOR-REMOVED TEXT
# ============================================================

positive_anchor_removed_text = {}

train_removed = 0
val_removed = 0
test_removed = 0


for username in sorted(
    all_positive_users
):

    anchor_removed_text, removed_count = (
        build_anchor_removed_text(
            username
        )
    )

    positive_anchor_removed_text[
        username
    ] = anchor_removed_text

    if username in train_positive_users:

        train_removed += removed_count

    elif username in val_positive_users:

        val_removed += removed_count

    elif username in test_positive_users:

        test_removed += removed_count


# ============================================================
# FINAL REMOVAL COUNTS
# ============================================================

print()
print("=" * 60)
print("FINAL ANCHOR REMOVAL COUNTS")
print("=" * 60)

print(
    "TRAIN removed:",
    train_removed
)

print(
    "VAL removed:",
    val_removed
)

print(
    "TEST removed:",
    test_removed
)

total_removed = (
    train_removed
    + val_removed
    + test_removed
)

print(
    "TOTAL removed:",
    total_removed
)

assert train_removed == 1535
assert val_removed == 199
assert test_removed == 186
assert total_removed == 1920

print(
    "✓ Record-level removal exactly matches previous experiments"
)


# ============================================================
# ASSIGN ANCHOR-REMOVED TEXT
# ============================================================

for df in [
    train_bert_aggressive_df,
    val_bert_aggressive_df,
    test_bert_aggressive_df
]:

    positive_mask = (
        df["label"] == 1
    )

    df.loc[
        positive_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            positive_mask,
            "username"
        ]
        .astype(str)
        .map(
            positive_anchor_removed_text
        )
        .fillna("")
        .values
    )

    negative_mask = (
        df["label"] == 0
    )

    df.loc[
        negative_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            negative_mask,
            "timeline_text"
        ]
        .fillna("")
        .astype(str)
        .values
    )


# ============================================================
# D2 PRESERVATION CHECK
# ============================================================

print()
print("=" * 60)
print("D2 PRESERVATION CHECK")
print("=" * 60)

for (
    original_df,
    result_df,
    split_name
) in [

    (
        train_df,
        train_bert_aggressive_df,
        "TRAIN"
    ),

    (
        val_df,
        val_bert_aggressive_df,
        "VALIDATION"
    ),

    (
        test_df,
        test_bert_aggressive_df,
        "TEST"
    )

]:

    original_d2 = (
        original_df.loc[
            original_df["label"] == 0,
            [
                "username",
                "timeline_text"
            ]
        ]
        .set_index("username")
        .sort_index()
    )

    result_d2 = (
        result_df.loc[
            result_df["label"] == 0,
            [
                "username",
                "anchor_removed_text"
            ]
        ]
        .rename(
            columns={
                "anchor_removed_text":
                    "timeline_text"
            }
        )
        .set_index("username")
        .sort_index()
    )

    assert original_d2.equals(
        result_d2
    )

    print(
        f"✓ {split_name} D2 unchanged"
    )


# ============================================================
# USER + LABEL PRESERVATION
# ============================================================

print()
print("=" * 60)
print("USER + LABEL PRESERVATION")
print("=" * 60)

for (
    original_df,
    result_df,
    split_name
) in [

    (
        train_df,
        train_bert_aggressive_df,
        "TRAIN"
    ),

    (
        val_df,
        val_bert_aggressive_df,
        "VALIDATION"
    ),

    (
        test_df,
        test_bert_aggressive_df,
        "TEST"
    )

]:

    assert (
        original_df["user_id"].tolist()
        ==
        result_df["user_id"].tolist()
    )

    assert (
        original_df["username"].tolist()
        ==
        result_df["username"].tolist()
    )

    assert (
        original_df["label"].tolist()
        ==
        result_df["label"].tolist()
    )

    print(
        f"✓ {split_name}: "
        "users and labels unchanged"
    )


# ============================================================
# ANCHOR-REMOVED TEXT INTEGRITY
# ============================================================

print()
print("=" * 60)
print("ANCHOR-REMOVED TEXT INTEGRITY")
print("=" * 60)

for (
    split_name,
    df
) in [

    (
        "TRAIN",
        train_bert_aggressive_df
    ),

    (
        "VALIDATION",
        val_bert_aggressive_df
    ),

    (
        "TEST",
        test_bert_aggressive_df
    )

]:

    empty_count = int(
        (
            df[
                "anchor_removed_text"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
            ==
            ""
        ).sum()
    )

    print(
        f"{split_name}: "
        f"empty anchor-removed texts = "
        f"{empty_count}"
    )


# ============================================================
# SAVE
# ============================================================

train_bert_aggressive_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "train_anchor_removed.pkl"
    )
)

val_bert_aggressive_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "val_anchor_removed.pkl"
    )
)

test_bert_aggressive_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "test_anchor_removed.pkl"
    )
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 4 COMPLETE")
print("=" * 60)

print(
    "✓ 1,847 unique D1 anchors removed"
)

print(
    "✓ 73 duplicate occurrences handled"
)

print(
    "✓ 1,920 total anchor records removed"
)

print(
    "✓ TRAIN = 1,535 removed"
)

print(
    "✓ VAL = 199 removed"
)

print(
    "✓ TEST = 186 removed"
)

print(
    "✓ D2 completely preserved"
)

print(
    "✓ User IDs preserved"
)

print(
    "✓ Labels preserved"
)

print(
    "✓ Anchor-removed source saved"
)
```

```python
# ============================================================
# CELL 5 — BERT + AGGRESSIVE
# ============================================================

import re
import html

print("=" * 60)
print("BERT + AGGRESSIVE — TEXT CLEANING")
print("=" * 60)

def aggressive_clean(text):
    text = str(text)
    text = html.unescape(text)
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"@\w+", " ", text)
    text = re.sub(r"#", " ", text)
    text = re.sub(r"&\w+;|<[^>]+>", " ", text)
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)
    text = re.sub(r"_+", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip().lower()

# Apply aggressive cleaning
train_bert_aggressive_df["model_text"] = (
    train_bert_aggressive_df["anchor_removed_text"]
    .fillna("")
    .map(aggressive_clean)
)

val_bert_aggressive_df["model_text"] = (
    val_bert_aggressive_df["anchor_removed_text"]
    .fillna("")
    .map(aggressive_clean)
)

test_bert_aggressive_df["model_text"] = (
    test_bert_aggressive_df["anchor_removed_text"]
    .fillna("")
    .map(aggressive_clean)
)

print()
print("Cleaning applied:")
print("✓ URLs removed")
print("✓ Mentions removed")
print("✓ Hashtag symbols removed")
print("✓ HTML / markup removed")
print("✓ Punctuation removed")
print("✓ Emojis / emoticons removed")
print("✓ Special symbols removed")
print("✓ Whitespace normalized")
print("✓ Text lowercased")

# Check text integrity
print()
print("=" * 60)
print("MODEL TEXT INTEGRITY")
print("=" * 60)

for name, df in [
    ("TRAIN", train_bert_aggressive_df),
    ("VALIDATION", val_bert_aggressive_df),
    ("TEST", test_bert_aggressive_df)
]:
    empty_count = int(
        df["model_text"].str.strip().eq("").sum()
    )
    print(f"{name}: empty model texts = {empty_count}")

# User + label preservation
for original, result, name in [
    (train_df, train_bert_aggressive_df, "TRAIN"),
    (val_df, val_bert_aggressive_df, "VALIDATION"),
    (test_df, test_bert_aggressive_df, "TEST")
]:
    assert original["user_id"].tolist() == result["user_id"].tolist()
    assert original["username"].tolist() == result["username"].tolist()
    assert original["label"].tolist() == result["label"].tolist()
    print(f"✓ {name}: users and labels unchanged")

# Save
train_bert_aggressive_df.to_pickle(
    os.path.join(RESULT_DIR, "train_cleaned.pkl")
)

val_bert_aggressive_df.to_pickle(
    os.path.join(RESULT_DIR, "val_cleaned.pkl")
)

test_bert_aggressive_df.to_pickle(
    os.path.join(RESULT_DIR, "test_cleaned.pkl")
)

print()
print("=" * 60)
print("CELL 5 COMPLETE")
print("=" * 60)

print("✓ Aggressive cleaning applied")
print("✓ Anchor-removed source used")
print("✓ Model text created")
print("✓ Users preserved")
print("✓ Labels preserved")
print("✓ Cleaned data saved")
```

```python
# ============================================================
# CELL 6 — BERT + AGGRESSIVE
# TOKEN LENGTH ANALYSIS
# ============================================================

from transformers import AutoTokenizer
import numpy as np
import pandas as pd

print("=" * 60)
print("BERT + AGGRESSIVE — TOKEN LENGTH ANALYSIS")
print("=" * 60)

print()
print("Tokenizer:", MODEL_NAME)
print("Experimental context length:", MAX_LENGTH)

from kaggle_secrets import UserSecretsClient

user_secrets = UserSecretsClient()
hf_token = user_secrets.get_secret("HF_TOKEN")

print()
print("✓ HF_TOKEN retrieved")

bert_aggressive_tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    token=hf_token
)

print()
print("Tokenizer max length:",
      bert_aggressive_tokenizer.model_max_length)
print("✓ BERT tokenizer loaded")


def get_token_lengths(texts, split_name):
    print()
    print(f"Processing {split_name}...")

    lengths = []

    for i, text in enumerate(texts):
        encoded = bert_aggressive_tokenizer(
            str(text),
            add_special_tokens=True,
            truncation=False,
            padding=False
        )

        lengths.append(len(encoded["input_ids"]))

        if (i + 1) % 1000 == 0 or (i + 1) == len(texts):
            print(f"{split_name}: {i + 1}/{len(texts)}")

    return np.array(lengths, dtype=np.int64)


train_token_lengths = get_token_lengths(
    train_bert_aggressive_df["model_text"].fillna("").tolist(),
    "TRAIN"
)

val_token_lengths = get_token_lengths(
    val_bert_aggressive_df["model_text"].fillna("").tolist(),
    "VALIDATION"
)

test_token_lengths = get_token_lengths(
    test_bert_aggressive_df["model_text"].fillna("").tolist(),
    "TEST"
)


def show_stats(lengths, split_name):
    print()
    print("=" * 60)
    print(f"{split_name} TOKEN STATISTICS")
    print("=" * 60)

    print("Minimum:", int(lengths.min()))
    print("Maximum:", int(lengths.max()))
    print("Mean:", round(float(lengths.mean()), 2))
    print("Median:", round(float(np.median(lengths)), 2))
    print("25th percentile:", round(float(np.percentile(lengths, 25)), 2))
    print("75th percentile:", round(float(np.percentile(lengths, 75)), 2))
    print("90th percentile:", round(float(np.percentile(lengths, 90)), 2))
    print("95th percentile:", round(float(np.percentile(lengths, 95)), 2))
    print("99th percentile:", round(float(np.percentile(lengths, 99)), 2))

    within_512 = int(np.sum(lengths <= 512))
    above_512 = int(np.sum(lengths > 512))

    print()
    print("512-TOKEN COVERAGE")
    print("Users <= 512:", within_512)
    print("Users > 512:", above_512)
    print(
        "Percentage <= 512:",
        round(100 * within_512 / len(lengths), 2),
        "%"
    )
    print(
        "Percentage > 512:",
        round(100 * above_512 / len(lengths), 2),
        "%"
    )

    within_4096 = int(np.sum(lengths <= 4096))
    above_4096 = int(np.sum(lengths > 4096))

    print()
    print("4096-TOKEN REFERENCE")
    print("Users <= 4096:", within_4096)
    print("Users > 4096:", above_4096)
    print(
        "Percentage <= 4096:",
        round(100 * within_4096 / len(lengths), 2),
        "%"
    )
    print(
        "Percentage > 4096:",
        round(100 * above_4096 / len(lengths), 2),
        "%"
    )


show_stats(train_token_lengths, "TRAIN")
show_stats(val_token_lengths, "VALIDATION")
show_stats(test_token_lengths, "TEST")


print()
print("=" * 60)
print("ESTIMATED CONTEXT CHUNKS")
print("=" * 60)

for context_length in [512, 4096]:
    print()
    print("Context length:", context_length)

    train_chunks = np.ceil(train_token_lengths / context_length)
    val_chunks = np.ceil(val_token_lengths / context_length)
    test_chunks = np.ceil(test_token_lengths / context_length)

    print(
        "TRAIN mean chunks:",
        round(float(train_chunks.mean()), 2)
    )
    print(
        "TRAIN maximum chunks:",
        int(train_chunks.max())
    )
    print(
        "VALIDATION mean chunks:",
        round(float(val_chunks.mean()), 2)
    )
    print(
        "VALIDATION maximum chunks:",
        int(val_chunks.max())
    )
    print(
        "TEST mean chunks:",
        round(float(test_chunks.mean()), 2)
    )
    print(
        "TEST maximum chunks:",
        int(test_chunks.max())
    )


token_summary = pd.DataFrame([
    {
        "Split": "Train",
        "Users": len(train_token_lengths),
        "Mean": train_token_lengths.mean(),
        "Median": np.median(train_token_lengths),
        "<=512 (%)": 100 * np.mean(train_token_lengths <= 512),
        "<=4096 (%)": 100 * np.mean(train_token_lengths <= 4096)
    },
    {
        "Split": "Validation",
        "Users": len(val_token_lengths),
        "Mean": val_token_lengths.mean(),
        "Median": np.median(val_token_lengths),
        "<=512 (%)": 100 * np.mean(val_token_lengths <= 512),
        "<=4096 (%)": 100 * np.mean(val_token_lengths <= 4096)
    },
    {
        "Split": "Test",
        "Users": len(test_token_lengths),
        "Mean": test_token_lengths.mean(),
        "Median": np.median(test_token_lengths),
        "<=512 (%)": 100 * np.mean(test_token_lengths <= 512),
        "<=4096 (%)": 100 * np.mean(test_token_lengths <= 4096)
    }
])

print()
print("=" * 60)
print("TOKEN-LENGTH SUMMARY")
print("=" * 60)

print(
    token_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.2f}"
    )
)


np.save(
    os.path.join(
        RESULT_DIR,
        "train_token_lengths.npy"
    ),
    train_token_lengths
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_token_lengths.npy"
    ),
    val_token_lengths
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_token_lengths.npy"
    ),
    test_token_lengths
)

token_summary.to_csv(
    os.path.join(
        RESULT_DIR,
        "token_length_summary.csv"
    ),
    index=False
)


print()
print("=" * 60)
print("CELL 6 COMPLETE")
print("=" * 60)

print("✓ Authenticated BERT tokenizer loaded")
print("✓ Aggressive cleaned text analyzed")
print("✓ No truncation used during analysis")
print("✓ 512-token coverage measured")
print("✓ 4096-token reference measured")
print("✓ Context chunk estimates calculated")
print("✓ Token-length files saved")
```

```python
# ============================================================
# CELL 11 — BERT + AGGRESSIVE
# BUILDING FROZEN BERT + METADATA MODEL
# ============================================================

from transformers import AutoModel
import torch.nn as nn

print("=" * 60)
print("BERT + AGGRESSIVE — BUILDING FROZEN BERT + METADATA MODEL")
print("=" * 60)

bert_encoder = AutoModel.from_pretrained(
    MODEL_NAME,
    token=hf_token
).to(device)

for p in bert_encoder.parameters():
    p.requires_grad = False


class BERTMetadataClassifier(nn.Module):
    def __init__(self, encoder, metadata_dim=53):
        super().__init__()

        self.encoder = encoder
        self.text_hidden_size = 768

        self.attention_projection = nn.Linear(
            768, 256
        )

        self.attention_score = nn.Linear(
            256, 1
        )

        self.metadata_branch = nn.Sequential(
            nn.Linear(metadata_dim, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, 64),
            nn.ReLU()
        )

        self.metadata_beta = nn.Parameter(
            torch.tensor(0.0)
        )

        self.classifier = nn.Sequential(
            nn.Linear(832, 512),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, 2)
        )

    def forward(
        self,
        input_ids,
        attention_mask,
        metadata
    ):
        with torch.no_grad():
            outputs = self.encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

        hidden_states = outputs.last_hidden_state

        projected = torch.tanh(
            self.attention_projection(hidden_states)
        )

        scores = self.attention_score(
            projected
        ).squeeze(-1)

        scores = scores.masked_fill(
            attention_mask == 0,
            -1e9
        )

        attention_weights = torch.softmax(
            scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        metadata_representation = (
            self.metadata_branch(metadata)
        )

        metadata_gate = torch.sigmoid(
            self.metadata_beta
        )

        gated_metadata = (
            metadata_gate *
            metadata_representation
        )

        fused = torch.cat(
            [
                text_representation,
                gated_metadata
            ],
            dim=1
        )

        logits = self.classifier(fused)

        return {
            "logits": logits,
            "text_representation":
                text_representation,
            "metadata_representation":
                metadata_representation,
            "attention_weights":
                attention_weights,
            "metadata_gate":
                metadata_gate
        }


model = BERTMetadataClassifier(
    bert_encoder,
    METADATA_DIM
).to(device)


# ============================================================
# PARAMETER COUNT
# ============================================================

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

frozen_params = sum(
    p.numel()
    for p in model.parameters()
    if not p.requires_grad
)

print()
print("=" * 60)
print("PARAMETER COUNT")
print("=" * 60)

print("Total parameters:", total_params)
print("Trainable parameters:", trainable_params)
print("Frozen parameters:", frozen_params)

print(
    "Total parameters (M):",
    round(total_params / 1e6, 3)
)

print(
    "Trainable parameters (M):",
    round(trainable_params / 1e6, 3)
)

print(
    "Frozen parameters (M):",
    round(frozen_params / 1e6, 3)
)

encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print("✓ All BERT backbone parameters frozen")


# ============================================================
# TRAINABLE COMPONENTS
# ============================================================

print()
print("=" * 60)
print("TRAINABLE COMPONENTS")
print("=" * 60)

for name, p in model.named_parameters():
    if p.requires_grad:
        print("✓", name)


# ============================================================
# ARCHITECTURE
# ============================================================

print()
print("=" * 60)
print("ARCHITECTURE CONFIGURATION")
print("=" * 60)

print("Transformer:", MODEL_NAME)
print("Base architecture: BERT-base-uncased")
print("Text hidden size: 768")
print("Attention projection: 256")
print("Metadata: 53 -> 128 -> 64")
print("Gate: g = sigmoid(beta)")
print("Initial beta: 0.0")
print("Initial gate: 0.5")
print("Fusion: 768 + 64 = 832")
print("Classifier: 832 -> 512 -> 256 -> 2")
print("Dropout: 0.3")
print("Transformer frozen: True")


# ============================================================
# FORWARD PASS SANITY CHECK
# ============================================================

print()
print("=" * 60)
print("FORWARD PASS SANITY CHECK")
print("=" * 60)

model.eval()

batch = next(iter(train_loader))

input_ids = batch["input_ids"].to(device)
attention_mask = batch["attention_mask"].to(device)
metadata = batch["metadata"].to(device)

with torch.no_grad():
    output = model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        metadata=metadata
    )

print(
    "Input IDs:",
    tuple(input_ids.shape)
)

print(
    "Attention mask:",
    tuple(attention_mask.shape)
)

print(
    "Metadata:",
    tuple(metadata.shape)
)

print(
    "Logits:",
    tuple(output["logits"].shape)
)

print(
    "Text representation:",
    tuple(
        output["text_representation"].shape
    )
)

print(
    "Metadata representation:",
    tuple(
        output["metadata_representation"].shape
    )
)

print(
    "Attention weights:",
    tuple(
        output["attention_weights"].shape
    )
)

gate = float(
    output["metadata_gate"].item()
)

print(
    "Metadata gate:",
    gate
)

padding_attention = output[
    "attention_weights"
][attention_mask == 0]

max_padding_attention = (
    float(padding_attention.max().item())
    if padding_attention.numel()
    else 0.0
)

print(
    "Maximum attention on padding:",
    max_padding_attention
)

assert output["logits"].shape == (16, 2)
assert output["text_representation"].shape == (16, 768)
assert output["metadata_representation"].shape == (16, 64)
assert output["attention_weights"].shape == (16, 512)
assert max_padding_attention == 0.0
assert abs(gate - 0.5) < 1e-6

encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print("✓ Transformer freeze verified")


# ============================================================
# SAVE CONFIG
# ============================================================

architecture_config = {
    "model_name": MODEL_NAME,
    "base_architecture": "BERT-base-uncased",
    "text_hidden_size": 768,
    "attention_projection": 256,
    "metadata_dimension": 53,
    "metadata_branch": "53 -> 128 -> 64",
    "initial_beta": 0.0,
    "initial_gate": 0.5,
    "fusion": "768 + 64 = 832",
    "classifier": "832 -> 512 -> 256 -> 2",
    "dropout": 0.3,
    "transformer_frozen": True
}

with open(
    os.path.join(
        RESULT_DIR,
        "architecture_config.json"
    ),
    "w"
) as f:
    json.dump(
        architecture_config,
        f,
        indent=2
    )


print()
print("=" * 60)
print("CELL 11 COMPLETE")
print("=" * 60)

print("✓ BERT-base-uncased loaded")
print("✓ Transformer backbone frozen")
print("✓ Attention pooling active")
print("✓ Metadata branch = 53 -> 128 -> 64")
print("✓ Single metadata gate active")
print("✓ Text representation = 768")
print("✓ Fusion = 768 + 64 = 832")
print("✓ Classifier = 832 -> 512 -> 256 -> 2")
print("✓ Padding attention masked")
print("✓ Forward pass successful")
print("✓ Frozen BERT + Aggressive model ready for training")
```

```python
# ============================================================
# CELL 12 — BERT + AGGRESSIVE
# MODEL TRAINING
# ============================================================

import os
import time
import copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

MAX_EPOCHS = 10
EARLY_STOPPING_PATIENCE = 2
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1

print("=" * 60)
print("STARTING BERT + AGGRESSIVE TRAINING")
print("=" * 60)

print()
print("Training users:", len(train_dataset))
print("Validation users:", len(val_dataset))
print("Test users:", len(test_dataset))
print("Maximum epochs:", MAX_EPOCHS)
print("Early stopping patience:", EARLY_STOPPING_PATIENCE)
print("Checkpoint metric: Validation F1")
print("Learning rate:", LEARNING_RATE)
print("Weight decay:", WEIGHT_DECAY)
print("Batch size:", BATCH_SIZE)
print("Transformer: FROZEN")

encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0
print()
print("✓ BERT backbone confirmed frozen")

trainable_params = [
    p for p in model.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_params,
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

total_steps = len(train_loader) * MAX_EPOCHS
warmup_steps = int(total_steps * WARMUP_RATIO)

def lr_lambda(step):
    if step < warmup_steps:
        return (step + 1) / max(1, warmup_steps)

    return max(
        0.0,
        (total_steps - step) /
        max(1, total_steps - warmup_steps)
    )

scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda
)

criterion = nn.CrossEntropyLoss()


def evaluate_model(model, loader):
    model.eval()

    total_loss = 0.0
    labels_all = []
    preds_all = []
    probs_all = []

    with torch.no_grad():
        for batch in loader:
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            metadata = batch["metadata"].to(device)
            labels = batch["labels"].to(device)

            output = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                metadata=metadata
            )

            logits = output["logits"]

            loss = criterion(
                logits,
                labels
            )

            total_loss += (
                loss.item() * labels.size(0)
            )

            probs = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            preds = torch.argmax(
                logits,
                dim=1
            )

            labels_all.extend(
                labels.cpu().numpy()
            )

            preds_all.extend(
                preds.cpu().numpy()
            )

            probs_all.extend(
                probs.cpu().numpy()
            )

    metrics = {
        "loss":
            total_loss / len(loader.dataset),

        "accuracy":
            accuracy_score(
                labels_all,
                preds_all
            ),

        "precision":
            precision_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "recall":
            recall_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "f1":
            f1_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "roc_auc":
            roc_auc_score(
                labels_all,
                probs_all
            )
    }

    return metrics


history = []
best_val_f1 = -np.inf
best_epoch = None
best_state_dict = None
epochs_without_improvement = 0

training_start = time.time()

for epoch in range(1, MAX_EPOCHS + 1):

    epoch_start = time.time()

    print()
    print("=" * 60)
    print(f"EPOCH {epoch}/{MAX_EPOCHS}")
    print("=" * 60)

    model.train()

    running_loss = 0.0
    train_labels = []
    train_preds = []

    for batch_idx, batch in enumerate(
        train_loader,
        start=1
    ):

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        metadata = batch["metadata"].to(device)
        labels = batch["labels"].to(device)

        optimizer.zero_grad()

        output = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = output["logits"]

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            trainable_params,
            max_norm=1.0
        )

        optimizer.step()
        scheduler.step()

        running_loss += (
            loss.item() * labels.size(0)
        )

        preds = torch.argmax(
            logits.detach(),
            dim=1
        )

        train_labels.extend(
            labels.detach().cpu().numpy()
        )

        train_preds.extend(
            preds.cpu().numpy()
        )

        if (
            batch_idx == 1
            or batch_idx % 50 == 0
            or batch_idx == len(train_loader)
        ):

            batch_acc = accuracy_score(
                labels.detach().cpu().numpy(),
                preds.cpu().numpy()
            )

            elapsed = time.time() - epoch_start

            print(
                f"Epoch {epoch} | "
                f"Batch {batch_idx}/{len(train_loader)} | "
                f"Loss: {loss.item():.4f} | "
                f"Accuracy: {batch_acc:.4f} | "
                f"LR: {optimizer.param_groups[0]['lr']:.2e} | "
                f"Time: {elapsed:.1f}s"
            )

    train_loss = (
        running_loss / len(train_loader.dataset)
    )

    train_accuracy = accuracy_score(
        train_labels,
        train_preds
    )

    train_precision = precision_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_recall = recall_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_f1 = f1_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    print()
    print(f"Epoch {epoch} — Validation")

    val_metrics = evaluate_model(
        model,
        val_loader
    )

    with torch.no_grad():
        metadata_gate = float(
            torch.sigmoid(
                model.metadata_beta
            ).item()
        )

    epoch_time = time.time() - epoch_start

    print()
    print("=" * 60)
    print(f"EPOCH {epoch} SUMMARY")
    print("=" * 60)

    print(f"Train Loss: {train_loss:.4f}")
    print(f"Train Accuracy: {train_accuracy:.4f}")
    print(f"Train Precision: {train_precision:.4f}")
    print(f"Train Recall: {train_recall:.4f}")
    print(f"Train F1: {train_f1:.4f}")

    print()
    print(f"Val Loss: {val_metrics['loss']:.4f}")
    print(f"Val Accuracy: {val_metrics['accuracy']:.4f}")
    print(f"Val Precision: {val_metrics['precision']:.4f}")
    print(f"Val Recall: {val_metrics['recall']:.4f}")
    print(f"Val F1: {val_metrics['f1']:.4f}")
    print(f"Val ROC-AUC: {val_metrics['roc_auc']:.4f}")
    print(f"Metadata gate: {metadata_gate:.6f}")
    print(f"Epoch Time: {epoch_time:.2f}s")

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "train_accuracy": train_accuracy,
        "train_precision": train_precision,
        "train_recall": train_recall,
        "train_f1": train_f1,
        "val_loss": val_metrics["loss"],
        "val_accuracy": val_metrics["accuracy"],
        "val_precision": val_metrics["precision"],
        "val_recall": val_metrics["recall"],
        "val_f1": val_metrics["f1"],
        "val_roc_auc": val_metrics["roc_auc"],
        "metadata_gate": metadata_gate,
        "learning_rate":
            optimizer.param_groups[0]["lr"],
        "epoch_time": epoch_time
    })

    if val_metrics["f1"] > best_val_f1:

        best_val_f1 = val_metrics["f1"]
        best_epoch = epoch

        best_state_dict = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }

        epochs_without_improvement = 0

        checkpoint_path = os.path.join(
            CHECKPOINT_DIR,
            "best_bert_aggressive.pt"
        )

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict":
                    best_state_dict,
                "optimizer_state_dict":
                    optimizer.state_dict(),
                "scheduler_state_dict":
                    scheduler.state_dict(),
                "best_val_f1":
                    best_val_f1,
                "experiment_name":
                    EXPERIMENT_NAME,
                "model_name":
                    MODEL_NAME,
                "cleaning_method":
                    CLEANING_METHOD
            },
            checkpoint_path
        )

        print()
        print("✓ NEW BEST CHECKPOINT")
        print(
            "Best Validation F1:",
            f"{best_val_f1:.4f}"
        )
        print(
            "Saved:",
            checkpoint_path
        )

    else:
        epochs_without_improvement += 1

        print()
        print("No validation F1 improvement.")
        print(
            "Epochs without improvement:",
            f"{epochs_without_improvement} / "
            f"{EARLY_STOPPING_PATIENCE}"
        )

    if (
        epochs_without_improvement
        >= EARLY_STOPPING_PATIENCE
    ):
        print()
        print("EARLY STOPPING TRIGGERED")
        break


assert best_state_dict is not None
assert best_epoch is not None

model.load_state_dict(best_state_dict)
model.to(device)
model.eval()

total_training_time = (
    time.time() - training_start
)

history_df = pd.DataFrame(history)

history_path = os.path.join(
    RESULT_DIR,
    "bert_aggressive_training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)

with torch.no_grad():
    learned_beta = float(
        model.metadata_beta.item()
    )

    learned_gate = float(
        torch.sigmoid(
            model.metadata_beta
        ).item()
    )

print()
print("=" * 60)
print("BERT + AGGRESSIVE TRAINING COMPLETE")
print("=" * 60)

print("Epochs completed:", len(history))
print("Best epoch:", best_epoch)
print(
    "Best Validation F1:",
    f"{best_val_f1:.4f}"
)
print(
    "Total training time:",
    f"{total_training_time:.2f} seconds"
)
print(
    "Best checkpoint:",
    os.path.join(
        CHECKPOINT_DIR,
        "best_bert_aggressive.pt"
    )
)
print(
    "Training history saved:",
    history_path
)

print()
print("Learned beta:", learned_beta)
print(
    "Metadata gate:",
    f"{learned_gate:.6f}"
)

print()
print("✓ BERT transformer remained frozen")
print("✓ Downstream layers trained")

print()
print("=" * 60)
print("CELL 12 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# CELL 13 — BERT + AGGRESSIVE
# FINAL TEST EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix,
    classification_report
)

print("=" * 60)
print("BERT + AGGRESSIVE — FINAL TEST EVALUATION")
print("=" * 60)

checkpoint_path = os.path.join(
    CHECKPOINT_DIR,
    "best_bert_aggressive.pt"
)

assert os.path.exists(checkpoint_path)

print()
print("✓ Best checkpoint found")

checkpoint = torch.load(
    checkpoint_path,
    map_location=device
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.to(device)
model.eval()

best_epoch = checkpoint["epoch"]
best_val_f1 = checkpoint["best_val_f1"]

print()
print("Best epoch:", best_epoch)
print(
    "Best validation F1:",
    f"{best_val_f1:.4f}"
)
print("Checkpoint:", checkpoint_path)

# ============================================================
# GATE + FREEZE CHECK
# ============================================================

with torch.no_grad():
    learned_beta = float(
        model.metadata_beta.item()
    )
    metadata_gate = float(
        torch.sigmoid(
            model.metadata_beta
        ).item()
    )

encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print("=" * 60)
print("GATE VERIFICATION")
print("=" * 60)

print("Learned beta:", learned_beta)
print(
    "Metadata gate g = sigmoid(beta):",
    f"{metadata_gate:.6f}"
)

print("✓ Single metadata gate verified")
print("✓ BERT backbone remains frozen")

# ============================================================
# TEST SET
# ============================================================

print()
print("=" * 60)
print("TEST SET INTEGRITY")
print("=" * 60)

assert len(test_dataset) == 767

print("Expected test users: 767")
print("Actual test users:", len(test_dataset))
print("✓ Test set size correct")

# ============================================================
# EVALUATION
# ============================================================

print()
print("=" * 60)
print("EVALUATING TEST SET")
print("=" * 60)

labels_all = []
preds_all = []
probs_all = []

total_test_loss = 0.0

with torch.no_grad():

    for batch_idx, batch in enumerate(
        test_loader,
        start=1
    ):

        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        metadata = batch["metadata"].to(device)
        labels = batch["labels"].to(device)

        output = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = output["logits"]

        loss = criterion(
            logits,
            labels
        )

        total_test_loss += (
            loss.item() * labels.size(0)
        )

        probs = torch.softmax(
            logits,
            dim=1
        )[:, 1]

        preds = torch.argmax(
            logits,
            dim=1
        )

        labels_all.extend(
            labels.cpu().numpy()
        )
        preds_all.extend(
            preds.cpu().numpy()
        )
        probs_all.extend(
            probs.cpu().numpy()
        )

        if (
            batch_idx == 1
            or batch_idx % 10 == 0
            or batch_idx == len(test_loader)
        ):
            print(
                f"Test batch "
                f"{batch_idx}/{len(test_loader)}"
            )

# ============================================================
# METRICS
# ============================================================

test_loss = (
    total_test_loss /
    len(test_loader.dataset)
)

test_accuracy = accuracy_score(
    labels_all,
    preds_all
)

test_precision = precision_score(
    labels_all,
    preds_all,
    zero_division=0
)

test_recall = recall_score(
    labels_all,
    preds_all,
    zero_division=0
)

test_f1 = f1_score(
    labels_all,
    preds_all,
    zero_division=0
)

test_roc_auc = roc_auc_score(
    labels_all,
    probs_all
)

cm = confusion_matrix(
    labels_all,
    preds_all
)

# ============================================================
# RESULTS
# ============================================================

print()
print("=" * 60)
print("BERT + AGGRESSIVE — FINAL TEST RESULTS")
print("=" * 60)

print(f"Test Loss: {test_loss:.4f}")
print(
    f"Test Accuracy: "
    f"{test_accuracy * 100:.2f}%"
)
print(
    f"Test Precision: "
    f"{test_precision * 100:.2f}%"
)
print(
    f"Test Recall: "
    f"{test_recall * 100:.2f}%"
)
print(
    f"Test F1: "
    f"{test_f1 * 100:.2f}%"
)
print(
    f"Test ROC-AUC: "
    f"{test_roc_auc * 100:.2f}%"
)

# ============================================================
# CONFUSION MATRIX
# ============================================================

print()
print("=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

print("                 Predicted")
print("                 0       1")
print(
    f"Actual 0        "
    f"{cm[0,0]:<7}{cm[0,1]}"
)
print(
    f"Actual 1        "
    f"{cm[1,0]:<7}{cm[1,1]}"
)

# ============================================================
# CLASSIFICATION REPORT
# ============================================================

print()
print("=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        labels_all,
        preds_all,
        target_names=[
            "Negative",
            "Positive"
        ],
        digits=4,
        zero_division=0
    )
)

# ============================================================
# INTEGRITY CHECK
# ============================================================

assert len(labels_all) == 767
assert len(preds_all) == 767
assert len(probs_all) == 767

print()
print("=" * 60)
print("FINAL TEST INTEGRITY CHECK")
print("=" * 60)

print("✓ All test users evaluated exactly once")
print("✓ Predictions aligned with labels")
print("✓ Best BERT Aggressive checkpoint used")
print("✓ Training data not used for evaluation")
print("✓ Validation data not used for evaluation")
print("✓ Transformer backbone remained frozen")

# ============================================================
# SAVE PREDICTIONS
# ============================================================

predictions_df = pd.DataFrame({
    "label": labels_all,
    "prediction": preds_all,
    "probability_positive": probs_all
})

predictions_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "bert_aggressive_test_predictions.csv"
    ),
    index=False
)

# ============================================================
# SAVE METRICS
# ============================================================

metrics = {
    "test_loss": float(test_loss),
    "accuracy": float(test_accuracy),
    "precision": float(test_precision),
    "recall": float(test_recall),
    "f1": float(test_f1),
    "roc_auc": float(test_roc_auc),
    "best_epoch": int(best_epoch),
    "best_validation_f1": float(best_val_f1),
    "learned_beta": float(learned_beta),
    "metadata_gate": float(metadata_gate),
    "test_users": 767,
    "confusion_matrix": cm.tolist()
}

with open(
    os.path.join(
        RESULT_DIR,
        "bert_aggressive_test_metrics.json"
    ),
    "w"
) as f:
    json.dump(
        metrics,
        f,
        indent=2
    )

# ============================================================
# SAVE CONFUSION MATRIX
# ============================================================

pd.DataFrame(
    cm,
    index=["Actual 0", "Actual 1"],
    columns=["Predicted 0", "Predicted 1"]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "bert_aggressive_confusion_matrix.csv"
    )
)

# ============================================================
# SAVE CLASSIFICATION REPORT
# ============================================================

report = classification_report(
    labels_all,
    preds_all,
    target_names=[
        "Negative",
        "Positive"
    ],
    output_dict=True,
    zero_division=0
)

pd.DataFrame(report).transpose().to_csv(
    os.path.join(
        RESULT_DIR,
        "bert_aggressive_classification_report.csv"
    )
)

# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 13 COMPLETE")
print("=" * 60)

print("✓ Best checkpoint evaluated")
print("✓ Held-out test evaluated exactly once")
print("✓ Final metrics calculated")
print("✓ Predictions saved")
print("✓ Metrics saved")
print("✓ Confusion matrix saved")
print("✓ Classification report saved")
print("✓ Complete test results saved")
print("✓ Frozen transformer verified")
```

```python
# ============================================================
# EXPLAINABILITY — CELL 1
# BERT + AGGRESSIVE
# t-SNE: FUSED REPRESENTATION → PCA → t-SNE
# ============================================================


import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE


print("=" * 70)
print("t-SNE REPRESENTATION ANALYSIS")
print("BERT + AGGRESSIVE")
print("=" * 70)


# ============================================================
# FIXED EXPLAINABILITY SETTINGS
# ============================================================

EXPLAINABILITY_SEED = 42

PCA_COMPONENTS = 50

TSNE_COMPONENTS = 2
TSNE_PERPLEXITY = 30
TSNE_LEARNING_RATE = 200
TSNE_MAX_ITER = 1000


np.random.seed(
    EXPLAINABILITY_SEED
)

torch.manual_seed(
    EXPLAINABILITY_SEED
)


# ============================================================
# OUTPUT DIRECTORIES
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

TSNE_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "tsne"
)

os.makedirs(
    TSNE_DIR,
    exist_ok=True
)


# ============================================================
# MODEL
# ============================================================

model.eval()


# ============================================================
# EXTRACT FUSED REPRESENTATIONS
# ============================================================

fused_representations = []
test_labels = []

print()
print(
    "Extracting fused model representations..."
)


with torch.no_grad():

    for batch_idx, batch in enumerate(
        test_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)


        # ----------------------------------------------------
        # BERT MODEL
        # ----------------------------------------------------

        output = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )


        # ----------------------------------------------------
        # ACTUAL OUTPUT NAMES FROM
        # BERT + AGGRESSIVE MODEL
        # ----------------------------------------------------

        text_representation = output[
            "text_representation"
        ]

        metadata_representation = output[
            "metadata_representation"
        ]


        # ----------------------------------------------------
        # SAME LEARNED METADATA GATE
        # ----------------------------------------------------

        metadata_gate = torch.sigmoid(
            model.metadata_beta
        )

        gated_metadata = (
            metadata_gate *
            metadata_representation
        )


        # ----------------------------------------------------
        # EXACT FUSED REPRESENTATION
        # ----------------------------------------------------

        fused = torch.cat(
            [
                text_representation,
                gated_metadata
            ],
            dim=1
        )


        fused_representations.append(
            fused.cpu().numpy()
        )

        test_labels.extend(
            labels.cpu().numpy().tolist()
        )


        if (
            batch_idx == 1
            or batch_idx % 10 == 0
            or batch_idx
            == len(
                test_loader
            )
        ):

            print(
                f"Processed batch "
                f"{batch_idx}/"
                f"{len(test_loader)}"
            )


# ============================================================
# COMBINE
# ============================================================

fused_representations = np.concatenate(
    fused_representations,
    axis=0
)

test_labels = np.array(
    test_labels
)


# ============================================================
# INTEGRITY
# ============================================================

print()
print("=" * 70)
print("FUSED REPRESENTATION")
print("=" * 70)

print(
    "Representation shape:",
    fused_representations.shape
)

print(
    "Number of test users:",
    len(test_labels)
)

print(
    "Representation dimensions:",
    fused_representations.shape[1]
)

print(
    "Negative users:",
    int(
        (test_labels == 0).sum()
    )
)

print(
    "Positive users:",
    int(
        (test_labels == 1).sum()
    )
)

print(
    "Metadata gate:",
    f"{float(metadata_gate.item()):.6f}"
)


assert (
    fused_representations.shape[0]
    ==
    len(test_labels)
)

assert (
    fused_representations.shape[1]
    ==
    832
)

assert np.isfinite(
    fused_representations
).all()


# ============================================================
# PCA
# ============================================================

print()
print("=" * 70)
print("PCA REDUCTION")
print("=" * 70)

pca_dimensions = min(
    PCA_COMPONENTS,
    fused_representations.shape[1]
)

pca = PCA(
    n_components=pca_dimensions,
    random_state=EXPLAINABILITY_SEED
)

pca_representation = pca.fit_transform(
    fused_representations
)

explained_variance = (
    pca.explained_variance_ratio_.sum()
)

print(
    "Original dimensions:",
    fused_representations.shape[1]
)

print(
    "PCA dimensions:",
    pca_representation.shape[1]
)

print(
    "Variance retained:",
    f"{explained_variance * 100:.2f}%"
)


# ============================================================
# t-SNE
# ============================================================

print()
print("=" * 70)
print("t-SNE REDUCTION")
print("=" * 70)

tsne = TSNE(
    n_components=TSNE_COMPONENTS,
    perplexity=TSNE_PERPLEXITY,
    learning_rate=TSNE_LEARNING_RATE,
    init="pca",
    random_state=EXPLAINABILITY_SEED,
    max_iter=TSNE_MAX_ITER
)

tsne_representation = tsne.fit_transform(
    pca_representation
)

print(
    "t-SNE output shape:",
    tsne_representation.shape
)


# ============================================================
# SAVE COORDINATES
# ============================================================

tsne_df = pd.DataFrame({

    "tsne_1":
        tsne_representation[:, 0],

    "tsne_2":
        tsne_representation[:, 1],

    "label":
        test_labels

})


tsne_df["class"] = tsne_df[
    "label"
].map({
    0: "Negative",
    1: "Positive"
})


tsne_csv_path = os.path.join(
    TSNE_DIR,
    "bert_aggressive_tsne_coordinates.csv"
)

tsne_df.to_csv(
    tsne_csv_path,
    index=False
)


# ============================================================
# PUBLICATION-QUALITY PLOT
# ============================================================

plt.figure(
    figsize=(10, 8)
)


negative_mask = (
    test_labels == 0
)

positive_mask = (
    test_labels == 1
)


plt.scatter(
    tsne_representation[
        negative_mask,
        0
    ],

    tsne_representation[
        negative_mask,
        1
    ],

    s=35,
    alpha=0.70,
    label="Negative",
    edgecolors="none"
)


plt.scatter(
    tsne_representation[
        positive_mask,
        0
    ],

    tsne_representation[
        positive_mask,
        1
    ],

    s=35,
    alpha=0.70,
    label="Positive",
    edgecolors="none"
)


plt.xlabel(
    "t-SNE Dimension 1",
    fontsize=12
)

plt.ylabel(
    "t-SNE Dimension 2",
    fontsize=12
)


plt.title(
    "BERT + Aggressive — "
    "t-SNE of Fused Test Representations",
    fontsize=14
)


plt.legend(
    fontsize=11
)


plt.grid(
    alpha=0.20
)

plt.tight_layout()


tsne_plot_path = os.path.join(
    TSNE_DIR,
    "bert_aggressive_tsne.png"
)


# SAVE
plt.savefig(
    tsne_plot_path,
    dpi=300,
    bbox_inches="tight"
)


# DISPLAY
plt.show()

plt.close()


# ============================================================
# SAVE CONFIGURATION
# ============================================================

tsne_config = {

    "model":
        "BERT",

    "cleaning":
        "Aggressive",

    "dataset":
        "Held-out test set",

    "test_users":
        int(
            len(test_labels)
        ),

    "input_representation":
        "Fused classifier representation",

    "text_representation_dimension":
        768,

    "metadata_representation_dimension":
        64,

    "fused_representation_dimension":
        int(
            fused_representations.shape[1]
        ),

    "pca_dimensions":
        int(
            pca_dimensions
        ),

    "pca_variance_retained":
        float(
            explained_variance
        ),

    "tsne_dimensions":
        int(
            TSNE_COMPONENTS
        ),

    "tsne_perplexity":
        int(
            TSNE_PERPLEXITY
        ),

    "tsne_learning_rate":
        TSNE_LEARNING_RATE,

    "tsne_max_iterations":
        int(
            TSNE_MAX_ITER
        ),

    "random_seed":
        EXPLAINABILITY_SEED,

    "metadata_gate":
        float(
            metadata_gate.item()
        )
}


with open(
    os.path.join(
        TSNE_DIR,
        "bert_aggressive_tsne_config.json"
    ),
    "w"
) as f:

    json.dump(
        tsne_config,
        f,
        indent=2
    )


# ============================================================
# FINAL OUTPUT
# ============================================================

print()
print("=" * 70)
print("t-SNE ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ BERT fused representation extracted"
)

print(
    "✓ Fused representation = 768 + 64 = 832"
)

print(
    "✓ PCA reduction completed"
)

print(
    "✓ t-SNE reduction completed"
)

print(
    "✓ Coordinates saved:"
)

print(
    tsne_csv_path
)

print(
    "✓ Graph saved:"
)

print(
    tsne_plot_path
)

print(
    "✓ Graph displayed above"
)

print(
    "✓ Configuration saved"
)
```

```python
# ============================================================
# EXPLAINABILITY — GRAD-CAM
# BERT + AGGRESSIVE
# TOKEN-LEVEL LOCAL EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from IPython.display import display, HTML


print("=" * 70)
print("GRAD-CAM ANALYSIS")
print("BERT + AGGRESSIVE")
print("=" * 70)


# ============================================================
# FIXED GRAD-CAM SETTINGS
# ============================================================

GRADCAM_SEED = 42
GRADCAM_TOP_TOKENS = 20

np.random.seed(
    GRADCAM_SEED
)

torch.manual_seed(
    GRADCAM_SEED
)


# ============================================================
# OUTPUT DIRECTORY
# ============================================================

GRADCAM_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "gradcam"
)

os.makedirs(
    GRADCAM_DIR,
    exist_ok=True
)


# ============================================================
# MODEL
# ============================================================

model.eval()

# Keep all model parameters frozen
for parameter in (
    model.parameters()
):

    parameter.requires_grad = False


print()
print("✓ BERT model loaded")
print("✓ Evaluation mode enabled")
print("✓ All model parameters remain frozen")


# ============================================================
# BUILD TEST PREDICTIONS
# DIRECTLY FROM BERT NOTEBOOK TEST LOADER
# ============================================================

test_labels = []
test_predictions = []
test_probabilities = []


with torch.no_grad():

    for batch in (
        test_loader
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)


        output = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )


        logits = output[
            "logits"
        ]


        probabilities = torch.softmax(
            logits.float(),
            dim=1
        )


        predictions = torch.argmax(
            logits,
            dim=1
        )


        test_labels.extend(
            labels
            .cpu()
            .numpy()
            .tolist()
        )

        test_predictions.extend(
            predictions
            .cpu()
            .numpy()
            .tolist()
        )

        test_probabilities.extend(
            probabilities[:, 1]
            .cpu()
            .numpy()
            .tolist()
        )


test_labels = np.array(
    test_labels
)

test_predictions = np.array(
    test_predictions
)

test_probabilities = np.array(
    test_probabilities
)


# ============================================================
# SELECT TP / TN / FP / FN
# SAME FIRST-MATCHING TEST ORDER
# ============================================================

selected_cases = {}


tp_candidates = np.where(
    (test_labels == 1)
    &
    (test_predictions == 1)
)[0]


tn_candidates = np.where(
    (test_labels == 0)
    &
    (test_predictions == 0)
)[0]


fp_candidates = np.where(
    (test_labels == 0)
    &
    (test_predictions == 1)
)[0]


fn_candidates = np.where(
    (test_labels == 1)
    &
    (test_predictions == 0)
)[0]


if len(tp_candidates) > 0:

    selected_cases["TP"] = int(
        tp_candidates[0]
    )


if len(tn_candidates) > 0:

    selected_cases["TN"] = int(
        tn_candidates[0]
    )


if len(fp_candidates) > 0:

    selected_cases["FP"] = int(
        fp_candidates[0]
    )


if len(fn_candidates) > 0:

    selected_cases["FN"] = int(
        fn_candidates[0]
    )


print()
print("=" * 70)
print("FIXED REPRESENTATIVE TEST CASES")
print("=" * 70)


for case_name, test_index in (
    selected_cases.items()
):

    print(
        f"{case_name}: "
        f"test index = {test_index} | "
        f"true = {test_labels[test_index]} | "
        f"pred = {test_predictions[test_index]} | "
        f"probability = "
        f"{test_probabilities[test_index]:.4f}"
    )


# ============================================================
# GRAD-CAM FUNCTION
# ============================================================

def compute_bert_gradcam(
    test_index
):

    sample = (
        test_dataset[
            int(test_index)
        ]
    )


    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
        .to(device)
    )


    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
        .to(device)
    )


    metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
    )


    # --------------------------------------------------------
    # BERT forward pass
    # --------------------------------------------------------
    # BERT parameters are frozen, but gradients are
    # enabled with respect to the hidden-state activation.
    # --------------------------------------------------------

    with torch.enable_grad():

        encoder_output = (
            model
            .encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )
        )


        hidden_states = (
            encoder_output.last_hidden_state
        )


        # Required because BERT is frozen
        hidden_states = (
            hidden_states.detach()
        )

        hidden_states.requires_grad_(True)

        hidden_states.retain_grad()


        # ----------------------------------------------------
        # EXACT ATTENTION POOLING FROM BERT MODEL
        # ----------------------------------------------------

        projected = (
            model
            .attention_projection(
                hidden_states
            )
        )


        scores = (
            model
            .attention_score(
                torch.tanh(
                    projected
                )
            )
            .squeeze(-1)
        )


        # Same padding masking as the notebook
        scores = scores.masked_fill(
            attention_mask == 0,
            torch.finfo(
                scores.dtype
            ).min
        )


        attention_weights = torch.softmax(
            scores,
            dim=1
        )


        text_representation = torch.sum(
            hidden_states *
            attention_weights.unsqueeze(-1),
            dim=1
        )


        # ----------------------------------------------------
        # EXACT METADATA BRANCH FROM MODEL
        # ----------------------------------------------------

        metadata_representation = (
            model
            .metadata_branch(
                metadata
            )
        )


        # ----------------------------------------------------
        # EXACT METADATA GATE
        # ----------------------------------------------------

        metadata_gate = torch.sigmoid(
            model
            .metadata_beta
        )


        gated_metadata = (
            metadata_gate *
            metadata_representation
        )


        # ----------------------------------------------------
        # EXACT FUSION + CLASSIFIER
        # ----------------------------------------------------

        fused = torch.cat(
            [
                text_representation,
                gated_metadata
            ],
            dim=1
        )


        logits = (
            model
            .classifier(
                fused
            )
        )


        # Positive/depression class
        target_score = logits[
            0,
            1
        ]


        # ----------------------------------------------------
        # BACKWARD
        # ----------------------------------------------------

        model.zero_grad(
            set_to_none=True
        )


        target_score.backward()


        gradients = (
            hidden_states.grad
            .detach()
            .clone()
        )


        activations = (
            hidden_states
            .detach()
            .clone()
        )


    # ========================================================
    # GRAD-CAM WEIGHTS
    # ========================================================

    # Global average pooling over token positions
    channel_weights = gradients.mean(
        dim=1
    )


    # Weighted combination of hidden-state channels
    cam = torch.sum(
        activations *
        channel_weights.unsqueeze(1),
        dim=-1
    )


    # Standard Grad-CAM ReLU
    cam = torch.relu(
        cam
    )


    cam = (
        cam
        .squeeze(0)
        .cpu()
        .numpy()
    )


    attention_weights_np = (
        attention_weights
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
    )


    input_ids_np = (
        input_ids
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
    )


    # --------------------------------------------------------
    # Normalize Grad-CAM
    # --------------------------------------------------------

    valid_mask = (
        attention_mask
        .detach()
        .squeeze(0)
        .cpu()
        .numpy()
        .astype(bool)
    )


    valid_cam = cam[
        valid_mask
    ]


    if (
        valid_cam.max()
        -
        valid_cam.min()
    ) > 1e-12:

        cam_normalized = (
            cam
            -
            valid_cam.min()
        ) / (
            valid_cam.max()
            -
            valid_cam.min()
        )

    else:

        cam_normalized = (
            np.zeros_like(cam)
        )


    # Padding tokens must never receive attribution
    cam_normalized[
        ~valid_mask
    ] = 0.0


    return {

        "cam":
            cam_normalized,

        "attention_weights":
            attention_weights_np,

        "input_ids":
            input_ids_np,

        "attention_mask":
            valid_mask,

        "positive_logit":
            float(
                target_score
                .detach()
                .cpu()
                .item()
            ),

        "logits":
            logits
            .detach()
            .cpu()
            .numpy()
            .squeeze(0)
    }


# ============================================================
# TOKEN PROCESSING
# ============================================================

def get_token_information(
    result
):

    token_ids = (
        result["input_ids"]
    )

    cam = (
        result["cam"]
    )

    attention_mask = (
        result["attention_mask"]
    )


    tokens = (
        bert_aggressive_tokenizer
        .convert_ids_to_tokens(
            token_ids.tolist()
        )
    )


    rows = []


    for position, (
        token,
        importance,
        valid
    ) in enumerate(
        zip(
            tokens,
            cam,
            attention_mask
        )
    ):

        if not valid:
            continue


        # Ignore special tokens
        if (
            token
            in
            bert_aggressive_tokenizer
            .all_special_tokens
        ):

            continue


        rows.append({

            "position":
                position,

            "token":
                token,

            "importance":
                float(
                    importance
                )
        })


    token_df = pd.DataFrame(
        rows
    )


    token_df = (
        token_df
        .sort_values(
            "importance",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )


    token_df["rank"] = (
        np.arange(
            len(token_df)
        )
        +
        1
    )


    return token_df


# ============================================================
# RUN GRAD-CAM FOR TP/TN/FP/FN
# ============================================================

gradcam_results = {}


print()
print("=" * 70)
print("GENERATING GRAD-CAM EXPLANATIONS")
print("=" * 70)


for case_name, test_index in (
    selected_cases.items()
):

    print()
    print(
        f"Processing {case_name} "
        f"(test index {test_index})..."
    )


    result = (
        compute_bert_gradcam(
            test_index
        )
    )


    token_df = (
        get_token_information(
            result
        )
    )


    gradcam_results[
        case_name
    ] = {

        "test_index":
            test_index,

        "result":
            result,

        "token_df":
            token_df
    }


    print(
        f"✓ {case_name} Grad-CAM complete"
    )


# ============================================================
# SAVE TOKEN IMPORTANCE
# ============================================================

all_gradcam_rows = []


for case_name, data in (
    gradcam_results.items()
):

    token_df = data[
        "token_df"
    ]


    for _, row in (
        token_df.iterrows()
    ):

        all_gradcam_rows.append({

            "case":
                case_name,

            "test_index":
                data["test_index"],

            "true_label":
                int(
                    test_labels[
                        data["test_index"]
                    ]
                ),

            "predicted_label":
                int(
                    test_predictions[
                        data["test_index"]
                    ]
                ),

            "positive_probability":
                float(
                    test_probabilities[
                        data["test_index"]
                    ]
                ),

            "position":
                int(
                    row["position"]
                ),

            "token":
                row["token"],

            "gradcam_importance":
                float(
                    row["importance"]
                ),

            "rank":
                int(
                    row["rank"]
                )
        })


gradcam_results_df = pd.DataFrame(
    all_gradcam_rows
)


gradcam_csv_path = os.path.join(
    GRADCAM_DIR,
    "bert_aggressive_gradcam_token_importance.csv"
)


gradcam_results_df.to_csv(
    gradcam_csv_path,
    index=False
)


# ============================================================
# PAPER-READY TOP TOKEN PLOTS
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM TOKEN IMPORTANCE")
print("=" * 70)


for case_name, data in (
    gradcam_results.items()
):

    token_df = (
        data["token_df"]
        .head(
            GRADCAM_TOP_TOKENS
        )
        .copy()
    )


    # Reverse order so most important is at top
    token_df = token_df.iloc[::-1]


    plt.figure(
        figsize=(10, 8)
    )


    plt.barh(
        token_df["token"],
        token_df["importance"]
    )


    plt.xlabel(
        "Grad-CAM Importance",
        fontsize=12
    )

    plt.ylabel(
        "Token",
        fontsize=12
    )


    plt.title(
        f"BERT + Aggressive — "
        f"Grad-CAM Token Importance ({case_name})",
        fontsize=14
    )


    plt.grid(
        axis="x",
        alpha=0.20
    )

    plt.tight_layout()


    plot_path = os.path.join(
        GRADCAM_DIR,
        f"bert_aggressive_gradcam_{case_name}.png"
    )


    # SAVE
    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )


    # DISPLAY
    plt.show()

    plt.close()


# ============================================================
# SAVE TOP TOKEN TABLE
# ============================================================

top_token_rows = []


for case_name, data in (
    gradcam_results.items()
):

    top_tokens = (
        data["token_df"]
        .head(
            GRADCAM_TOP_TOKENS
        )
    )


    for _, row in (
        top_tokens.iterrows()
    ):

        top_token_rows.append({

            "case":
                case_name,

            "rank":
                int(
                    row["rank"]
                ),

            "token":
                row["token"],

            "gradcam_importance":
                float(
                    row["importance"]
                )
        })


top_token_df = pd.DataFrame(
    top_token_rows
)


top_token_df.to_csv(
    os.path.join(
        GRADCAM_DIR,
        "bert_aggressive_gradcam_top_tokens.csv"
    ),
    index=False
)


# ============================================================
# SAVE CONFIGURATION
# ============================================================

gradcam_config = {

    "model":
        "BERT",

    "cleaning":
        "Aggressive",

    "dataset":
        "Held-out test set",

    "target":
        "Positive/depression class logit",

    "layer":
        "Final Transformer hidden states",

    "hidden_dimension":
        768,

    "method":
        "Transformer Grad-CAM",

    "cases":
        [
            "TP",
            "TN",
            "FP",
            "FN"
        ],

    "case_selection":
        "First matching test example in fixed test order",

    "top_tokens_displayed":
        GRADCAM_TOP_TOKENS,

    "random_seed":
        GRADCAM_SEED,

    "model_training":
        "No additional training",

    "transformer_weights":
        "Frozen"
}


with open(
    os.path.join(
        GRADCAM_DIR,
        "bert_aggressive_gradcam_config.json"
    ),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        gradcam_config,
        f,
        indent=2
    )


# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM ANALYSIS COMPLETE")
print("=" * 70)


print(
    "✓ Same TP/TN/FP/FN selection logic used"
)

print(
    "✓ Final BERT hidden states used"
)

print(
    "✓ Positive-class gradients computed"
)

print(
    "✓ BERT weights remained frozen"
)

print(
    "✓ No additional model training"
)

print(
    "✓ Token importance calculated"
)

print(
    "✓ Top-20 token plots saved and displayed"
)

print(
    "✓ Token importance CSV saved"
)

print(
    "✓ Configuration saved"
)

print()

print(
    "Grad-CAM output directory:",
    GRADCAM_DIR
)
```

```python
# ============================================================
# EXPLAINABILITY — LIME
# BERT + AGGRESSIVE
# TOKEN / WORD-LEVEL LOCAL EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from lime.lime_text import LimeTextExplainer
from IPython.display import display, HTML


print("=" * 70)
print("LIME ANALYSIS")
print("BERT + AGGRESSIVE")
print("=" * 70)


# ============================================================
# FIXED LIME SETTINGS
# ============================================================

LIME_SEED = 42
LIME_NUM_SAMPLES = 1000
LIME_NUM_FEATURES = 15
LIME_BATCH_SIZE = 8


np.random.seed(
    LIME_SEED
)

torch.manual_seed(
    LIME_SEED
)


# ============================================================
# OUTPUT DIRECTORY
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

LIME_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "lime"
)

os.makedirs(
    LIME_DIR,
    exist_ok=True
)


# ============================================================
# MODEL
# ============================================================

model.eval()


print()
print(
    "✓ BERT model loaded"
)

print(
    "✓ Evaluation mode enabled"
)


# ============================================================
# BUILD TEST PREDICTIONS
# DIRECTLY FROM BERT TEST LOADER
# ============================================================

labels_np = []
preds_np = []
probs_np = []


with torch.no_grad():

    for batch in (
        test_loader
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)


        output = (
            model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                metadata=metadata
            )
        )


        logits = output[
            "logits"
        ]


        probabilities = torch.softmax(
            logits,
            dim=1
        )


        predictions = torch.argmax(
            logits,
            dim=1
        )


        labels_np.extend(
            labels
            .cpu()
            .numpy()
            .tolist()
        )


        preds_np.extend(
            predictions
            .cpu()
            .numpy()
            .tolist()
        )


        probs_np.extend(
            probabilities[:, 1]
            .cpu()
            .numpy()
            .tolist()
        )


labels_np = np.array(
    labels_np
)

preds_np = np.array(
    preds_np
)

probs_np = np.array(
    probs_np
)


# ============================================================
# SELECT TP / TN / FP / FN
# FIRST MATCHING TEST EXAMPLE
# ============================================================

selected_cases = {}


tp_candidates = np.where(
    (labels_np == 1)
    &
    (preds_np == 1)
)[0]


tn_candidates = np.where(
    (labels_np == 0)
    &
    (preds_np == 0)
)[0]


fp_candidates = np.where(
    (labels_np == 0)
    &
    (preds_np == 1)
)[0]


fn_candidates = np.where(
    (labels_np == 1)
    &
    (preds_np == 0)
)[0]


if len(tp_candidates) > 0:

    selected_cases["TP"] = int(
        tp_candidates[0]
    )


if len(tn_candidates) > 0:

    selected_cases["TN"] = int(
        tn_candidates[0]
    )


if len(fp_candidates) > 0:

    selected_cases["FP"] = int(
        fp_candidates[0]
    )


if len(fn_candidates) > 0:

    selected_cases["FN"] = int(
        fn_candidates[0]
    )


print()
print("=" * 70)
print("FIXED REPRESENTATIVE TEST CASES")
print("=" * 70)


for case_name, test_index in (
    selected_cases.items()
):

    print(
        f"{case_name}: "
        f"test index = {test_index} | "
        f"true = {labels_np[test_index]} | "
        f"pred = {preds_np[test_index]} | "
        f"probability = "
        f"{probs_np[test_index]:.4f}"
    )


# ============================================================
# GET TEST TEXT
# ============================================================

def get_test_text(
    test_index
):

    return str(
        test_bert_aggressive_df.iloc[
            int(test_index)
        ][
            "model_text"
        ]
    )


# ============================================================
# MAKE LIME PREDICTOR
# METADATA REMAINS FIXED FOR EACH USER
# TEXT IS PERTURBED BY LIME
# ============================================================

def make_lime_predictor(
    test_index
):

    sample = (
        test_dataset[
            int(test_index)
        ]
    )


    fixed_metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
        .float()
    )


    def predictor(
        texts
    ):

        probability_batches = []


        # ----------------------------------------------------
        # Process perturbations in small batches
        # to avoid GPU memory overflow.
        # ----------------------------------------------------

        for start in range(
            0,
            len(texts),
            LIME_BATCH_SIZE
        ):

            batch_texts = texts[
                start:
                start + LIME_BATCH_SIZE
            ]


            encoded = (
                tokenizer(
                    batch_texts,
                    padding="max_length",
                    truncation=True,
                    max_length=MAX_LENGTH,
                    return_attention_mask=True,
                    return_tensors="pt"
                )
            )


            lime_input_ids = (
                encoded[
                    "input_ids"
                ]
                .to(device)
            )


            lime_attention_mask = (
                encoded[
                    "attention_mask"
                ]
                .to(device)
            )


            batch_metadata = (
                fixed_metadata.expand(
                    len(batch_texts),
                    -1
                )
            )


            with torch.no_grad():

                output = (
                    model(
                        input_ids=lime_input_ids,
                        attention_mask=lime_attention_mask,
                        metadata=batch_metadata
                    )
                )


                probabilities = torch.softmax(
                    output["logits"],
                    dim=1
                )


            probability_batches.append(
                probabilities
                .cpu()
                .numpy()
            )


            del (
                lime_input_ids,
                lime_attention_mask,
                batch_metadata,
                output,
                probabilities
            )


            if torch.cuda.is_available():

                torch.cuda.empty_cache()


        return np.vstack(
            probability_batches
        )


    return predictor


# ============================================================
# LIME EXPLAINER
# ============================================================

explainer = LimeTextExplainer(
    class_names=[
        "Negative",
        "Positive"
    ],
    split_expression=r"\s+",
    random_state=LIME_SEED
)


# ============================================================
# RUN LIME FOR FOUR CASES
# ============================================================

all_explanations = {}


print()
print("=" * 70)
print("GENERATING LIME EXPLANATIONS")
print("=" * 70)


for case_name, test_index in (
    selected_cases.items()
):

    print()
    print(
        f"Processing {case_name} "
        f"(test index {test_index})..."
    )


    text = get_test_text(
        test_index
    )


    predictor = make_lime_predictor(
        test_index
    )


    explanation = (
        explainer.explain_instance(
            text,
            predictor,
            labels=(1,),
            num_features=LIME_NUM_FEATURES,
            num_samples=LIME_NUM_SAMPLES
        )
    )


    all_explanations[
        case_name
    ] = explanation


    print(
        f"✓ {case_name} explanation complete"
    )


# ============================================================
# SAVE LIME RESULTS
# ============================================================

all_lime_rows = []


for case_name, explanation in (
    all_explanations.items()
):

    explanation_list = (
        explanation.as_list(
            label=1
        )
    )


    for rank, (
        word,
        weight
    ) in enumerate(
        explanation_list,
        start=1
    ):

        all_lime_rows.append({

            "case":
                case_name,

            "test_index":
                selected_cases[
                    case_name
                ],

            "true_label":
                int(
                    labels_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "predicted_label":
                int(
                    preds_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "positive_probability":
                float(
                    probs_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "rank":
                rank,

            "word":
                word,

            "lime_weight":
                float(
                    weight
                )
        })


lime_results_df = pd.DataFrame(
    all_lime_rows
)


lime_csv_path = os.path.join(
    LIME_DIR,
    "bert_aggressive_lime_results.csv"
)


lime_results_df.to_csv(
    lime_csv_path,
    index=False
)


# ============================================================
# DISPLAY + SAVE HTML
# ============================================================

print()
print("=" * 70)
print("LIME EXPLANATIONS")
print("=" * 70)


for case_name, explanation in (
    all_explanations.items()
):

    print()
    print(
        "=" * 60
    )


    print(
        f"{case_name} — "
        f"LIME LOCAL EXPLANATION"
    )


    print(
        "=" * 60
    )


    display(
        HTML(
            explanation.as_html(
                labels=[1]
            )
        )
    )


    html_path = os.path.join(
        LIME_DIR,
        f"bert_aggressive_lime_{case_name}.html"
    )


    with open(
        html_path,
        "w",
        encoding="utf-8"
    ) as f:

        f.write(
            explanation.as_html(
                labels=[1]
            )
        )


# ============================================================
# PAPER-FRIENDLY LIME BAR PLOTS
# ============================================================

for case_name, explanation in (
    all_explanations.items()
):

    explanation_list = (
        explanation.as_list(
            label=1
        )
    )


    plot_df = pd.DataFrame(
        explanation_list,
        columns=[
            "word",
            "weight"
        ]
    )


    plot_df = plot_df.sort_values(
        "weight"
    )


    plt.figure(
        figsize=(10, 7)
    )


    plt.barh(
        plot_df["word"],
        plot_df["weight"]
    )


    plt.axvline(
        0,
        linewidth=0.8
    )


    plt.xlabel(
        "LIME Weight",
        fontsize=12
    )


    plt.ylabel(
        "Token / Word",
        fontsize=12
    )


    plt.title(
        f"BERT + Aggressive — "
        f"LIME Explanation ({case_name})",
        fontsize=14
    )


    plt.tight_layout()


    plot_path = os.path.join(
        LIME_DIR,
        f"bert_aggressive_lime_{case_name}.png"
    )


    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()

    plt.close()


# ============================================================
# SAVE CASE INFORMATION
# ============================================================

case_info_rows = []


for case_name, test_index in (
    selected_cases.items()
):

    case_info_rows.append({

        "case":
            case_name,

        "test_index":
            int(test_index),

        "true_label":
            int(
                labels_np[
                    test_index
                ]
            ),

        "predicted_label":
            int(
                preds_np[
                    test_index
                ]
            ),

        "positive_probability":
            float(
                probs_np[
                    test_index
                ]
            ),

        "text_length_characters":
            len(
                get_test_text(
                    test_index
                )
            )
    })


case_info_df = pd.DataFrame(
    case_info_rows
)


case_info_df.to_csv(
    os.path.join(
        LIME_DIR,
        "bert_aggressive_lime_case_information.csv"
    ),
    index=False
)


# ============================================================
# SAVE CONFIGURATION
# ============================================================

lime_config = {

    "model":
        "BERT",

    "cleaning":
        "Aggressive",

    "dataset":
        "Held-out test set",

    "target":
        "Positive/depression class",

    "representative_cases":
        [
            "TP",
            "TN",
            "FP",
            "FN"
        ],

    "case_selection":
        "First matching test example in fixed test order",

    "random_seed":
        LIME_SEED,

    "num_samples":
        LIME_NUM_SAMPLES,

    "num_features":
        LIME_NUM_FEATURES,

    "batch_size":
        LIME_BATCH_SIZE,

    "metadata":
        "Fixed for each explained user",

    "text_perturbation":
        "LIME word-level perturbations",

    "model_training":
        "No additional training"
}


with open(
    os.path.join(
        LIME_DIR,
        "bert_aggressive_lime_config.json"
    ),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        lime_config,
        f,
        indent=2
    )


# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("LIME ANALYSIS COMPLETE")
print("=" * 70)


print(
    "✓ TP, TN, FP, FN cases explained"
)

print(
    "✓ Same trained BERT model used"
)

print(
    "✓ Metadata held fixed per user"
)

print(
    f"✓ {LIME_NUM_SAMPLES} LIME perturbations per case"
)

print(
    f"✓ Top {LIME_NUM_FEATURES} features extracted"
)

print(
    "✓ LIME results saved"
)

print(
    "✓ HTML explanations saved"
)

print(
    "✓ Paper-ready LIME plots saved and displayed"
)

print()

print(
    "LIME output directory:",
    LIME_DIR
)
```
