```python
# ============================================================
# CELL 3 — BERT + EMOTION-BASED
# ============================================================

print("=" * 60)
print("BERT + EMOTION-BASED — EXACT D1 ANCHOR DETECTION")
print("=" * 60)

import os
import re
import json
import glob
from collections import defaultdict

DATASET_ROOT = (
    "/kaggle/input/datasets/"
    "vedantnawghare35/"
    "depression-detection-mddl-final-pbl/"
    "Dataset_MDDL"
)

D1_TIMELINE_DIR = os.path.join(
    DATASET_ROOT,
    "labeled",
    "positive",
    "data",
    "timeline"
)

assert os.path.exists(D1_TIMELINE_DIR)

d1_timeline_files = glob.glob(
    os.path.join(D1_TIMELINE_DIR, "*.json")
)

print()
print("D1 timeline files found:", len(d1_timeline_files))

assert len(d1_timeline_files) == 2626

REFINED_ANCHOR_PATTERNS = [
    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",
    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",
    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depression\b",
    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",
    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",
    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",
    r"\bi\s+have\s+(?:major\s+)?depression\b",
    r"\bi\s+suffer\s+from\s+(?:major\s+)?depression\b",
    r"\bi\s+(?:was|am|'m)\s+diagnosed\b.{0,80}\bdepress",
    r"\b(?:my|a)\s+(?:doctor|psychiatrist)\s+(?:has\s+)?diagnosed\s+me\b.{0,80}\bdepress"
]

EXCLUSION_PATTERNS = [
    r"\bmy\s+friend\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+sister\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+brother\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+mother\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+father\s+(?:was|is|has\s+been)\s+diagnosed",
    r"\bmy\s+answer\s+to\s*:",
    r"\banswer\s+to\s*:",
    r"\bquestion\s*:",
    r"\bq\s*:"
]

anchor_patterns = [
    re.compile(
        p,
        re.IGNORECASE | re.DOTALL
    )
    for p in REFINED_ANCHOR_PATTERNS
]

exclusion_patterns = [
    re.compile(
        p,
        re.IGNORECASE | re.DOTALL
    )
    for p in EXCLUSION_PATTERNS
]


def is_refined_anchor_tweet(text):
    if text is None:
        return False

    text = str(text).strip()

    if not text:
        return False

    for pattern in exclusion_patterns:
        if pattern.search(text):
            return False

    for pattern in anchor_patterns:
        if pattern.search(text):
            return True

    return False


train_positive_users = set(
    train_df.loc[
        train_df["label"] == 1,
        "username"
    ].astype(str)
)

val_positive_users = set(
    val_df.loc[
        val_df["label"] == 1,
        "username"
    ].astype(str)
)

test_positive_users = set(
    test_df.loc[
        test_df["label"] == 1,
        "username"
    ].astype(str)
)

all_positive_users = (
    train_positive_users
    | val_positive_users
    | test_positive_users
)

print()
print("=" * 60)
print("POSITIVE USERS BY SPLIT")
print("=" * 60)

print("TRAIN:", len(train_positive_users))
print("VALIDATION:", len(val_positive_users))
print("TEST:", len(test_positive_users))

assert len(train_positive_users) == 2013
assert len(val_positive_users) == 252
assert len(test_positive_users) == 251
assert len(all_positive_users) == 2516


timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:
    filename = os.path.basename(timeline_file)
    username = filename.split("created_at")[0]

    timeline_files_by_user[
        username
    ].append(timeline_file)

assert len(timeline_files_by_user) == 2516


anchor_records = []
users_processed = set()
tweets_scanned = 0

for username in sorted(all_positive_users):

    if username not in timeline_files_by_user:
        raise KeyError(
            f"Missing D1 timeline for user: {username}"
        )

    users_processed.add(username)

    for timeline_file in timeline_files_by_user[username]:

        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:
                line = line.strip()

                if not line:
                    continue

                record = json.loads(line)
                tweets_scanned += 1

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get("id", "")
                    )
                )

                tweet_text = str(
                    record.get("text", "")
                )

                if (
                    tweet_id
                    and
                    is_refined_anchor_tweet(
                        tweet_text
                    )
                ):

                    anchor_records.append({
                        "username": username,
                        "tweet_id": tweet_id,
                        "created_at":
                            record.get("created_at"),
                        "text": tweet_text
                    })


anchor_pairs = {
    (
        record["username"],
        record["tweet_id"]
    )
    for record in anchor_records
}

train_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in train_positive_users
}

val_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in val_positive_users
}

test_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in test_positive_users
}


print()
print("=" * 60)
print("EXACT ANCHOR SUMMARY")
print("=" * 60)

print(
    "Positive users processed:",
    len(users_processed)
)

print(
    "Tweets scanned:",
    tweets_scanned
)

print(
    "Raw anchor records:",
    len(anchor_records)
)

print(
    "Unique anchor pairs:",
    len(anchor_pairs)
)

print()
print(
    "TRAIN unique anchors:",
    len(train_anchor_pairs)
)

print(
    "VAL unique anchors:",
    len(val_anchor_pairs)
)

print(
    "TEST unique anchors:",
    len(test_anchor_pairs)
)

assert len(users_processed) == 2516
assert tweets_scanned == 508806
assert len(anchor_records) == 1920
assert len(anchor_pairs) == 1847
assert len(train_anchor_pairs) == 1476
assert len(val_anchor_pairs) == 190
assert len(test_anchor_pairs) == 181


# Save anchor records
bert_emotion_anchor_df = pd.DataFrame(
    anchor_records
)

bert_emotion_anchor_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "exact_anchor_records.csv"
    ),
    index=False
)

pd.DataFrame(
    list(anchor_pairs),
    columns=["username", "tweet_id"]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "unique_anchor_pairs.csv"
    ),
    index=False
)


print()
print("=" * 60)
print("CELL 3 COMPLETE")
print("=" * 60)

print("✓ Exact anchor detector reproduced")
print("✓ 1,847 unique anchors confirmed")
print("✓ 1,920 anchor records confirmed")
print("✓ TRAIN = 1,476 unique anchors")
print("✓ VAL = 190 unique anchors")
print("✓ TEST = 181 unique anchors")
print("✓ No D2 users included")
print("✓ Anchor records saved")
print("✓ No timelines modified yet")
```

```python
# ============================================================
# CELL 4 — BERT + EMOTION-BASED
# ============================================================

print("=" * 60)
print("BERT + EMOTION-BASED — EXACT D1 ANCHOR REMOVAL")
print("=" * 60)

EXACT_ANCHOR_SET = set(anchor_pairs)

train_anchor_records = [
    r for r in anchor_records
    if r["username"] in train_positive_users
]

val_anchor_records = [
    r for r in anchor_records
    if r["username"] in val_positive_users
]

test_anchor_records = [
    r for r in anchor_records
    if r["username"] in test_positive_users
]

print()
print("=" * 60)
print("PRE-REMOVAL ANCHOR VERIFICATION")
print("=" * 60)

duplicate_anchor_occurrences = (
    len(anchor_records) - len(anchor_pairs)
)

print("Raw anchor records:", len(anchor_records))
print("Unique anchor pairs:", len(anchor_pairs))
print(
    "Duplicate anchor occurrences:",
    duplicate_anchor_occurrences
)

assert len(anchor_records) == 1920
assert len(anchor_pairs) == 1847
assert duplicate_anchor_occurrences == 73

print("✓ 1,847 unique anchors verified")
print("✓ 73 duplicate occurrences verified")
print("✓ 1,920 total anchor records verified")

print()
print("=" * 60)
print("SPLIT-LEVEL ANCHOR RECORD COUNTS")
print("=" * 60)

print("TRAIN:", len(train_anchor_records), "| expected: 1535")
print("VAL:", len(val_anchor_records), "| expected: 199")
print("TEST:", len(test_anchor_records), "| expected: 186")

assert len(train_anchor_records) == 1535
assert len(val_anchor_records) == 199
assert len(test_anchor_records) == 186

# Timeline index
timeline_files_by_user = defaultdict(list)

for timeline_file in d1_timeline_files:
    filename = os.path.basename(timeline_file)
    username = filename.split("created_at")[0]
    timeline_files_by_user[username].append(
        timeline_file
    )

assert len(timeline_files_by_user) == 2516


def build_anchor_removed_text(username):
    kept_texts = []
    removed_count = 0

    for timeline_file in timeline_files_by_user[username]:
        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:
                line = line.strip()

                if not line:
                    continue

                record = json.loads(line)

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get("id", "")
                    )
                )

                if (
                    str(username),
                    tweet_id
                ) in EXACT_ANCHOR_SET:

                    removed_count += 1
                    continue

                tweet_text = str(
                    record.get("text", "")
                ).strip()

                if tweet_text:
                    kept_texts.append(tweet_text)

    return " ".join(kept_texts), removed_count


# Copy source data
train_bert_emotion_df = train_df.copy(deep=True)
val_bert_emotion_df = val_df.copy(deep=True)
test_bert_emotion_df = test_df.copy(deep=True)

positive_anchor_removed_text = {}

train_removed = 0
val_removed = 0
test_removed = 0

for username in sorted(all_positive_users):

    text, removed = build_anchor_removed_text(username)

    positive_anchor_removed_text[username] = text

    if username in train_positive_users:
        train_removed += removed
    elif username in val_positive_users:
        val_removed += removed
    elif username in test_positive_users:
        test_removed += removed

print()
print("=" * 60)
print("FINAL ANCHOR REMOVAL COUNTS")
print("=" * 60)

print("TRAIN removed:", train_removed)
print("VAL removed:", val_removed)
print("TEST removed:", test_removed)
print(
    "TOTAL removed:",
    train_removed + val_removed + test_removed
)

assert train_removed == 1535
assert val_removed == 199
assert test_removed == 186
assert (
    train_removed +
    val_removed +
    test_removed
) == 1920

print(
    "✓ Record-level removal exactly matches previous experiments"
)

# Assign anchor-removed text
for df in [
    train_bert_emotion_df,
    val_bert_emotion_df,
    test_bert_emotion_df
]:

    positive_mask = df["label"] == 1

    df.loc[
        positive_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            positive_mask,
            "username"
        ]
        .astype(str)
        .map(positive_anchor_removed_text)
        .fillna("")
        .values
    )

    negative_mask = df["label"] == 0

    df.loc[
        negative_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            negative_mask,
            "timeline_text"
        ]
        .fillna("")
        .astype(str)
        .values
    )


print()
print("=" * 60)
print("D2 PRESERVATION CHECK")
print("=" * 60)

for original_df, result_df, name in [
    (train_df, train_bert_emotion_df, "TRAIN"),
    (val_df, val_bert_emotion_df, "VALIDATION"),
    (test_df, test_bert_emotion_df, "TEST")
]:

    original_d2 = (
        original_df.loc[
            original_df["label"] == 0,
            ["username", "timeline_text"]
        ]
        .set_index("username")
        .sort_index()
    )

    result_d2 = (
        result_df.loc[
            result_df["label"] == 0,
            ["username", "anchor_removed_text"]
        ]
        .rename(
            columns={
                "anchor_removed_text":
                    "timeline_text"
            }
        )
        .set_index("username")
        .sort_index()
    )

    assert original_d2.equals(result_d2)
    print(f"✓ {name} D2 unchanged")


print()
print("=" * 60)
print("USER + LABEL PRESERVATION")
print("=" * 60)

for original_df, result_df, name in [
    (train_df, train_bert_emotion_df, "TRAIN"),
    (val_df, val_bert_emotion_df, "VALIDATION"),
    (test_df, test_bert_emotion_df, "TEST")
]:

    assert (
        original_df["user_id"].tolist()
        ==
        result_df["user_id"].tolist()
    )

    assert (
        original_df["username"].tolist()
        ==
        result_df["username"].tolist()
    )

    assert (
        original_df["label"].tolist()
        ==
        result_df["label"].tolist()
    )

    print(
        f"✓ {name}: users and labels unchanged"
    )


print()
print("=" * 60)
print("ANCHOR-REMOVED TEXT INTEGRITY")
print("=" * 60)

for name, df in [
    ("TRAIN", train_bert_emotion_df),
    ("VALIDATION", val_bert_emotion_df),
    ("TEST", test_bert_emotion_df)
]:

    empty_count = int(
        (
            df["anchor_removed_text"]
            .fillna("")
            .astype(str)
            .str.strip()
            .eq("")
        ).sum()
    )

    print(
        f"{name}: empty anchor-removed texts = "
        f"{empty_count}"
    )


train_bert_emotion_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "train_anchor_removed.pkl"
    )
)

val_bert_emotion_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "val_anchor_removed.pkl"
    )
)

test_bert_emotion_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "test_anchor_removed.pkl"
    )
)

print()
print("=" * 60)
print("CELL 4 COMPLETE")
print("=" * 60)

print("✓ 1,847 unique D1 anchors removed")
print("✓ 73 duplicate occurrences handled")
print("✓ 1,920 total anchor records removed")
print("✓ TRAIN = 1,535 removed")
print("✓ VAL = 199 removed")
print("✓ TEST = 186 removed")
print("✓ D2 completely preserved")
print("✓ User IDs preserved")
print("✓ Labels preserved")
print("✓ Anchor-removed source saved")
```

```python
# ============================================================
# CELL 5 — BERT + EMOTION-BASED
# ============================================================

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)
from kaggle_secrets import UserSecretsClient
import torch.nn.functional as F
from tqdm.auto import tqdm

print("=" * 60)
print("BERT + EMOTION-BASED — EMOTION EXTRACTION")
print("=" * 60)

EMOTION_MODEL_NAME = (
    "cardiffnlp/twitter-roberta-base-emotion-latest"
)

user_secrets = UserSecretsClient()
hf_token = user_secrets.get_secret("HF_TOKEN")

print()
print("Emotion model:", EMOTION_MODEL_NAME)
print("Emotion features:", len(EMOTION_LABELS))
print("✓ HF_TOKEN retrieved")

emotion_tokenizer = AutoTokenizer.from_pretrained(
    EMOTION_MODEL_NAME,
    token=hf_token
)

emotion_model = (
    AutoModelForSequenceClassification
    .from_pretrained(
        EMOTION_MODEL_NAME,
        token=hf_token
    )
    .to(device)
)

emotion_model.eval()

for p in emotion_model.parameters():
    p.requires_grad = False

print("✓ Emotion tokenizer loaded")
print("✓ Emotion model loaded")
print("✓ Emotion model frozen")

model_labels = [
    emotion_model.config.id2label[i]
    for i in range(
        emotion_model.config.num_labels
    )
]

print()
print("Model emotion labels:")
print(model_labels)

assert emotion_model.config.num_labels == 11
assert model_labels == EMOTION_LABELS

def extract_emotion_features(
    texts,
    split_name,
    batch_size=32
):
    features = []
    empty_count = 0
    nonempty_indices = []

    for i, text in enumerate(texts):
        if str(text).strip() == "":
            empty_count += 1
        else:
            nonempty_indices.append(i)

    output = np.zeros(
        (len(texts), 11),
        dtype=np.float32
    )

    print()
    print(f"Processing {split_name}...")
    print("Users:", len(texts))
    print("Empty texts:", empty_count)
    print("Non-empty texts:", len(nonempty_indices))

    total_batches = (
        len(nonempty_indices) + batch_size - 1
    ) // batch_size

    with torch.no_grad():

        for start in tqdm(
            range(
                0,
                len(nonempty_indices),
                batch_size
            ),
            desc=split_name
        ):

            batch_indices = nonempty_indices[
                start:start + batch_size
            ]

            batch_texts = [
                str(texts[i])
                for i in batch_indices
            ]

            encoded = emotion_tokenizer(
                batch_texts,
                padding=True,
                truncation=True,
                max_length=512,
                return_tensors="pt"
            )

            encoded = {
                k: v.to(device)
                for k, v in encoded.items()
            }

            logits = emotion_model(
                **encoded
            ).logits.float()

            # CardiffNLP emotion model is multilabel.
            probabilities = torch.sigmoid(logits)

            output[
                batch_indices
            ] = probabilities.cpu().numpy()

    assert output.shape == (
        len(texts),
        11
    )

    assert np.isfinite(output).all()
    assert output.min() >= 0.0
    assert output.max() <= 1.0

    return output


train_emotion_features = extract_emotion_features(
    train_bert_emotion_df[
        "anchor_removed_text"
    ].fillna("").tolist(),
    "TRAIN"
)

val_emotion_features = extract_emotion_features(
    val_bert_emotion_df[
        "anchor_removed_text"
    ].fillna("").tolist(),
    "VALIDATION"
)

test_emotion_features = extract_emotion_features(
    test_bert_emotion_df[
        "anchor_removed_text"
    ].fillna("").tolist(),
    "TEST"
)

print()
print("=" * 60)
print("EMOTION FEATURE SHAPES")
print("=" * 60)

print("Train:", train_emotion_features.shape)
print("Validation:", val_emotion_features.shape)
print("Test:", test_emotion_features.shape)

assert train_emotion_features.shape == (6135, 11)
assert val_emotion_features.shape == (767, 11)
assert test_emotion_features.shape == (767, 11)

print("✓ Exact 11 emotion features confirmed")

print()
print("=" * 60)
print("EMOTION FEATURE INTEGRITY")
print("=" * 60)

for features, name in [
    (train_emotion_features, "TRAIN"),
    (val_emotion_features, "VALIDATION"),
    (test_emotion_features, "TEST")
]:
    print(
        f"{name}: "
        f"NaN={np.isnan(features).sum()}, "
        f"Inf={np.isinf(features).sum()}"
    )

    print(
        f"{name}: "
        f"min={features.min():.6f}, "
        f"max={features.max():.6f}, "
        f"mean={features.mean():.6f}"
    )

    assert np.isnan(features).sum() == 0
    assert np.isinf(features).sum() == 0

emotion_columns = [
    f"emotion_{label}"
    for label in EMOTION_LABELS
]

print()
print("=" * 60)
print("EMOTION FEATURE COLUMNS")
print("=" * 60)

for i, column in enumerate(
    emotion_columns,
    start=1
):
    print(
        f"{i:02d}. {column}"
    )

# Save raw emotion features
np.save(
    os.path.join(
        RESULT_DIR,
        "train_emotion_features.npy"
    ),
    train_emotion_features
)

np.save(
    os.path.join(
        RESULT_DIR,
        "val_emotion_features.npy"
    ),
    val_emotion_features
)

np.save(
    os.path.join(
        RESULT_DIR,
        "test_emotion_features.npy"
    ),
    test_emotion_features
)

pd.DataFrame(
    train_emotion_features,
    columns=emotion_columns
).to_csv(
    os.path.join(
        RESULT_DIR,
        "train_emotion_features.csv"
    ),
    index=False
)

pd.DataFrame(
    val_emotion_features,
    columns=emotion_columns
).to_csv(
    os.path.join(
        RESULT_DIR,
        "val_emotion_features.csv"
    ),
    index=False
)

pd.DataFrame(
    test_emotion_features,
    columns=emotion_columns
).to_csv(
    os.path.join(
        RESULT_DIR,
        "test_emotion_features.csv"
    ),
    index=False
)

print()
print("=" * 60)
print("CELL 5 COMPLETE")
print("=" * 60)

print("✓ Emotion model applied to anchor-removed text")
print("✓ Multilabel sigmoid probabilities used")
print("✓ 11 emotion features extracted")
print("✓ Train = 6135 × 11")
print("✓ Validation = 767 × 11")
print("✓ Test = 767 × 11")
print("✓ No NaN / Inf")
print("✓ Emotion features saved")
```

```python
# ============================================================
# CELL 9 — BERT + EMOTION-BASED
# ============================================================

from torch.utils.data import Dataset, DataLoader

print("=" * 60)
print("BERT + EMOTION-BASED — BUILDING PYTORCH DATASETS")
print("=" * 60)

# Map original TRAIN users to their row positions
user_to_index = {
    user_id: i
    for i, user_id in enumerate(
        train_bert_emotion_df["user_id"]
    )
}

balanced_indices = np.array(
    [
        user_to_index[user_id]
        for user_id in
        balanced_train_bert_emotion_df["user_id"]
    ],
    dtype=np.int64
)

# Balanced TRAIN inputs
train_input_ids_balanced = train_input_ids[
    balanced_indices
]

train_attention_masks_balanced = (
    train_attention_masks[
        balanced_indices
    ]
)

train_metadata_balanced = (
    train_bert_emotion_metadata_scaled[
        balanced_indices
    ]
)

train_emotion_balanced = (
    train_emotion_scaled[
        balanced_indices
    ]
)

train_labels_balanced = (
    balanced_train_bert_emotion_df["label"]
    .to_numpy(dtype=np.int64)
)

# Validation labels
val_labels = (
    val_bert_emotion_df["label"]
    .to_numpy(dtype=np.int64)
)

# Test labels
test_labels = (
    test_bert_emotion_df["label"]
    .to_numpy(dtype=np.int64)
)


class EmotionBERTDataset(Dataset):

    def __init__(
        self,
        input_ids,
        attention_masks,
        metadata,
        emotions,
        labels
    ):
        self.input_ids = torch.tensor(
            input_ids,
            dtype=torch.long
        )

        self.attention_masks = torch.tensor(
            attention_masks,
            dtype=torch.long
        )

        self.metadata = torch.tensor(
            metadata,
            dtype=torch.float32
        )

        self.emotions = torch.tensor(
            emotions,
            dtype=torch.float32
        )

        self.labels = torch.tensor(
            labels,
            dtype=torch.long
        )

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, index):
        return {
            "input_ids": self.input_ids[index],
            "attention_mask": self.attention_masks[index],
            "metadata": self.metadata[index],
            "emotions": self.emotions[index],
            "labels": self.labels[index]
        }


train_dataset = EmotionBERTDataset(
    train_input_ids_balanced,
    train_attention_masks_balanced,
    train_metadata_balanced,
    train_emotion_balanced,
    train_labels_balanced
)

val_dataset = EmotionBERTDataset(
    val_input_ids,
    val_attention_masks,
    val_bert_emotion_metadata_scaled,
    val_emotion_scaled,
    val_labels
)

test_dataset = EmotionBERTDataset(
    test_input_ids,
    test_attention_masks,
    test_bert_emotion_metadata_scaled,
    test_emotion_scaled,
    test_labels
)


BATCH_SIZE = 16

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)


print()
print("=" * 60)
print("DATASET SIZES")
print("=" * 60)

print("Balanced Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))

assert len(train_dataset) == 4026
assert len(val_dataset) == 767
assert len(test_dataset) == 767


print()
print("=" * 60)
print("DATALOADER CONFIGURATION")
print("=" * 60)

print("Batch size:", BATCH_SIZE)
print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))

assert len(train_loader) == 252
assert len(val_loader) == 48
assert len(test_loader) == 48


# Batch sanity check
batch = next(iter(train_loader))

print()
print("=" * 60)
print("BATCH SANITY CHECK")
print("=" * 60)

print(
    "Input IDs:",
    tuple(batch["input_ids"].shape)
)

print(
    "Attention mask:",
    tuple(batch["attention_mask"].shape)
)

print(
    "Metadata:",
    tuple(batch["metadata"].shape)
)

print(
    "Emotions:",
    tuple(batch["emotions"].shape)
)

print(
    "Labels:",
    tuple(batch["labels"].shape)
)

assert (
    tuple(batch["input_ids"].shape)
    == (16, 512)
)

assert (
    tuple(batch["attention_mask"].shape)
    == (16, 512)
)

assert (
    tuple(batch["metadata"].shape)
    == (16, 53)
)

assert (
    tuple(batch["emotions"].shape)
    == (16, 11)
)

assert (
    tuple(batch["labels"].shape)
    == (16,)
)


print()
print("=" * 60)
print("BATCH DATA TYPES")
print("=" * 60)

print(
    "Input IDs:",
    batch["input_ids"].dtype
)

print(
    "Attention mask:",
    batch["attention_mask"].dtype
)

print(
    "Metadata:",
    batch["metadata"].dtype
)

print(
    "Emotions:",
    batch["emotions"].dtype
)

print(
    "Labels:",
    batch["labels"].dtype
)

assert batch["input_ids"].dtype == torch.int64
assert batch["attention_mask"].dtype == torch.int64
assert batch["metadata"].dtype == torch.float32
assert batch["emotions"].dtype == torch.float32
assert batch["labels"].dtype == torch.int64


print()
print("=" * 60)
print("TRAINING LABEL BALANCE")
print("=" * 60)

balanced_distribution = np.bincount(
    train_labels_balanced,
    minlength=2
)

print(
    "Balanced TRAIN:",
    balanced_distribution
)

assert np.array_equal(
    balanced_distribution,
    np.array([2013, 2013])
)


print()
print("=" * 60)
print("USER ALIGNMENT")
print("=" * 60)

assert (
    len(
        balanced_train_bert_emotion_df
    )
    ==
    len(train_dataset)
)

assert (
    len(train_emotion_balanced)
    ==
    len(train_labels_balanced)
)

print(
    "Balanced training users:",
    len(balanced_train_bert_emotion_df)
)

print("✓ User alignment verified")


dataset_info = {
    "train_users": 4026,
    "validation_users": 767,
    "test_users": 767,
    "batch_size": 16,
    "train_batches": 252,
    "validation_batches": 48,
    "test_batches": 48,
    "context_length": 512,
    "metadata_features": 53,
    "emotion_features": 11
}

with open(
    os.path.join(
        RESULT_DIR,
        "dataset_info.json"
    ),
    "w"
) as f:
    json.dump(
        dataset_info,
        f,
        indent=2
    )


print()
print("=" * 60)
print("CELL 9 COMPLETE")
print("=" * 60)

print("✓ Balanced TRAIN Dataset created")
print("✓ Validation Dataset unchanged")
print("✓ Test Dataset unchanged")
print("✓ TRAIN = 4,026 users")
print("✓ TRAIN = 2,013 negative + 2,013 positive")
print("✓ Batch size = 16")
print("✓ Training shuffle enabled")
print("✓ Validation shuffle disabled")
print("✓ Test shuffle disabled")
print("✓ 512-token inputs aligned")
print("✓ 53 metadata features aligned")
print("✓ 11 emotion features aligned")
print("✓ Labels aligned")
print("✓ DataLoader sanity check passed")
```

```python
# ============================================================
# CELL 10 — BERT + EMOTION-BASED
# ============================================================

from transformers import AutoModel
import torch.nn as nn

print("=" * 60)
print("BERT + EMOTION-BASED — BUILDING FROZEN BERT + METADATA + EMOTION MODEL")
print("=" * 60)

bert_encoder = AutoModel.from_pretrained(
    MODEL_NAME,
    token=hf_token
).to(device)

for param in bert_encoder.parameters():
    param.requires_grad = False


class BERTEmotionClassifier(nn.Module):

    def __init__(
        self,
        encoder,
        metadata_dim=53,
        emotion_dim=11
    ):
        super().__init__()

        self.encoder = encoder
        self.text_hidden_size = 768

        # Text attention pooling
        self.attention_projection = nn.Linear(
            768,
            256
        )

        self.attention_score = nn.Linear(
            256,
            1
        )

        # Metadata branch
        self.metadata_branch = nn.Sequential(
            nn.Linear(
                metadata_dim,
                128
            ),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(
                128,
                64
            ),
            nn.ReLU()
        )

        # Emotion branch
        self.emotion_branch = nn.Sequential(
            nn.Linear(
                emotion_dim,
                64
            ),
            nn.ReLU(),
            nn.Dropout(0.3)
        )

        # Learnable gates
        self.metadata_beta = nn.Parameter(
            torch.tensor(0.0)
        )

        self.emotion_beta = nn.Parameter(
            torch.tensor(0.0)
        )

        # 768 text + 64 metadata + 64 emotion
        self.classifier = nn.Sequential(
            nn.Linear(
                896,
                512
            ),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(
                512,
                256
            ),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(
                256,
                2
            )
        )

    def forward(
        self,
        input_ids,
        attention_mask,
        metadata,
        emotions
    ):

        # Frozen BERT
        with torch.no_grad():

            outputs = self.encoder(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

        hidden_states = outputs.last_hidden_state

        # Attention pooling
        projected = torch.tanh(
            self.attention_projection(
                hidden_states
            )
        )

        scores = self.attention_score(
            projected
        ).squeeze(-1)

        scores = scores.masked_fill(
            attention_mask == 0,
            -1e9
        )

        attention_weights = torch.softmax(
            scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # Metadata
        metadata_representation = (
            self.metadata_branch(metadata)
        )

        metadata_gate = torch.sigmoid(
            self.metadata_beta
        )

        gated_metadata = (
            metadata_gate *
            metadata_representation
        )

        # Emotion
        emotion_representation = (
            self.emotion_branch(emotions)
        )

        emotion_gate = torch.sigmoid(
            self.emotion_beta
        )

        gated_emotion = (
            emotion_gate *
            emotion_representation
        )

        # Fusion
        fused = torch.cat(
            [
                text_representation,
                gated_metadata,
                gated_emotion
            ],
            dim=1
        )

        logits = self.classifier(
            fused
        )

        return {
            "logits": logits,
            "text_representation":
                text_representation,
            "metadata_representation":
                metadata_representation,
            "emotion_representation":
                emotion_representation,
            "attention_weights":
                attention_weights,
            "metadata_gate":
                metadata_gate,
            "emotion_gate":
                emotion_gate
        }


model = BERTEmotionClassifier(
    bert_encoder,
    metadata_dim=53,
    emotion_dim=11
).to(device)


# ============================================================
# PARAMETER COUNT
# ============================================================

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

frozen_params = sum(
    p.numel()
    for p in model.parameters()
    if not p.requires_grad
)

print()
print("=" * 60)
print("PARAMETER COUNT")
print("=" * 60)

print(
    "Total parameters:",
    total_params
)

print(
    "Trainable parameters:",
    trainable_params
)

print(
    "Frozen parameters:",
    frozen_params
)

print(
    "Total parameters (M):",
    round(
        total_params / 1e6,
        3
    )
)

print(
    "Trainable parameters (M):",
    round(
        trainable_params / 1e6,
        3
    )
)

print(
    "Frozen parameters (M):",
    round(
        frozen_params / 1e6,
        3
    )
)

encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print(
    "✓ All BERT backbone parameters frozen"
)


# ============================================================
# TRAINABLE COMPONENTS
# ============================================================

print()
print("=" * 60)
print("TRAINABLE COMPONENTS")
print("=" * 60)

for name, param in model.named_parameters():

    if param.requires_grad:
        print(
            "✓",
            name
        )


# ============================================================
# ARCHITECTURE
# ============================================================

print()
print("=" * 60)
print("ARCHITECTURE CONFIGURATION")
print("=" * 60)

print(
    "Transformer:",
    MODEL_NAME
)

print(
    "Base architecture: BERT-base-uncased"
)

print(
    "Text hidden size: 768"
)

print(
    "Attention projection: 256"
)

print(
    "Metadata: 53 -> 128 -> 64"
)

print(
    "Emotion: 11 -> 64"
)

print(
    "Metadata gate: g_m = sigmoid(beta_m)"
)

print(
    "Emotion gate: g_e = sigmoid(beta_e)"
)

print(
    "Initial metadata beta: 0.0"
)

print(
    "Initial emotion beta: 0.0"
)

print(
    "Initial metadata gate: 0.5"
)

print(
    "Initial emotion gate: 0.5"
)

print(
    "Fusion: 768 + 64 + 64 = 896"
)

print(
    "Classifier: 896 -> 512 -> 256 -> 2"
)

print(
    "Dropout: 0.3"
)

print(
    "Transformer frozen: True"
)


# ============================================================
# FORWARD PASS SANITY CHECK
# ============================================================

print()
print("=" * 60)
print("FORWARD PASS SANITY CHECK")
print("=" * 60)

model.eval()

batch = next(
    iter(train_loader)
)

input_ids = batch[
    "input_ids"
].to(device)

attention_mask = batch[
    "attention_mask"
].to(device)

metadata = batch[
    "metadata"
].to(device)

emotions = batch[
    "emotions"
].to(device)

with torch.no_grad():

    output = model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        metadata=metadata,
        emotions=emotions
    )

print(
    "Input IDs:",
    tuple(
        input_ids.shape
    )
)

print(
    "Attention mask:",
    tuple(
        attention_mask.shape
    )
)

print(
    "Metadata:",
    tuple(
        metadata.shape
    )
)

print(
    "Emotions:",
    tuple(
        emotions.shape
    )
)

print(
    "Logits:",
    tuple(
        output["logits"].shape
    )
)

print(
    "Text representation:",
    tuple(
        output[
            "text_representation"
        ].shape
    )
)

print(
    "Metadata representation:",
    tuple(
        output[
            "metadata_representation"
        ].shape
    )
)

print(
    "Emotion representation:",
    tuple(
        output[
            "emotion_representation"
        ].shape
    )
)

print(
    "Attention weights:",
    tuple(
        output[
            "attention_weights"
        ].shape
    )
)

metadata_gate = float(
    output[
        "metadata_gate"
    ].item()
)

emotion_gate = float(
    output[
        "emotion_gate"
    ].item()
)

print(
    "Metadata gate:",
    metadata_gate
)

print(
    "Emotion gate:",
    emotion_gate
)

padding_attention = output[
    "attention_weights"
][attention_mask == 0]

max_padding_attention = (
    float(
        padding_attention.max().item()
    )
    if padding_attention.numel()
    else 0.0
)

print(
    "Maximum attention on padding:",
    max_padding_attention
)

assert output[
    "logits"
].shape == (16, 2)

assert output[
    "text_representation"
].shape == (16, 768)

assert output[
    "metadata_representation"
].shape == (16, 64)

assert output[
    "emotion_representation"
].shape == (16, 64)

assert output[
    "attention_weights"
].shape == (16, 512)

assert abs(
    metadata_gate - 0.5
) < 1e-6

assert abs(
    emotion_gate - 0.5
) < 1e-6

assert max_padding_attention == 0.0

assert encoder_trainable == 0

print()
print(
    "✓ Transformer freeze verified"
)

print(
    "✓ Metadata gate verified"
)

print(
    "✓ Emotion gate verified"
)


# ============================================================
# SAVE CONFIGURATION
# ============================================================

architecture_config = {
    "model_name": MODEL_NAME,
    "base_architecture": "BERT-base-uncased",
    "text_hidden_size": 768,
    "attention_projection": 256,
    "metadata_branch": "53 -> 128 -> 64",
    "emotion_branch": "11 -> 64",
    "metadata_gate": "sigmoid(metadata_beta)",
    "emotion_gate": "sigmoid(emotion_beta)",
    "initial_metadata_beta": 0.0,
    "initial_emotion_beta": 0.0,
    "initial_metadata_gate": 0.5,
    "initial_emotion_gate": 0.5,
    "fusion": "768 + 64 + 64 = 896",
    "classifier": "896 -> 512 -> 256 -> 2",
    "dropout": 0.3,
    "transformer_frozen": True,
    "total_parameters": total_params,
    "trainable_parameters": trainable_params,
    "frozen_parameters": frozen_params
}

with open(
    os.path.join(
        RESULT_DIR,
        "architecture_config.json"
    ),
    "w"
) as f:

    json.dump(
        architecture_config,
        f,
        indent=2
    )


print()
print("=" * 60)
print("CELL 10 COMPLETE")
print("=" * 60)

print(
    "✓ BERT-base-uncased loaded"
)

print(
    "✓ Transformer backbone frozen"
)

print(
    "✓ Attention pooling active"
)

print(
    "✓ Metadata branch = 53 -> 128 -> 64"
)

print(
    "✓ Emotion branch = 11 -> 64"
)

print(
    "✓ Metadata gate active"
)

print(
    "✓ Emotion gate active"
)

print(
    "✓ Text representation = 768"
)

print(
    "✓ Fusion = 768 + 64 + 64 = 896"
)

print(
    "✓ Classifier = 896 -> 512 -> 256 -> 2"
)

print(
    "✓ Padding attention masked"
)

print(
    "✓ Forward pass successful"
)

print(
    "✓ Frozen BERT + metadata + emotion model ready"
)
```

```python
# ============================================================
# CELL 11 — BERT + EMOTION-BASED
# ============================================================

import os
import time
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

MAX_EPOCHS = 10
EARLY_STOPPING_PATIENCE = 2
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1

print("=" * 60)
print("STARTING BERT + EMOTION-BASED TRAINING")
print("=" * 60)

print()
print("Training users:", len(train_dataset))
print("Validation users:", len(val_dataset))
print("Test users:", len(test_dataset))
print("Maximum epochs:", MAX_EPOCHS)
print("Early stopping patience:", EARLY_STOPPING_PATIENCE)
print("Checkpoint metric: Validation F1")
print("Learning rate:", LEARNING_RATE)
print("Weight decay:", WEIGHT_DECAY)
print("Batch size:", BATCH_SIZE)
print("Transformer: FROZEN")

# Verify BERT remains frozen
encoder_trainable = sum(
    p.numel()
    for p in model.encoder.parameters()
    if p.requires_grad
)

assert encoder_trainable == 0

print()
print("✓ BERT backbone confirmed frozen")

trainable_params = [
    p for p in model.parameters()
    if p.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_params,
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

total_steps = (
    len(train_loader)
    * MAX_EPOCHS
)

warmup_steps = int(
    total_steps * WARMUP_RATIO
)


def lr_lambda(step):
    if step < warmup_steps:
        return (
            (step + 1)
            / max(1, warmup_steps)
        )

    return max(
        0.0,
        (total_steps - step)
        / max(
            1,
            total_steps - warmup_steps
        )
    )


scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda
)

criterion = nn.CrossEntropyLoss()


def evaluate_model(model, loader):

    model.eval()

    total_loss = 0.0
    labels_all = []
    preds_all = []
    probs_all = []

    with torch.no_grad():

        for batch in loader:

            input_ids = batch[
                "input_ids"
            ].to(device)

            attention_mask = batch[
                "attention_mask"
            ].to(device)

            metadata = batch[
                "metadata"
            ].to(device)

            emotions = batch[
                "emotions"
            ].to(device)

            labels = batch[
                "labels"
            ].to(device)

            output = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                metadata=metadata,
                emotions=emotions
            )

            logits = output["logits"]

            loss = criterion(
                logits,
                labels
            )

            total_loss += (
                loss.item()
                * labels.size(0)
            )

            probs = torch.softmax(
                logits,
                dim=1
            )[:, 1]

            preds = torch.argmax(
                logits,
                dim=1
            )

            labels_all.extend(
                labels.cpu().numpy()
            )

            preds_all.extend(
                preds.cpu().numpy()
            )

            probs_all.extend(
                probs.cpu().numpy()
            )

    return {
        "loss":
            total_loss
            / len(loader.dataset),

        "accuracy":
            accuracy_score(
                labels_all,
                preds_all
            ),

        "precision":
            precision_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "recall":
            recall_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "f1":
            f1_score(
                labels_all,
                preds_all,
                zero_division=0
            ),

        "roc_auc":
            roc_auc_score(
                labels_all,
                probs_all
            )
    }


history = []

best_val_f1 = -np.inf
best_epoch = None
best_state_dict = None

epochs_without_improvement = 0

training_start = time.time()


for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    epoch_start = time.time()

    print()
    print("=" * 60)
    print(
        f"EPOCH {epoch}/{MAX_EPOCHS}"
    )
    print("=" * 60)

    model.train()

    running_loss = 0.0
    train_labels = []
    train_preds = []

    for batch_idx, batch in enumerate(
        train_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        emotions = batch[
            "emotions"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)

        optimizer.zero_grad()

        output = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata,
            emotions=emotions
        )

        logits = output["logits"]

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            trainable_params,
            max_norm=1.0
        )

        optimizer.step()
        scheduler.step()

        running_loss += (
            loss.item()
            * labels.size(0)
        )

        preds = torch.argmax(
            logits.detach(),
            dim=1
        )

        train_labels.extend(
            labels.detach()
            .cpu()
            .numpy()
        )

        train_preds.extend(
            preds.cpu().numpy()
        )

        if (
            batch_idx == 1
            or batch_idx % 50 == 0
            or batch_idx == len(train_loader)
        ):

            batch_accuracy = accuracy_score(
                labels.detach()
                .cpu()
                .numpy(),
                preds.cpu().numpy()
            )

            elapsed = (
                time.time()
                - epoch_start
            )

            print(
                f"Epoch {epoch} | "
                f"Batch "
                f"{batch_idx}/"
                f"{len(train_loader)} | "
                f"Loss: "
                f"{loss.item():.4f} | "
                f"Accuracy: "
                f"{batch_accuracy:.4f} | "
                f"LR: "
                f"{optimizer.param_groups[0]['lr']:.2e} | "
                f"Time: "
                f"{elapsed:.1f}s"
            )

    train_loss = (
        running_loss
        / len(train_loader.dataset)
    )

    train_accuracy = accuracy_score(
        train_labels,
        train_preds
    )

    train_precision = precision_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_recall = recall_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    train_f1 = f1_score(
        train_labels,
        train_preds,
        zero_division=0
    )

    print()
    print(
        f"Epoch {epoch} — Validation"
    )

    val_metrics = evaluate_model(
        model,
        val_loader
    )

    with torch.no_grad():

        metadata_gate = float(
            torch.sigmoid(
                model.metadata_beta
            ).item()
        )

        emotion_gate = float(
            torch.sigmoid(
                model.emotion_beta
            ).item()
        )

    epoch_time = (
        time.time()
        - epoch_start
    )

    print()
    print("=" * 60)
    print(f"EPOCH {epoch} SUMMARY")
    print("=" * 60)

    print(
        f"Train Loss: "
        f"{train_loss:.4f}"
    )

    print(
        f"Train Accuracy: "
        f"{train_accuracy:.4f}"
    )

    print(
        f"Train Precision: "
        f"{train_precision:.4f}"
    )

    print(
        f"Train Recall: "
        f"{train_recall:.4f}"
    )

    print(
        f"Train F1: "
        f"{train_f1:.4f}"
    )

    print()
    print(
        f"Val Loss: "
        f"{val_metrics['loss']:.4f}"
    )

    print(
        f"Val Accuracy: "
        f"{val_metrics['accuracy']:.4f}"
    )

    print(
        f"Val Precision: "
        f"{val_metrics['precision']:.4f}"
    )

    print(
        f"Val Recall: "
        f"{val_metrics['recall']:.4f}"
    )

    print(
        f"Val F1: "
        f"{val_metrics['f1']:.4f}"
    )

    print(
        f"Val ROC-AUC: "
        f"{val_metrics['roc_auc']:.4f}"
    )

    print(
        f"Metadata gate: "
        f"{metadata_gate:.6f}"
    )

    print(
        f"Emotion gate: "
        f"{emotion_gate:.6f}"
    )

    print(
        f"Epoch Time: "
        f"{epoch_time:.2f}s"
    )

    history.append({
        "epoch": epoch,

        "train_loss":
            train_loss,

        "train_accuracy":
            train_accuracy,

        "train_precision":
            train_precision,

        "train_recall":
            train_recall,

        "train_f1":
            train_f1,

        "val_loss":
            val_metrics["loss"],

        "val_accuracy":
            val_metrics["accuracy"],

        "val_precision":
            val_metrics["precision"],

        "val_recall":
            val_metrics["recall"],

        "val_f1":
            val_metrics["f1"],

        "val_roc_auc":
            val_metrics["roc_auc"],

        "metadata_gate":
            metadata_gate,

        "emotion_gate":
            emotion_gate,

        "learning_rate":
            optimizer.param_groups[0]["lr"],

        "epoch_time":
            epoch_time
    })

    # --------------------------------------------------------
    # BEST CHECKPOINT
    # --------------------------------------------------------

    if val_metrics["f1"] > best_val_f1:

        best_val_f1 = (
            val_metrics["f1"]
        )

        best_epoch = epoch

        best_state_dict = {
            key:
                value.detach()
                .cpu()
                .clone()
            for key, value
            in model.state_dict().items()
        }

        epochs_without_improvement = 0

        checkpoint_path = os.path.join(
            CHECKPOINT_DIR,
            "best_bert_emotion.pt"
        )

        torch.save(
            {
                "epoch":
                    epoch,

                "model_state_dict":
                    best_state_dict,

                "optimizer_state_dict":
                    optimizer.state_dict(),

                "scheduler_state_dict":
                    scheduler.state_dict(),

                "best_val_f1":
                    best_val_f1,

                "experiment_name":
                    EXPERIMENT_NAME,

                "model_name":
                    MODEL_NAME,

                "cleaning_method":
                    CLEANING_METHOD
            },
            checkpoint_path
        )

        print()
        print("✓ NEW BEST CHECKPOINT")
        print(
            "Best Validation F1:",
            f"{best_val_f1:.4f}"
        )

        print(
            "Saved:",
            checkpoint_path
        )

    else:

        epochs_without_improvement += 1

        print()
        print(
            "No validation F1 improvement."
        )

        print(
            "Epochs without improvement:",
            f"{epochs_without_improvement} / "
            f"{EARLY_STOPPING_PATIENCE}"
        )

    # --------------------------------------------------------
    # EARLY STOPPING
    # --------------------------------------------------------

    if (
        epochs_without_improvement
        >= EARLY_STOPPING_PATIENCE
    ):

        print()
        print(
            "EARLY STOPPING TRIGGERED"
        )

        break


assert best_state_dict is not None
assert best_epoch is not None

model.load_state_dict(
    best_state_dict
)

model.to(device)
model.eval()

total_training_time = (
    time.time()
    - training_start
)

history_df = pd.DataFrame(
    history
)

history_path = os.path.join(
    RESULT_DIR,
    "bert_emotion_training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)

with torch.no_grad():

    learned_metadata_beta = float(
        model.metadata_beta.item()
    )

    learned_emotion_beta = float(
        model.emotion_beta.item()
    )

    learned_metadata_gate = float(
        torch.sigmoid(
            model.metadata_beta
        ).item()
    )

    learned_emotion_gate = float(
        torch.sigmoid(
            model.emotion_beta
        ).item()
    )


print()
print("=" * 60)
print("BERT + EMOTION-BASED TRAINING COMPLETE")
print("=" * 60)

print(
    "Epochs completed:",
    len(history)
)

print(
    "Best epoch:",
    best_epoch
)

print(
    "Best Validation F1:",
    f"{best_val_f1:.4f}"
)

print(
    "Total training time:",
    f"{total_training_time:.2f} seconds"
)

print(
    "Best checkpoint:",
    os.path.join(
        CHECKPOINT_DIR,
        "best_bert_emotion.pt"
    )
)

print(
    "Training history saved:",
    history_path
)

print()
print(
    "Learned metadata beta:",
    learned_metadata_beta
)

print(
    "Learned emotion beta:",
    learned_emotion_beta
)

print(
    "Metadata gate:",
    f"{learned_metadata_gate:.6f}"
)

print(
    "Emotion gate:",
    f"{learned_emotion_gate:.6f}"
)

print()
print("✓ BERT transformer remained frozen")
print("✓ Metadata branch trained")
print("✓ Emotion branch trained")
print("✓ Metadata gate learned")
print("✓ Emotion gate learned")

print()
print("=" * 60)
print("CELL 11 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# BERT + EMOTION-BASED — COMPLETE GRAPH GENERATION
# ============================================================

import os
import time
import numpy as np
import matplotlib.pyplot as plt

from tqdm.auto import tqdm

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve,
    precision_recall_curve,
    average_precision_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# ============================================================
# GRAPH DIRECTORY
# ============================================================

GRAPH_DIR = "/kaggle/working/bert_emotion/results/graphs"

os.makedirs(
    GRAPH_DIR,
    exist_ok=True
)

print("=" * 60)
print("BERT + EMOTION-BASED — COMPLETE GRAPH GENERATION")
print("=" * 60)

print()
print("Graph directory:")
print(GRAPH_DIR)

print("✓ Graph directory ready")


# ============================================================
# STEP 1 — RECREATE TEST PREDICTIONS
# ============================================================

print()
print("=" * 60)
print("RECREATING TEST PREDICTIONS")
print("=" * 60)


# ------------------------------------------------------------
# LOAD BEST CHECKPOINT
# ------------------------------------------------------------

best_checkpoint_path = (
    "/kaggle/working/bert_emotion/"
    "checkpoints/best_bert_emotion.pt"
)

assert os.path.exists(
    best_checkpoint_path
), (
    f"Checkpoint not found:\n"
    f"{best_checkpoint_path}"
)


checkpoint = torch.load(
    best_checkpoint_path,
    map_location=device,
    weights_only=False
)


model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.to(device)

model.eval()


print()
print("✓ Best checkpoint loaded")

print(
    "Best epoch:",
    checkpoint.get("epoch", "Unknown")
)

print(
    "Best validation F1:",
    round(
        checkpoint.get(
            "best_val_f1",
            0
        ),
        4
    )
)


# ------------------------------------------------------------
# PREDICTION ARRAYS
# ------------------------------------------------------------

all_labels = []

all_predictions = []

all_probabilities = []


print()
print("Evaluating test set...")


# ------------------------------------------------------------
# TEST EVALUATION
# ------------------------------------------------------------

with torch.no_grad():

    for batch in tqdm(
        test_loader,
        desc="TEST"
    ):

        input_ids = (
            batch["input_ids"]
            .to(device)
        )

        attention_mask = (
            batch["attention_mask"]
            .to(device)
        )

        metadata = (
            batch["metadata"]
            .to(device)
        )

        emotions = (
            batch["emotions"]
            .to(device)
        )

        labels = (
            batch["labels"]
            .to(device)
        )


        # ----------------------------------------------------
        # FORWARD PASS
        # ----------------------------------------------------

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata,
            emotions=emotions
        )


        # ----------------------------------------------------
        # MODEL RETURNS DICTIONARY
        # ----------------------------------------------------

        logits = outputs["logits"]


        # ----------------------------------------------------
        # PROBABILITIES
        # ----------------------------------------------------

        probabilities = torch.softmax(
            logits,
            dim=1
        )


        # Positive class probability
        positive_probabilities = (
            probabilities[:, 1]
        )


        # Predicted class
        predictions = torch.argmax(
            probabilities,
            dim=1
        )


        # ----------------------------------------------------
        # STORE RESULTS
        # ----------------------------------------------------

        all_labels.extend(
            labels.cpu().numpy()
        )

        all_predictions.extend(
            predictions.cpu().numpy()
        )

        all_probabilities.extend(
            positive_probabilities.cpu().numpy()
        )


# ============================================================
# CONVERT TO NUMPY
# ============================================================

y_true = np.array(
    all_labels
)

y_pred = np.array(
    all_predictions
)

y_prob = np.array(
    all_probabilities
)


print()
print("Prediction shapes:")

print(
    "y_true:",
    y_true.shape
)

print(
    "y_pred:",
    y_pred.shape
)

print(
    "y_prob:",
    y_prob.shape
)


assert len(y_true) == 767

assert len(y_pred) == 767

assert len(y_prob) == 767


print()
print("✓ Test predictions recreated")
print("✓ All 767 users evaluated")
print("✓ Probabilities ready for ROC-AUC")


# ============================================================
# STEP 2 — CALCULATE METRICS
# ============================================================

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_true,
    y_prob
)

average_precision = average_precision_score(
    y_true,
    y_prob
)


print()
print("=" * 60)
print("RECALCULATED TEST METRICS")
print("=" * 60)

print(
    f"Accuracy: {accuracy * 100:.2f}%"
)

print(
    f"Precision: {precision * 100:.2f}%"
)

print(
    f"Recall: {recall * 100:.2f}%"
)

print(
    f"F1 Score: {f1 * 100:.2f}%"
)

print(
    f"ROC-AUC: {roc_auc * 100:.2f}%"
)

print(
    f"Average Precision: "
    f"{average_precision * 100:.2f}%"
)


# ============================================================
# GRAPH 1 — PERFORMANCE METRICS
# ============================================================

print()
print("=" * 60)
print("GRAPH 1 — PERFORMANCE METRICS")
print("=" * 60)


metric_names = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"
]

metric_values = [
    accuracy * 100,
    precision * 100,
    recall * 100,
    f1 * 100,
    roc_auc * 100
]


plt.figure(
    figsize=(10, 6)
)

bars = plt.bar(
    metric_names,
    metric_values
)


plt.ylim(
    0,
    100
)

plt.ylabel(
    "Score (%)"
)

plt.title(
    "BERT + Emotion-Based — Test Performance Metrics"
)

plt.grid(
    axis="y",
    alpha=0.3
)


for bar, value in zip(
    bars,
    metric_values
):

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center"
    )


plt.tight_layout()


performance_path = os.path.join(
    GRAPH_DIR,
    "performance_metrics.png"
)


plt.savefig(
    performance_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print("✓ Performance Metrics graph displayed")
print("✓ Saved:")
print(performance_path)


# ============================================================
# GRAPH 2 — ROC CURVE
# ============================================================

print()
print("=" * 60)
print("GRAPH 2 — ROC CURVE")
print("=" * 60)


fpr, tpr, thresholds = roc_curve(
    y_true,
    y_prob
)


plt.figure(
    figsize=(8, 7)
)


plt.plot(
    fpr,
    tpr,
    linewidth=2,
    label=(
        f"ROC Curve "
        f"(AUC = {roc_auc:.4f})"
    )
)


plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "BERT + Emotion-Based — ROC Curve"
)

plt.legend(
    loc="lower right"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()


roc_path = os.path.join(
    GRAPH_DIR,
    "roc_curve.png"
)


plt.savefig(
    roc_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print("✓ ROC Curve displayed")
print("✓ ROC-AUC:", round(roc_auc, 4))

print("✓ Saved:")
print(roc_path)


# ============================================================
# GRAPH 3 — PRECISION-RECALL CURVE
# ============================================================

print()
print("=" * 60)
print("GRAPH 3 — PRECISION-RECALL CURVE")
print("=" * 60)


pr_precision, pr_recall, pr_thresholds = (
    precision_recall_curve(
        y_true,
        y_prob
    )
)


plt.figure(
    figsize=(8, 7)
)


plt.plot(
    pr_recall,
    pr_precision,
    linewidth=2,
    label=(
        f"PR Curve "
        f"(AP = {average_precision:.4f})"
    )
)


plt.xlabel(
    "Recall"
)

plt.ylabel(
    "Precision"
)

plt.title(
    "BERT + Emotion-Based — Precision-Recall Curve"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


pr_path = os.path.join(
    GRAPH_DIR,
    "precision_recall_curve.png"
)


plt.savefig(
    pr_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print("✓ Precision-Recall Curve displayed")

print("✓ Average Precision:",
      round(
          average_precision,
          4
      )
)

print("✓ Saved:")
print(pr_path)


# ============================================================
# GRAPH 4 — CONFUSION MATRIX
# ============================================================

print()
print("=" * 60)
print("GRAPH 4 — CONFUSION MATRIX")
print("=" * 60)


cm = confusion_matrix(
    y_true,
    y_pred
)


print()
print("Confusion Matrix:")

print(cm)


fig, ax = plt.subplots(
    figsize=(7, 6)
)


display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "Negative",
        "Positive"
    ]
)


display.plot(
    ax=ax,
    values_format="d"
)


plt.title(
    "BERT + Emotion-Based — Confusion Matrix"
)

plt.tight_layout()


cm_path = os.path.join(
    GRAPH_DIR,
    "confusion_matrix.png"
)


plt.savefig(
    cm_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print("✓ Confusion Matrix displayed")

print("✓ Saved:")
print(cm_path)


# ============================================================
# GRAPH 5 — NORMALIZED CONFUSION MATRIX
# ============================================================

print()
print("=" * 60)
print("GRAPH 5 — NORMALIZED CONFUSION MATRIX")
print("=" * 60)


cm_normalized = confusion_matrix(
    y_true,
    y_pred,
    normalize="true"
)


fig, ax = plt.subplots(
    figsize=(7, 6)
)


display = ConfusionMatrixDisplay(
    confusion_matrix=cm_normalized,
    display_labels=[
        "Negative",
        "Positive"
    ]
)


display.plot(
    ax=ax,
    values_format=".3f"
)


plt.title(
    "BERT + Emotion-Based — Normalized Confusion Matrix"
)

plt.tight_layout()


normalized_cm_path = os.path.join(
    GRAPH_DIR,
    "normalized_confusion_matrix.png"
)


plt.savefig(
    normalized_cm_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print("✓ Normalized Confusion Matrix displayed")

print("✓ Saved:")
print(normalized_cm_path)


# ============================================================
# GRAPH 6 — PRECISION / RECALL / F1 COMPARISON
# ============================================================

print()
print("=" * 60)
print("GRAPH 6 — PRECISION, RECALL & F1")
print("=" * 60)


classification_names = [
    "Precision",
    "Recall",
    "F1 Score"
]


classification_values = [
    precision * 100,
    recall * 100,
    f1 * 100
]


plt.figure(
    figsize=(8, 6)
)


bars = plt.bar(
    classification_names,
    classification_values
)


plt.ylim(
    0,
    100
)

plt.ylabel(
    "Score (%)"
)

plt.title(
    "BERT + Emotion-Based — Positive Class Performance"
)

plt.grid(
    axis="y",
    alpha=0.3
)


for bar, value in zip(
    bars,
    classification_values
):

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center"
    )


plt.tight_layout()


prf_path = os.path.join(
    GRAPH_DIR,
    "precision_recall_f1.png"
)


plt.savefig(
    prf_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print("✓ Precision/Recall/F1 graph displayed")

print("✓ Saved:")
print(prf_path)


# ============================================================
# GRAPH 7 — CLASS DISTRIBUTION
# ============================================================

print()
print("=" * 60)
print("GRAPH 7 — TEST CLASS DISTRIBUTION")
print("=" * 60)


negative_count = int(
    np.sum(
        y_true == 0
    )
)

positive_count = int(
    np.sum(
        y_true == 1
    )
)


class_names = [
    "Negative",
    "Positive"
]

class_counts = [
    negative_count,
    positive_count
]


plt.figure(
    figsize=(8, 6)
)


bars = plt.bar(
    class_names,
    class_counts
)


plt.ylabel(
    "Number of Users"
)

plt.title(
    "BERT + Emotion-Based — Test Set Class Distribution"
)

plt.grid(
    axis="y",
    alpha=0.3
)


for bar, value in zip(
    bars,
    class_counts
):

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,

        value + 5,

        str(value),

        ha="center"
    )


plt.tight_layout()


class_distribution_path = os.path.join(
    GRAPH_DIR,
    "test_class_distribution.png"
)


plt.savefig(
    class_distribution_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print("✓ Test Class Distribution graph displayed")

print("✓ Saved:")
print(class_distribution_path)


# ============================================================
# GRAPH 8 — PREDICTED CLASS DISTRIBUTION
# ============================================================

print()
print("=" * 60)
print("GRAPH 8 — PREDICTED CLASS DISTRIBUTION")
print("=" * 60)


predicted_negative = int(
    np.sum(
        y_pred == 0
    )
)

predicted_positive = int(
    np.sum(
        y_pred == 1
    )
)


predicted_names = [
    "Predicted Negative",
    "Predicted Positive"
]


predicted_counts = [
    predicted_negative,
    predicted_positive
]


plt.figure(
    figsize=(8, 6)
)


bars = plt.bar(
    predicted_names,
    predicted_counts
)


plt.ylabel(
    "Number of Users"
)

plt.title(
    "BERT + Emotion-Based — Predicted Class Distribution"
)

plt.grid(
    axis="y",
    alpha=0.3
)


for bar, value in zip(
    bars,
    predicted_counts
):

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,

        value + 5,

        str(value),

        ha="center"
    )


plt.tight_layout()


prediction_distribution_path = os.path.join(
    GRAPH_DIR,
    "predicted_class_distribution.png"
)


plt.savefig(
    prediction_distribution_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print("✓ Predicted Class Distribution graph displayed")

print("✓ Saved:")
print(prediction_distribution_path)


# ============================================================
# GRAPH 9 — PROBABILITY DISTRIBUTION
# ============================================================

print()
print("=" * 60)
print("GRAPH 9 — POSITIVE CLASS PROBABILITY DISTRIBUTION")
print("=" * 60)


negative_probabilities = y_prob[
    y_true == 0
]

positive_probabilities = y_prob[
    y_true == 1
]


plt.figure(
    figsize=(10, 6)
)


plt.hist(
    negative_probabilities,
    bins=30,
    alpha=0.6,
    label="Actual Negative"
)


plt.hist(
    positive_probabilities,
    bins=30,
    alpha=0.6,
    label="Actual Positive"
)


plt.xlabel(
    "Predicted Positive Probability"
)

plt.ylabel(
    "Number of Users"
)

plt.title(
    "BERT + Emotion-Based — Positive Class Probability Distribution"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()


probability_path = os.path.join(
    GRAPH_DIR,
    "positive_probability_distribution.png"
)


plt.savefig(
    probability_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print("✓ Probability Distribution graph displayed")

print("✓ Saved:")
print(probability_path)


# ============================================================
# GRAPH 10 — TRAINING HISTORY DETECTION
# ============================================================

print()
print("=" * 60)
print("TRAINING HISTORY GRAPHS")
print("=" * 60)


# ------------------------------------------------------------
# FIND COMMON TRAINING HISTORY VARIABLE NAMES
# ------------------------------------------------------------

def find_variable(
    possible_names
):

    for name in possible_names:

        if name in globals():

            value = globals()[name]

            if value is not None:

                return value, name

    return None, None


# ------------------------------------------------------------
# TRAIN LOSS
# ------------------------------------------------------------

train_loss_history, train_loss_name = (
    find_variable([
        "train_losses",
        "train_loss_history",
        "history_train_loss",
        "epoch_train_losses"
    ])
)


# ------------------------------------------------------------
# VALIDATION LOSS
# ------------------------------------------------------------

val_loss_history, val_loss_name = (
    find_variable([
        "val_losses",
        "validation_losses",
        "val_loss_history",
        "history_val_loss",
        "epoch_val_losses"
    ])
)


# ------------------------------------------------------------
# TRAIN F1
# ------------------------------------------------------------

train_f1_history, train_f1_name = (
    find_variable([
        "train_f1_scores",
        "train_f1_history",
        "train_f1",
        "epoch_train_f1"
    ])
)


# ------------------------------------------------------------
# VALIDATION F1
# ------------------------------------------------------------

val_f1_history, val_f1_name = (
    find_variable([
        "val_f1_scores",
        "validation_f1_scores",
        "val_f1_history",
        "val_f1",
        "epoch_val_f1"
    ])
)


# ------------------------------------------------------------
# EPOCH TIMES
# ------------------------------------------------------------

epoch_times, epoch_times_name = (
    find_variable([
        "epoch_times",
        "training_times",
        "epoch_training_times",
        "epoch_durations"
    ])
)


print()

print(
    "Train loss history:",
    train_loss_name
)

print(
    "Validation loss history:",
    val_loss_name
)

print(
    "Train F1 history:",
    train_f1_name
)

print(
    "Validation F1 history:",
    val_f1_name
)

print(
    "Epoch times:",
    epoch_times_name
)


# ============================================================
# GRAPH 10A — TRAINING / VALIDATION LOSS
# ============================================================

if (
    train_loss_history is not None
    and
    val_loss_history is not None
):

    print()
    print("Generating Training & Validation Loss graph...")


    epochs_loss = np.arange(
        1,
        len(train_loss_history) + 1
    )


    plt.figure(
        figsize=(10, 6)
    )


    plt.plot(
        epochs_loss,
        train_loss_history,
        marker="o",
        label="Training Loss"
    )


    plt.plot(
        epochs_loss,
        val_loss_history,
        marker="o",
        label="Validation Loss"
    )


    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Loss"
    )

    plt.title(
        "BERT + Emotion-Based — Training and Validation Loss"
    )

    plt.xticks(
        epochs_loss
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()


    loss_path = os.path.join(
        GRAPH_DIR,
        "training_validation_loss.png"
    )


    plt.savefig(
        loss_path,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()

    plt.close()


    print("✓ Loss graph displayed")

    print("✓ Saved:")
    print(loss_path)


else:

    print()
    print(
        "⚠ Training/Validation loss history "
        "not found in memory."
    )

    print(
        "Skipping loss graph."
    )


# ============================================================
# GRAPH 10B — TRAINING / VALIDATION F1
# ============================================================

if (
    train_f1_history is not None
    and
    val_f1_history is not None
):

    print()
    print("Generating Training & Validation F1 graph...")


    epochs_f1 = np.arange(
        1,
        len(train_f1_history) + 1
    )


    plt.figure(
        figsize=(10, 6)
    )


    plt.plot(
        epochs_f1,
        train_f1_history,
        marker="o",
        label="Training F1"
    )


    plt.plot(
        epochs_f1,
        val_f1_history,
        marker="o",
        label="Validation F1"
    )


    # Best epoch
    best_epoch_index = int(
        np.argmax(
            val_f1_history
        )
    )


    best_epoch_number = (
        best_epoch_index + 1
    )


    best_f1_value = (
        val_f1_history[
            best_epoch_index
        ]
    )


    plt.scatter(
        best_epoch_number,
        best_f1_value,
        s=100,
        label=(
            f"Best Validation F1 "
            f"(Epoch {best_epoch_number})"
        )
    )


    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "F1 Score"
    )

    plt.title(
        "BERT + Emotion-Based — Training and Validation F1"
    )

    plt.xticks(
        epochs_f1
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()


    f1_history_path = os.path.join(
        GRAPH_DIR,
        "training_validation_f1.png"
    )


    plt.savefig(
        f1_history_path,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()

    plt.close()


    print("✓ Training/Validation F1 graph displayed")

    print("✓ Saved:")
    print(f1_history_path)


else:

    print()
    print(
        "⚠ Training/Validation F1 history "
        "not found in memory."
    )

    print(
        "Skipping F1 history graph."
    )


# ============================================================
# GRAPH 10C — TRAINING TIME PER EPOCH
# ============================================================

if epoch_times is not None:

    print()
    print("Generating Training Time per Epoch graph...")


    epoch_numbers = np.arange(
        1,
        len(epoch_times) + 1
    )


    plt.figure(
        figsize=(10, 6)
    )


    bars = plt.bar(
        epoch_numbers,
        epoch_times
    )


    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Training Time (seconds)"
    )

    plt.title(
        "BERT + Emotion-Based — Training Time per Epoch"
    )

    plt.xticks(
        epoch_numbers
    )

    plt.grid(
        axis="y",
        alpha=0.3
    )


    for bar, value in zip(
        bars,
        epoch_times
    ):

        plt.text(
            bar.get_x()
            + bar.get_width() / 2,

            value,

            f"{value:.1f}s",

            ha="center",
            va="bottom"
        )


    plt.tight_layout()


    training_time_path = os.path.join(
        GRAPH_DIR,
        "training_time_per_epoch.png"
    )


    plt.savefig(
        training_time_path,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()

    plt.close()


    total_training_time = float(
        np.sum(
            epoch_times
        )
    )


    print(
        "Total training time:",
        f"{total_training_time:.2f} seconds"
    )

    print(
        "Total training time:",
        f"{total_training_time / 60:.2f} minutes"
    )

    print("✓ Training time graph displayed")

    print("✓ Saved:")
    print(training_time_path)


else:

    print()
    print(
        "⚠ Epoch timing data "
        "not found in memory."
    )

    print(
        "Skipping training time graph."
    )


# ============================================================
# SAVE FINAL METRICS
# ============================================================

print()
print("=" * 60)
print("SAVING GRAPH METRICS")
print("=" * 60)


graph_metrics = {
    "accuracy": float(accuracy),
    "precision": float(precision),
    "recall": float(recall),
    "f1_score": float(f1),
    "roc_auc": float(roc_auc),
    "average_precision": float(
        average_precision
    )
}


import json


graph_metrics_path = os.path.join(
    GRAPH_DIR,
    "graph_metrics.json"
)


with open(
    graph_metrics_path,
    "w"
) as file:

    json.dump(
        graph_metrics,
        file,
        indent=4
    )


print("✓ Graph metrics saved:")
print(graph_metrics_path)


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 60)
print("GRAPH GENERATION COMPLETE")
print("=" * 60)

print()

print("CORE TEST GRAPHS GENERATED:")

print("✓ 1. Performance Metrics")
print("✓ 2. ROC Curve")
print("✓ 3. Precision-Recall Curve")
print("✓ 4. Confusion Matrix")
print("✓ 5. Normalized Confusion Matrix")
print("✓ 6. Precision / Recall / F1 Comparison")
print("✓ 7. Test Class Distribution")
print("✓ 8. Predicted Class Distribution")
print("✓ 9. Positive Probability Distribution")

print()

print("TRAINING GRAPHS:")

if (
    train_loss_history is not None
    and
    val_loss_history is not None
):

    print(
        "✓ Training / Validation Loss"
    )

else:

    print(
        "⚠ Loss graph unavailable "
        "(history variable missing)"
    )


if (
    train_f1_history is not None
    and
    val_f1_history is not None
):

    print(
        "✓ Training / Validation F1"
    )

else:

    print(
        "⚠ F1 history graph unavailable "
        "(history variable missing)"
    )


if epoch_times is not None:

    print(
        "✓ Training Time per Epoch"
    )

else:

    print(
        "⚠ Training Time graph unavailable "
        "(timing variable missing)"
    )


print()

print("All available graphs saved to:")

print(GRAPH_DIR)

print()
print("✓ COMPLETE")
```

```python
# ============================================================
# EXPLAINABILITY — SHAP
# BERT + EMOTION-BASED
# GLOBAL METADATA + EMOTION FEATURE EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import shap

print("=" * 70)
print("SHAP ANALYSIS")
print("BERT + EMOTION-BASED")
print("=" * 70)

# ============================================================
# FIXED SHAP SETTINGS
# ============================================================

SHAP_SEED = 42
SHAP_BACKGROUND_SIZE = 50
SHAP_TEST_SIZE_PER_CLASS = 50
SHAP_TOP_FEATURES = 20

np.random.seed(SHAP_SEED)
torch.manual_seed(SHAP_SEED)

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

SHAP_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "shap"
)

os.makedirs(
    SHAP_DIR,
    exist_ok=True
)

# ============================================================
# FEATURE NAMES
# ============================================================

metadata_feature_names = list(
    selected_metadata_names
)

assert len(metadata_feature_names) == 53

emotion_feature_names = list(
    emotion_columns
)

assert len(emotion_feature_names) == 11

feature_names = (
    metadata_feature_names
    +
    emotion_feature_names
)

assert len(feature_names) == 64

print()
print(
    "Metadata features:",
    len(metadata_feature_names)
)

print(
    "Emotion features:",
    len(emotion_feature_names)
)

print(
    "Total auxiliary features:",
    len(feature_names)
)

# ============================================================
# FIXED BACKGROUND DATA
# TRAINING USERS ONLY
# ============================================================

background_metadata = (
    train_dataset.metadata[
        :SHAP_BACKGROUND_SIZE
    ]
    .clone()
    .detach()
)

background_emotions = (
    train_dataset.emotions[
        :SHAP_BACKGROUND_SIZE
    ]
    .clone()
    .detach()
)

background_auxiliary = torch.cat(
    [
        background_metadata,
        background_emotions
    ],
    dim=1
).to(device)

print()
print(
    "SHAP background samples:",
    background_auxiliary.shape[0]
)

print(
    "Background dimensions:",
    background_auxiliary.shape[1]
)

assert background_auxiliary.shape == (
    SHAP_BACKGROUND_SIZE,
    64
)

# ============================================================
# FIXED TEST USERS
# 50 NEGATIVE + 50 POSITIVE
# ============================================================

test_labels_tensor = (
    test_dataset.labels
    .detach()
    .cpu()
)

negative_indices = torch.where(
    test_labels_tensor == 0
)[0].numpy()

positive_indices = torch.where(
    test_labels_tensor == 1
)[0].numpy()

assert len(
    negative_indices
) >= SHAP_TEST_SIZE_PER_CLASS

assert len(
    positive_indices
) >= SHAP_TEST_SIZE_PER_CLASS

shap_negative_indices = (
    negative_indices[
        :SHAP_TEST_SIZE_PER_CLASS
    ]
)

shap_positive_indices = (
    positive_indices[
        :SHAP_TEST_SIZE_PER_CLASS
    ]
)

shap_test_indices = np.concatenate(
    [
        shap_negative_indices,
        shap_positive_indices
    ]
)

print()
print(
    "SHAP test samples:",
    len(shap_test_indices)
)

print(
    "Negative:",
    len(shap_negative_indices)
)

print(
    "Positive:",
    len(shap_positive_indices)
)

# ============================================================
# MODEL WRAPPER
# ============================================================
# Text input remains fixed for each explained user.
# Metadata + emotion features are explained together.
# ============================================================

class AuxiliarySHAPWrapper(
    nn.Module
):

    def __init__(
        self,
        base_model,
        fixed_input_ids,
        fixed_attention_mask
    ):

        super().__init__()

        self.base_model = base_model

        self.fixed_input_ids = (
            fixed_input_ids
            .to(device)
        )

        self.fixed_attention_mask = (
            fixed_attention_mask
            .to(device)
        )

    def forward(
        self,
        auxiliary_features
    ):

        batch_size = (
            auxiliary_features.size(0)
        )

        input_ids = (
            self.fixed_input_ids
            .expand(
                batch_size,
                -1
            )
        )

        attention_mask = (
            self.fixed_attention_mask
            .expand(
                batch_size,
                -1
            )
        )

        metadata = (
            auxiliary_features[:, :53]
        )

        emotions = (
            auxiliary_features[:, 53:]
        )

        output = self.base_model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata,
            emotions=emotions
        )

        positive_logit = output[
            "logits"
        ][:, 1:2]

        return positive_logit


# ============================================================
# MODEL PREPARATION
# ============================================================

model.eval()

for parameter in model.parameters():
    parameter.requires_grad = False

print()
print(
    "✓ BERT Emotion-Based model set to evaluation mode"
)

print(
    "✓ Model parameters frozen for SHAP"
)

# ============================================================
# COMPUTE SHAP VALUES
# ============================================================

all_shap_values = []
all_auxiliary_values = []
all_explained_labels = []

print()
print("=" * 70)
print("COMPUTING SHAP VALUES")
print("=" * 70)

for count, test_index in enumerate(
    shap_test_indices,
    start=1
):

    sample = test_dataset[
        int(test_index)
    ]

    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
    )

    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
    )

    metadata_sample = (
        sample["metadata"]
        .unsqueeze(0)
        .float()
        .to(device)
    )

    emotion_sample = (
        sample["emotions"]
        .unsqueeze(0)
        .float()
        .to(device)
    )

    auxiliary_sample = torch.cat(
        [
            metadata_sample,
            emotion_sample
        ],
        dim=1
    )

    label = int(
        sample["labels"].item()
    )

    # --------------------------------------------------------
    # USER-SPECIFIC WRAPPER
    # --------------------------------------------------------

    wrapper = AuxiliarySHAPWrapper(
        model,
        input_ids,
        attention_mask
    ).to(device)

    wrapper.eval()

    # --------------------------------------------------------
    # GRADIENT SHAP
    # --------------------------------------------------------

    explainer = shap.GradientExplainer(
        wrapper,
        background_auxiliary
    )

    shap_result = explainer.shap_values(
        auxiliary_sample
    )

    # --------------------------------------------------------
    # HANDLE SHAP OUTPUT FORMAT
    # --------------------------------------------------------

    if isinstance(
        shap_result,
        list
    ):

        shap_array = np.asarray(
            shap_result[0]
        )

    else:

        shap_array = np.asarray(
            shap_result
        )

    shap_array = np.squeeze(
        shap_array
    )

    assert shap_array.shape == (
        64,
    )

    all_shap_values.append(
        shap_array
    )

    all_auxiliary_values.append(
        auxiliary_sample
        .detach()
        .cpu()
        .numpy()
        .squeeze(0)
    )

    all_explained_labels.append(
        label
    )

    if (
        count == 1
        or count % 10 == 0
        or count == len(
            shap_test_indices
        )
    ):

        print(
            f"Explained "
            f"{count}/"
            f"{len(shap_test_indices)} "
            f"test users"
        )

# ============================================================
# COMBINE RESULTS
# ============================================================

all_shap_values = np.vstack(
    all_shap_values
)

all_auxiliary_values = np.vstack(
    all_auxiliary_values
)

all_explained_labels = np.array(
    all_explained_labels
)

print()
print("=" * 70)
print("SHAP RESULT SUMMARY")
print("=" * 70)

print(
    "SHAP matrix:",
    all_shap_values.shape
)

print(
    "Auxiliary feature matrix:",
    all_auxiliary_values.shape
)

assert all_shap_values.shape == (
    len(shap_test_indices),
    64
)

assert all_auxiliary_values.shape == (
    len(shap_test_indices),
    64
)

assert np.isfinite(
    all_shap_values
).all()

# ============================================================
# GLOBAL FEATURE IMPORTANCE
# ============================================================

mean_abs_shap = np.mean(
    np.abs(
        all_shap_values
    ),
    axis=0
)

mean_signed_shap = np.mean(
    all_shap_values,
    axis=0
)

feature_types = (
    ["metadata"] * 53
    +
    ["emotion"] * 11
)

feature_importance_df = pd.DataFrame({

    "feature":
        feature_names,

    "feature_type":
        feature_types,

    "mean_abs_shap":
        mean_abs_shap,

    "mean_signed_shap":
        mean_signed_shap
})

feature_importance_df = (
    feature_importance_df
    .sort_values(
        "mean_abs_shap",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)

print()
print("=" * 70)
print("TOP 20 SHAP FEATURES")
print("=" * 70)

print(
    feature_importance_df
    .head(SHAP_TOP_FEATURES)
    .to_string(
        index=False
    )
)

# ============================================================
# SAVE SHAP VALUES
# ============================================================

shap_values_df = pd.DataFrame(
    all_shap_values,
    columns=feature_names
)

shap_values_df.to_csv(
    os.path.join(
        SHAP_DIR,
        "bert_emotion_shap_values.csv"
    ),
    index=False
)

# ============================================================
# SAVE FEATURE IMPORTANCE
# ============================================================

feature_importance_df.to_csv(
    os.path.join(
        SHAP_DIR,
        "bert_emotion_shap_feature_importance.csv"
    ),
    index=False
)

# ============================================================
# SAVE TEST USER INFORMATION
# ============================================================

explained_users_df = pd.DataFrame({

    "test_index":
        shap_test_indices,

    "label":
        all_explained_labels

})

explained_users_df.to_csv(
    os.path.join(
        SHAP_DIR,
        "bert_emotion_shap_test_users.csv"
    ),
    index=False
)

# ============================================================
# SHAP BEESWARM
# ============================================================

plt.figure(
    figsize=(11, 8)
)

shap.summary_plot(
    all_shap_values,
    all_auxiliary_values,
    feature_names=feature_names,
    max_display=SHAP_TOP_FEATURES,
    show=False
)

plt.title(
    "BERT + Emotion-Based — "
    "SHAP Feature Importance",
    fontsize=14
)

plt.tight_layout()

shap_beeswarm_path = os.path.join(
    SHAP_DIR,
    "bert_emotion_shap_beeswarm.png"
)

plt.savefig(
    shap_beeswarm_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

# ============================================================
# SHAP BAR
# ============================================================

top_features = (
    feature_importance_df
    .head(
        SHAP_TOP_FEATURES
    )
    .iloc[::-1]
)

plt.figure(
    figsize=(10, 8)
)

plt.barh(
    top_features["feature"],
    top_features["mean_abs_shap"]
)

plt.xlabel(
    "Mean Absolute SHAP Value",
    fontsize=12
)

plt.ylabel(
    "Metadata / Emotion Feature",
    fontsize=12
)

plt.title(
    "BERT + Emotion-Based — "
    "Global SHAP Feature Importance",
    fontsize=14
)

plt.tight_layout()

shap_bar_path = os.path.join(
    SHAP_DIR,
    "bert_emotion_shap_bar.png"
)

plt.savefig(
    shap_bar_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

# ============================================================
# SAVE CONFIG
# ============================================================

shap_config = {

    "model":
        "BERT",

    "cleaning":
        "Emotion-Based",

    "dataset":
        "Held-out test set",

    "metadata_features":
        53,

    "emotion_features":
        11,

    "total_explained_features":
        64,

    "background_samples":
        SHAP_BACKGROUND_SIZE,

    "test_samples_per_class":
        SHAP_TEST_SIZE_PER_CLASS,

    "total_explained_users":
        len(shap_test_indices),

    "target":
        "Positive/depression class logit",

    "explainer":
        "GradientExplainer",

    "text_fixed_per_user":
        True,

    "random_seed":
        SHAP_SEED,

    "top_features":
        SHAP_TOP_FEATURES
}

with open(
    os.path.join(
        SHAP_DIR,
        "bert_emotion_shap_config.json"
    ),
    "w"
) as f:

    json.dump(
        shap_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("SHAP ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ 53 metadata + 11 emotion features explained"
)

print(
    "✓ 50 negative + 50 positive users"
)

print(
    "✓ Training-only background used"
)

print(
    "✓ Positive-class logit explained"
)

print(
    "✓ SHAP values saved"
)

print(
    "✓ Feature ranking saved"
)

print(
    "✓ Beeswarm saved and displayed"
)

print(
    "✓ Bar plot saved and displayed"
)

print(
    "✓ Configuration saved"
)

print()
print(
    "SHAP output directory:",
    SHAP_DIR
)
```

```python
# ============================================================
# BERT + EMOTION-BASED — t-SNE
# FUSED REPRESENTATION ANALYSIS
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

print("=" * 70)
print("t-SNE REPRESENTATION ANALYSIS")
print("BERT + EMOTION-BASED")
print("=" * 70)

# ============================================================
# FIXED SETTINGS
# ============================================================

EXPLAINABILITY_SEED = 42

PCA_COMPONENTS = 50

TSNE_COMPONENTS = 2
TSNE_PERPLEXITY = 30
TSNE_LEARNING_RATE = 200
TSNE_MAX_ITER = 1000

np.random.seed(
    EXPLAINABILITY_SEED
)

torch.manual_seed(
    EXPLAINABILITY_SEED
)

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

TSNE_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "tsne"
)

os.makedirs(
    TSNE_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model.eval()

print()
print(
    "✓ BERT Emotion-Based model in evaluation mode"
)

# ============================================================
# EXTRACT EXACT FUSED REPRESENTATION
# ============================================================

fused_representations = []
test_labels = []

print()
print("=" * 70)
print("EXTRACTING FUSED REPRESENTATIONS")
print("=" * 70)

with torch.no_grad():

    for batch_idx, batch in enumerate(
        test_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        emotions = batch[
            "emotions"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)

        # ----------------------------------------------------
        # BERT ENCODER
        # ----------------------------------------------------

        encoder_output = model.encoder(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        hidden_states = (
            encoder_output.last_hidden_state
        )

        # ----------------------------------------------------
        # EXACT TEXT ATTENTION POOLING
        # ----------------------------------------------------

        projected = torch.tanh(
            model.attention_projection(
                hidden_states
            )
        )

        attention_scores = (
            model.attention_score(
                projected
            )
            .squeeze(-1)
        )

        attention_scores = (
            attention_scores.masked_fill(
                attention_mask == 0,
                -1e9
            )
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states
            *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # ----------------------------------------------------
        # EXACT METADATA BRANCH
        # ----------------------------------------------------

        metadata_representation = (
            model.metadata_branch(
                metadata
            )
        )

        metadata_gate = torch.sigmoid(
            model.metadata_beta
        )

        gated_metadata = (
            metadata_gate
            *
            metadata_representation
        )

        # ----------------------------------------------------
        # EXACT EMOTION BRANCH
        # ----------------------------------------------------

        emotion_representation = (
            model.emotion_branch(
                emotions
            )
        )

        emotion_gate = torch.sigmoid(
            model.emotion_beta
        )

        gated_emotion = (
            emotion_gate
            *
            emotion_representation
        )

        # ----------------------------------------------------
        # EXACT FUSED REPRESENTATION
        # ----------------------------------------------------

        fused = torch.cat(
            [
                text_representation,
                gated_metadata,
                gated_emotion
            ],
            dim=1
        )

        fused_representations.append(
            fused.cpu().numpy()
        )

        test_labels.extend(
            labels.cpu().numpy().tolist()
        )

        if (
            batch_idx == 1
            or batch_idx % 10 == 0
            or batch_idx == len(test_loader)
        ):

            print(
                f"Processed batch "
                f"{batch_idx}/{len(test_loader)}"
            )

# ============================================================
# COMBINE
# ============================================================

fused_representations = np.concatenate(
    fused_representations,
    axis=0
)

test_labels = np.array(
    test_labels
)

print()
print("=" * 70)
print("FUSED REPRESENTATION SUMMARY")
print("=" * 70)

print(
    "Fused representation shape:",
    fused_representations.shape
)

print(
    "Test users:",
    len(test_labels)
)

print(
    "Fused dimensions:",
    fused_representations.shape[1]
)

print(
    "Negative users:",
    int(
        (test_labels == 0).sum()
    )
)

print(
    "Positive users:",
    int(
        (test_labels == 1).sum()
    )
)

print(
    "Metadata gate:",
    f"{float(metadata_gate.item()):.6f}"
)

print(
    "Emotion gate:",
    f"{float(emotion_gate.item()):.6f}"
)

assert (
    fused_representations.shape[0]
    ==
    len(test_labels)
)

assert (
    fused_representations.shape[1]
    ==
    896
)

assert np.isfinite(
    fused_representations
).all()

# ============================================================
# PCA
# ============================================================

print()
print("=" * 70)
print("PCA REDUCTION")
print("=" * 70)

pca_dimensions = min(
    PCA_COMPONENTS,
    fused_representations.shape[1]
)

pca = PCA(
    n_components=pca_dimensions,
    random_state=EXPLAINABILITY_SEED
)

pca_representation = pca.fit_transform(
    fused_representations
)

pca_variance_retained = (
    pca.explained_variance_ratio_.sum()
)

print(
    "Original dimensions:",
    fused_representations.shape[1]
)

print(
    "PCA dimensions:",
    pca_representation.shape[1]
)

print(
    "Variance retained:",
    f"{pca_variance_retained * 100:.2f}%"
)

# ============================================================
# t-SNE
# ============================================================

print()
print("=" * 70)
print("t-SNE REDUCTION")
print("=" * 70)

tsne = TSNE(
    n_components=TSNE_COMPONENTS,
    perplexity=TSNE_PERPLEXITY,
    learning_rate=TSNE_LEARNING_RATE,
    init="pca",
    random_state=EXPLAINABILITY_SEED,
    max_iter=TSNE_MAX_ITER
)

tsne_representation = tsne.fit_transform(
    pca_representation
)

print(
    "t-SNE output shape:",
    tsne_representation.shape
)

# ============================================================
# SAVE COORDINATES
# ============================================================

tsne_df = pd.DataFrame({

    "tsne_1":
        tsne_representation[:, 0],

    "tsne_2":
        tsne_representation[:, 1],

    "label":
        test_labels

})

tsne_df["class"] = tsne_df[
    "label"
].map({
    0: "Negative",
    1: "Positive"
})

tsne_csv_path = os.path.join(
    TSNE_DIR,
    "bert_emotion_tsne_coordinates.csv"
)

tsne_df.to_csv(
    tsne_csv_path,
    index=False
)

# ============================================================
# PLOT
# ============================================================

plt.figure(
    figsize=(10, 8)
)

negative_mask = (
    test_labels == 0
)

positive_mask = (
    test_labels == 1
)

plt.scatter(
    tsne_representation[
        negative_mask,
        0
    ],
    tsne_representation[
        negative_mask,
        1
    ],
    s=40,
    alpha=0.72,
    label="Negative",
    color="#4C78A8",
    edgecolors="white",
    linewidths=0.4
)

plt.scatter(
    tsne_representation[
        positive_mask,
        0
    ],
    tsne_representation[
        positive_mask,
        1
    ],
    s=40,
    alpha=0.72,
    label="Positive",
    color="#E45756",
    edgecolors="white",
    linewidths=0.4
)

plt.xlabel(
    "t-SNE Dimension 1",
    fontsize=12
)

plt.ylabel(
    "t-SNE Dimension 2",
    fontsize=12
)

plt.title(
    "BERT + Emotion-Based — "
    "t-SNE of Fused Test Representations",
    fontsize=14,
    fontweight="bold"
)

plt.legend(
    fontsize=11
)

plt.grid(
    alpha=0.20
)

plt.tight_layout()

tsne_plot_path = os.path.join(
    TSNE_DIR,
    "bert_emotion_tsne.png"
)

plt.savefig(
    tsne_plot_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

# ============================================================
# SAVE CONFIGURATION
# ============================================================

tsne_config = {

    "model":
        "BERT",

    "cleaning":
        "Emotion-Based",

    "dataset":
        "Held-out test set",

    "test_users":
        int(len(test_labels)),

    "input_representation":
        "Fused classifier representation",

    "text_representation_dimension":
        768,

    "metadata_representation_dimension":
        64,

    "emotion_representation_dimension":
        64,

    "fused_representation_dimension":
        int(
            fused_representations.shape[1]
        ),

    "pca_dimensions":
        int(
            pca_dimensions
        ),

    "pca_variance_retained":
        float(
            pca_variance_retained
        ),

    "tsne_dimensions":
        int(
            TSNE_COMPONENTS
        ),

    "tsne_perplexity":
        int(
            TSNE_PERPLEXITY
        ),

    "tsne_learning_rate":
        TSNE_LEARNING_RATE,

    "tsne_max_iterations":
        int(
            TSNE_MAX_ITER
        ),

    "random_seed":
        EXPLAINABILITY_SEED,

    "metadata_gate":
        float(
            metadata_gate.item()
        ),

    "emotion_gate":
        float(
            emotion_gate.item()
        )
}

with open(
    os.path.join(
        TSNE_DIR,
        "bert_emotion_tsne_config.json"
    ),
    "w"
) as f:

    json.dump(
        tsne_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("t-SNE ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ Exact BERT + metadata + emotion fused representation extracted"
)

print(
    "✓ 896-dimensional fused representation verified"
)

print(
    "✓ PCA reduction completed"
)

print(
    "✓ t-SNE reduction completed"
)

print(
    "✓ t-SNE graph saved:",
    tsne_plot_path
)

print(
    "✓ Coordinates saved:",
    tsne_csv_path
)

print(
    "✓ Configuration saved"
)
```

```python
# ============================================================
# EXPLAINABILITY — LIME
# BERT + EMOTION-BASED
# LOCAL TOKEN-LEVEL EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from lime.lime_text import LimeTextExplainer
from IPython.display import display, HTML

print("=" * 70)
print("LIME ANALYSIS")
print("BERT + EMOTION-BASED")
print("=" * 70)

# ============================================================
# FIXED LIME SETTINGS
# ============================================================

LIME_SEED = 42
LIME_NUM_FEATURES = 15
LIME_NUM_SAMPLES = 1000
LIME_BATCH_SIZE = 8

np.random.seed(
    LIME_SEED
)

torch.manual_seed(
    LIME_SEED
)

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

LIME_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "lime"
)

os.makedirs(
    LIME_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model.eval()

for parameter in model.parameters():
    parameter.requires_grad = False

print()
print(
    "✓ BERT Emotion-Based model ready"
)

# ============================================================
# TEST PREDICTIONS
# ============================================================

labels_np = []
preds_np = []
probs_np = []

with torch.no_grad():

    for batch in test_loader:

        input_ids = (
            batch["input_ids"]
            .to(device)
        )

        attention_mask = (
            batch["attention_mask"]
            .to(device)
        )

        metadata = (
            batch["metadata"]
            .to(device)
        )

        emotions = (
            batch["emotions"]
            .to(device)
        )

        labels = (
            batch["labels"]
            .to(device)
        )

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata,
            emotions=emotions
        )

        logits = outputs[
            "logits"
        ]

        probabilities = torch.softmax(
            logits.float(),
            dim=1
        )[:, 1]

        predictions = (
            probabilities >= 0.5
        ).long()

        labels_np.extend(
            labels.cpu()
            .numpy()
            .tolist()
        )

        preds_np.extend(
            predictions.cpu()
            .numpy()
            .tolist()
        )

        probs_np.extend(
            probabilities.cpu()
            .numpy()
            .tolist()
        )

labels_np = np.asarray(
    labels_np
)

preds_np = np.asarray(
    preds_np
)

probs_np = np.asarray(
    probs_np
)

print()
print(
    "Test users:",
    len(labels_np)
)

# ============================================================
# DETERMINISTIC TP / TN / FP / FN
# ============================================================

tp_candidates = np.where(
    (labels_np == 1) &
    (preds_np == 1)
)[0]

tn_candidates = np.where(
    (labels_np == 0) &
    (preds_np == 0)
)[0]

fp_candidates = np.where(
    (labels_np == 0) &
    (preds_np == 1)
)[0]

fn_candidates = np.where(
    (labels_np == 1) &
    (preds_np == 0)
)[0]

assert len(tp_candidates) > 0
assert len(tn_candidates) > 0
assert len(fp_candidates) > 0
assert len(fn_candidates) > 0

selected_cases = {

    "TP":
        int(
            tp_candidates[0]
        ),

    "TN":
        int(
            tn_candidates[0]
        ),

    "FP":
        int(
            fp_candidates[0]
        ),

    "FN":
        int(
            fn_candidates[0]
        )
}

print()
print("=" * 70)
print("SELECTED CASES")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print(
        f"{case_name}: "
        f"Test Index = {test_index} | "
        f"True = {labels_np[test_index]} | "
        f"Pred = {preds_np[test_index]} | "
        f"Positive Probability = "
        f"{probs_np[test_index]:.4f}"
    )

# ============================================================
# LIME TEXT ACCESS
# ============================================================
# The same anchor-removed text used by the
# BERT Emotion-Based pipeline is used for LIME.
# ============================================================

def get_test_text(
    test_index
):

    return str(
        test_bert_emotion_df.iloc[
            int(test_index)
        ]["anchor_removed_text"]
    )

# ============================================================
# LIME PREDICTOR
# ============================================================
# Text is perturbed by LIME.
# Metadata and emotion features remain fixed
# for the explained user.
# ============================================================

def make_lime_predictor(
    test_index
):

    sample = test_dataset[
        int(test_index)
    ]

    fixed_metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .float()
        .to(device)
    )

    fixed_emotions = (
        sample["emotions"]
        .unsqueeze(0)
        .float()
        .to(device)
    )

    def lime_predict(
        texts
    ):

        all_probabilities = []

        for start in range(
            0,
            len(texts),
            LIME_BATCH_SIZE
        ):

            text_batch = texts[
                start:
                start + LIME_BATCH_SIZE
            ]

            encoded = tokenizer(
                text_batch,
                max_length=MAX_LENGTH,
                truncation=True,
                padding="max_length",
                return_attention_mask=True
            )

            input_ids = torch.tensor(
                encoded["input_ids"],
                dtype=torch.long
            ).to(device)

            attention_mask = torch.tensor(
                encoded["attention_mask"],
                dtype=torch.long
            ).to(device)

            metadata_batch = (
                fixed_metadata
                .expand(
                    len(text_batch),
                    -1
                )
            )

            emotion_batch = (
                fixed_emotions
                .expand(
                    len(text_batch),
                    -1
                )
            )

            with torch.no_grad():

                outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    metadata=metadata_batch,
                    emotions=emotion_batch
                )

                logits = outputs[
                    "logits"
                ]

                probabilities = torch.softmax(
                    logits.float(),
                    dim=1
                )

            all_probabilities.append(
                probabilities
                .cpu()
                .numpy()
            )

            del input_ids
            del attention_mask
            del metadata_batch
            del emotion_batch
            del outputs
            del logits
            del probabilities

            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        return np.vstack(
            all_probabilities
        )

    return lime_predict

# ============================================================
# LIME EXPLAINER
# ============================================================

explainer = LimeTextExplainer(
    class_names=[
        "Negative",
        "Positive"
    ],
    split_expression=r"\s+",
    random_state=LIME_SEED
)

print()
print(
    "✓ LIME explainer ready"
)

# ============================================================
# RUN LIME FOR FOUR CASES
# ============================================================

all_explanations = {}

print()
print("=" * 70)
print("GENERATING LIME EXPLANATIONS")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print()
    print(
        f"Processing {case_name} "
        f"(test index {test_index})..."
    )

    text = get_test_text(
        test_index
    )

    predictor = make_lime_predictor(
        test_index
    )

    explanation = explainer.explain_instance(
        text,
        predictor,
        labels=(1,),
        num_features=LIME_NUM_FEATURES,
        num_samples=LIME_NUM_SAMPLES
    )

    all_explanations[
        case_name
    ] = explanation

    print(
        f"✓ {case_name} explanation complete"
    )

# ============================================================
# SAVE LIME RESULTS
# ============================================================

all_lime_rows = []

for case_name, explanation in all_explanations.items():

    explanation_list = explanation.as_list(
        label=1
    )

    for rank, (
        word,
        weight
    ) in enumerate(
        explanation_list,
        start=1
    ):

        all_lime_rows.append({

            "case":
                case_name,

            "test_index":
                selected_cases[
                    case_name
                ],

            "true_label":
                int(
                    labels_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "predicted_label":
                int(
                    preds_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "positive_probability":
                float(
                    probs_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "rank":
                rank,

            "word":
                word,

            "lime_weight":
                float(weight)

        })

lime_results_df = pd.DataFrame(
    all_lime_rows
)

lime_csv_path = os.path.join(
    LIME_DIR,
    "bert_emotion_lime_results.csv"
)

lime_results_df.to_csv(
    lime_csv_path,
    index=False
)

# ============================================================
# DISPLAY + SAVE HTML
# ============================================================

print()
print("=" * 70)
print("LIME EXPLANATIONS")
print("=" * 70)

for case_name, explanation in all_explanations.items():

    print()
    print(
        "=" * 60
    )

    print(
        f"{case_name} — "
        f"LIME LOCAL EXPLANATION"
    )

    print(
        "=" * 60
    )

    display(
        HTML(
            explanation.as_html(
                labels=[1]
            )
        )
    )

    html_path = os.path.join(
        LIME_DIR,
        f"bert_emotion_lime_{case_name}.html"
    )

    with open(
        html_path,
        "w",
        encoding="utf-8"
    ) as f:

        f.write(
            explanation.as_html(
                labels=[1]
            )
        )

# ============================================================
# PAPER-FRIENDLY LIME BAR PLOTS
# ============================================================

for case_name, explanation in all_explanations.items():

    explanation_list = explanation.as_list(
        label=1
    )

    plot_df = pd.DataFrame(
        explanation_list,
        columns=[
            "word",
            "weight"
        ]
    )

    plot_df = plot_df.sort_values(
        "weight"
    )

    plt.figure(
        figsize=(10, 7)
    )

    bars = plt.barh(
        plot_df["word"],
        plot_df["weight"]
    )

    for bar, weight in zip(
        bars,
        plot_df["weight"]
    ):

        if weight > 0:

            bar.set_color(
                "#E45756"
            )

        else:

            bar.set_color(
                "#4C78A8"
            )

    plt.axvline(
        0,
        linewidth=0.8
    )

    plt.xlabel(
        "LIME Weight",
        fontsize=12
    )

    plt.ylabel(
        "Token / Word",
        fontsize=12
    )

    plt.title(
        f"BERT + Emotion-Based — "
        f"LIME Explanation ({case_name})",
        fontsize=14
    )

    plt.grid(
        axis="x",
        alpha=0.20
    )

    plt.tight_layout()

    plot_path = os.path.join(
        LIME_DIR,
        f"bert_emotion_lime_{case_name}.png"
    )

    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()

# ============================================================
# SAVE CASE INFORMATION
# ============================================================

case_info_rows = []

for case_name, test_index in selected_cases.items():

    case_info_rows.append({

        "case":
            case_name,

        "test_index":
            int(test_index),

        "true_label":
            int(
                labels_np[
                    test_index
                ]
            ),

        "predicted_label":
            int(
                preds_np[
                    test_index
                ]
            ),

        "positive_probability":
            float(
                probs_np[
                    test_index
                ]
            ),

        "text_length_characters":
            len(
                get_test_text(
                    test_index
                )
            )

    })

case_info_df = pd.DataFrame(
    case_info_rows
)

case_info_df.to_csv(
    os.path.join(
        LIME_DIR,
        "bert_emotion_lime_case_information.csv"
    ),
    index=False
)

# ============================================================
# SAVE CONFIGURATION
# ============================================================

lime_config = {

    "model":
        "BERT",

    "cleaning":
        "Emotion-Based",

    "dataset":
        "Held-out test set",

    "target":
        "Positive/depression class",

    "representative_cases":
        [
            "TP",
            "TN",
            "FP",
            "FN"
        ],

    "case_selection":
        "First matching test example in fixed test order",

    "random_seed":
        LIME_SEED,

    "num_samples":
        LIME_NUM_SAMPLES,

    "num_features":
        LIME_NUM_FEATURES,

    "batch_size":
        LIME_BATCH_SIZE,

    "metadata":
        "Fixed for each explained user",

    "emotion_features":
        "Fixed for each explained user",

    "text_perturbation":
        "LIME word-level perturbations",

    "model_training":
        "No additional training"
}

with open(
    os.path.join(
        LIME_DIR,
        "bert_emotion_lime_config.json"
    ),
    "w"
) as f:

    json.dump(
        lime_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("LIME ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ TP, TN, FP, FN cases explained"
)

print(
    "✓ Same trained BERT Emotion-Based model used"
)

print(
    "✓ Metadata held fixed per user"
)

print(
    "✓ Emotion features held fixed per user"
)

print(
    "✓ 1000 LIME perturbations per case"
)

print(
    "✓ Top 15 features extracted"
)

print(
    "✓ LIME results saved"
)

print(
    "✓ HTML explanations saved"
)

print(
    "✓ Paper-ready LIME plots saved and displayed"
)

print()
print(
    "LIME output directory:",
    LIME_DIR
)
```

```python
# ============================================================
# EXPLAINABILITY — GRAD-CAM
# BERT + EMOTION-BASED
# TOKEN-LEVEL EXPLANATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

print("=" * 70)
print("GRAD-CAM ANALYSIS")
print("BERT + EMOTION-BASED")
print("=" * 70)

# ============================================================
# FIXED SETTINGS
# ============================================================

GRADCAM_SEED = 42
TOP_TOKENS = 20

np.random.seed(
    GRADCAM_SEED
)

torch.manual_seed(
    GRADCAM_SEED
)

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

GRADCAM_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "gradcam"
)

os.makedirs(
    GRADCAM_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model.eval()

for parameter in model.parameters():
    parameter.requires_grad = False

print()
print(
    "✓ BERT Emotion-Based model ready"
)

# ============================================================
# COLLECT TEST PREDICTIONS
# ============================================================

test_true = []
test_pred = []
test_prob = []

with torch.no_grad():

    for batch in test_loader:

        input_ids = (
            batch["input_ids"]
            .to(device)
        )

        attention_mask = (
            batch["attention_mask"]
            .to(device)
        )

        metadata = (
            batch["metadata"]
            .to(device)
        )

        emotions = (
            batch["emotions"]
            .to(device)
        )

        labels = (
            batch["labels"]
            .to(device)
        )

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata,
            emotions=emotions
        )

        logits = outputs[
            "logits"
        ]

        probabilities = torch.softmax(
            logits.float(),
            dim=1
        )[:, 1]

        predictions = (
            probabilities >= 0.5
        ).long()

        test_true.extend(
            labels.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        test_pred.extend(
            predictions.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        test_prob.extend(
            probabilities.detach()
            .cpu()
            .numpy()
            .tolist()
        )

test_true = np.asarray(
    test_true
)

test_pred = np.asarray(
    test_pred
)

test_prob = np.asarray(
    test_prob
)

print()
print(
    "Test users:",
    len(test_true)
)

# ============================================================
# DETERMINISTIC TP / TN / FP / FN
# ============================================================

tp_candidates = np.where(
    (test_true == 1) &
    (test_pred == 1)
)[0]

tn_candidates = np.where(
    (test_true == 0) &
    (test_pred == 0)
)[0]

fp_candidates = np.where(
    (test_true == 0) &
    (test_pred == 1)
)[0]

fn_candidates = np.where(
    (test_true == 1) &
    (test_pred == 0)
)[0]

assert len(tp_candidates) > 0
assert len(tn_candidates) > 0
assert len(fp_candidates) > 0
assert len(fn_candidates) > 0

selected_cases = {

    "TP":
        int(tp_candidates[0]),

    "TN":
        int(tn_candidates[0]),

    "FP":
        int(fp_candidates[0]),

    "FN":
        int(fn_candidates[0])
}

print()
print("=" * 70)
print("SELECTED CASES")
print("=" * 70)

for case_name, index in selected_cases.items():

    print(
        f"{case_name}: "
        f"Test Index = {index} | "
        f"True = {test_true[index]} | "
        f"Pred = {test_pred[index]} | "
        f"Positive Probability = {test_prob[index]:.4f}"
    )

# ============================================================
# GRAD-CAM FUNCTION
# ============================================================

def compute_gradcam(
    sample_index
):

    sample = test_dataset[
        sample_index
    ]

    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
        .to(device)
    )

    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
        .to(device)
    )

    metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
    )

    emotions = (
        sample["emotions"]
        .unsqueeze(0)
        .to(device)
    )

    # --------------------------------------------------------
    # ENABLE GRADIENTS
    # --------------------------------------------------------

    model.zero_grad(
        set_to_none=True
    )

    with torch.enable_grad():

        # ----------------------------------------------------
        # BERT ENCODER
        # ----------------------------------------------------

        encoder_output = model.encoder(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        hidden_states = (
            encoder_output.last_hidden_state
        )

        hidden_states = hidden_states.detach()
        hidden_states.requires_grad_(True)
        hidden_states.retain_grad()

        # ----------------------------------------------------
        # ATTENTION POOLING
        # ----------------------------------------------------

        projected = torch.tanh(
            model.attention_projection(
                hidden_states
            )
        )

        attention_scores = (
            model.attention_score(
                projected
            )
            .squeeze(-1)
        )

        attention_scores = (
            attention_scores.masked_fill(
                attention_mask == 0,
                -1e9
            )
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        text_representation = torch.sum(
            hidden_states *
            attention_weights.unsqueeze(-1),
            dim=1
        )

        # ----------------------------------------------------
        # METADATA BRANCH
        # ----------------------------------------------------

        metadata_representation = (
            model.metadata_branch(
                metadata
            )
        )

        metadata_gate = torch.sigmoid(
            model.metadata_beta
        )

        gated_metadata = (
            metadata_gate *
            metadata_representation
        )

        # ----------------------------------------------------
        # EMOTION BRANCH
        # ----------------------------------------------------

        emotion_representation = (
            model.emotion_branch(
                emotions
            )
        )

        emotion_gate = torch.sigmoid(
            model.emotion_beta
        )

        gated_emotion = (
            emotion_gate *
            emotion_representation
        )

        # ----------------------------------------------------
        # FUSION
        # ----------------------------------------------------

        fused = torch.cat(
            [
                text_representation,
                gated_metadata,
                gated_emotion
            ],
            dim=1
        )

        # ----------------------------------------------------
        # CLASSIFIER
        # ----------------------------------------------------

        logits = model.classifier(
            fused
        )

        positive_logit = (
            logits[:, 1]
        )

        positive_logit.backward()

        # ----------------------------------------------------
        # GRAD-CAM
        # ----------------------------------------------------

        gradients = hidden_states.grad

        channel_weights = gradients.mean(
            dim=1
        )

        cam = (
            (
                hidden_states *
                channel_weights.unsqueeze(1)
            )
            .sum(dim=2)
            .squeeze(0)
        )

        cam = torch.relu(
            cam
        )

        cam = (
            cam /
            (
                cam.max() +
                1e-12
            )
        )

    cam = (
        cam
        .detach()
        .cpu()
        .numpy()
    )

    tokens = tokenizer.convert_ids_to_tokens(
        input_ids.squeeze(0)
        .detach()
        .cpu()
        .tolist()
    )

    mask = (
        attention_mask.squeeze(0)
        .detach()
        .cpu()
        .numpy()
        .astype(bool)
    )

    tokens = [
        token
        for token, valid in zip(
            tokens,
            mask
        )
        if valid
    ]

    cam = cam[
        mask
    ]

    # --------------------------------------------------------
    # REMOVE SPECIAL TOKENS
    # --------------------------------------------------------

    valid_pairs = []

    special_tokens = {
        tokenizer.cls_token,
        tokenizer.sep_token,
        tokenizer.pad_token
    }

    for token, importance in zip(
        tokens,
        cam
    ):

        if token in special_tokens:
            continue

        valid_pairs.append(
            (
                token,
                float(importance)
            )
        )

    # --------------------------------------------------------
    # TOP TOKENS
    # --------------------------------------------------------

    valid_pairs = sorted(
        valid_pairs,
        key=lambda x: x[1],
        reverse=True
    )

    top_pairs = valid_pairs[
        :TOP_TOKENS
    ]

    return (
        top_pairs,
        tokens,
        cam,
        int(test_true[sample_index]),
        int(test_pred[sample_index]),
        float(test_prob[sample_index])
    )


# ============================================================
# RUN FOUR CASES
# ============================================================

gradcam_results = {}

print()
print("=" * 70)
print("COMPUTING GRAD-CAM")
print("=" * 70)

for case_name, sample_index in selected_cases.items():

    print()
    print(
        f"Processing {case_name} "
        f"(Test Index {sample_index})"
    )

    result = compute_gradcam(
        sample_index
    )

    gradcam_results[
        case_name
    ] = result

    print(
        f"✓ {case_name} complete"
    )

# ============================================================
# SAVE TOKEN IMPORTANCE CSV
# ============================================================

all_rows = []

for case_name, result in gradcam_results.items():

    top_pairs = result[0]

    true_label = result[3]
    predicted_label = result[4]
    positive_probability = result[5]

    for rank, (
        token,
        importance
    ) in enumerate(
        top_pairs,
        start=1
    ):

        all_rows.append({

            "case":
                case_name,

            "rank":
                rank,

            "token":
                token,

            "importance":
                importance,

            "true_label":
                true_label,

            "predicted_label":
                predicted_label,

            "positive_probability":
                positive_probability
        })

gradcam_df = pd.DataFrame(
    all_rows
)

gradcam_csv_path = os.path.join(
    GRADCAM_DIR,
    "bert_emotion_gradcam_token_importance.csv"
)

gradcam_df.to_csv(
    gradcam_csv_path,
    index=False
)

# ============================================================
# PLOT FOUR CASES
# ============================================================

case_titles = {

    "TP":
        "True Positive",

    "TN":
        "True Negative",

    "FP":
        "False Positive",

    "FN":
        "False Negative"
}

for case_name, result in gradcam_results.items():

    top_pairs = result[0]

    tokens_plot = [
        pair[0]
        for pair in top_pairs
    ][::-1]

    importance_plot = [
        pair[1]
        for pair in top_pairs
    ][::-1]

    plt.figure(
        figsize=(10, 7)
    )

    bars = plt.barh(
        tokens_plot,
        importance_plot
    )

    if len(bars) > 0:

        max_index = int(
            np.argmax(
                importance_plot
            )
        )

        bars[max_index].set_color(
            "#F28E2B"
        )

        for index, bar in enumerate(
            bars
        ):

            if index != max_index:

                bar.set_color(
                    "#59A14F"
                )

    plt.xlabel(
        "Grad-CAM Token Importance"
    )

    plt.ylabel(
        "Token"
    )

    plt.title(
        "BERT + Emotion-Based — "
        f"{case_titles[case_name]}"
    )

    plt.grid(
        axis="x",
        alpha=0.20
    )

    plt.tight_layout()

    plot_path = os.path.join(
        GRADCAM_DIR,
        f"bert_emotion_gradcam_{case_name.lower()}.png"
    )

    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()

# ============================================================
# SAVE CONFIGURATION
# ============================================================

gradcam_config = {

    "model":
        "BERT",

    "cleaning":
        "Emotion-Based",

    "method":
        "Gradient-weighted Class Activation Mapping",

    "target":
        "Positive/depression-class logit",

    "representation":
        "Final Transformer hidden states",

    "metadata_branch":
        "Included in classifier forward pass",

    "emotion_branch":
        "Included in classifier forward pass",

    "channel_weight":
        "Mean gradient across token positions",

    "activation":
        "ReLU",

    "normalization":
        "Per-user max normalization",

    "top_tokens":
        TOP_TOKENS,

    "seed":
        GRADCAM_SEED,

    "cases":
        selected_cases,

    "test_users":
        len(test_true)

}

with open(
    os.path.join(
        GRADCAM_DIR,
        "bert_emotion_gradcam_config.json"
    ),
    "w"
) as f:

    json.dump(
        gradcam_config,
        f,
        indent=2
    )

# ============================================================
# PRINT RESULTS
# ============================================================

print()
print("=" * 70)
print("GRAD-CAM ANALYSIS COMPLETE")
print("=" * 70)

for case_name, result in gradcam_results.items():

    print()
    print(
        f"{case_name} — "
        f"True: {result[3]} | "
        f"Predicted: {result[4]} | "
        f"P(Positive): {result[5]:.4f}"
    )

    print(
        "Top tokens:",
        result[0]
    )

print()
print(
    "✓ TP, TN, FP, FN cases explained"
)

print(
    "✓ Final BERT hidden states used"
)

print(
    "✓ Metadata + emotion branches included"
)

print(
    "✓ Positive/depression-class logit used"
)

print(
    "✓ Top 20 tokens extracted"
)

print(
    "✓ Token importance CSV saved"
)

print(
    "✓ Four paper-ready plots saved and displayed"
)

print(
    "✓ Configuration saved"
)

print()
print(
    "Grad-CAM output directory:",
    GRADCAM_DIR
)
```
