```python
# ============================================================
# CELL 3 — EMOBERTA-BASE + RAW
# EXACT D1 ANCHOR DETECTION
# ============================================================

import os
import re
import json
import glob

from collections import defaultdict


print("=" * 60)
print("EMOBERTA-BASE + RAW — EXACT D1 ANCHOR DETECTION")
print("=" * 60)


# ============================================================
# D1 TIMELINE DIRECTORY
# ============================================================

DATASET_ROOT = (
    "/kaggle/input/datasets/"
    "vedantnawghare35/"
    "depression-detection-mddl-final-pbl/"
    "Dataset_MDDL"
)

D1_TIMELINE_DIR = os.path.join(
    DATASET_ROOT,
    "labeled",
    "positive",
    "data",
    "timeline"
)

assert os.path.exists(
    D1_TIMELINE_DIR
), (
    f"D1 timeline directory not found: "
    f"{D1_TIMELINE_DIR}"
)


# ============================================================
# FIND D1 TIMELINE FILES
# ============================================================

d1_timeline_files = glob.glob(
    os.path.join(
        D1_TIMELINE_DIR,
        "*.json"
    )
)

print()
print(
    "D1 timeline files found:",
    len(d1_timeline_files)
)

assert len(
    d1_timeline_files
) == 2626


# ============================================================
# EXACT ANCHOR PATTERNS
# SAME AS ALL PREVIOUS EXPERIMENTS
# ============================================================

REFINED_ANCHOR_PATTERNS = [

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depression\b",

    r"\bi(?:'m| am)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+(?:was|were)\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi(?:'ve| have)\s+been\s+diagnosed\s+with\s+(?:major\s+)?depressive\s+disorder\b",

    r"\bi\s+have\s+(?:major\s+)?depression\b",

    r"\bi\s+suffer\s+from\s+(?:major\s+)?depression\b",

    r"\bi\s+(?:was|am|'m)\s+diagnosed\b.{0,80}\bdepress",

    (
        r"\b(?:my|a)\s+(?:doctor|psychiatrist)"
        r"\s+(?:has\s+)?diagnosed\s+me\b"
        r".{0,80}\bdepress"
    )
]


compiled_anchor_patterns = [

    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )

    for pattern
    in REFINED_ANCHOR_PATTERNS
]


# ============================================================
# EXCLUSION PATTERNS
# ============================================================

EXCLUSION_PATTERNS = [

    r"\bmy\s+friend\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+sister\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+brother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+mother\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+father\s+(?:was|is|has\s+been)\s+diagnosed",

    r"\bmy\s+answer\s+to\s*:",

    r"\banswer\s+to\s*:",

    r"\bquestion\s*:",

    r"\bq\s*:"
]


compiled_exclusion_patterns = [

    re.compile(
        pattern,
        flags=re.IGNORECASE | re.DOTALL
    )

    for pattern
    in EXCLUSION_PATTERNS
]


# ============================================================
# EXACT ANCHOR DETECTOR
# ============================================================

def is_refined_anchor_tweet(
    text
):

    if text is None:
        return False

    text = str(
        text
    ).strip()

    if not text:
        return False

    # --------------------------------------------------------
    # Exclusions first
    # --------------------------------------------------------

    for pattern in compiled_exclusion_patterns:

        if pattern.search(text):
            return False

    # --------------------------------------------------------
    # Anchor patterns
    # --------------------------------------------------------

    for pattern in compiled_anchor_patterns:

        if pattern.search(text):
            return True

    return False


# ============================================================
# POSITIVE USERS BY SPLIT
# ============================================================

train_positive_users = set(
    train_df.loc[
        train_df["label"] == 1,
        "username"
    ]
    .astype(str)
)

val_positive_users = set(
    val_df.loc[
        val_df["label"] == 1,
        "username"
    ]
    .astype(str)
)

test_positive_users = set(
    test_df.loc[
        test_df["label"] == 1,
        "username"
    ]
    .astype(str)
)


all_positive_users = (
    train_positive_users
    |
    val_positive_users
    |
    test_positive_users
)


print()
print("=" * 60)
print("POSITIVE USERS BY SPLIT")
print("=" * 60)

print(
    "TRAIN:",
    len(train_positive_users)
)

print(
    "VALIDATION:",
    len(val_positive_users)
)

print(
    "TEST:",
    len(test_positive_users)
)


assert len(
    train_positive_users
) == 2013

assert len(
    val_positive_users
) == 252

assert len(
    test_positive_users
) == 251

assert len(
    all_positive_users
) == 2516


# ============================================================
# INDEX TIMELINE FILES
# ============================================================

timeline_files_by_user = (
    defaultdict(list)
)


for timeline_file in d1_timeline_files:

    filename = os.path.basename(
        timeline_file
    )

    username = filename.split(
        "created_at"
    )[0]

    timeline_files_by_user[
        username
    ].append(
        timeline_file
    )


assert len(
    timeline_files_by_user
) == 2516


# ============================================================
# SCAN ALL POSITIVE D1 TIMELINES
# ============================================================

anchor_records = []

users_processed = set()

tweets_scanned = 0


for username in sorted(
    all_positive_users
):

    if username not in timeline_files_by_user:

        raise KeyError(
            f"Missing D1 timeline for user: "
            f"{username}"
        )

    users_processed.add(
        username
    )


    for timeline_file in (
        timeline_files_by_user[
            username
        ]
    ):

        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:

                line = line.strip()

                if not line:
                    continue

                record = json.loads(
                    line
                )

                tweets_scanned += 1

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get(
                            "id",
                            ""
                        )
                    )
                )

                tweet_text = str(
                    record.get(
                        "text",
                        ""
                    )
                )


                if (
                    tweet_id
                    and
                    is_refined_anchor_tweet(
                        tweet_text
                    )
                ):

                    anchor_records.append(
                        {
                            "username":
                                username,

                            "tweet_id":
                                tweet_id,

                            "created_at":
                                record.get(
                                    "created_at"
                                ),

                            "text":
                                tweet_text
                        }
                    )


# ============================================================
# UNIQUE ANCHOR PAIRS
# ============================================================

anchor_pairs = {

    (
        record["username"],
        record["tweet_id"]
    )

    for record
    in anchor_records
}


# ============================================================
# SPLIT-WISE UNIQUE ANCHORS
# ============================================================

train_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in train_positive_users
}

val_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in val_positive_users
}

test_anchor_pairs = {
    pair
    for pair in anchor_pairs
    if pair[0] in test_positive_users
}


# ============================================================
# SUMMARY
# ============================================================

print()
print("=" * 60)
print("EXACT ANCHOR SUMMARY")
print("=" * 60)

print(
    "Positive users processed:",
    len(users_processed)
)

print(
    "Tweets scanned:",
    tweets_scanned
)

print(
    "Raw anchor records:",
    len(anchor_records)
)

print(
    "Unique anchor pairs:",
    len(anchor_pairs)
)

print()
print(
    "TRAIN unique anchors:",
    len(train_anchor_pairs)
)

print(
    "VAL unique anchors:",
    len(val_anchor_pairs)
)

print(
    "TEST unique anchors:",
    len(test_anchor_pairs)
)


# ============================================================
# EXACT CONTROL ASSERTIONS
# ============================================================

assert len(
    users_processed
) == 2516

assert len(
    anchor_records
) == 1920

assert len(
    anchor_pairs
) == 1847

assert len(
    train_anchor_pairs
) == 1476

assert len(
    val_anchor_pairs
) == 190

assert len(
    test_anchor_pairs
) == 181


# ============================================================
# SAVE ANCHOR RECORDS
# ============================================================

emoberta_raw_anchor_df = pd.DataFrame(
    anchor_records
)

emoberta_raw_anchor_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "exact_anchor_records.csv"
    ),
    index=False
)


# ============================================================
# SAVE UNIQUE PAIRS
# ============================================================

pd.DataFrame(
    list(anchor_pairs),
    columns=[
        "username",
        "tweet_id"
    ]
).to_csv(
    os.path.join(
        RESULT_DIR,
        "unique_anchor_pairs.csv"
    ),
    index=False
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 3 COMPLETE")
print("=" * 60)

print(
    "✓ Exact anchor detector reproduced"
)

print(
    "✓ 1,847 unique anchors confirmed"
)

print(
    "✓ 1,920 anchor records confirmed"
)

print(
    "✓ TRAIN = 1,476 unique anchors"
)

print(
    "✓ VAL = 190 unique anchors"
)

print(
    "✓ TEST = 181 unique anchors"
)

print(
    "✓ No D2 users included"
)

print(
    "✓ Anchor records saved"
)

print(
    "✓ No timelines modified yet"
)
```

```python
# ============================================================
# CELL 4 — EMOBERTA-BASE + RAW
# EXACT D1 ANCHOR REMOVAL
# ============================================================

print("=" * 60)
print("EMOBERTA-BASE + RAW — EXACT D1 ANCHOR REMOVAL")
print("=" * 60)


# ============================================================
# EXPECTED COUNTS
# ============================================================

EXPECTED_TRAIN_REMOVED = 1535
EXPECTED_VAL_REMOVED = 199
EXPECTED_TEST_REMOVED = 186
EXPECTED_TOTAL_REMOVED = 1920


# ============================================================
# PRE-REMOVAL VERIFICATION
# ============================================================

print()
print("=" * 60)
print("PRE-REMOVAL ANCHOR VERIFICATION")
print("=" * 60)

duplicate_anchor_occurrences = (
    len(anchor_records)
    - len(anchor_pairs)
)

print(
    "Raw anchor records:",
    len(anchor_records)
)

print(
    "Unique anchor pairs:",
    len(anchor_pairs)
)

print(
    "Duplicate anchor occurrences:",
    duplicate_anchor_occurrences
)

assert len(anchor_records) == 1920
assert len(anchor_pairs) == 1847
assert duplicate_anchor_occurrences == 73

print("✓ 1,847 unique anchors verified")
print("✓ 73 duplicate occurrences verified")
print("✓ 1,920 total anchor records verified")


# ============================================================
# SPLIT-LEVEL RECORD COUNTS
# ============================================================

train_anchor_records = [
    r
    for r in anchor_records
    if r["username"] in train_positive_users
]

val_anchor_records = [
    r
    for r in anchor_records
    if r["username"] in val_positive_users
]

test_anchor_records = [
    r
    for r in anchor_records
    if r["username"] in test_positive_users
]


print()
print("=" * 60)
print("SPLIT-LEVEL ANCHOR RECORD COUNTS")
print("=" * 60)

print(
    "TRAIN:",
    len(train_anchor_records),
    "| expected:",
    EXPECTED_TRAIN_REMOVED
)

print(
    "VAL:",
    len(val_anchor_records),
    "| expected:",
    EXPECTED_VAL_REMOVED
)

print(
    "TEST:",
    len(test_anchor_records),
    "| expected:",
    EXPECTED_TEST_REMOVED
)

assert len(train_anchor_records) == EXPECTED_TRAIN_REMOVED
assert len(val_anchor_records) == EXPECTED_VAL_REMOVED
assert len(test_anchor_records) == EXPECTED_TEST_REMOVED


# ============================================================
# EXACT ANCHOR SET
# ============================================================

EXACT_ANCHOR_SET = set(
    anchor_pairs
)


# ============================================================
# BUILD ANCHOR-REMOVED TEXT
# ============================================================

def build_anchor_removed_text(
    username
):

    kept_texts = []
    removed_count = 0

    for timeline_file in (
        timeline_files_by_user[
            username
        ]
    ):

        with open(
            timeline_file,
            "r",
            encoding="utf-8"
        ) as f:

            for line in f:

                line = line.strip()

                if not line:
                    continue

                record = json.loads(line)

                tweet_id = str(
                    record.get(
                        "id_str",
                        record.get(
                            "id",
                            ""
                        )
                    )
                )

                pair = (
                    str(username),
                    tweet_id
                )

                if pair in EXACT_ANCHOR_SET:

                    removed_count += 1
                    continue

                tweet_text = str(
                    record.get(
                        "text",
                        ""
                    )
                ).strip()

                if tweet_text:

                    kept_texts.append(
                        tweet_text
                    )

    return (
        " ".join(kept_texts),
        removed_count
    )


# ============================================================
# COPY ORIGINAL DATAFRAMES
# ============================================================

train_emoberta_raw_df = (
    train_df.copy(
        deep=True
    )
)

val_emoberta_raw_df = (
    val_df.copy(
        deep=True
    )
)

test_emoberta_raw_df = (
    test_df.copy(
        deep=True
    )
)


# ============================================================
# BUILD POSITIVE USER TEXT
# ============================================================

positive_anchor_removed_text = {}

train_removed = 0
val_removed = 0
test_removed = 0


for username in sorted(
    all_positive_users
):

    (
        anchor_removed_text,
        removed_count
    ) = build_anchor_removed_text(
        username
    )

    positive_anchor_removed_text[
        username
    ] = anchor_removed_text


    if username in train_positive_users:

        train_removed += removed_count

    elif username in val_positive_users:

        val_removed += removed_count

    elif username in test_positive_users:

        test_removed += removed_count


# ============================================================
# VERIFY REMOVAL COUNTS
# ============================================================

print()
print("=" * 60)
print("FINAL ANCHOR REMOVAL COUNTS")
print("=" * 60)

print(
    "TRAIN removed:",
    train_removed
)

print(
    "VAL removed:",
    val_removed
)

print(
    "TEST removed:",
    test_removed
)

print(
    "TOTAL removed:",
    train_removed
    + val_removed
    + test_removed
)


assert train_removed == EXPECTED_TRAIN_REMOVED
assert val_removed == EXPECTED_VAL_REMOVED
assert test_removed == EXPECTED_TEST_REMOVED

assert (
    train_removed
    + val_removed
    + test_removed
    ==
    EXPECTED_TOTAL_REMOVED
)

print(
    "✓ Record-level removal exactly matches "
    "previous experiments"
)


# ============================================================
# ASSIGN ANCHOR-REMOVED TEXT
# ============================================================

for df in [

    train_emoberta_raw_df,
    val_emoberta_raw_df,
    test_emoberta_raw_df

]:

    positive_mask = (
        df["label"] == 1
    )

    df.loc[
        positive_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            positive_mask,
            "username"
        ]
        .astype(str)
        .map(
            positive_anchor_removed_text
        )
        .fillna("")
        .values
    )


    # D2 / negative users remain unchanged

    negative_mask = (
        df["label"] == 0
    )

    df.loc[
        negative_mask,
        "anchor_removed_text"
    ] = (
        df.loc[
            negative_mask,
            "timeline_text"
        ]
        .fillna("")
        .astype(str)
        .values
    )


# ============================================================
# D2 PRESERVATION CHECK
# ============================================================

print()
print("=" * 60)
print("D2 PRESERVATION CHECK")
print("=" * 60)

for original_df, result_df, split_name in [

    (
        train_df,
        train_emoberta_raw_df,
        "TRAIN"
    ),

    (
        val_df,
        val_emoberta_raw_df,
        "VALIDATION"
    ),

    (
        test_df,
        test_emoberta_raw_df,
        "TEST"
    )

]:

    original_d2 = (
        original_df.loc[
            original_df["label"] == 0,
            [
                "username",
                "timeline_text"
            ]
        ]
        .set_index(
            "username"
        )
        .sort_index()
    )

    result_d2 = (
        result_df.loc[
            result_df["label"] == 0,
            [
                "username",
                "anchor_removed_text"
            ]
        ]
        .rename(
            columns={
                "anchor_removed_text":
                    "timeline_text"
            }
        )
        .set_index(
            "username"
        )
        .sort_index()
    )

    assert original_d2.equals(
        result_d2
    )

    print(
        f"✓ {split_name} D2 unchanged"
    )


# ============================================================
# USER + LABEL PRESERVATION
# ============================================================

print()
print("=" * 60)
print("USER + LABEL PRESERVATION")
print("=" * 60)

for original_df, result_df, split_name in [

    (
        train_df,
        train_emoberta_raw_df,
        "TRAIN"
    ),

    (
        val_df,
        val_emoberta_raw_df,
        "VALIDATION"
    ),

    (
        test_df,
        test_emoberta_raw_df,
        "TEST"
    )

]:

    assert (
        original_df["user_id"].tolist()
        ==
        result_df["user_id"].tolist()
    )

    assert (
        original_df["username"].tolist()
        ==
        result_df["username"].tolist()
    )

    assert (
        original_df["label"].tolist()
        ==
        result_df["label"].tolist()
    )

    print(
        f"✓ {split_name}: users and labels unchanged"
    )


# ============================================================
# EMPTY TEXT CHECK
# ============================================================

print()
print("=" * 60)
print("ANCHOR-REMOVED TEXT INTEGRITY")
print("=" * 60)

for split_name, df in [

    (
        "TRAIN",
        train_emoberta_raw_df
    ),

    (
        "VALIDATION",
        val_emoberta_raw_df
    ),

    (
        "TEST",
        test_emoberta_raw_df
    )

]:

    empty_count = int(
        (
            df[
                "anchor_removed_text"
            ]
            .fillna("")
            .astype(str)
            .str.strip()
            ==
            ""
        ).sum()
    )

    print(
        f"{split_name}: "
        f"empty anchor-removed texts = "
        f"{empty_count}"
    )


# ============================================================
# PREVIEW
# ============================================================

print()
print("=" * 60)
print("ANCHOR-REMOVED TEXT PREVIEW")
print("=" * 60)

for _, row in (
    train_emoberta_raw_df
    .head(3)
    .iterrows()
):

    print()
    print(
        "User:",
        row["username"]
    )

    print(
        "Label:",
        row["label"]
    )

    print(
        row[
            "anchor_removed_text"
        ][:1200]
    )


# ============================================================
# SAVE
# ============================================================

train_emoberta_raw_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "train_anchor_removed.pkl"
    )
)

val_emoberta_raw_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "val_anchor_removed.pkl"
    )
)

test_emoberta_raw_df.to_pickle(
    os.path.join(
        RESULT_DIR,
        "test_anchor_removed.pkl"
    )
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 4 COMPLETE")
print("=" * 60)

print("✓ 1,847 unique D1 anchors removed")
print("✓ 73 duplicate occurrences handled")
print("✓ 1,920 total anchor records removed")
print("✓ TRAIN = 1,535 removed")
print("✓ VAL = 199 removed")
print("✓ TEST = 186 removed")
print("✓ D2 completely preserved")
print("✓ User IDs preserved")
print("✓ Labels preserved")
print("✓ RAW text retained after anchor removal")
print("✓ Anchor-removed source saved")
```

```python
# ============================================================
# CELL 6 — EMOBERTA-BASE + RAW
# TOKEN LENGTH ANALYSIS
# ============================================================

from transformers import AutoTokenizer
import numpy as np
import pandas as pd

print("=" * 60)
print("EMOBERTA-BASE + RAW — TOKEN LENGTH ANALYSIS")
print("=" * 60)


# ============================================================
# CONFIGURATION
# ============================================================

print()
print("Tokenizer:", MODEL_NAME)
print("Experimental context length:", MAX_LENGTH)


# ============================================================
# HUGGING FACE AUTHENTICATION
# ============================================================

from kaggle_secrets import UserSecretsClient

user_secrets = UserSecretsClient()

hf_token = user_secrets.get_secret(
    "HF_TOKEN"
)

print()
print("✓ HF_TOKEN retrieved")


# ============================================================
# LOAD TOKENIZER
# ============================================================

emoberta_raw_tokenizer = (
    AutoTokenizer.from_pretrained(
        MODEL_NAME,
        token=hf_token
    )
)

print()
print(
    "Tokenizer max length:",
    emoberta_raw_tokenizer.model_max_length
)

print(
    "✓ EmoBERTa tokenizer loaded"
)


# ============================================================
# TOKEN LENGTH FUNCTION
# ============================================================

def calculate_token_lengths(
    texts,
    split_name
):

    print()
    print(
        f"Processing {split_name}..."
    )

    lengths = []

    for idx, text in enumerate(
        texts
    ):

        encoded = (
            emoberta_raw_tokenizer(
                str(text),
                add_special_tokens=True,
                truncation=False,
                padding=False
            )
        )

        lengths.append(
            len(
                encoded["input_ids"]
            )
        )

        if (
            (idx + 1) % 1000 == 0
            or
            (idx + 1) == len(texts)
        ):

            print(
                f"{split_name}: "
                f"{idx + 1}/{len(texts)}"
            )

    return np.asarray(
        lengths,
        dtype=np.int64
    )


# ============================================================
# TRAIN
# ============================================================

emoberta_raw_train_token_lengths = (
    calculate_token_lengths(
        train_emoberta_raw_df[
            "model_text"
        ]
        .fillna("")
        .astype(str)
        .tolist(),

        "TRAIN"
    )
)


# ============================================================
# VALIDATION
# ============================================================

emoberta_raw_val_token_lengths = (
    calculate_token_lengths(
        val_emoberta_raw_df[
            "model_text"
        ]
        .fillna("")
        .astype(str)
        .tolist(),

        "VALIDATION"
    )
)


# ============================================================
# TEST
# ============================================================

emoberta_raw_test_token_lengths = (
    calculate_token_lengths(
        test_emoberta_raw_df[
            "model_text"
        ]
        .fillna("")
        .astype(str)
        .tolist(),

        "TEST"
    )
)


# ============================================================
# STATISTICS
# ============================================================

def print_token_statistics(
    lengths,
    split_name
):

    print()
    print("=" * 60)
    print(
        f"{split_name} TOKEN STATISTICS"
    )
    print("=" * 60)

    print(
        "Minimum:",
        int(lengths.min())
    )

    print(
        "Maximum:",
        int(lengths.max())
    )

    print(
        "Mean:",
        round(
            float(lengths.mean()),
            2
        )
    )

    print(
        "Median:",
        round(
            float(np.median(lengths)),
            2
        )
    )

    print(
        "25th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    25
                )
            ),
            2
        )
    )

    print(
        "75th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    75
                )
            ),
            2
        )
    )

    print(
        "90th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    90
                )
            ),
            2
        )
    )

    print(
        "95th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    95
                )
            ),
            2
        )
    )

    print(
        "99th percentile:",
        round(
            float(
                np.percentile(
                    lengths,
                    99
                )
            ),
            2
        )
    )


    # --------------------------------------------------------
    # 512 COVERAGE
    # --------------------------------------------------------

    within_512 = int(
        np.sum(
            lengths <= 512
        )
    )

    above_512 = int(
        np.sum(
            lengths > 512
        )
    )

    print()
    print(
        "512-TOKEN COVERAGE"
    )

    print(
        "Users <= 512:",
        within_512
    )

    print(
        "Users > 512:",
        above_512
    )

    print(
        "Percentage <= 512:",
        round(
            100 * within_512 / len(lengths),
            2
        ),
        "%"
    )

    print(
        "Percentage > 512:",
        round(
            100 * above_512 / len(lengths),
            2
        ),
        "%"
    )


    # --------------------------------------------------------
    # 4096 REFERENCE
    # --------------------------------------------------------

    within_4096 = int(
        np.sum(
            lengths <= 4096
        )
    )

    above_4096 = int(
        np.sum(
            lengths > 4096
        )
    )

    print()
    print(
        "4096-TOKEN REFERENCE"
    )

    print(
        "Users <= 4096:",
        within_4096
    )

    print(
        "Users > 4096:",
        above_4096
    )

    print(
        "Percentage <= 4096:",
        round(
            100 * within_4096 / len(lengths),
            2
        ),
        "%"
    )

    print(
        "Percentage > 4096:",
        round(
            100 * above_4096 / len(lengths),
            2
        ),
        "%"
    )


# ============================================================
# PRINT STATISTICS
# ============================================================

print_token_statistics(
    emoberta_raw_train_token_lengths,
    "TRAIN"
)

print_token_statistics(
    emoberta_raw_val_token_lengths,
    "VALIDATION"
)

print_token_statistics(
    emoberta_raw_test_token_lengths,
    "TEST"
)


# ============================================================
# ESTIMATED CHUNKS
# ============================================================

print()
print("=" * 60)
print("ESTIMATED CONTEXT CHUNKS")
print("=" * 60)

for context_length in [
    512,
    4096
]:

    print()
    print(
        "Context length:",
        context_length
    )

    train_chunks = np.ceil(
        emoberta_raw_train_token_lengths
        / context_length
    )

    val_chunks = np.ceil(
        emoberta_raw_val_token_lengths
        / context_length
    )

    test_chunks = np.ceil(
        emoberta_raw_test_token_lengths
        / context_length
    )

    print(
        "TRAIN mean chunks:",
        round(
            float(train_chunks.mean()),
            2
        )
    )

    print(
        "TRAIN maximum chunks:",
        int(train_chunks.max())
    )

    print(
        "VALIDATION mean chunks:",
        round(
            float(val_chunks.mean()),
            2
        )
    )

    print(
        "VALIDATION maximum chunks:",
        int(val_chunks.max())
    )

    print(
        "TEST mean chunks:",
        round(
            float(test_chunks.mean()),
            2
        )
    )

    print(
        "TEST maximum chunks:",
        int(test_chunks.max())
    )


# ============================================================
# SUMMARY TABLE
# ============================================================

emoberta_raw_token_summary = pd.DataFrame([

    {
        "Split": "Train",
        "Users": 6135,
        "Mean":
            np.mean(
                emoberta_raw_train_token_lengths
            ),
        "Median":
            np.median(
                emoberta_raw_train_token_lengths
            ),
        "<=512 (%)":
            100 * np.mean(
                emoberta_raw_train_token_lengths
                <= 512
            ),
        "<=4096 (%)":
            100 * np.mean(
                emoberta_raw_train_token_lengths
                <= 4096
            )
    },

    {
        "Split": "Validation",
        "Users": 767,
        "Mean":
            np.mean(
                emoberta_raw_val_token_lengths
            ),
        "Median":
            np.median(
                emoberta_raw_val_token_lengths
            ),
        "<=512 (%)":
            100 * np.mean(
                emoberta_raw_val_token_lengths
                <= 512
            ),
        "<=4096 (%)":
            100 * np.mean(
                emoberta_raw_val_token_lengths
                <= 4096
            )
    },

    {
        "Split": "Test",
        "Users": 767,
        "Mean":
            np.mean(
                emoberta_raw_test_token_lengths
            ),
        "Median":
            np.median(
                emoberta_raw_test_token_lengths
            ),
        "<=512 (%)":
            100 * np.mean(
                emoberta_raw_test_token_lengths
                <= 512
            ),
        "<=4096 (%)":
            100 * np.mean(
                emoberta_raw_test_token_lengths
                <= 4096
            )
    }
])


print()
print("=" * 60)
print("TOKEN-LENGTH SUMMARY")
print("=" * 60)

print(
    emoberta_raw_token_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.2f}"
    )
)


# ============================================================
# SAVE TOKEN LENGTHS
# ============================================================

pd.DataFrame({
    "token_length":
        emoberta_raw_train_token_lengths
}).to_csv(
    os.path.join(
        RESULT_DIR,
        "train_token_lengths.csv"
    ),
    index=False
)

pd.DataFrame({
    "token_length":
        emoberta_raw_val_token_lengths
}).to_csv(
    os.path.join(
        RESULT_DIR,
        "val_token_lengths.csv"
    ),
    index=False
)

pd.DataFrame({
    "token_length":
        emoberta_raw_test_token_lengths
}).to_csv(
    os.path.join(
        RESULT_DIR,
        "test_token_lengths.csv"
    ),
    index=False
)

emoberta_raw_token_summary.to_csv(
    os.path.join(
        RESULT_DIR,
        "token_length_summary.csv"
    ),
    index=False
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 6 COMPLETE")
print("=" * 60)

print(
    "✓ Authenticated EmoBERTa tokenizer loaded"
)

print(
    "✓ RAW text analyzed"
)

print(
    "✓ No truncation used during analysis"
)

print(
    "✓ 512-token coverage measured"
)

print(
    "✓ 4096-token reference measured"
)

print(
    "✓ Context chunk estimates calculated"
)

print(
    "✓ Token-length files saved"
)
```

```python
# ============================================================
# CELL 11 — EMOBERTA-BASE + RAW
# BUILDING FROZEN EMOBERTA + METADATA MODEL
# ============================================================

import os
import json
import torch
import torch.nn as nn

from transformers import AutoModel


print("=" * 60)
print("EMOBERTA-BASE + RAW — BUILDING FROZEN EMOBERTA + METADATA MODEL")
print("=" * 60)


# ============================================================
# MODEL CONFIGURATION
# ============================================================

TEXT_HIDDEN_SIZE = 768

ATTENTION_PROJECTION = 256

METADATA_INPUT = 53
METADATA_HIDDEN_1 = 128
METADATA_HIDDEN_2 = 64

DROPOUT = 0.3

NUM_CLASSES = 2

FREEZE_TRANSFORMER = True


# ============================================================
# MODEL DEFINITION
# ============================================================

class FrozenEmoBERTaMetadataClassifier(
    nn.Module
):

    def __init__(
        self,
        model_name,
        metadata_dim=53,
        attention_dim=256,
        dropout=0.3
    ):

        super().__init__()


        # ----------------------------------------------------
        # EmoBERTa backbone
        # ----------------------------------------------------

        self.emoberta = AutoModel.from_pretrained(
            model_name,
            token=hf_token
        )


        hidden_size = (
            self.emoberta.config.hidden_size
        )

        assert hidden_size == 768


        # ----------------------------------------------------
        # FREEZE TRANSFORMER
        # ----------------------------------------------------

        for parameter in (
            self.emoberta.parameters()
        ):

            parameter.requires_grad = False


        # Keep backbone in evaluation mode
        # to disable dropout inside transformer.

        self.emoberta.eval()


        # ----------------------------------------------------
        # Attention pooling
        # ----------------------------------------------------

        self.attention_projection = nn.Linear(
            hidden_size,
            attention_dim
        )

        self.attention_score = nn.Linear(
            attention_dim,
            1
        )


        # ----------------------------------------------------
        # Metadata branch
        # 53 -> 128 -> 64
        # ----------------------------------------------------

        self.metadata_branch = nn.Sequential(

            nn.Linear(
                metadata_dim,
                128
            ),

            nn.ReLU(),

            nn.Dropout(
                dropout
            ),

            nn.Linear(
                128,
                64
            ),

            nn.ReLU()
        )


        # ----------------------------------------------------
        # Single metadata gate
        # g = sigmoid(beta)
        # ----------------------------------------------------

        self.metadata_beta = nn.Parameter(
            torch.tensor(
                0.0,
                dtype=torch.float32
            )
        )


        # ----------------------------------------------------
        # Classifier
        # 832 -> 512 -> 256 -> 2
        # ----------------------------------------------------

        self.classifier = nn.Sequential(

            nn.Linear(
                832,
                512
            ),

            nn.ReLU(),

            nn.Dropout(
                dropout
            ),

            nn.Linear(
                512,
                256
            ),

            nn.ReLU(),

            nn.Dropout(
                dropout
            ),

            nn.Linear(
                256,
                NUM_CLASSES
            )
        )


    # ========================================================
    # FORWARD
    # ========================================================

    def forward(
        self,
        input_ids,
        attention_mask,
        metadata
    ):

        # ----------------------------------------------------
        # Frozen EmoBERTa
        # ----------------------------------------------------

        with torch.no_grad():

            outputs = self.emoberta(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

        hidden_states = (
            outputs.last_hidden_state
        )


        # ----------------------------------------------------
        # Attention pooling
        # ----------------------------------------------------

        projected = (
            self.attention_projection(
                hidden_states
            )
        )

        attention_scores = (
            self.attention_score(
                torch.tanh(projected)
            )
            .squeeze(-1)
        )


        # ----------------------------------------------------
        # Mask padding
        # ----------------------------------------------------

        attention_scores = (
            attention_scores.masked_fill(
                attention_mask == 0,
                torch.finfo(
                    attention_scores.dtype
                ).min
            )
        )


        # ----------------------------------------------------
        # Softmax
        # ----------------------------------------------------

        attention_weights = (
            torch.softmax(
                attention_scores,
                dim=1
            )
        )


        # ----------------------------------------------------
        # Text representation
        # ----------------------------------------------------

        text_representation = (
            hidden_states
            *
            attention_weights.unsqueeze(-1)
        ).sum(
            dim=1
        )


        # ----------------------------------------------------
        # Metadata representation
        # ----------------------------------------------------

        metadata_representation = (
            self.metadata_branch(
                metadata
            )
        )


        # ----------------------------------------------------
        # Metadata gate
        # ----------------------------------------------------

        metadata_gate = torch.sigmoid(
            self.metadata_beta
        )


        # ----------------------------------------------------
        # Fusion
        # ----------------------------------------------------

        fused_representation = torch.cat(
            [
                text_representation,

                metadata_gate
                *
                metadata_representation
            ],
            dim=1
        )


        # ----------------------------------------------------
        # Classification
        # ----------------------------------------------------

        logits = self.classifier(
            fused_representation
        )


        return {

            "logits":
                logits,

            "text_representation":
                text_representation,

            "metadata_representation":
                metadata_representation,

            "attention_weights":
                attention_weights,

            "metadata_gate":
                metadata_gate
        }


# ============================================================
# BUILD MODEL
# ============================================================

model_emoberta_raw = (
    FrozenEmoBERTaMetadataClassifier(
        model_name=MODEL_NAME,
        metadata_dim=METADATA_INPUT,
        attention_dim=ATTENTION_PROJECTION,
        dropout=DROPOUT
    )
    .to(device)
)


# ============================================================
# VERIFY FROZEN BACKBONE
# ============================================================

total_parameters = sum(
    p.numel()
    for p in model_emoberta_raw.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in model_emoberta_raw.parameters()
    if p.requires_grad
)

frozen_parameters = sum(
    p.numel()
    for p in model_emoberta_raw.parameters()
    if not p.requires_grad
)


print()
print("=" * 60)
print("PARAMETER COUNT")
print("=" * 60)

print(
    "Total parameters:",
    total_parameters
)

print(
    "Trainable parameters:",
    trainable_parameters
)

print(
    "Frozen parameters:",
    frozen_parameters
)

print(
    "Total parameters (M):",
    round(
        total_parameters / 1e6,
        3
    )
)

print(
    "Trainable parameters (M):",
    round(
        trainable_parameters / 1e6,
        3
    )
)

print(
    "Frozen parameters (M):",
    round(
        frozen_parameters / 1e6,
        3
    )
)


# ============================================================
# FREEZE ASSERTIONS
# ============================================================

emoberta_parameters = list(
    model_emoberta_raw
    .emoberta
    .parameters()
)

assert len(
    emoberta_parameters
) > 0

assert all(
    not p.requires_grad
    for p in emoberta_parameters
)

print()
print(
    "✓ All EmoBERTa backbone parameters frozen"
)


# ============================================================
# TRAINABLE COMPONENTS
# ============================================================

trainable_component_names = [
    name
    for name, parameter
    in model_emoberta_raw.named_parameters()
    if parameter.requires_grad
]

print()
print("=" * 60)
print("TRAINABLE COMPONENTS")
print("=" * 60)

for name in trainable_component_names:

    print(
        "✓",
        name
    )


# ============================================================
# ARCHITECTURE CONFIGURATION
# ============================================================

initial_gate = float(
    torch.sigmoid(
        model_emoberta_raw.metadata_beta
    )
    .detach()
    .cpu()
)


print()
print("=" * 60)
print("ARCHITECTURE CONFIGURATION")
print("=" * 60)

print(
    "Transformer:",
    MODEL_NAME
)

print(
    "Base architecture:",
    "RoBERTa-base"
)

print(
    "Text hidden size:",
    TEXT_HIDDEN_SIZE
)

print(
    "Attention projection:",
    ATTENTION_PROJECTION
)

print(
    "Metadata:",
    "53 -> 128 -> 64"
)

print(
    "Gate:",
    "g = sigmoid(beta)"
)

print(
    "Initial beta:",
    0.0
)

print(
    "Initial gate:",
    initial_gate
)

print(
    "Fusion:",
    "768 + 64 = 832"
)

print(
    "Classifier:",
    "832 -> 512 -> 256 -> 2"
)

print(
    "Dropout:",
    DROPOUT
)

print(
    "Transformer frozen:",
    FREEZE_TRANSFORMER
)


# ============================================================
# FORWARD PASS SANITY CHECK
# ============================================================

print()
print("=" * 60)
print("FORWARD PASS SANITY CHECK")
print("=" * 60)


model_emoberta_raw.eval()

with torch.no_grad():

    sample_input_ids = (
        batch["input_ids"]
        .to(device)
    )

    sample_attention_mask = (
        batch["attention_mask"]
        .to(device)
    )

    sample_metadata = (
        batch["metadata"]
        .to(device)
    )

    outputs = model_emoberta_raw(
        input_ids=sample_input_ids,
        attention_mask=sample_attention_mask,
        metadata=sample_metadata
    )


# ============================================================
# OUTPUT SHAPES
# ============================================================

print(
    "Input IDs:",
    tuple(
        sample_input_ids.shape
    )
)

print(
    "Attention mask:",
    tuple(
        sample_attention_mask.shape
    )
)

print(
    "Metadata:",
    tuple(
        sample_metadata.shape
    )
)

print(
    "Logits:",
    tuple(
        outputs["logits"].shape
    )
)

print(
    "Text representation:",
    tuple(
        outputs[
            "text_representation"
        ].shape
    )
)

print(
    "Metadata representation:",
    tuple(
        outputs[
            "metadata_representation"
        ].shape
    )
)

print(
    "Attention weights:",
    tuple(
        outputs[
            "attention_weights"
        ].shape
    )
)

print(
    "Metadata gate:",
    float(
        outputs[
            "metadata_gate"
        ]
        .detach()
        .cpu()
    )
)


# ============================================================
# PADDING ATTENTION CHECK
# ============================================================

padding_mask = (
    sample_attention_mask == 0
)

padding_attention_values = (
    outputs[
        "attention_weights"
    ][padding_mask]
)


if padding_attention_values.numel() > 0:

    maximum_padding_attention = float(
        padding_attention_values
        .max()
        .detach()
        .cpu()
    )

else:

    maximum_padding_attention = 0.0


print(
    "Maximum attention on padding:",
    maximum_padding_attention
)


# ============================================================
# ASSERTIONS
# ============================================================

assert outputs[
    "logits"
].shape == (
    16,
    2
)

assert outputs[
    "text_representation"
].shape == (
    16,
    768
)

assert outputs[
    "metadata_representation"
].shape == (
    16,
    64
)

assert outputs[
    "attention_weights"
].shape == (
    16,
    512
)

assert abs(
    maximum_padding_attention
) < 1e-6


# ============================================================
# VERIFY BACKBONE REMAINS FROZEN
# ============================================================

assert all(
    not p.requires_grad
    for p in model_emoberta_raw
    .emoberta
    .parameters()
)

print()
print(
    "✓ Transformer freeze verified"
)


# ============================================================
# SAVE MODEL CONFIGURATION
# ============================================================

model_config = {

    "model_name":
        MODEL_NAME,

    "base_architecture":
        "RoBERTa-base",

    "text_hidden_size":
        TEXT_HIDDEN_SIZE,

    "attention_projection":
        ATTENTION_PROJECTION,

    "metadata_input":
        METADATA_INPUT,

    "metadata_hidden_1":
        METADATA_HIDDEN_1,

    "metadata_hidden_2":
        METADATA_HIDDEN_2,

    "initial_beta":
        0.0,

    "initial_gate":
        initial_gate,

    "fusion_dimension":
        832,

    "classifier":
        "832 -> 512 -> 256 -> 2",

    "dropout":
        DROPOUT,

    "transformer_frozen":
        FREEZE_TRANSFORMER,

    "total_parameters":
        total_parameters,

    "trainable_parameters":
        trainable_parameters,

    "frozen_parameters":
        frozen_parameters
}


with open(
    os.path.join(
        RESULT_DIR,
        "model_config.json"
    ),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        model_config,
        f,
        indent=2
    )


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 11 COMPLETE")
print("=" * 60)

print("✓ EmoBERTa-base loaded")
print("✓ Transformer backbone frozen")
print("✓ Attention pooling active")
print("✓ Metadata branch = 53 -> 128 -> 64")
print("✓ Single metadata gate active")
print("✓ Text representation = 768")
print("✓ Fusion = 768 + 64 = 832")
print("✓ Classifier = 832 -> 512 -> 256 -> 2")
print("✓ Padding attention masked")
print("✓ Forward pass successful")
print("✓ Frozen EmoBERTa + RAW model ready for training")
```

```python
# ============================================================
# CELL 12 — EMOBERTA-BASE + RAW
# FROZEN TRANSFORMER TRAINING
# ============================================================

import os
import time
import json
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.optim import AdamW
from torch.optim.lr_scheduler import OneCycleLR

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


print("=" * 60)
print("STARTING EMOBERTA-BASE + RAW TRAINING")
print("=" * 60)


# ============================================================
# FIXED TRAINING CONFIGURATION
# ============================================================

MAX_EPOCHS = 10
EARLY_STOPPING_PATIENCE = 2

LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01

print()
print("Training users:", 4026)
print("Validation users:", 767)
print("Test users:", 767)

print(
    "Maximum epochs:",
    MAX_EPOCHS
)

print(
    "Early stopping patience:",
    EARLY_STOPPING_PATIENCE
)

print(
    "Checkpoint metric:",
    "Validation F1"
)

print(
    "Learning rate:",
    LEARNING_RATE
)

print(
    "Weight decay:",
    WEIGHT_DECAY
)

print(
    "Batch size:",
    BATCH_SIZE
)

print(
    "Transformer:",
    "FROZEN"
)


# ============================================================
# LOSS
# ============================================================

criterion = nn.CrossEntropyLoss()


# ============================================================
# VERIFY BACKBONE IS FROZEN
# ============================================================

assert all(
    not parameter.requires_grad
    for parameter
    in model_emoberta_raw
    .emoberta
    .parameters()
)

print()
print(
    "✓ EmoBERTa backbone confirmed frozen"
)


# ============================================================
# OPTIMIZER
# ONLY TRAINABLE PARAMETERS
# ============================================================

trainable_parameters_list = [
    parameter
    for parameter
    in model_emoberta_raw.parameters()
    if parameter.requires_grad
]

assert len(
    trainable_parameters_list
) > 0


optimizer = AdamW(
    trainable_parameters_list,
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)


# ============================================================
# SCHEDULER
# ============================================================

total_training_steps = (
    len(
        train_emoberta_raw_loader
    )
    *
    MAX_EPOCHS
)

scheduler = OneCycleLR(
    optimizer,
    max_lr=LEARNING_RATE,
    total_steps=total_training_steps,
    pct_start=0.1,
    anneal_strategy="cos"
)


# ============================================================
# METRIC FUNCTION
# ============================================================

def calculate_metrics(
    labels,
    probabilities
):

    labels = np.asarray(
        labels
    )

    probabilities = np.asarray(
        probabilities
    )

    predictions = (
        probabilities >= 0.5
    ).astype(int)


    return {

        "accuracy":
            accuracy_score(
                labels,
                predictions
            ),

        "precision":
            precision_score(
                labels,
                predictions,
                zero_division=0
            ),

        "recall":
            recall_score(
                labels,
                predictions,
                zero_division=0
            ),

        "f1":
            f1_score(
                labels,
                predictions,
                zero_division=0
            ),

        "roc_auc":
            roc_auc_score(
                labels,
                probabilities
            )
    }


# ============================================================
# VALIDATION FUNCTION
# ============================================================

def evaluate_model(
    loader
):

    model_emoberta_raw.eval()

    losses = []

    all_labels = []
    all_probabilities = []


    with torch.no_grad():

        for batch in loader:

            input_ids = (
                batch[
                    "input_ids"
                ]
                .to(device)
            )

            attention_mask = (
                batch[
                    "attention_mask"
                ]
                .to(device)
            )

            metadata = (
                batch[
                    "metadata"
                ]
                .to(device)
            )

            labels = (
                batch[
                    "labels"
                ]
                .to(device)
            )


            outputs = (
                model_emoberta_raw(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    metadata=metadata
                )
            )

            logits = outputs[
                "logits"
            ]


            loss = criterion(
                logits,
                labels
            )


            probabilities = (
                torch.softmax(
                    logits.float(),
                    dim=1
                )[:, 1]
            )


            losses.append(
                float(
                    loss.detach().cpu()
                )
            )

            all_labels.extend(
                labels.detach()
                .cpu()
                .numpy()
                .tolist()
            )

            all_probabilities.extend(
                probabilities.detach()
                .cpu()
                .numpy()
                .tolist()
            )


    metrics = calculate_metrics(
        all_labels,
        all_probabilities
    )

    metrics["loss"] = float(
        np.mean(losses)
    )

    return metrics


# ============================================================
# TRAINING HISTORY
# ============================================================

training_history = []

best_val_f1 = -np.inf
best_epoch = 0
epochs_without_improvement = 0

training_start_time = time.time()


# ============================================================
# TRAINING LOOP
# ============================================================

for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    print()
    print("=" * 60)
    print(
        f"EPOCH {epoch}/{MAX_EPOCHS}"
    )
    print("=" * 60)


    model_emoberta_raw.train()

    # --------------------------------------------------------
    # Keep frozen transformer in eval mode
    # --------------------------------------------------------

    model_emoberta_raw.emoberta.eval()


    epoch_start_time = time.time()

    running_loss = 0.0

    train_labels = []
    train_probabilities = []


    # ========================================================
    # TRAIN BATCHES
    # ========================================================

    for batch_idx, batch in enumerate(
        train_emoberta_raw_loader,
        start=1
    ):

        input_ids = (
            batch[
                "input_ids"
            ]
            .to(device)
        )

        attention_mask = (
            batch[
                "attention_mask"
            ]
            .to(device)
        )

        metadata = (
            batch[
                "metadata"
            ]
            .to(device)
        )

        labels = (
            batch[
                "labels"
            ]
            .to(device)
        )


        # ----------------------------------------------------
        # RESET GRADIENTS
        # ----------------------------------------------------

        optimizer.zero_grad(
            set_to_none=True
        )


        # ----------------------------------------------------
        # FORWARD
        # ----------------------------------------------------

        outputs = model_emoberta_raw(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = outputs[
            "logits"
        ]


        # ----------------------------------------------------
        # LOSS
        # ----------------------------------------------------

        loss = criterion(
            logits,
            labels
        )


        # ----------------------------------------------------
        # BACKPROPAGATION
        # ----------------------------------------------------

        loss.backward()


        # ----------------------------------------------------
        # GRADIENT CLIPPING
        # ----------------------------------------------------

        torch.nn.utils.clip_grad_norm_(
            trainable_parameters_list,
            max_norm=1.0
        )


        # ----------------------------------------------------
        # OPTIMIZER
        # ----------------------------------------------------

        optimizer.step()


        # ----------------------------------------------------
        # LR SCHEDULER
        # ----------------------------------------------------

        scheduler.step()


        # ----------------------------------------------------
        # LOSS
        # ----------------------------------------------------

        running_loss += (
            loss.item()
        )


        # ----------------------------------------------------
        # PROBABILITIES
        # ----------------------------------------------------

        probabilities = (
            torch.softmax(
                logits.detach().float(),
                dim=1
            )[:, 1]
        )


        train_labels.extend(
            labels.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        train_probabilities.extend(
            probabilities.detach()
            .cpu()
            .numpy()
            .tolist()
        )


        # ====================================================
        # PROGRESS
        # ====================================================

        if (
            batch_idx == 1
            or batch_idx % 50 == 0
            or batch_idx
            == len(
                train_emoberta_raw_loader
            )
        ):

            temp_predictions = (
                np.asarray(
                    train_probabilities
                )
                >= 0.5
            ).astype(int)


            temp_accuracy = (
                accuracy_score(
                    train_labels,
                    temp_predictions
                )
            )


            current_lr = (
                optimizer
                .param_groups[0]
                ["lr"]
            )


            elapsed = (
                time.time()
                -
                epoch_start_time
            )


            print(
                f"Epoch {epoch} | "
                f"Batch {batch_idx}/"
                f"{len(train_emoberta_raw_loader)} | "
                f"Loss: {loss.item():.4f} | "
                f"Accuracy: {temp_accuracy:.4f} | "
                f"LR: {current_lr:.2e} | "
                f"Time: {elapsed:.1f}s"
            )


    # ========================================================
    # TRAIN METRICS
    # ========================================================

    train_metrics = calculate_metrics(
        train_labels,
        train_probabilities
    )

    train_loss = (
        running_loss
        /
        len(
            train_emoberta_raw_loader
        )
    )


    # ========================================================
    # VALIDATION
    # ========================================================

    print()
    print(
        f"Epoch {epoch} — Validation"
    )


    val_metrics = evaluate_model(
        val_emoberta_raw_loader
    )


    # ========================================================
    # METADATA GATE
    # ========================================================

    metadata_gate = float(
        torch.sigmoid(
            model_emoberta_raw.metadata_beta
        )
        .detach()
        .cpu()
    )


    # ========================================================
    # EPOCH TIME
    # ========================================================

    epoch_time = (
        time.time()
        -
        epoch_start_time
    )


    # ========================================================
    # HISTORY RECORD
    # ========================================================

    epoch_record = {

        "epoch":
            epoch,

        "train_loss":
            train_loss,

        "train_accuracy":
            train_metrics["accuracy"],

        "train_precision":
            train_metrics["precision"],

        "train_recall":
            train_metrics["recall"],

        "train_f1":
            train_metrics["f1"],

        "val_loss":
            val_metrics["loss"],

        "val_accuracy":
            val_metrics["accuracy"],

        "val_precision":
            val_metrics["precision"],

        "val_recall":
            val_metrics["recall"],

        "val_f1":
            val_metrics["f1"],

        "val_roc_auc":
            val_metrics["roc_auc"],

        "metadata_gate":
            metadata_gate,

        "learning_rate":
            optimizer.param_groups[0]["lr"],

        "epoch_time_seconds":
            epoch_time
    }


    training_history.append(
        epoch_record
    )


    # ========================================================
    # SUMMARY
    # ========================================================

    print()
    print("=" * 60)
    print(
        f"EPOCH {epoch} SUMMARY"
    )
    print("=" * 60)


    print(
        "Train Loss:",
        f"{train_loss:.4f}"
    )

    print(
        "Train Accuracy:",
        f"{train_metrics['accuracy']:.4f}"
    )

    print(
        "Train Precision:",
        f"{train_metrics['precision']:.4f}"
    )

    print(
        "Train Recall:",
        f"{train_metrics['recall']:.4f}"
    )

    print(
        "Train F1:",
        f"{train_metrics['f1']:.4f}"
    )

    print()

    print(
        "Val Loss:",
        f"{val_metrics['loss']:.4f}"
    )

    print(
        "Val Accuracy:",
        f"{val_metrics['accuracy']:.4f}"
    )

    print(
        "Val Precision:",
        f"{val_metrics['precision']:.4f}"
    )

    print(
        "Val Recall:",
        f"{val_metrics['recall']:.4f}"
    )

    print(
        "Val F1:",
        f"{val_metrics['f1']:.4f}"
    )

    print(
        "Val ROC-AUC:",
        f"{val_metrics['roc_auc']:.4f}"
    )

    print(
        "Metadata gate:",
        f"{metadata_gate:.6f}"
    )

    print(
        "Epoch Time:",
        f"{epoch_time:.2f}s"
    )


    # ========================================================
    # BEST CHECKPOINT
    # ========================================================

    if (
        val_metrics["f1"]
        >
        best_val_f1
    ):

        best_val_f1 = (
            val_metrics["f1"]
        )

        best_epoch = epoch

        epochs_without_improvement = 0


        best_checkpoint = os.path.join(
            CHECKPOINT_DIR,
            "best_emoberta_raw.pt"
        )


        torch.save(
            {

                "epoch":
                    epoch,

                "model_state_dict":
                    model_emoberta_raw
                    .state_dict(),

                "optimizer_state_dict":
                    optimizer
                    .state_dict(),

                "scheduler_state_dict":
                    scheduler
                    .state_dict(),

                "best_val_f1":
                    best_val_f1,

                "metadata_beta":
                    float(
                        model_emoberta_raw
                        .metadata_beta
                        .detach()
                        .cpu()
                    ),

                "history":
                    training_history
            },

            best_checkpoint
        )


        print()
        print(
            "✓ NEW BEST CHECKPOINT"
        )

        print(
            "Best Validation F1:",
            f"{best_val_f1:.4f}"
        )

        print(
            "Saved:",
            best_checkpoint
        )


    else:

        epochs_without_improvement += 1


        print()
        print(
            "No validation F1 improvement."
        )

        print(
            "Epochs without improvement:",
            f"{epochs_without_improvement} / "
            f"{EARLY_STOPPING_PATIENCE}"
        )


        if (
            epochs_without_improvement
            >=
            EARLY_STOPPING_PATIENCE
        ):

            print()
            print(
                "EARLY STOPPING TRIGGERED"
            )

            break


# ============================================================
# HISTORY DATAFRAME
# ============================================================

history_df = pd.DataFrame(
    training_history
)


# ============================================================
# SAVE HISTORY
# ============================================================

history_path = os.path.join(
    RESULT_DIR,
    "emoberta_raw_training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)


# ============================================================
# SAVE TRAINING SUMMARY
# ============================================================

training_summary = {

    "model":
        "EmoBERTa-base",

    "cleaning_method":
        "RAW",

    "transformer_frozen":
        True,

    "epochs_completed":
        len(history_df),

    "best_epoch":
        best_epoch,

    "best_validation_f1":
        best_val_f1,

    "total_training_time_seconds":
        time.time()
        -
        training_start_time,

    "learning_rate":
        LEARNING_RATE,

    "weight_decay":
        WEIGHT_DECAY,

    "batch_size":
        BATCH_SIZE,

    "seed":
        SEED
}


with open(
    os.path.join(
        RESULT_DIR,
        "training_summary.json"
    ),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        training_summary,
        f,
        indent=2
    )


# ============================================================
# FINAL BACKBONE CHECK
# ============================================================

assert all(
    not parameter.requires_grad
    for parameter
    in model_emoberta_raw
    .emoberta
    .parameters()
)


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("EMOBERTA-BASE + RAW TRAINING COMPLETE")
print("=" * 60)

print(
    "Epochs completed:",
    len(history_df)
)

print(
    "Best epoch:",
    best_epoch
)

print(
    "Best Validation F1:",
    f"{best_val_f1:.4f}"
)

print(
    "Total training time:",
    f"{time.time() - training_start_time:.2f} seconds"
)

print(
    "Best checkpoint:",
    best_checkpoint
)

print(
    "Training history saved:",
    history_path
)

print()
print(
    "✓ EmoBERTa transformer remained frozen"
)

print(
    "✓ Downstream layers trained"
)

print()
print("=" * 60)
print("CELL 12 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# CELL 13 — EMOBERTA-BASE + RAW
# FINAL TEST EVALUATION
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)


print("=" * 60)
print("EMOBERTA-BASE + RAW — FINAL TEST EVALUATION")
print("=" * 60)


# ============================================================
# BEST CHECKPOINT
# ============================================================

best_checkpoint = os.path.join(
    CHECKPOINT_DIR,
    "best_emoberta_raw.pt"
)

print()
print(
    "Checking best EmoBERTa RAW checkpoint..."
)

assert os.path.exists(
    best_checkpoint
), (
    f"Checkpoint not found: "
    f"{best_checkpoint}"
)

print(
    "✓ Best checkpoint found"
)


# ============================================================
# LOAD BEST CHECKPOINT
# ============================================================

print()
print(
    "Loading BEST EmoBERTa RAW checkpoint..."
)

checkpoint = torch.load(
    best_checkpoint,
    map_location=device,
    weights_only=False
)

model_emoberta_raw.load_state_dict(
    checkpoint["model_state_dict"]
)

model_emoberta_raw.eval()

# Keep the backbone frozen and in eval mode.
model_emoberta_raw.emoberta.eval()


best_epoch_loaded = int(
    checkpoint["epoch"]
)

best_val_f1_loaded = float(
    checkpoint["best_val_f1"]
)


print(
    "✓ Best checkpoint loaded"
)

print(
    "Best epoch:",
    best_epoch_loaded
)

print(
    "Best validation F1:",
    f"{best_val_f1_loaded:.4f}"
)

print(
    "Checkpoint:",
    best_checkpoint
)


# ============================================================
# GATE VERIFICATION
# ============================================================

print()
print("=" * 60)
print("GATE VERIFICATION")
print("=" * 60)

learned_beta = float(
    model_emoberta_raw
    .metadata_beta
    .detach()
    .cpu()
)

metadata_gate = float(
    torch.sigmoid(
        model_emoberta_raw
        .metadata_beta
    )
    .detach()
    .cpu()
)

print(
    "Learned beta:",
    learned_beta
)

print(
    "Metadata gate g = sigmoid(beta):",
    f"{metadata_gate:.6f}"
)

print(
    "✓ Single metadata gate verified"
)


# ============================================================
# FROZEN BACKBONE VERIFICATION
# ============================================================

assert all(
    not parameter.requires_grad
    for parameter
    in model_emoberta_raw
    .emoberta
    .parameters()
)

print(
    "✓ EmoBERTa backbone remains frozen"
)


# ============================================================
# TEST SET INTEGRITY
# ============================================================

print()
print("=" * 60)
print("TEST SET INTEGRITY")
print("=" * 60)

expected_test_users = 767

actual_test_users = len(
    test_emoberta_raw_dataset
)

print(
    "Expected test users:",
    expected_test_users
)

print(
    "Actual test users:",
    actual_test_users
)

assert actual_test_users == (
    expected_test_users
)

print(
    "✓ Test set size correct"
)


# ============================================================
# EVALUATING TEST SET
# ============================================================

print()
print("=" * 60)
print("EVALUATING TEST SET")
print("=" * 60)


test_losses = []

test_labels_all = []
test_probabilities_all = []
test_predictions_all = []

evaluated_test_users = 0


with torch.no_grad():

    for batch_idx, batch in enumerate(
        test_emoberta_raw_loader,
        start=1
    ):

        input_ids = (
            batch[
                "input_ids"
            ]
            .to(device)
        )

        attention_mask = (
            batch[
                "attention_mask"
            ]
            .to(device)
        )

        metadata = (
            batch[
                "metadata"
            ]
            .to(device)
        )

        labels = (
            batch[
                "labels"
            ]
            .to(device)
        )


        outputs = model_emoberta_raw(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = outputs[
            "logits"
        ]


        # ----------------------------------------------------
        # LOSS
        # ----------------------------------------------------

        loss = criterion(
            logits,
            labels
        )

        test_losses.append(
            float(
                loss.detach().cpu()
            )
        )


        # ----------------------------------------------------
        # POSITIVE CLASS PROBABILITY
        # ----------------------------------------------------

        probabilities = torch.softmax(
            logits.float(),
            dim=1
        )[:, 1]


        # ----------------------------------------------------
        # BINARY PREDICTION
        # ----------------------------------------------------

        predictions = (
            probabilities >= 0.5
        ).long()


        # ----------------------------------------------------
        # COLLECT
        # ----------------------------------------------------

        test_labels_all.extend(
            labels.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        test_probabilities_all.extend(
            probabilities.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        test_predictions_all.extend(
            predictions.detach()
            .cpu()
            .numpy()
            .tolist()
        )

        evaluated_test_users += len(
            labels
        )


        # ----------------------------------------------------
        # PROGRESS
        # ----------------------------------------------------

        if (
            batch_idx == 1
            or batch_idx % 10 == 0
            or batch_idx
            == len(
                test_emoberta_raw_loader
            )
        ):

            print(
                f"Test batch "
                f"{batch_idx}/"
                f"{len(test_emoberta_raw_loader)}"
            )


# ============================================================
# TEST METRICS
# ============================================================

test_loss = float(
    np.mean(
        test_losses
    )
)

test_accuracy = accuracy_score(
    test_labels_all,
    test_predictions_all
)

test_precision = precision_score(
    test_labels_all,
    test_predictions_all,
    zero_division=0
)

test_recall = recall_score(
    test_labels_all,
    test_predictions_all,
    zero_division=0
)

test_f1 = f1_score(
    test_labels_all,
    test_predictions_all,
    zero_division=0
)

test_roc_auc = roc_auc_score(
    test_labels_all,
    test_probabilities_all
)


# ============================================================
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    test_labels_all,
    test_predictions_all,
    labels=[0, 1]
)

tn, fp, fn, tp = cm.ravel()


# ============================================================
# CLASSIFICATION REPORT
# ============================================================

classification_report_text = (
    classification_report(
        test_labels_all,
        test_predictions_all,
        target_names=[
            "Negative",
            "Positive"
        ],
        digits=4,
        zero_division=0
    )
)


# ============================================================
# FINAL TEST RESULTS
# ============================================================

print()
print("=" * 60)
print(
    "EMOBERTA-BASE + RAW — FINAL TEST RESULTS"
)
print("=" * 60)

print(
    "Test Loss:",
    f"{test_loss:.4f}"
)

print(
    "Test Accuracy:",
    f"{test_accuracy * 100:.2f}%"
)

print(
    "Test Precision:",
    f"{test_precision * 100:.2f}%"
)

print(
    "Test Recall:",
    f"{test_recall * 100:.2f}%"
)

print(
    "Test F1:",
    f"{test_f1 * 100:.2f}%"
)

print(
    "Test ROC-AUC:",
    f"{test_roc_auc * 100:.2f}%"
)


# ============================================================
# CONFUSION MATRIX
# ============================================================

print()
print("=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

print(
    "                 Predicted"
)

print(
    "                 0       1"
)

print(
    f"Actual 0        "
    f"{tn:<7}"
    f"{fp}"
)

print(
    f"Actual 1        "
    f"{fn:<7}"
    f"{tp}"
)


# ============================================================
# CLASSIFICATION REPORT
# ============================================================

print()
print("=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report_text
)


# ============================================================
# FINAL TEST INTEGRITY
# ============================================================

print()
print("=" * 60)
print("FINAL TEST INTEGRITY CHECK")
print("=" * 60)

print(
    "Expected test users:",
    expected_test_users
)

print(
    "Evaluated test users:",
    evaluated_test_users
)

assert (
    evaluated_test_users
    ==
    expected_test_users
)

assert len(
    test_labels_all
) == expected_test_users

assert len(
    test_probabilities_all
) == expected_test_users

assert len(
    test_predictions_all
) == expected_test_users


print(
    "✓ All test users evaluated exactly once"
)

print(
    "✓ Predictions aligned with labels"
)

print(
    "✓ Best EmoBERTa RAW checkpoint used"
)

print(
    "✓ Training data not used for evaluation"
)

print(
    "✓ Validation data not used for evaluation"
)

print(
    "✓ Transformer backbone remained frozen"
)


# ============================================================
# SAVE TEST PREDICTIONS
# ============================================================

predictions_df = pd.DataFrame({

    "user_id":
        test_emoberta_raw_df[
            "user_id"
        ].values,

    "username":
        test_emoberta_raw_df[
            "username"
        ].values,

    "label":
        np.asarray(
            test_labels_all
        ),

    "probability":
        np.asarray(
            test_probabilities_all
        ),

    "prediction":
        np.asarray(
            test_predictions_all
        )
})


predictions_path = os.path.join(
    RESULT_DIR,
    "emoberta_raw_test_predictions.csv"
)

predictions_df.to_csv(
    predictions_path,
    index=False
)


# ============================================================
# SAVE METRICS
# ============================================================

metrics_df = pd.DataFrame([{

    "Model Name":
        "EmoBERTa-base",

    "Cleaning Method":
        "RAW",

    "Test Loss":
        test_loss,

    "Accuracy (%)":
        test_accuracy * 100,

    "Precision (%)":
        test_precision * 100,

    "Recall (%)":
        test_recall * 100,

    "F1 (%)":
        test_f1 * 100,

    "ROC-AUC (%)":
        test_roc_auc * 100,

    "Best Epoch":
        best_epoch_loaded,

    "Best Val F1 (%)":
        best_val_f1_loaded * 100,

    "Parameters (M)":
        total_parameters / 1e6,

    "Trainable Parameters (M)":
        trainable_parameters / 1e6,

    "Frozen Parameters (M)":
        frozen_parameters / 1e6,

    "TN":
        int(tn),

    "FP":
        int(fp),

    "FN":
        int(fn),

    "TP":
        int(tp),

    "Metadata Beta":
        learned_beta,

    "Metadata Gate":
        metadata_gate
}])


metrics_path = os.path.join(
    RESULT_DIR,
    "emoberta_raw_test_metrics.csv"
)

metrics_df.to_csv(
    metrics_path,
    index=False
)


# ============================================================
# SAVE CONFUSION MATRIX
# ============================================================

cm_df = pd.DataFrame(
    cm,
    index=[
        "Actual 0",
        "Actual 1"
    ],
    columns=[
        "Predicted 0",
        "Predicted 1"
    ]
)

cm_path = os.path.join(
    RESULT_DIR,
    "emoberta_raw_confusion_matrix.csv"
)

cm_df.to_csv(
    cm_path
)


# ============================================================
# SAVE CLASSIFICATION REPORT
# ============================================================

report_path = os.path.join(
    RESULT_DIR,
    "emoberta_raw_classification_report.txt"
)

with open(
    report_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        classification_report_text
    )


# ============================================================
# SAVE COMPLETE FINAL RESULTS
# ============================================================

final_results = {

    "model_name":
        "EmoBERTa-base",

    "cleaning_method":
        "RAW",

    "transformer_frozen":
        True,

    "test_loss":
        test_loss,

    "accuracy":
        test_accuracy,

    "precision":
        test_precision,

    "recall":
        test_recall,

    "f1":
        test_f1,

    "roc_auc":
        test_roc_auc,

    "best_epoch":
        best_epoch_loaded,

    "best_validation_f1":
        best_val_f1_loaded,

    "parameters_m":
        total_parameters / 1e6,

    "trainable_parameters_m":
        trainable_parameters / 1e6,

    "frozen_parameters_m":
        frozen_parameters / 1e6,

    "tn":
        int(tn),

    "fp":
        int(fp),

    "fn":
        int(fn),

    "tp":
        int(tp),

    "metadata_beta":
        learned_beta,

    "metadata_gate":
        metadata_gate,

    "evaluated_test_users":
        evaluated_test_users
}


with open(
    os.path.join(
        RESULT_DIR,
        "emoberta_raw_final_results.json"
    ),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        final_results,
        f,
        indent=2
    )


# ============================================================
# FINAL
# ============================================================

print()
print("=" * 60)
print("CELL 13 COMPLETE")
print("=" * 60)

print("✓ Best checkpoint evaluated")
print("✓ Held-out test evaluated exactly once")
print("✓ Final metrics calculated")
print("✓ Predictions saved")
print("✓ Metrics saved")
print("✓ Confusion matrix saved")
print("✓ Classification report saved")
print("✓ Complete final results saved")
print("✓ Frozen transformer verified")
```

```python
# ============================================================
# EXPLAINABILITY — CELL 1
# EMOBERTA-BASE + RAW
# t-SNE: FUSED REPRESENTATION → PCA → t-SNE
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

print("=" * 70)
print("t-SNE REPRESENTATION ANALYSIS")
print("EMOBERTA-BASE + RAW")
print("=" * 70)

# ============================================================
# FIXED EXPLAINABILITY SETTINGS
# ============================================================

EXPLAINABILITY_SEED = 42
PCA_COMPONENTS = 50
TSNE_COMPONENTS = 2
TSNE_PERPLEXITY = 30
TSNE_LEARNING_RATE = 200
TSNE_MAX_ITER = 1000

np.random.seed(
    EXPLAINABILITY_SEED
)

torch.manual_seed(
    EXPLAINABILITY_SEED
)

# ============================================================
# OUTPUT DIRECTORIES
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

TSNE_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "tsne"
)

os.makedirs(
    TSNE_DIR,
    exist_ok=True
)

# ============================================================
# MODEL
# ============================================================

model_emoberta_raw.eval()

print()
print(
    "✓ EmoBERTa RAW model in evaluation mode"
)

# ============================================================
# EXTRACT FUSED REPRESENTATIONS
# ============================================================

fused_representations = []
test_labels = []

print()
print("Extracting fused model representations...")

with torch.no_grad():

    for batch_idx, batch in enumerate(
        test_emoberta_raw_loader,
        start=1
    ):

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        labels = batch[
            "labels"
        ].to(device)

        output = model_emoberta_raw(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        # ----------------------------------------------------
        # EXACT MODEL OUTPUTS
        # ----------------------------------------------------

        text_representation = output[
            "text_representation"
        ]

        metadata_representation = output[
            "metadata_representation"
        ]

        # ----------------------------------------------------
        # SAME LEARNED METADATA GATE
        # ----------------------------------------------------

        metadata_gate = torch.sigmoid(
            model_emoberta_raw.metadata_beta
        )

        gated_metadata = (
            metadata_gate
            *
            metadata_representation
        )

        # ----------------------------------------------------
        # EXACT FUSED REPRESENTATION
        # ----------------------------------------------------

        fused = torch.cat(
            [
                text_representation,
                gated_metadata
            ],
            dim=1
        )

        fused_representations.append(
            fused.cpu().numpy()
        )

        test_labels.extend(
            labels.cpu().numpy().tolist()
        )

        if (
            batch_idx == 1
            or batch_idx % 10 == 0
            or batch_idx == len(
                test_emoberta_raw_loader
            )
        ):

            print(
                f"Processed batch "
                f"{batch_idx}/"
                f"{len(test_emoberta_raw_loader)}"
            )

# ============================================================
# COMBINE
# ============================================================

fused_representations = np.concatenate(
    fused_representations,
    axis=0
)

test_labels = np.array(
    test_labels
)

# ============================================================
# INTEGRITY
# ============================================================

print()
print("=" * 70)
print("FUSED REPRESENTATION")
print("=" * 70)

print(
    "Representation shape:",
    fused_representations.shape
)

print(
    "Number of test users:",
    len(test_labels)
)

print(
    "Representation dimensions:",
    fused_representations.shape[1]
)

print(
    "Negative users:",
    int(
        (test_labels == 0).sum()
    )
)

print(
    "Positive users:",
    int(
        (test_labels == 1).sum()
    )
)

print(
    "Metadata gate:",
    f"{float(metadata_gate.item()):.6f}"
)

assert (
    fused_representations.shape[0]
    ==
    len(test_labels)
)

# EmoBERTa RAW = 768 text + 64 metadata
assert (
    fused_representations.shape[1]
    ==
    832
)

assert np.isfinite(
    fused_representations
).all()

# ============================================================
# PCA
# ============================================================

print()
print("=" * 70)
print("PCA REDUCTION")
print("=" * 70)

pca_dimensions = min(
    PCA_COMPONENTS,
    fused_representations.shape[1]
)

pca = PCA(
    n_components=pca_dimensions,
    random_state=EXPLAINABILITY_SEED
)

pca_representation = pca.fit_transform(
    fused_representations
)

explained_variance = (
    pca.explained_variance_ratio_.sum()
)

print(
    "Original dimensions:",
    fused_representations.shape[1]
)

print(
    "PCA dimensions:",
    pca_representation.shape[1]
)

print(
    "Variance retained:",
    f"{explained_variance * 100:.2f}%"
)

# ============================================================
# t-SNE
# ============================================================

print()
print("=" * 70)
print("t-SNE REDUCTION")
print("=" * 70)

tsne = TSNE(
    n_components=TSNE_COMPONENTS,
    perplexity=TSNE_PERPLEXITY,
    learning_rate=TSNE_LEARNING_RATE,
    init="pca",
    random_state=EXPLAINABILITY_SEED,
    max_iter=TSNE_MAX_ITER
)

tsne_representation = tsne.fit_transform(
    pca_representation
)

print(
    "t-SNE output shape:",
    tsne_representation.shape
)

# ============================================================
# SAVE COORDINATES
# ============================================================

tsne_df = pd.DataFrame({

    "tsne_1":
        tsne_representation[:, 0],

    "tsne_2":
        tsne_representation[:, 1],

    "label":
        test_labels

})

tsne_df["class"] = tsne_df[
    "label"
].map({
    0: "Negative",
    1: "Positive"
})

tsne_csv_path = os.path.join(
    TSNE_DIR,
    "emoberta_raw_tsne_coordinates.csv"
)

tsne_df.to_csv(
    tsne_csv_path,
    index=False
)

# ============================================================
# PUBLICATION-QUALITY PLOT
# ============================================================

plt.figure(
    figsize=(10, 8)
)

negative_mask = (
    test_labels == 0
)

positive_mask = (
    test_labels == 1
)

plt.scatter(
    tsne_representation[
        negative_mask,
        0
    ],
    tsne_representation[
        negative_mask,
        1
    ],
    s=35,
    alpha=0.70,
    label="Negative",
    edgecolors="none"
)

plt.scatter(
    tsne_representation[
        positive_mask,
        0
    ],
    tsne_representation[
        positive_mask,
        1
    ],
    s=35,
    alpha=0.70,
    label="Positive",
    edgecolors="none"
)

plt.xlabel(
    "t-SNE Dimension 1",
    fontsize=12
)

plt.ylabel(
    "t-SNE Dimension 2",
    fontsize=12
)

plt.title(
    "EmoBERTa-base + RAW — "
    "t-SNE of Fused Test Representations",
    fontsize=14
)

plt.legend(
    fontsize=11
)

plt.grid(
    alpha=0.20
)

plt.tight_layout()

tsne_plot_path = os.path.join(
    TSNE_DIR,
    "emoberta_raw_tsne.png"
)

# SAVE
plt.savefig(
    tsne_plot_path,
    dpi=300,
    bbox_inches="tight"
)

# DISPLAY
plt.show()

plt.close()

# ============================================================
# SAVE CONFIGURATION
# ============================================================

tsne_config = {

    "model":
        "EmoBERTa-base",

    "model_name":
        MODEL_NAME,

    "cleaning":
        "RAW",

    "dataset":
        "Held-out test set",

    "test_users":
        int(
            len(test_labels)
        ),

    "input_representation":
        "Fused classifier representation",

    "text_representation_dimension":
        int(
            text_representation.shape[1]
        ),

    "metadata_representation_dimension":
        int(
            metadata_representation.shape[1]
        ),

    "fused_representation_dimension":
        int(
            fused_representations.shape[1]
        ),

    "metadata_gate":
        float(
            metadata_gate.item()
        ),

    "pca_dimensions":
        int(
            pca_dimensions
        ),

    "pca_variance_retained":
        float(
            explained_variance
        ),

    "tsne_dimensions":
        int(
            TSNE_COMPONENTS
        ),

    "tsne_perplexity":
        int(
            TSNE_PERPLEXITY
        ),

    "tsne_learning_rate":
        TSNE_LEARNING_RATE,

    "tsne_max_iterations":
        int(
            TSNE_MAX_ITER
        ),

    "random_seed":
        EXPLAINABILITY_SEED

}

with open(
    os.path.join(
        TSNE_DIR,
        "emoberta_raw_tsne_config.json"
    ),
    "w"
) as f:

    json.dump(
        tsne_config,
        f,
        indent=2
    )

# ============================================================
# FINAL OUTPUT
# ============================================================

print()
print("=" * 70)
print("t-SNE ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ Fused representation extracted"
)

print(
    "✓ EmoBERTa text representation = 768"
)

print(
    "✓ Gated metadata representation = 64"
)

print(
    "✓ Fused representation = 832"
)

print(
    "✓ PCA reduction completed"
)

print(
    "✓ t-SNE reduction completed"
)

print(
    "✓ Coordinates saved:"
)

print(
    tsne_csv_path
)

print(
    "✓ Graph saved:"
)

print(
    tsne_plot_path
)

print(
    "✓ Graph displayed above"
)

print(
    "✓ Configuration saved"
)
```

```python
# ============================================================
# EXPLAINABILITY — CELL 3
# EMOBERTA-BASE + RAW
# SHAP: METADATA-LEVEL GLOBAL EXPLANATION
# ============================================================

import os
import json
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import shap

print("=" * 70)
print("SHAP ANALYSIS")
print("EMOBERTA-BASE + RAW")
print("=" * 70)

# ============================================================
# FIXED SHAP SETTINGS
# ============================================================

SHAP_SEED = 42
SHAP_BACKGROUND_SIZE = 50
SHAP_TEST_SIZE_PER_CLASS = 50
SHAP_TOP_FEATURES = 20

np.random.seed(
    SHAP_SEED
)

random.seed(
    SHAP_SEED
)

torch.manual_seed(
    SHAP_SEED
)

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

SHAP_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "shap"
)

os.makedirs(
    SHAP_DIR,
    exist_ok=True
)

# ============================================================
# EXACT 53 METADATA FEATURE NAMES
# ============================================================

behavioral_features = [
    "posting_frequency",
    "follower_friend_ratio",
    "night_posting_ratio",
    "engagement",
    "followers_count",
    "friends_count",
    "statuses_count",
    "favourites_count"
]

linguistic_features = [
    "first_person_ratio",
    "negation_ratio",
    "uppercase_ratio",
    "punctuation_freq"
]

readability_features = [
    "flesch_reading_ease",
    "gunning_fog",
    "flesch_kincaid_grade"
]

emotional_features = [
    "sentiment_score",
    "emotion_score",
    "pos_emoji_count",
    "neg_emoji_count",
    "emoji_density",
    "exclamation_count"
]

psychological_features = [
    "depression_lexicon_score",
    "hopelessness_score",
    "loneliness_score",
    "anxiety_score",
    "sleep_words_ratio",
    "fatigue_words_ratio",
    "selfharm_keywords_ratio"
]

writing_style_features = [
    "avg_sentence_length",
    "lexical_diversity",
    "word_count",
    "avg_paragraph_length",
    "repeated_letters"
]

lda_features = [
    f"lda_topic_{i}"
    for i in range(1, 21)
]

raw_metadata_features = (
    behavioral_features
    + linguistic_features
    + readability_features
    + emotional_features
    + psychological_features
    + writing_style_features
    + lda_features
)

assert len(
    raw_metadata_features
) == 53

print(
    f"✓ Metadata features confirmed: "
    f"{len(raw_metadata_features)}"
)

# ============================================================
# MODEL
# ============================================================

model_emoberta_raw.eval()

print(
    "✓ EmoBERTa + RAW model loaded"
)

print(
    "✓ Evaluation mode enabled"
)

# ============================================================
# SHAP WRAPPER
# FIX TEXT INPUT FOR EACH USER
# VARY ONLY METADATA
# ============================================================

class MetadataSHAPWrapper(nn.Module):

    def __init__(
        self,
        model,
        input_ids,
        attention_mask
    ):

        super().__init__()

        self.model = model

        self.register_buffer(
            "input_ids",
            input_ids
        )

        self.register_buffer(
            "attention_mask",
            attention_mask
        )

    def forward(
        self,
        metadata
    ):

        batch_size = metadata.shape[0]

        input_ids = self.input_ids.expand(
            batch_size,
            -1
        )

        attention_mask = self.attention_mask.expand(
            batch_size,
            -1
        )

        output = self.model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        # Positive / depression class logit
        return output["logits"][:, 1:2]


# ============================================================
# BACKGROUND METADATA
# ============================================================

background_metadata = (
    train_emoberta_raw_dataset.metadata[
        :SHAP_BACKGROUND_SIZE
    ]
    .clone()
    .detach()
    .float()
    .to(device)
)

print()
print("=" * 70)
print("SHAP BACKGROUND")
print("=" * 70)

print(
    "Background samples:",
    background_metadata.shape[0]
)

print(
    "Metadata dimensions:",
    background_metadata.shape[1]
)

assert background_metadata.shape == (
    SHAP_BACKGROUND_SIZE,
    53
)

# ============================================================
# TEST USERS
# 50 NEGATIVE + 50 POSITIVE
# ============================================================

test_labels = np.asarray(
    test_emoberta_raw_dataset.labels
)

negative_indices = np.where(
    test_labels == 0
)[0]

positive_indices = np.where(
    test_labels == 1
)[0]

np.random.seed(
    SHAP_SEED
)

negative_indices = np.random.choice(
    negative_indices,
    size=min(
        SHAP_TEST_SIZE_PER_CLASS,
        len(negative_indices)
    ),
    replace=False
)

positive_indices = np.random.choice(
    positive_indices,
    size=min(
        SHAP_TEST_SIZE_PER_CLASS,
        len(positive_indices)
    ),
    replace=False
)

selected_shap_indices = np.concatenate(
    [
        negative_indices,
        positive_indices
    ]
)

print()
print("=" * 70)
print("SHAP TEST USERS")
print("=" * 70)

print(
    "Negative users:",
    len(negative_indices)
)

print(
    "Positive users:",
    len(positive_indices)
)

print(
    "Total users:",
    len(selected_shap_indices)
)

# ============================================================
# COMPUTE SHAP VALUES
# ============================================================

all_shap_rows = []
all_user_importance = []

print()
print("=" * 70)
print("GENERATING SHAP EXPLANATIONS")
print("=" * 70)

for count, test_index in enumerate(
    selected_shap_indices,
    start=1
):

    test_index = int(
        test_index
    )

    sample = test_emoberta_raw_dataset[
        test_index
    ]

    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
        .to(device)
    )

    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
        .to(device)
    )

    metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
        .float()
    )

    true_label = int(
        sample["labels"].item()
    )

    wrapper = MetadataSHAPWrapper(
        model=model_emoberta_raw,
        input_ids=input_ids,
        attention_mask=attention_mask
    ).to(device)

    wrapper.eval()

    explainer = shap.GradientExplainer(
        wrapper,
        background_metadata
    )

    shap_values = explainer.shap_values(
        metadata
    )

    # --------------------------------------------------------
    # SHAP VERSION HANDLING
    # --------------------------------------------------------

    if isinstance(
        shap_values,
        list
    ):

        shap_array = np.asarray(
            shap_values[0]
        )

    else:

        shap_array = np.asarray(
            shap_values
        )

    shap_array = np.squeeze(
        shap_array
    )

    assert shap_array.shape == (
        53,
    )

    # --------------------------------------------------------
    # SAVE USER-LEVEL IMPORTANCE
    # --------------------------------------------------------

    user_importance = pd.DataFrame({

        "feature":
            raw_metadata_features,

        "shap_value":
            shap_array,

        "abs_shap_value":
            np.abs(shap_array)

    })

    user_importance = user_importance.sort_values(
        "abs_shap_value",
        ascending=False
    ).reset_index(
        drop=True
    )

    user_importance["test_index"] = test_index
    user_importance["true_label"] = true_label

    all_user_importance.append(
        user_importance
    )

    for rank, row in enumerate(
        user_importance.itertuples(
            index=False
        ),
        start=1
    ):

        all_shap_rows.append({

            "test_index":
                test_index,

            "true_label":
                true_label,

            "feature":
                row.feature,

            "shap_value":
                float(
                    row.shap_value
                ),

            "abs_shap_value":
                float(
                    row.abs_shap_value
                ),

            "rank":
                rank
        })

    if (
        count == 1
        or count % 10 == 0
        or count == len(
            selected_shap_indices
        )
    ):

        print(
            f"Processed "
            f"{count}/"
            f"{len(selected_shap_indices)}"
        )

# ============================================================
# COMBINE SHAP RESULTS
# ============================================================

shap_results_df = pd.DataFrame(
    all_shap_rows
)

shap_csv_path = os.path.join(
    SHAP_DIR,
    "emoberta_raw_shap_values.csv"
)

shap_results_df.to_csv(
    shap_csv_path,
    index=False
)

# ============================================================
# GLOBAL FEATURE IMPORTANCE
# ============================================================

global_importance = (
    shap_results_df
    .groupby("feature")[
        "abs_shap_value"
    ]
    .mean()
    .sort_values(
        ascending=False
    )
    .reset_index()
)

global_importance.columns = [
    "feature",
    "mean_abs_shap_value"
]

global_importance["rank"] = (
    np.arange(
        len(global_importance)
    ) + 1
)

global_importance_csv = os.path.join(
    SHAP_DIR,
    "emoberta_raw_shap_feature_importance.csv"
)

global_importance.to_csv(
    global_importance_csv,
    index=False
)

# ============================================================
# TOP FEATURES
# ============================================================

top_features = global_importance.head(
    SHAP_TOP_FEATURES
).copy()

print()
print("=" * 70)
print("TOP SHAP FEATURES")
print("=" * 70)

print(
    top_features[
        [
            "rank",
            "feature",
            "mean_abs_shap_value"
        ]
    ].to_string(
        index=False
    )
)

# ============================================================
# PAPER-READY GLOBAL BAR PLOT
# ============================================================

plot_df = top_features.iloc[::-1]

plt.figure(
    figsize=(10, 8)
)

plt.barh(
    plot_df["feature"],
    plot_df["mean_abs_shap_value"]
)

plt.xlabel(
    "Mean |SHAP Value|",
    fontsize=12
)

plt.ylabel(
    "Metadata Feature",
    fontsize=12
)

plt.title(
    "EmoBERTa-base + RAW — "
    "SHAP Metadata Feature Importance",
    fontsize=14
)

plt.grid(
    axis="x",
    alpha=0.20
)

plt.tight_layout()

global_plot_path = os.path.join(
    SHAP_DIR,
    "emoberta_raw_shap_global_feature_importance.png"
)

plt.savefig(
    global_plot_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

# ============================================================
# SHAP SUMMARY / BEESWARM
# ============================================================

# Reconstruct matrix:
# users × features

shap_matrix = np.zeros(
    (
        len(selected_shap_indices),
        len(raw_metadata_features)
    ),
    dtype=np.float32
)

for row_index, test_index in enumerate(
    selected_shap_indices
):

    user_rows = shap_results_df[
        shap_results_df["test_index"]
        == int(test_index)
    ]

    feature_map = dict(
        zip(
            user_rows["feature"],
            user_rows["shap_value"]
        )
    )

    for feature_index, feature in enumerate(
        raw_metadata_features
    ):

        shap_matrix[
            row_index,
            feature_index
        ] = feature_map.get(
            feature,
            0.0
        )

metadata_matrix = np.asarray(
    [
        test_emoberta_raw_dataset.metadata[
            int(test_index)
        ].cpu().numpy()
        for test_index in selected_shap_indices
    ],
    dtype=np.float32
)

plt.figure(
    figsize=(10, 8)
)

shap.summary_plot(
    shap_matrix,
    metadata_matrix,
    feature_names=raw_metadata_features,
    max_display=SHAP_TOP_FEATURES,
    show=False
)

plt.tight_layout()

summary_plot_path = os.path.join(
    SHAP_DIR,
    "emoberta_raw_shap_summary.png"
)

plt.savefig(
    summary_plot_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

# ============================================================
# SAVE USER INFORMATION
# ============================================================

user_information = []

for test_index in selected_shap_indices:

    test_index = int(
        test_index
    )

    user_information.append({

        "test_index":
            test_index,

        "true_label":
            int(
                test_emoberta_raw_dataset.labels[
                    test_index
                ].item()
            )
    })

user_information_df = pd.DataFrame(
    user_information
)

user_information_path = os.path.join(
    SHAP_DIR,
    "emoberta_raw_shap_user_information.csv"
)

user_information_df.to_csv(
    user_information_path,
    index=False
)

# ============================================================
# SAVE CONFIGURATION
# ============================================================

shap_config = {

    "model":
        "EmoBERTa-base",

    "model_name":
        MODEL_NAME,

    "cleaning":
        "RAW / MINIMAL CLEANING",

    "dataset":
        "Held-out test set",

    "method":
        "SHAP GradientExplainer",

    "explanation_target":
        "Positive/depression class logit",

    "explanation_type":
        "Metadata-level explanation",

    "metadata_features":
        53,

    "background_size":
        SHAP_BACKGROUND_SIZE,

    "test_size_per_class":
        SHAP_TEST_SIZE_PER_CLASS,

    "total_test_users":
        int(
            len(selected_shap_indices)
        ),

    "top_features_displayed":
        SHAP_TOP_FEATURES,

    "random_seed":
        SHAP_SEED,

    "model_training":
        "No additional training",

    "transformer_weights":
        "Frozen"
}

config_path = os.path.join(
    SHAP_DIR,
    "emoberta_raw_shap_config.json"
)

with open(
    config_path,
    "w"
) as f:

    json.dump(
        shap_config,
        f,
        indent=2
    )

# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("SHAP ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ 53 metadata features used"
)

print(
    "✓ 50 negative + 50 positive test users targeted"
)

print(
    "✓ Metadata-only SHAP explanations generated"
)

print(
    "✓ Global feature importance calculated"
)

print(
    "✓ SHAP summary plot saved"
)

print(
    "✓ Top feature plot saved"
)

print(
    "✓ SHAP CSV saved"
)

print(
    "✓ Configuration saved"
)

print()
print(
    "SHAP output directory:",
    SHAP_DIR
)
```

```python
# ============================================================
# EXPLAINABILITY — LIME
# EMOBERTA-BASE + RAW
# TOKEN / WORD-LEVEL LOCAL EXPLANATION
# ============================================================


import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from lime.lime_text import LimeTextExplainer
from IPython.display import display, HTML


print("=" * 70)
print("LIME EXPLAINABILITY")
print("EmoBERTa-base + RAW")
print("=" * 70)


# ============================================================
# FIXED LIME SETTINGS
# ============================================================

LIME_SEED = 42
LIME_NUM_SAMPLES = 1000
LIME_NUM_FEATURES = 15
LIME_BATCH_SIZE = 8

np.random.seed(LIME_SEED)
torch.manual_seed(LIME_SEED)


# ============================================================
# OUTPUT DIRECTORY
# ============================================================

EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

LIME_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "lime"
)

os.makedirs(
    LIME_DIR,
    exist_ok=True
)


# ============================================================
# MODEL
# ============================================================

model_emoberta_raw.eval()

print("✓ EmoBERTa-base + RAW model loaded")
print("✓ Evaluation mode enabled")


# ============================================================
# TEST PREDICTIONS
# ============================================================

labels_np = []
preds_np = []
probs_np = []

with torch.no_grad():

    for batch in test_emoberta_raw_loader:

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        output = model_emoberta_raw(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = output["logits"]

        probabilities = torch.softmax(
            logits,
            dim=1
        )

        predictions = torch.argmax(
            logits,
            dim=1
        )

        labels_np.extend(
            batch["labels"]
            .cpu()
            .numpy()
            .tolist()
        )

        preds_np.extend(
            predictions
            .cpu()
            .numpy()
            .tolist()
        )

        probs_np.extend(
            probabilities[:, 1]
            .cpu()
            .numpy()
            .tolist()
        )


labels_np = np.array(
    labels_np
)

preds_np = np.array(
    preds_np
)

probs_np = np.array(
    probs_np
)


print()
print("=" * 70)
print("TEST PREDICTIONS READY")
print("=" * 70)

print(
    "Test samples:",
    len(labels_np)
)


# ============================================================
# SELECT TP / TN / FP / FN
# ============================================================

selected_cases = {}

tp_candidates = np.where(
    (labels_np == 1)
    &
    (preds_np == 1)
)[0]

tn_candidates = np.where(
    (labels_np == 0)
    &
    (preds_np == 0)
)[0]

fp_candidates = np.where(
    (labels_np == 0)
    &
    (preds_np == 1)
)[0]

fn_candidates = np.where(
    (labels_np == 1)
    &
    (preds_np == 0)
)[0]


if len(tp_candidates) > 0:
    selected_cases["TP"] = int(
        tp_candidates[0]
    )

if len(tn_candidates) > 0:
    selected_cases["TN"] = int(
        tn_candidates[0]
    )

if len(fp_candidates) > 0:
    selected_cases["FP"] = int(
        fp_candidates[0]
    )

if len(fn_candidates) > 0:
    selected_cases["FN"] = int(
        fn_candidates[0]
    )


print()
print("=" * 70)
print("SELECTED CASES")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print(
        f"{case_name} : "
        f"test index {test_index} | "
        f"true={labels_np[test_index]} | "
        f"pred={preds_np[test_index]} | "
        f"probability={probs_np[test_index]:.4f}"
    )


# ============================================================
# GET TEST TEXT
# ============================================================

def get_test_text(test_index):

    return str(
        test_emoberta_raw_df.iloc[
            test_index
        ]["model_text"]
    )


# ============================================================
# CREATE LIME PREDICTOR
# TEXT VARIES
# METADATA REMAINS FIXED
# ============================================================

def make_lime_predictor(test_index):

    sample = test_emoberta_raw_dataset[
        test_index
    ]

    fixed_metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
        .float()
    )

    def predictor(texts):

        all_probabilities = []

        # ----------------------------------------------------
        # PROCESS LIME PERTURBATIONS IN SMALL BATCHES
        # TO AVOID GPU OUT-OF-MEMORY
        # ----------------------------------------------------

        for start in range(
            0,
            len(texts),
            LIME_BATCH_SIZE
        ):

            batch_texts = texts[
                start:
                start + LIME_BATCH_SIZE
            ]

            encoded = emoberta_raw_tokenizer(
                batch_texts,
                padding="max_length",
                truncation=True,
                max_length=MAX_LENGTH,
                return_attention_mask=True,
                return_tensors="pt"
            )

            lime_input_ids = (
                encoded["input_ids"]
                .to(device)
            )

            lime_attention_mask = (
                encoded["attention_mask"]
                .to(device)
            )

            batch_metadata = (
                fixed_metadata.expand(
                    len(batch_texts),
                    -1
                )
            )

            with torch.no_grad():

                output = model_emoberta_raw(
                    input_ids=lime_input_ids,
                    attention_mask=lime_attention_mask,
                    metadata=batch_metadata
                )

                probabilities = torch.softmax(
                    output["logits"],
                    dim=1
                )

            all_probabilities.append(
                probabilities
                .cpu()
                .numpy()
            )

            del (
                lime_input_ids,
                lime_attention_mask,
                batch_metadata,
                output,
                probabilities
            )

            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        return np.vstack(
            all_probabilities
        )

    return predictor


# ============================================================
# LIME EXPLAINER
# ============================================================

explainer = LimeTextExplainer(
    class_names=[
        "Negative",
        "Positive"
    ],
    split_expression=r"\s+",
    random_state=LIME_SEED
)


# ============================================================
# RUN LIME FOR FOUR CASES
# ============================================================

all_explanations = {}

print()
print("=" * 70)
print("GENERATING LIME EXPLANATIONS")
print("=" * 70)

for case_name, test_index in selected_cases.items():

    print()
    print(
        f"Processing {case_name} "
        f"(test index {test_index})..."
    )

    text = get_test_text(
        test_index
    )

    predictor = make_lime_predictor(
        test_index
    )

    explanation = explainer.explain_instance(
        text,
        predictor,
        labels=(1,),
        num_features=LIME_NUM_FEATURES,
        num_samples=LIME_NUM_SAMPLES
    )

    all_explanations[
        case_name
    ] = explanation

    print(
        f"✓ {case_name} explanation complete"
    )


# ============================================================
# SAVE LIME RESULTS
# ============================================================

all_lime_rows = []

for case_name, explanation in all_explanations.items():

    explanation_list = (
        explanation.as_list(
            label=1
        )
    )

    for rank, (
        word,
        weight
    ) in enumerate(
        explanation_list,
        start=1
    ):

        all_lime_rows.append({

            "case":
                case_name,

            "test_index":
                selected_cases[
                    case_name
                ],

            "true_label":
                int(
                    labels_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "predicted_label":
                int(
                    preds_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "positive_probability":
                float(
                    probs_np[
                        selected_cases[
                            case_name
                        ]
                    ]
                ),

            "rank":
                rank,

            "word":
                word,

            "lime_weight":
                float(weight)

        })


lime_results_df = pd.DataFrame(
    all_lime_rows
)


lime_csv_path = os.path.join(
    LIME_DIR,
    "emoberta_raw_lime_results.csv"
)

lime_results_df.to_csv(
    lime_csv_path,
    index=False
)


# ============================================================
# DISPLAY + SAVE HTML
# ============================================================

print()
print("=" * 70)
print("LIME EXPLANATIONS")
print("=" * 70)

for case_name, explanation in all_explanations.items():

    print()
    print(
        "=" * 60
    )

    print(
        f"{case_name} — "
        f"LIME LOCAL EXPLANATION"
    )

    print(
        "=" * 60
    )

    display(
        HTML(
            explanation.as_html(
                labels=[1]
            )
        )
    )

    html_path = os.path.join(
        LIME_DIR,
        f"emoberta_raw_lime_{case_name}.html"
    )

    with open(
        html_path,
        "w",
        encoding="utf-8"
    ) as f:

        f.write(
            explanation.as_html(
                labels=[1]
            )
        )


# ============================================================
# PAPER-FRIENDLY LIME BAR PLOTS
# ============================================================

for case_name, explanation in all_explanations.items():

    explanation_list = (
        explanation.as_list(
            label=1
        )
    )

    plot_df = pd.DataFrame(
        explanation_list,
        columns=[
            "word",
            "weight"
        ]
    )

    plot_df = plot_df.sort_values(
        "weight"
    )

    plt.figure(
        figsize=(10, 7)
    )

    plt.barh(
        plot_df["word"],
        plot_df["weight"]
    )

    plt.axvline(
        0,
        linewidth=0.8
    )

    plt.xlabel(
        "LIME Weight",
        fontsize=12
    )

    plt.ylabel(
        "Token / Word",
        fontsize=12
    )

    plt.title(
        f"EmoBERTa-base + RAW — "
        f"LIME Explanation ({case_name})",
        fontsize=14
    )

    plt.tight_layout()

    plot_path = os.path.join(
        LIME_DIR,
        f"emoberta_raw_lime_{case_name}.png"
    )

    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# ============================================================
# SAVE CASE INFORMATION
# ============================================================

case_info_rows = []

for case_name, test_index in selected_cases.items():

    case_info_rows.append({

        "case":
            case_name,

        "test_index":
            int(test_index),

        "true_label":
            int(
                labels_np[
                    test_index
                ]
            ),

        "predicted_label":
            int(
                preds_np[
                    test_index
                ]
            ),

        "positive_probability":
            float(
                probs_np[
                    test_index
                ]
            ),

        "text_length_characters":
            len(
                get_test_text(
                    test_index
                )
            )

    })


case_info_df = pd.DataFrame(
    case_info_rows
)


case_info_df.to_csv(
    os.path.join(
        LIME_DIR,
        "emoberta_raw_lime_case_information.csv"
    ),
    index=False
)


# ============================================================
# SAVE CONFIGURATION
# ============================================================

lime_config = {

    "model":
        "EmoBERTa-base",

    "cleaning":
        "RAW",

    "dataset":
        "Held-out test set",

    "target":
        "Positive/depression class",

    "representative_cases":
        [
            "TP",
            "TN",
            "FP",
            "FN"
        ],

    "case_selection":
        "First matching test example in fixed test order",

    "random_seed":
        LIME_SEED,

    "num_samples":
        LIME_NUM_SAMPLES,

    "num_features":
        LIME_NUM_FEATURES,

    "batch_size":
        LIME_BATCH_SIZE,

    "metadata":
        "Fixed for each explained user",

    "text_perturbation":
        "LIME word-level perturbations",

    "model_training":
        "No additional training",

    "transformer_weights":
        "Frozen"
}


with open(
    os.path.join(
        LIME_DIR,
        "emoberta_raw_lime_config.json"
    ),
    "w"
) as f:

    json.dump(
        lime_config,
        f,
        indent=2
    )


# ============================================================
# COMPLETE
# ============================================================

print()
print("=" * 70)
print("LIME ANALYSIS COMPLETE")
print("=" * 70)

print(
    "✓ TP, TN, FP, FN cases explained"
)

print(
    "✓ Same trained EmoBERTa-base + RAW model used"
)

print(
    "✓ Metadata held fixed per user"
)

print(
    f"✓ {LIME_NUM_SAMPLES} LIME perturbations per case"
)

print(
    f"✓ Top {LIME_NUM_FEATURES} features extracted"
)

print(
    "✓ LIME results saved"
)

print(
    "✓ HTML explanations saved"
)

print(
    "✓ Paper-ready LIME plots saved and displayed"
)

print()
print(
    "LIME output directory:",
    LIME_DIR
)
```

```python
# ============================================================
# EXPLAINABILITY — CELL 2
# EMOBERTA-BASE + RAW
# GRAD-CAM: TOKEN-LEVEL LOCAL EXPLANATION
# ============================================================


import os
import json
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt


print("=" * 70)
print("GRAD-CAM EXPLAINABILITY")
print("EmoBERTa-base + RAW")
print("=" * 70)


GRADCAM_SEED = 42
GRADCAM_TOP_TOKENS = 20

np.random.seed(GRADCAM_SEED)
torch.manual_seed(GRADCAM_SEED)


EXPLAINABILITY_DIR = os.path.join(
    BASE_RESULT_DIR,
    "explainability"
)

GRADCAM_DIR = os.path.join(
    EXPLAINABILITY_DIR,
    "gradcam"
)

os.makedirs(
    GRADCAM_DIR,
    exist_ok=True
)

model_emoberta_raw.eval()


# ============================================================
# STEP 1 — CALCULATE TEST PREDICTIONS
# ============================================================

labels_np = []
preds_np = []
probs_np = []

with torch.no_grad():

    for batch in test_emoberta_raw_loader:

        input_ids = batch[
            "input_ids"
        ].to(device)

        attention_mask = batch[
            "attention_mask"
        ].to(device)

        metadata = batch[
            "metadata"
        ].to(device)

        output = model_emoberta_raw(
            input_ids=input_ids,
            attention_mask=attention_mask,
            metadata=metadata
        )

        logits = output["logits"]

        probabilities = torch.softmax(
            logits,
            dim=1
        )

        predictions = torch.argmax(
            logits,
            dim=1
        )

        labels_np.extend(
            batch["labels"]
            .cpu()
            .numpy()
            .tolist()
        )

        preds_np.extend(
            predictions
            .cpu()
            .numpy()
            .tolist()
        )

        probs_np.extend(
            probabilities[:, 1]
            .cpu()
            .numpy()
            .tolist()
        )


labels_np = np.array(labels_np)
preds_np = np.array(preds_np)
probs_np = np.array(probs_np)


print("=" * 70)
print("TEST PREDICTIONS READY")
print(f"Test samples : {len(labels_np)}")
print("=" * 70)


# ============================================================
# STEP 2 — SELECT TP / TN / FP / FN CASES
# ============================================================

selected_cases = {}

tp_candidates = np.where(
    (labels_np == 1) &
    (preds_np == 1)
)[0]

tn_candidates = np.where(
    (labels_np == 0) &
    (preds_np == 0)
)[0]

fp_candidates = np.where(
    (labels_np == 0) &
    (preds_np == 1)
)[0]

fn_candidates = np.where(
    (labels_np == 1) &
    (preds_np == 0)
)[0]


if len(tp_candidates) > 0:
    selected_cases["TP"] = int(
        tp_candidates[0]
    )

if len(tn_candidates) > 0:
    selected_cases["TN"] = int(
        tn_candidates[0]
    )

if len(fp_candidates) > 0:
    selected_cases["FP"] = int(
        fp_candidates[0]
    )

if len(fn_candidates) > 0:
    selected_cases["FN"] = int(
        fn_candidates[0]
    )


print("=" * 70)
print("SELECTED CASES")
print("=" * 70)

for case_type, test_index in selected_cases.items():

    print(
        f"{case_type} : test index {test_index} | "
        f"true={labels_np[test_index]} | "
        f"pred={preds_np[test_index]} | "
        f"probability={probs_np[test_index]:.4f}"
    )


# ============================================================
# STEP 3 — GRAD-CAM COMPUTATION
# ============================================================

def compute_emoberta_raw_gradcam(test_index):

    sample = test_emoberta_raw_dataset[
        test_index
    ]

    input_ids = (
        sample["input_ids"]
        .unsqueeze(0)
        .to(device)
    )

    attention_mask = (
        sample["attention_mask"]
        .unsqueeze(0)
        .to(device)
    )

    metadata = (
        sample["metadata"]
        .unsqueeze(0)
        .to(device)
    )


    # --------------------------------------------------------
    # EMOBERTA ENCODER
    # --------------------------------------------------------

    with torch.no_grad():

        emoberta_output = model_emoberta_raw.emoberta(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

    hidden_states = (
        emoberta_output
        .last_hidden_state
        .detach()
    )

    hidden_states.requires_grad_(True)
    hidden_states.retain_grad()


    # --------------------------------------------------------
    # ATTENTION POOLING
    # --------------------------------------------------------

    attention_features = (
        model_emoberta_raw.attention_projection(
            hidden_states
        )
    )

    attention_scores = (
        model_emoberta_raw.attention_score(
            torch.tanh(
                attention_features
            )
        )
        .squeeze(-1)
    )

    attention_scores = attention_scores.float()

    attention_scores = (
        attention_scores.masked_fill(
            attention_mask == 0,
            torch.finfo(
                attention_scores.dtype
            ).min
        )
    )

    attention_weights = torch.softmax(
        attention_scores,
        dim=1
    )

    text_representation = torch.sum(
        hidden_states *
        attention_weights.unsqueeze(-1),
        dim=1
    )


    # --------------------------------------------------------
    # METADATA BRANCH
    # --------------------------------------------------------

    metadata_representation = (
        model_emoberta_raw.metadata_branch(
            metadata
        )
    )

    metadata_gate = torch.sigmoid(
        model_emoberta_raw.metadata_beta
    )

    gated_metadata = (
        metadata_gate *
        metadata_representation
    )


    # --------------------------------------------------------
    # FUSION + CLASSIFIER
    # --------------------------------------------------------

    fused = torch.cat(
        [
            text_representation,
            gated_metadata
        ],
        dim=1
    )

    logits = model_emoberta_raw.classifier(
        fused
    )

    positive_logit = logits[:, 1]


    # --------------------------------------------------------
    # BACKWARD
    # --------------------------------------------------------

    model_emoberta_raw.zero_grad(
        set_to_none=True
    )

    positive_logit.backward()

    gradients = hidden_states.grad


    # --------------------------------------------------------
    # GRAD-CAM
    # --------------------------------------------------------

    channel_weights = gradients.mean(
        dim=1,
        keepdim=True
    )

    gradcam = (
        channel_weights *
        hidden_states
    ).sum(dim=-1)

    gradcam = torch.relu(
        gradcam
    )

    gradcam = gradcam.squeeze(0)

    attention_weights = (
        attention_weights
        .squeeze(0)
    )

    valid_mask = (
        attention_mask
        .squeeze(0)
    )

    gradcam = (
        gradcam
        .detach()
        .cpu()
        .numpy()
    )

    attention_weights = (
        attention_weights
        .detach()
        .cpu()
        .numpy()
    )

    valid_mask = (
        valid_mask
        .detach()
        .cpu()
        .numpy()
    )

    gradcam = (
        gradcam *
        valid_mask
    )

    if gradcam.max() > 0:

        gradcam = (
            gradcam /
            gradcam.max()
        )

    return {
        "gradcam": gradcam,
        "attention_weights": attention_weights,
        "input_ids": sample[
            "input_ids"
        ].cpu().numpy(),
        "attention_mask": sample[
            "attention_mask"
        ].cpu().numpy(),
        "true_label": int(
            labels_np[test_index]
        ),
        "predicted_label": int(
            preds_np[test_index]
        ),
        "positive_probability": float(
            probs_np[test_index]
        )
    }


# ============================================================
# STEP 4 — TOKEN INFORMATION
# ============================================================

def get_token_information(
    input_ids,
    attention_mask,
    gradcam,
    tokenizer,
    top_tokens=20
):

    token_ids = input_ids[
        attention_mask.astype(bool)
    ]

    importance = gradcam[
        attention_mask.astype(bool)
    ]

    tokens = tokenizer.convert_ids_to_tokens(
        token_ids.tolist()
    )

    token_information = []

    for token, score in zip(
        tokens,
        importance
    ):

        if token in tokenizer.all_special_tokens:
            continue

        token_information.append(
            {
                "token": token,
                "importance": float(score)
            }
        )


    token_information = sorted(
        token_information,
        key=lambda x: x["importance"],
        reverse=True
    )


    for rank, item in enumerate(
        token_information,
        start=1
    ):

        item["rank"] = rank


    return token_information[
        :top_tokens
    ]


# ============================================================
# STEP 5 — RUN TP / TN / FP / FN
# ============================================================

all_gradcam_results = []

for case_type, test_index in selected_cases.items():

    print("=" * 70)
    print(f"PROCESSING {case_type}")
    print(f"Test index : {test_index}")
    print("=" * 70)

    result = compute_emoberta_raw_gradcam(
        test_index
    )

    token_information = get_token_information(
        result["input_ids"],
        result["attention_mask"],
        result["gradcam"],
        emoberta_raw_tokenizer,
        GRADCAM_TOP_TOKENS
    )

    token_df = pd.DataFrame(
        token_information
    )


    token_csv_path = os.path.join(
        GRADCAM_DIR,
        f"emoberta_raw_gradcam_{case_type.lower()}_tokens.csv"
    )

    token_df.to_csv(
        token_csv_path,
        index=False
    )


    # --------------------------------------------------------
    # TOP TOKEN PLOT
    # --------------------------------------------------------

    plot_df = token_df.sort_values(
        "importance",
        ascending=True
    )

    plt.figure(
        figsize=(10, 8)
    )

    plt.barh(
        plot_df["token"],
        plot_df["importance"]
    )

    plt.xlabel(
        "Grad-CAM Importance"
    )

    plt.ylabel(
        "Token"
    )

    plt.title(
        f"EmoBERTa-base + RAW — "
        f"Grad-CAM — {case_type}"
    )

    plt.tight_layout()

    plot_path = os.path.join(
        GRADCAM_DIR,
        f"emoberta_raw_gradcam_{case_type.lower()}.png"
    )

    plt.savefig(
        plot_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


    all_gradcam_results.append(
        {
            "case_type": case_type,
            "test_index": test_index,
            "true_label": result[
                "true_label"
            ],
            "predicted_label": result[
                "predicted_label"
            ],
            "positive_probability": result[
                "positive_probability"
            ],
            "top_tokens": token_information
        }
    )

    print(
        f"Saved: {token_csv_path}"
    )


# ============================================================
# STEP 6 — SAVE CONFIGURATION
# ============================================================

gradcam_config = {

    "model":
        "tae898/emoberta-base",

    "architecture":
        "EmoBERTa-base + Metadata Branch",

    "cleaning":
        "RAW",

    "dataset":
        "Held-out test set",

    "method":
        "Transformer Grad-CAM",

    "target":
        "Positive class logit",

    "hidden_dimension":
        int(
            model_emoberta_raw.emoberta.config.hidden_size
        ),

    "cases":
        [
            "TP",
            "TN",
            "FP",
            "FN"
        ],

    "top_tokens_displayed":
        GRADCAM_TOP_TOKENS,

    "random_seed":
        GRADCAM_SEED,

    "model_training":
        "No additional training",

    "transformer_weights":
        "Frozen"
}


config_path = os.path.join(
    GRADCAM_DIR,
    "emoberta_raw_gradcam_config.json"
)

with open(
    config_path,
    "w"
) as f:

    json.dump(
        gradcam_config,
        f,
        indent=4
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("=" * 70)
print("GRAD-CAM COMPLETE")
print("=" * 70)

print(
    f"Output directory:\n{GRADCAM_DIR}"
)

print(
    f"Configuration saved:\n{config_path}"
)

print(
    "Cases processed:",
    list(selected_cases.keys())
)

print("=" * 70)
```
